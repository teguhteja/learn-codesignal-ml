Next, we need an `ARCHITECTURE.md`. While the `README` is for users, the `ARCHITECTURE` file is for developers. It explains the "how" behind your system. Here is how we might explain our data model:

```markdown
## Data Model
RecipeBox uses 8 entities with specific relationships:
- Recipe → RecipeIngredients ← Ingredient (A many-to-many relationship)
- MealPlan → MealPlanItems → Recipe (A plan refers to multiple recipes)

```

In this documentation, we explain that our shopping list isn't just a list of items; it’s an **aggregation**. If three recipes in your meal plan use eggs, the system must sum those eggs together into one single shopping list item. Writing this down helps you (and others) understand the logic you implemented.

## Capturing Decisions With ADRs

An Architectural Decision Record (ADR) is a short document that captures a specific choice you made during development. This is a standard practice in production environments because it prevents teams from having the same argument twice.

Let's look at how to build an ADR for our database choice. We start with the **Context** (why we needed to make a choice).

```markdown
# ADR-001: SQLite for Development Database

## Context
We need a database that is easy to set up for a learning project without requiring a full server installation like PostgreSQL.

```

Next, we state the **Decision** and the **Alternatives** we considered.

```markdown
## Decision
Use SQLite for local development, while keeping the code compatible with PostgreSQL for production.

## Alternatives
- Full PostgreSQL: Rejected because it adds setup complexity for students.
- MongoDB: Rejected because our data (Recipes, Ingredients) is highly relational.

```

Finally, we list the **Consequences**. This is the most important part because it shows you understand the trade-offs.

```markdown
## Consequences
- (+) Simple setup with no extra software.
- (-) Some advanced PostgreSQL search features aren't available.
- (Neutral) We use SQLAlchemy, so switching to PostgreSQL later reduces some migration effort; however, dialect differences (e.g. full-text search, constraints, locking), schema migration behavior, and feature parity still require explicit work and testing.

```

By following this pattern, you show that your technical choices are deliberate and calculated, not accidental.

## The Capstone Demo: Telling The Story

The final piece of your portfolio is the demo. You have built complex logic, like the Serving Scaling Algorithm, and you want to show it off. A good demo script follows a logical flow.

First, you show the basic creation (all endpoints require a JWT token and JSON):

```bash
# Obtain token first (login endpoint), then:
export TOKEN="your_jwt_token_here"

# Create a recipe for 4 people
curl -X POST http://localhost:8000/api/recipes \
  -H "Authorization: Bearer $TOKEN" \
  -H "Content-Type: application/json" \
  -d '{"name": "Pasta", "servings": 4}'

```

Then, you show the "Magic" (the business logic):

```bash
# Add that recipe to a meal plan, but for 8 people (servings scaled)
curl -X POST http://localhost:8000/api/meal-plans/1/meals \
  -H "Authorization: Bearer $TOKEN" \
  -H "Content-Type: application/json" \
  -d '{"recipe_id": 1, "date": "2025-03-17", "meal_type": "dinner", "servings": 8}'

```

Finally, you show the result (standard API response; then GET the meal plan to show scaled ingredients):

```text
Response (201 Created): 
{ 
  "id": 42, 
  "meal_plan_id": 1, 
  "recipe_id": 1, 
  "date": "2025-03-17", 
  "meal_type": "dinner", 
  "servings": 8 
}

To show scaling: GET /api/meal-plans/1 with Bearer token returns the plan; the recipe's ingredients are scaled by your backend when generating shopping lists or meal details.

```

By showing the inputs and the outputs, you demonstrate that your API isn't just a database; it’s a tool that solves real problems for users.

### 📋 1. User-Facing Documentation (`README.md`)

```markdown
# RecipeBox

RecipeBox is a production-ready recipe management application that helps home cooks organize their culinary life. Built with modern software development practices, it provides a complete solution for managing recipes, planning meals, tracking nutrition, and creating shopping lists.

## Features

### Recipe Management
Users can create, store, and organize their recipe collection with detailed information including ingredients, instructions, and cooking times. Recipes can be easily edited or deleted, giving users full control over their personal cookbook.

### Meal Planning
Home cooks can effortlessly organize their weekly cooking schedule using an interactive seven-day calendar grid. The interface allows users to assign selected recipes to specific days and explicit meal slots—such as Breakfast, Lunch, or Dinner—while enabling dynamic serving adjustments that automatically recalculate ingredient amounts across the entire menu plan.

### Shopping Lists
The application eliminates duplicate grocery items by automatically compiling a unified shopping list directly from a user's scheduled weekly meal plans. The underlying engine converts varying metric fractions, aggregates identical ingredients into a single consolidated total, and groups items by supermarket department blocks—like Produce, Meat, or Dairy—with interactive checkbox tracking to streamline trips.

### Nutrition Tracking
Users can easily monitor their dietary and health goals through automated caloric and macronutrient calculation readouts. The system aggregates nutritional profiles (calories, protein, carbohydrates, and fats) per serving for individual recipes and dynamically summarizes total daily intake levels across an entire weekly menu grid, highlighting balance thresholds in real time.

### Search
Finding the perfect meal is seamless with a full-text search capability that scans titles, instructions, and descriptions instantly. Users can narrow down their options by applying multi-dimensional query filters—such as maximum preparation time constraints, cooking duration limits, or serving size ranges—to discover recipes that fit their current schedule.

## Installation

### Prerequisites
* Python 3.11+
* PostgreSQL 15+
* Redis 7+

### Steps
1. Clone the repository and navigate to the root folder:
   ```bash
   git clone [https://github.com/yourorg/recipebox-api.git](https://github.com/yourorg/recipebox-api.git)
   cd recipebox-api

```

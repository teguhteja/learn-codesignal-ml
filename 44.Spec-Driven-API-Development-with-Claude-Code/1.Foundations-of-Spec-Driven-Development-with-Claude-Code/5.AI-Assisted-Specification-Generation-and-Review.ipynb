# AI-Assisted Specification Generation and Review

## Introduction: The Human-AI Partnership in SDD

Welcome to Unit 5! You have learned how to evaluate specifications in Units 1–4. Now you will learn the actual Spec-Driven Development (SDD) workflow: collaborating with AI to generate specifications.

### The Traditional Mistake

Many developers think SDD means *"write detailed specifications by hand before coding."* That is the old way—and it is why specification-driven approaches failed historically. They were simply too time-consuming.

### The SDD Reality with AI

In modern SDD with Claude Code:

1. **Humans** provide requirements and intent (structured prompts).
2. **AI** generates formal specifications using templates.
3. **Humans** review for quality and business correctness.
4. **AI** refines based on feedback.
5. **Humans** approve when the spec meets the quality threshold.

### Your Role in the Partnership

You are **not** a specification writer. You are a **specification architect**:

* Define **WHAT** you want (business intent).
* Review AI-generated specs critically.
* Catch domain knowledge the AI does not have.
* Guide refinement through targeted feedback.
* Approve when the quality threshold is met ($\ge 75/100$).

### AI's Role

Claude Code is your **specification generator**:

* Systematically covers all template sections.
* Considers edge cases you might forget.
* Maintains a consistent format.
* Generates complete technical plans.
* *But it always needs your domain expertise to validate.*

> 💡 **Core Principle: Division of Labor**
> Humans define **WHAT** and **WHY**. AI generates **HOW**.
> Humans excel at understanding business needs, domain constraints, and user value, while AI excels at systematic documentation, comprehensive coverage, and consistent formatting.

With AI assistance, specification creation becomes faster than jumping straight to code while producing significantly better architectural outcomes.

## The AI-Accelerated SDD Workflow

This workflow transforms specifications from a development bottleneck into an accelerator:

```text
[Human Prompt] ➔ [AI Generates Spec] ➔ [Human Critical Review] ➔ [AI Refines Spec] ➔ [Human Approval]

```

1. **Human provides structured requirements:** Outlines context, requirements, user perspective, and constraints.
2. **AI generates comprehensive specification:** Follows the template and covers all sections systematically.
3. **Human reviews critically:** Validates business logic, checks domain accuracy, and scores the spec across 5 quality dimensions.
4. **AI refines based on feedback:** Addresses specific issues and adds missing details.
5. **Human approves final spec:** Confirms a score of $\ge 75/100$ and verifies technical readiness.
6. **AI generates tests and implementation:** Creates test suites and production code matching the specification.
7. **Human validates against business needs:** Runs the generated tests and checks application behavior.

## The Specification Generation Prompt Pattern

When you want to add a feature to TaskMaster, start by giving Claude Code requirements using a structured prompt pattern.

### Essential Prompt Structure

```markdown
Generate functional specification for: [Feature Name]

CONTEXT:
- System: TaskMaster API (see CLAUDE.md for tech stack and standards)
- Existing features: [List related features]
- Current user workflow: [Describe current state]

ANALYZE EXISTING CODE:
- File: [specific file and line range]
  * Extract: [pattern to match, e.g., response format, error handling]

REQUIREMENTS:
- [Requirement 1 with specific values]
- [Requirement 2 with specific values]
- [Key constraint with specific values]

USER PERSPECTIVE:
- Who needs this: [User type]
- Problem it solves: [Pain point]
- Success looks like: [Measurable outcome]

FORMAT:
Follow specs/_templates/functional-spec-template.md

EXAMPLE REFERENCE:
See [specific high-quality spec] for quality benchmark

CONSTRAINTS:
- No implementation details (no database, no code)
- Specific values (not "reasonable limit", say "10 max")
- Examples must use real data formats (UUID4, ISO-8601 timestamps)

```

### Three Critical Sections for Quality:

* **`ANALYZE EXISTING CODE`** — Forces the AI to extract working design patterns directly from your codebase.
* **`EXAMPLE REFERENCE`** — Mandates that the AI match your existing specification quality and formatting style.
* **`CONSTRAINTS`** — Guards against implementation leaks (spoilers) and vague language.

## The 5 Dimensions of Specification Quality

Score specifications out of 100 across five rigid core dimensions:

### 1. Clarity (0–25 Points)

Terms must be defined precisely with zero ambiguity.

* ❌ *Vague:* "Valid format"
* ✅ *Clear:* "Regex: `^[a-z0-9-]{1,30}$`"

### 2. Completeness (0–25 Points)

All template sections must be present with substance. No `[TBD]` or "will be determined" placeholders are allowed.

### 3. Testability (0–20 Points)

All acceptance scenarios should use a clear **Given/When/Then** format. Concrete execution examples must use real data formats (such as UUID4 or ISO-8601 timestamps) instead of placeholders.

### 4. Consistency (0–20 Points)

The specification must perfectly match all `CLAUDE.md` conventions, including naming styles, error envelope layouts, and API response structures.

### 5. Appropriate Abstraction (0–10 Points)

The document must focus purely on **WHAT** and **WHY**, not HOW. It should contain no code syntax or low-level database details.

## Calibrating Your Scores

Use these strict rubrics to score consistently:

### Clarity (0–25)

* **20–25:** All terms defined precisely with formats/patterns. Zero ambiguous language.
* **15–19:** Most terms defined; 1–2 need precision adjustments (e.g., changing "reasonable limit" to "10 max").
* **10–14:** Multiple vague terms are present, but core concepts remain understandable.
* **0–9:** Severe ambiguity prevents understanding the operational requirements.

### Completeness (0–25)

* **20–25:** All template sections present with substantive content. No `[TBD]` markers.
* **15–19:** All sections present, but 1–2 are thin (e.g., a single sentence where a paragraph is needed).
* **10–14:** Missing 1 entire section OR multiple sections are empty stubs.
* **0–9:** Multiple sections are missing or completely empty.

### Testability (0–20)

* **16–20:** All scenarios structured in Given/When/Then blocks. Examples use real formats (UUID4, ISO-8601).
* **12–15:** Structured scenarios are present, but some examples use placeholders (like `"some-id"` or `"timestamp"`).
* **8–11:** Scenarios are present but lack concrete structure or realistic data.
* **0–7:** No structured scenarios exist; relies entirely on narrative descriptions.

### Consistency (0–20)

* **16–20:** Matches all project conventions (naming, error formats, response structures).
* **12–15:** Follows most conventions; 1–2 minor deviations.
* **8–11:** Multiple inconsistencies with existing API patterns.
* **0–7:** Introduces new patterns that directly contradict established project conventions.

### Appropriate Abstraction (0–10)

* **8–10:** Pure behavioral focus. Zero implementation or code execution details.
* **5–7:** Mostly behavioral, but mentions implementation details 1–2 times (e.g., "store this in the database table").
* **3–4:** Multiple technical implementation references are mixed with requirements.
* **0–2:** Reads like code documentation rather than a behavioral user specification.

> 🎯 **The 75-Point Threshold:** A specification scoring **above 75** is officially considered **implementation-ready**. Anything scoring **below 75** requires immediate refinement.

## Consolidating Multiple AI Outputs

Sometimes you will have more than one AI-generated specification for the same feature—perhaps you asked Claude Code to generate multiple approaches for comparison, or different team members prompted independent sessions. When specs diverge, use this systematic conflict-resolution rubric to build a single source of truth.

### The Conflict-Resolution Rubric

When specs disagree on the same requirement, resolve the conflict using this exact priority order:

1. **Constitution First:** If your project has a constitution or `CLAUDE.md` with an explicit rule, it always wins—regardless of how well-written the conflicting spec is.
* *Example:* `CLAUDE.md` mandates ISO-8601 timestamps. A specification proposing Unix epoch timestamps loses immediately.


2. **Existing Code Conventions:** When the constitution is silent, defer to how the current codebase already handles similar cases. Consistency reduces integration risk.
* *Example:* Existing endpoints return errors as `{"detail": "message"}`. A spec proposing `{"success": false, "error_text": "message"}` loses to maintain system consistency.


3. **Better UX / Testability:** When neither the constitution nor the existing code dictates an answer, prefer the version with the stronger user experience or more testable acceptance criteria (concrete Given/When/Then scenarios with real data formats).
* *Example:* One spec limits pagination to "a reasonable number," while another specifies "max 50 items, default 20." The second wins on testability.


4. **Merge Complementary Content:** Not every difference is a true conflict. Specs often cover different edge cases or add complementary details—merge these sections instead of discarding a spec wholesale.
* *Example:* Spec A documents rate-limiting behaviors; Spec B documents concurrent-edit conflicts. Combine both into the unified document.



```text
CONSOLIDATION FLOW:
[Identify Diffs] ➔ Apply: Constitution ➔ Code Conventions ➔ UX/Testability ➔ Merge Content ➔ [Re-score Spec]

```

> 📝 **Pro Tip:** Record a one-line rationale for each conflict resolution. This helps future reviewers understand your engineering reasoning and prevents arbitrary decisions.

**AI Tool B might generate:**

```python
# Binary interpretation with 413 Payload Too Large error
MAX_BYTES = 5 * 1024 * 1024  # 5,242,880 bytes
if file.size > MAX_BYTES:
    raise HTTPException(status_code=413, detail="File size exceeds maximum threshold")

```

**AI Tool C might generate:**

```python
# Content-Length header approximation with 422 error
if content_length > 5000000:
    return JSONResponse(status_code=422, content={"error": "Upload error"})

```

**Result**: These differences create architectural drift and inconsistent validation. A file that is exactly 5,100,000 bytes will pass successfully on AI Tool B's endpoint but crash with a generic error on AI Tool A's endpoint. Frontend applications cannot predict the error format or the exact HTTP status code.

### Required Specification

```markdown
### File Size Constraint Validation
- **Hard Upper Limit:** Max allowable file size is strictly bounded at 5,242,880 bytes (5 MiB calculated as binary $5 \times 1,048,576$ bytes).
- **Error Behavior:** If an uploaded file stream exceeds this boundary, the endpoint must abort processing and immediately return an HTTP `413 Payload Too Large` status code with this standardized JSON body contract:
```json
{
  "detail": "File size exceeds the maximum limit of 5,242,880 bytes."
}

```

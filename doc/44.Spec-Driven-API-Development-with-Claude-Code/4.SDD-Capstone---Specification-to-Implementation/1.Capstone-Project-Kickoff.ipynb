## Recall: Spec-Driven Development (SDD) and Orchestration

Before we dive into the code, let’s quickly remember the workflow we have been practicing. In SDD, we never start by writing code. Instead, we follow a strict sequence:

1. **Requirements:** Define what we want.
2. **Specification:** Detail exactly how it should work.
3. **Implementation:** Use AI agents to write the code based on the spec.

In our previous lessons, you learned about **Orchestration**. This is the process where you act as the manager. You don't just ask an AI to "build the whole app." You break the project into small, atomic tasks and use a task-executor agent to finish them one by one. This ensures the AI stays focused and doesn't get "confused" by too much information at once.

## Phase 1: From Informal Requirements to a PRD and Domain Model

Every great project starts with an idea. We begin by creating a file called `informal-requirements.txt`. It contains a few paragraphs describing the app in plain English. For RecipeBox, it might look like this:

```text
Users need to manage recipes and plan meals for the week. Recipes have ingredients with amounts and units. Meal plans assign recipes to days. Shopping lists should be generated automatically from meal plans.

```

Next, we use an AI to turn this into a **Product Requirements Document (PRD)**. This document is a formal version of your idea. It includes "Success Metrics" (how we know the app works) and "Constraints" (rules we must follow, like using PostgreSQL).

The most important part of this phase is the **Domain Model**. This is a map of how your data fits together. Let's look at how we build the relationship between Recipes and Ingredients.

> ⚠️ **Note:** The snippets below are conceptual pseudo-code to illustrate entities and relationships. They are not runnable Python (e.g., `uuid` and `decimal` are placeholders for "unique identifier" and "precise number"); your actual implementation will use proper types (e.g., SQLAlchemy columns or spec field types).

First, we define a `Recipe`:

```python
# A Recipe has a name and basic info 
# (conceptual pseudo-code; id might be UUID or int in implementation)
class Recipe:
    id: uuid
    name: str
    description: str

```

Next, we need `Ingredients`. But a recipe doesn't just "have" an ingredient; it needs a specific amount of that ingredient:

```python
# An Ingredient is a general item like "Flour"
class Ingredient:
    id: uuid
    name: str
    category: str  # e.g., "Baking"

```

Finally, we connect them using a "join table" called `RecipeIngredient`. This stores the connection and the specific amount:

```python
# This connects a specific Recipe to a specific Ingredient
class RecipeIngredient:
    recipe_id: uuid
    ingredient_id: uuid
    amount: decimal
    unit: str  # e.g., "cups"

```

By defining these relationships early, we ensure our database is organized and efficient.

## Phase 2: The Functional Specification (Aiming for 85/100)

Now that we have a PRD and a data map, we need a **Functional Specification**. Think of this as the "Instruction Manual" for the AI. It lists every API endpoint (like `POST /recipes`) and the exact rules for calculation.

For RecipeBox, we have complex rules like **Serving Scaling**. If a recipe is for 4 people, but you want to cook for 2, the app must divide every ingredient amount by 2.

To make sure our spec is high quality, we grade it on five dimensions:

* **Clarity:** Is every field type exact (e.g., `decimal(10,2)`)?
* **Completeness:** Are all 7 data models and 15+ endpoints listed?
* **Testability:** Are there "Given/When/Then" examples?
* **Consistency:** Do all endpoints use the same naming style?
* **Appropriate Abstraction:** Does it describe *what* the system does, not *how* to write the code?

In the capstone, we aim for a score of **85/100**. If your spec is vague, you must refine it. For example, instead of saying "calculate nutrition," you should specify: *"Sum the calories of all ingredients, multiplied by the serving factor."*

## Phase 3: The Technical Blueprint and CLAUDE.md

With the "what" decided, we move to the "how." We create a Technical Plan that defines our architecture. For RecipeBox, we use the **Repository Pattern**. This means our API doesn't talk to the database directly. Instead, it talks to a "Repository" file that handles the data access layer. This keeps our code clean and easy to test.

The most critical file you will create is `CLAUDE.md`. This is your project’s Constitution. It contains the guidelines that every AI agent must follow.

```markdown
# CLAUDE.md Guidelines

- **Tech Stack:** FastAPI, SQLAlchemy, PostgreSQL.
- **Code Style:** Use type hints and Google-style docstrings.
- **Architecture:** All database access must go through the /repositories directory.
- **Testing:** Every task must have 90%+ test coverage.

```

When an agent starts a task, it reads this file first. This ensures that even if different agents work on different parts of the app, the code remains highly uniform.

## Phase 4: Specialized Agents and Task Decomposition

In a large project, one AI isn't enough. We create a workforce of specialized agents in the `.claude/agents/` directory, each with a specific job:

* **task-executor:** The builder who writes the core production code.
* **test-enhancer:** The auditor who adds extra tests to find edge-case bugs.
* **recipe-validator:** The domain expert who checks if the math for servings and units is correct.
* **doc-updater:** The writer who keeps the system documentation current.

Finally, we break the project into **23 atomic tasks**. An atomic task is a small piece of work that takes 30–90 minutes and affects 3 or fewer files.

We also identify **Parallel Workflows**. For example, once the "Foundation" (database setup) is complete, one agent can work on the Recipe API while another agent works on the Meal Planning system at the same time. This is how professional teams build software quickly.

### 📋 2. Comprehensive Requirements Foundation (`docs/prds/recipebox-v1.0.md`)

```markdown
# RecipeBox Product Requirements Document

**Version:** 1.0  
**Date:** July 14, 2026  
**Status:** Approved  

---

## Problem Statement
Modern home cooks face extreme friction when managing their kitchens. Recipes are scattered across disconnected websites, paper books, and social media media streams, leading to inefficient meal planning. Planning dinner involves manually checking ingredients, calculating serving scales, and manually writing out shopping lists, which often results in grocery waste or forgotten items. 

Furthermore, tracking nutritional goals (like macros and daily calories) is tedious, requiring manual input into separate applications. RecipeBox solves this by providing an all-in-one digital platform that unifies recipe management, dynamic meal scheduling, smart grocery list aggregation, and automated third-party nutritional analysis into a secure, multi-tenant workspace.

---

## Target Users

### Primary Persona: The Busy Meal Planner
* **Profile:** Working professionals or parents managing busy weekly schedules.
* **Needs:** A streamlined way to store favorite meals, organize weekly menus in under 10 minutes, and auto-generate clean, categorized grocery lists that reduce shopping time.
* **Friction Points:** Frustrated by manual ingredient scaling and separate nutrition tracking apps.

### Secondary Persona: The Macro-Conscious Cook
* **Profile:** Individuals tracking strict fitness, weight, or health goals.
* **Needs:** Transparent, real-time insight into the caloric and macronutrient values of their home-cooked meals without manual calculations.
* **Friction Points:** Distrusts generic entry approximations and requires precise per-serving data.

---

## Core Requirements

### 1. Recipe Management
* **Full CRUD Operations:** Authenticated users can create, read, update, and delete recipes.
* **Structured Parameters:** Each recipe must capture a distinct title, optional description, preparation time, cooking time, an array of structured instruction steps, and multiple category tags.
* **Relational Ingredient Mapping:** Recipes must bind explicitly to an ingredient master index table, storing the precise text name, decimal volume amount, and standard unit token (e.g., grams, pieces, cups, tablespoons).
* **Dynamic Serving Scaling:** The user interface and underlying API must calculate and scale all ingredient volumes instantly when a user adjusts the base serving factor (e.g., scaling a 4-serving recipe down to 2).

### 2. Meal Planning
* **Weekly Grid Allocation:** Users can build a weekly plan by assigning recipes to a specific calendar date and meal tier category (`Breakfast`, `Lunch`, `Dinner`, `Snack`).
* **Interactive Operations:** Users can view their weekly overview, swap recipes out, add additional items, or delete entries smoothly.
* **Ownership Boundary Guarding:** Users are strictly blocked from viewing, modifying, or copying meal plans belonging to other user IDs.

### 3. Shopping List Generation
* **Automated Compilation:** The engine must inspect a defined date range within a meal plan, extract all required recipe ingredients, and compile them into a unified shopping list.
* **Smart Unit Conversion & Aggregation:** The engine must automatically normalize and convert compatible fractions (e.g., adding `0.5 cups` of milk and `2 tablespoons` of milk into a single, standardized total decimal measurement).
* **Grocery Category Grouping:** Line items must group automatically by target department tags (e.g., `Produce`, `Baking`, `Meat`, `Dairy`) to streamline the shopping experience.
* **Interactive Check-off Matrix:** Users can toggle an item's status to track what has already been collected, with checked items persisting through session changes.

### 4. Nutrition Tracking
* **Automated Macro Calculations:** Every recipe entry must automatically display calculated calorie metrics, protein, carbohydrates, and fats.
* **External Integration Interface:** The service layer will securely connect to the Edamam Nutrition Analysis API via asynchronous requests to fetch ingredient details.
* **Meal Plan Aggregations:** The system must sum daily caloric and macronutrient values across an active meal plan, displaying clear warning highlights if a plan exceeds or falls short of user-defined healthy limits.
* **Caching Optimization:** To minimize third-party API costs and latency, calculated ingredient profiles must be cached in Redis with a 30-day expiration window.

### 5. Recipe Search & Discovery
* **Full-Text Search Engine:** Users can query their recipe collection by typing matching letters or full words found in titles or descriptions.
* **Multi-Dimensional Filtering:** Supports filtering results by preparation time caps, nutritional limits, or exact tag selections (e.g., `Gluten-Free`).
* **Sequential Sorting:** Users can sort results chronologically by creation date, alphabetized title, or total preparation time.

---

## Technical Constraints

### Technology Stack
* **Web Framework:** FastAPI (Python 3.11+) utilizing asynchronous route controllers.
* **Database Engine:** PostgreSQL 15+ for robust relational data persistence.
* **Object-Relational Mapper (ORM):** SQLAlchemy 2.0 utilizing Declarative Mapping conventions.
* **Cache & Message Broker:** Redis 7.0 for handling service data caching and background task states.
* **Database Migration Driver:** Alembic for structured, linear schema tracking.
* **Security & Auth:** Stateless JWT (JSON Web Tokens) signing using the `HS256` symmetric algorithm.

### Data Requirements
* **Decimal Precision:** All ingredient amounts must use the `NUMERIC(10, 2)` decimal data type to avoid floating-point rounding errors in conversions.
* **Time-Zone Standardization:** All timestamp data fields must enforce ISO 8601 formatting and normalize to UTC before storage.
* **Third-Party API Fail-safes:** Network timeouts for Edamam API connections are capped at 4.0 seconds. If a connection fails, the system must fall back to cached values or log a clean error rather than crashing.

### Architecture Patterns
* **Decoupled Repository Pattern:** Route controllers are strictly blocked from executing inline SQL or managing direct ORM queries. All database access must be handled through specialized classes in the `/repositories` layer.
* **Layered Business Services:** All calculations, conversions, and API integrations must live within an independent `/services` layer.
* **Dependency Injection (DI):** Utilize FastAPI's `Depends` engine to securely inject database sessions, repositories, and authentication contexts.

---

## Success Metrics

### User Engagement
* **Planning Efficiency Target:** Primary users must be able to generate a complete 7-day meal plan and its matching categorized shopping list in under **3.0 minutes**.
* **Adoption Goal:** Achieve a milestone of $\ge 70\%$ of active beta testers successfully generating at least two consecutive weekly meal plans within their first 14 days of onboarding.

### Technical Performance
* **API Route Response Time:** Core database read routes (`GET /api/recipes`) must maintain a p95 latency under **200ms** under normal load conditions.
* **List Aggregation Performance:** Complex list compilation operations across a full 7-day calendar view must complete execution under **400ms**.
* **Cache Hit Optimization:** Maintain an operational Redis database cache hit ratio $\ge 85\%$ for all common nutritional ingredient queries after 30 days of standard system use.

### Quality
* **Test Suite Line Coverage:** Enforce a strict minimum **95% statement coverage floor** across all core repository, service, and routing modules.
* **Vulnerability Target:** Maintain **zero unmitigated HIGH or CRITICAL security flaws** (such as BOLA or injection vulnerabilities) within production environments.
* **Operational Uptime:** Maintain a stable target availability metrics of **99.9%** monthly uptime (excluding planned maintenance windows).

---

## Out of Scope (Version 1.0)
* **Social Sharing Interactions:** Collaborative group meal plans, public link sharing, and community recipe review walls.
* **Automated Grocery Integration:** Direct cart syncing or order placement with third-party delivery APIs (such as Instacart or Walmart).
* **Smart Device Synchronizations:** Voice assistant compatibility features (e.g., Alexa Skill support) or native smart-refrigerator screen integrations.

---

## Domain Model Overview



The data layer structure is built around a secure multi-tenant core, ensuring all data links back to an explicit user entity:
* A `User` maintains a one-to-many relationship with `Recipe` and `MealPlan` objects.
* A `Recipe` holds a many-to-many relationship with `Ingredient` records, managed via the `RecipeIngredient` junction association table. This table stores the specific amount and unit text for that recipe.
* A `MealPlan` contains a one-to-many relationship with `MealPlanItem` entries, where each item links back to a target `Recipe` and specifies a target day and meal type.
* A `ShoppingList` is derived dynamically from a `MealPlan`, orchestrating a one-to-many relationship with `ShoppingListItem` data rows grouped by grocery category.

---

## Approval
* **Product Management:** Signed Off (2026-07-14)
* **Lead Engineering Architect:** Signed Off (2026-07-14)
* **Security & Compliance Lead:** Signed Off (2026-07-14)

```

```markdown
# RecipeBox Domain Model

## Entity Definitions

### User
**Purpose:** Represents an authenticated user account within the multi-tenant system, serving as the root tenant isolation anchor for recipes and meal plans.

**Fields:**
* `id`: `UUID` (Primary Key, Generated Uuid4)
* `email`: `VARCHAR(255)` (Unique, Indexed, Case-Insensitive, Not Null)
* `password_hash`: `VARCHAR(255)` (Not Null)
* `name`: `VARCHAR(100)` (Not Null)
* `created_at`: `TIMESTAMP WITH TIME ZONE` (Not Null, Default: `CURRENT_TIMESTAMP`)

**Relationships:**
* `recipes`: One-to-Many relationship mapping to the `Recipe` model (`cascade="all, delete-orphan"`).
* `meal_plans`: One-to-Many relationship mapping to the `MealPlan` model (`cascade="all, delete-orphan"`).

---

### Recipe
**Purpose:** Represents a structured cooking recipe managed and owned by a specific user account.

**Fields:**
* `id`: `UUID` (Primary Key, Generated Uuid4)
* `user_id`: `UUID` (Foreign Key targeting `users.id`, Indexed, Not Null)
* `name`: `VARCHAR(200)` (Not Null)
* `description`: `VARCHAR(1000)` (Nullable)
* `instructions`: `TEXT` (Not Null)
* `prep_time_minutes`: `INTEGER` (Not Null)
* `cook_time_minutes`: `INTEGER` (Not Null)
* `servings`: `INTEGER` (Not Null, Default: 4)
* `tags`: `VARCHAR(50)[]` (Array field of string tags, Nullable)
* `created_at`: `TIMESTAMP WITH TIME ZONE` (Not Null, Default: `CURRENT_TIMESTAMP`)
* `updated_at`: `TIMESTAMP WITH TIME ZONE` (Not Null, Default: `CURRENT_TIMESTAMP`, OnUpdate: `CURRENT_TIMESTAMP`)

**Constraints:**
* `name`: Character length minimum 1, maximum 200.
* `servings`: Integer value range strictly between 1 and 50 inclusive.
* `prep_time_minutes`: Integer value range $\ge 0$ and $\le 1440$.
* `cook_time_minutes`: Integer value range $\ge 0$ and $\le 1440$.

**Relationships:**
* `user`: Many-to-One mapping linking back to the parent `User` record.
* `recipe_ingredients`: One-to-Many relationship mapping to `RecipeIngredient` (`cascade="all, delete-orphan"`).

---

### Ingredient
**Purpose:** A global master index catalog storing unique individual raw culinary items shared across all tenants.

**Fields:**
* `id`: `UUID` (Primary Key, Generated Uuid4)
* `name`: `VARCHAR(200)` (Unique, Indexed, Not Null)
* `category`: `VARCHAR(50)` (Not Null)
  - *Valid Categories:* `Produce`, `Dairy`, `Meat`, `Baking`, `Pantry`, `Frozen`, `Beverages`
* `calories_per_100g`: `NUMERIC(10, 2)` (Not Null, Default: 0.00)
* `protein_per_100g`: `NUMERIC(10, 2)` (Not Null, Default: 0.00)
* `carbs_per_100g`: `NUMERIC(10, 2)` (Not Null, Default: 0.00)
* `fat_per_100g`: `NUMERIC(10, 2)` (Not Null, Default: 0.00)

**Constraints:**
* `name`: Stripped, case-insensitive string matching.
* Nutritional fields (`calories`, `protein`, `carbs`, `fat`): `NUMERIC(10, 2)` type $\ge 0.00$.

**Relationships:**
* `recipe_ingredients`: One-to-Many reference tracking usage across various recipes.

---

### RecipeIngredient
**Purpose:** A contextual junction table mapping ingredients into recipes, detailing precise quantitative volume measures.

**Fields:**
* `id`: `UUID` (Primary Key, Generated Uuid4)
* `recipe_id`: `UUID` (Foreign Key targeting `recipes.id`, Not Null)
* `ingredient_id`: `UUID` (Foreign Key targeting `ingredients.id`, Not Null)
* `amount`: `NUMERIC(10, 2)` (Not Null)
* `unit`: `VARCHAR(20)` (Not Null)
  - *Valid Units:* `cup`, `tablespoon`, `teaspoon`, `gram`, `ounce`, `piece`

**Constraints:**
* `amount`: Numeric value strictly $> 0.00$ and $\le 10000.00$.
* `UniqueConstraint`: Combined allocation pairing (`recipe_id`, `ingredient_id`) must be unique.

**Relationships:**
* `recipe`: Many-to-One cross-reference link mapping to the parent `Recipe`.
* `ingredient`: Many-to-One mapping linking to the targeted global master `Ingredient`.

---

### MealPlan
**Purpose:** A time-bound container managing a user's scheduled recipe allocations across a specific calendar week.

**Fields:**
* `id`: `UUID` (Primary Key, Generated Uuid4)
* `user_id`: `UUID` (Foreign Key targeting `users.id`, Indexed, Not Null)
* `week_start_date`: `DATE` (Not Null)
* `created_at`: `TIMESTAMP WITH TIME ZONE` (Not Null, Default: `CURRENT_TIMESTAMP`)

**Constraints:**
* `week_start_date`: Date must explicitly land on a Monday.
* `UniqueConstraint`: Dual combination pairing (`user_id`, `week_start_date`) must be unique per user.

**Relationships:**
* `user`: Many-to-One connection mapping back to the owning `User`.
* `items`: One-to-Many mapping linking to `MealPlanItem` entries (`cascade="all, delete-orphan"`).
* `shopping_list`: One-to-One relationship mapping to the derived `ShoppingList` (`cascade="all, delete-orphan"`).

---

### MealPlanItem
**Purpose:** Maps a specific recipe execution instance onto a single day slot within a broader parent weekly meal plan container.

**Fields:**
* `id`: `UUID` (Primary Key, Generated Uuid4)
* `meal_plan_id`: `UUID` (Foreign Key targeting `meal_plans.id`, Not Null)
* `recipe_id`: `UUID` (Foreign Key targeting `recipes.id`, Not Null)
* `date`: `DATE` (Not Null)
* `meal_type`: `VARCHAR(20)` (Not Null)
  - *Valid Meal Types:* `Breakfast`, `Lunch`, `Dinner`, `Snack`
* `target_servings`: `INTEGER` (Not Null)

**Constraints:**
* `target_servings`: Integer value range strictly between 1 and 100 inclusive.
* `date`: Must fall within the boundary range: $[\text{week\_start\_date}, \text{week\_start\_date} + 6\text{ days}]$.

**Relationships:**
* `meal_plan`: Many-to-One relationship linking back to the parent `MealPlan`.
* `recipe`: Many-to-One reference pointing to the scheduled target `Recipe`.

---

### ShoppingList
**Purpose:** The central container managing aggregated shopping requirements extracted dynamically from a user's weekly meal plan grid.

**Fields:**
* `id`: `UUID` (Primary Key, Generated Uuid4)
* `meal_plan_id`: `UUID` (Foreign Key targeting `meal_plans.id`, Unique, Not Null)
* `updated_at`: `TIMESTAMP WITH TIME ZONE` (Not Null, Default: `CURRENT_TIMESTAMP`)

**Relationships:**
* `meal_plan`: Many-to-One relationship mapping back onto the generating parent `MealPlan`.
* `items`: One-to-Many relationship linking to `ShoppingListItem` lines (`cascade="all, delete-orphan"`).

---

### ShoppingListItem
**Purpose:** Represents an individual aggregated ingredient item within a shopping list, consolidated across all scheduled recipes.

**Fields:**
* `id`: `UUID` (Primary Key, Generated Uuid4)
* `shopping_list_id`: `UUID` (Foreign Key targeting `shopping_lists.id`, Not Null)
* `ingredient_name`: `VARCHAR(200)` (Not Null)
* `category`: `VARCHAR(50)` (Not Null)
* `total_amount`: `NUMERIC(10, 2)` (Not Null)
* `unit`: `VARCHAR(20)` (Not Null)
* `checked`: `BOOLEAN` (Not Null, Default: `FALSE`)

**Constraints:**
* `total_amount`: Numeric metric value range strictly $> 0.00$.
* `UniqueConstraint`: Multi-column matching pairing (`shopping_list_id`, `ingredient_name`, `unit`) must be unique.

**Relationships:**
* `shopping_list`: Many-to-One relationship linking back to the parent `ShoppingList` container.

---

## Key Business Logic Calculations

### 1. Serving Scaling Formula
To scale ingredient amounts for a specific meal plan item, calculate the `Scaling Factor` ($SF$) and apply it to the base ingredient amount:
$$SF = \frac{\text{MealPlanItem.target\_servings}}{\text{Recipe.servings}}$$
$$\text{Scaled Amount} = \text{RecipeIngredient.amount} \times SF$$

### 2. 6-Step Shopping List Aggregation Algorithm
1. **Initialize Map:** Create an empty hash map structured as `(ingredient_name, unit) -> {amount: Decimal, category: String}`.
2. **Collect Items:** Query all `MealPlanItem` entries within the designated time range for the active `MealPlan`.
3. **Fetch Relational Data:** For each item, fetch its linked `Recipe` and all corresponding `RecipeIngredient` rows.
4. **Scale Quantities:** Calculate the `Scaled Amount` using the Serving Scaling Formula for each ingredient.
5. **Convert & Accumulate:** Check if the `(ingredient_name, unit)` key exists in the map. If it exists, add the scaled amount to the running total. If the units are compatible but different (e.g., teaspoons and tablespoons), normalize them to the primary unit first before adding.
6. **Persist Records:** Iterate through the final map and batch-insert rows into the `ShoppingListItem` table, grouping items by category.

### 3. Unit Conversion & Equivalence Rules
The system handles unit conversions using the following exact multipliers:
* $1\text{ cup} = 16\text{ tablespoons}$
* $1\text{ tablespoon} = 3\text{ teaspoons}$
* $1\text{ ounce} = 28.35\text{ grams}$

> ⚠️ **Note:** Incompatible units (e.g., `gram` and `cup`) cannot be combined automatically. They will be listed as separate lines in the final shopping list to ensure accuracy.

### 4. Nutrition Calculation Logic
To calculate the nutritional values for a recipe, sum the macros of its component ingredients based on their scaled amounts:
$$\text{Recipe Calories} = \sum \left( \frac{\text{Scaled Ingredient Weight in Grams}}{100} \times \text{Ingredient.calories\_per\_100g} \right)$$

---

## Database Indexes Blueprint
To maintain low latency under concurrent user load, the following explicit database indexes are implemented:

```sql
-- Multi-Tenant Tenant Isolation Optimizations
CREATE INDEX idx_recipes_user_id ON recipes(user_id);
CREATE INDEX idx_meal_plans_user_id ON meal_plans(user_id);

-- Eager Loading & Join Path Query Optimization
CREATE INDEX idx_recipe_ingredients_recipe_id ON recipe_ingredients(recipe_id);
CREATE INDEX idx_meal_plan_items_meal_plan_id ON meal_plan_items(meal_plan_id);
CREATE INDEX idx_shopping_list_items_list_id ON shopping_list_items(shopping_list_id);

-- Global Master Index Lookups
CREATE INDEX idx_ingredients_name_lower ON ingredients(LOWER(name));

```

```
# CLAUDE.md
# RecipeBox Project Constitution

## Tech Stack

- **Python 3.11+** - Modern Python with improved type hints and performance
- **FastAPI 0.100+** - Async support and automatic OpenAPI generation
- **PostgreSQL 15+** - Reliable relational database for recipe data
- **SQLAlchemy 2.0** - Modern ORM with improved async support
- **Redis 7+** - Background job queue for nutrition API calls (simpler than Celery)
- **pytest** - Testing framework with ≥90% coverage requirement

## Architectural Principles

### Repository Pattern
TODO: Describe how database access should be organized

### Service Layer
TODO: Describe where business logic should live

### Background Jobs
TODO: Describe how to handle external API calls

### API-First Design
TODO: Describe the role of OpenAPI specification

## Code Standards

### Type Hints
TODO: Explain type hint requirements and provide example

### Docstrings
TODO: Specify docstring style and provide example

### Error Handling
TODO: Describe error handling approach with HTTPException

### Formatting
TODO: List formatting tools to use

## Agent Orchestration Rules

### Task Delegation
TODO: Explain how main session delegates to agents

### Checkpoint Strategy
TODO: Describe when reviews happen

### Parallel Workflows
TODO: Identify which phases can run in parallel

## Development Workflow

TODO: List the 5-step development workflow

## Testing Requirements

TODO: Specify coverage targets and testing approach

## Error Handling Strategy

TODO: Describe how service layer handles repository exceptions

## Migration Strategy

TODO: Describe Alembic migration approach

## Caching Strategy

TODO: Describe Redis caching for nutrition API


# technical-plan.md
# RecipeBox Technical Plan

## Stack

- Python 3.11+
- FastAPI 0.100+
- PostgreSQL 15+
- SQLAlchemy 2.0
- Redis 7+ (background jobs)
- pytest with ≥90% coverage

## Architecture

### Repository Pattern
All database access goes through repository classes (e.g. RecipeRepository, MealPlanRepository). Routes and services call repository methods only; no raw session.query() in API or service layer. Repositories return domain objects or raise domain exceptions.

### Dependency Injection
FastAPI Depends() injects get_db() and repository instances into route handlers. Services receive repositories via constructor injection. Single place to swap implementations (e.g. test doubles).

### API-First
OpenAPI spec (or Pydantic schemas) define request/response shapes and status codes before implementation. Routes validate with Pydantic; services enforce business rules; repositories handle persistence.

### Background Jobs
Redis queue for nutrition API calls. NutritionFetchWorker pops jobs, calls USDA (or mock), caches result, updates DB. Sync/mocked implementation acceptable for local practice.

## Data Models (SQLAlchemy)

User (id, email, hashed_password, username, created_at, updated_at). Recipe (id, user_id FK, name, description, prep_time, cook_time, servings, instructions, tags, created_at, updated_at). Ingredient (id, name, category enum). RecipeIngredient (id, recipe_id FK, ingredient_id FK, amount Numeric(10,2), unit enum). MealPlan (id, user_id FK, week_start_date, name, created_at, updated_at). MealPlanItem (id, meal_plan_id FK, recipe_id FK, date, meal_type enum, servings). ShoppingList (id, meal_plan_id FK unique, generated_at). ShoppingListItem (id, shopping_list_id FK, ingredient_id FK, total_amount Numeric(10,2), unit, checked, category). Relationships and indexes per specification.

## Repositories

RecipeRepository: create, get_by_id, get_all_by_user, update, delete. IngredientRepository: create, get_by_id, get_by_name, list_all. MealPlanRepository: create, get_by_id, get_by_user_week, update, delete. ShoppingListRepository: create, get_by_meal_plan_id, add_item, update_item. All with appropriate error handling (e.g. not found).

## Services

RecipeService: CRUD via repository; scale_ingredients(recipe_id, new_servings) multiplies amounts by factor. MealPlanService: CRUD; date validation (0-30 days ahead); add_meal_plan_item, remove_meal_plan_item. ShoppingListService: generate_from_meal_plan (6-step aggregation: collect items, scale, group by ingredient, convert units, sum, create ShoppingListItems). NutritionService: fetch per ingredient (USDA/mock), cache, aggregate per recipe/meal plan.

## API Routes

Recipe: POST/GET/PATCH/DELETE /api/recipes, GET /api/recipes (list). Meal plan: POST/GET/PATCH/DELETE /api/meal-plans, POST/GET/DELETE /api/meal-plans/{id}/meals. Shopping list: POST /api/meal-plans/{id}/shopping-list, GET /api/shopping-lists/{id}. Search: GET /api/search?q=... Status codes: 201 created, 200 ok, 204 delete, 404 not found, 422 validation.

## Background Worker

NutritionFetchWorker: Poll Redis queue for (ingredient_id or recipe_id); call nutrition API; store in NutritionIngredient or cache; TTL 7 days.

## Testing Strategy

Unit tests: pytest for each repository and service; mock DB or use in-memory SQLite. Integration tests: FastAPI TestClient against app; seed DB, assert status and JSON. Target ≥90% coverage; test-enhancer phase to 95%. Directory: tests/unit/, tests/integration/.


# tasks.md
# RecipeBox Implementation Tasks

## Phase 1: Foundation (T001-T006)

### T001: Create User Model and Repository
**Estimate:** 45 min **Dependencies:** None **Files:** `src/models/user.py`, `src/repositories/user_repository.py`, `tests/unit/test_user_model.py`
**Acceptance Criteria:** User model (id, email, hashed_password, created_at); UserRepository create, get_by_email; unit tests; coverage ≥90%.

### T002: Create Recipe Model
**Estimate:** 45 min **Dependencies:** T001 **Files:** `src/models/recipe.py`, `tests/unit/test_recipe_model.py`
**Acceptance Criteria:** Recipe model with FKs, relationships to User and RecipeIngredient; unit tests.

### T003: Create Ingredient and RecipeIngredient Models
**Estimate:** 60 min **Dependencies:** T002 **Files:** `src/models/ingredient.py`, `src/models/recipe_ingredient.py`, tests
**Acceptance Criteria:** Ingredient, RecipeIngredient (amount Numeric(10,2)), relationships; unit tests.

### T004: Create MealPlan and MealPlanItem Models
**Estimate:** 45 min **Dependencies:** T002 **Files:** `src/models/meal_plan.py`, `src/models/meal_plan_item.py`, tests
**Acceptance Criteria:** MealPlan, MealPlanItem with FKs and relationships; unit tests.

### T005: Create ShoppingList and ShoppingListItem Models
**Estimate:** 45 min **Dependencies:** T004 **Files:** `src/models/shopping_list.py`, `src/models/shopping_list_item.py`, tests
**Acceptance Criteria:** ShoppingList, ShoppingListItem (total_amount Numeric(10,2)); unit tests.

### T006: Alembic Migration and Model Integration
**Estimate:** 30 min **Dependencies:** T001-T005 **Files:** Alembic migration, `src/models/__init__.py`
**Acceptance Criteria:** Single migration creates all tables; Base metadata consistent; upgrade/downgrade run.

## Phase 2: Recipe API (T007-T010)

### T007: RecipeRepository and RecipeService
**Estimate:** 60 min **Dependencies:** T006 **Files:** repository, service, unit tests
**Acceptance Criteria:** RecipeRepository CRUD; RecipeService with scale_ingredients; tests ≥90%.

### T008: Recipe Pydantic Schemas
**Estimate:** 30 min **Dependencies:** T007 **Files:** schemas/recipe.py
**Acceptance Criteria:** Request/response schemas for create, update, get; validation rules.

### T009: Recipe API Endpoints
**Estimate:** 60 min **Dependencies:** T008 **Files:** routers/recipes.py, main app include
**Acceptance Criteria:** POST/GET/PATCH/DELETE /api/recipes, GET list; 201/200/204/404/422.

### T010: Recipe API Integration Tests
**Estimate:** 45 min **Dependencies:** T009 **Files:** tests/integration/test_recipes_api.py
**Acceptance Criteria:** TestClient tests for all endpoints; assert status and body.

## Phase 3: Meal Planning (T011-T014)

### T011: MealPlanRepository and MealPlanService
**Estimate:** 60 min **Dependencies:** T006 **Files:** repository, service, date validation, tests
**Acceptance Criteria:** MealPlanRepository CRUD; MealPlanService add/remove item, date rules.

### T012: MealPlan Pydantic Schemas
**Estimate:** 30 min **Dependencies:** T011 **Files:** schemas/meal_plan.py
**Acceptance Criteria:** Schemas for plan and meal item; date validation.

### T013: Meal Plan API Endpoints
**Estimate:** 60 min **Dependencies:** T012 **Files:** routers/meal_plans.py
**Acceptance Criteria:** Create plan, add/remove meal, list, get, delete; 7 endpoints.

### T014: Meal Plan API Integration Tests
**Estimate:** 45 min **Dependencies:** T013 **Files:** tests/integration/test_meal_plans_api.py
**Acceptance Criteria:** Full request/response tests for all endpoints.

## Phase 4: Shopping Lists (T015-T017)

### T015: ShoppingListService Aggregation
**Estimate:** 75 min **Dependencies:** T007, T011 **Files:** shopping_list_service.py, unit tests
**Acceptance Criteria:** generate_from_meal_plan implements 6-step algorithm; unit tests.

### T016: Shopping List API Endpoints
**Estimate:** 45 min **Dependencies:** T015 **Files:** routers/shopping_lists.py
**Acceptance Criteria:** POST generate, GET list/items; 201/200/404.

### T017: Shopping List Integration Tests
**Estimate:** 45 min **Dependencies:** T016 **Files:** tests/integration/test_shopping_lists_api.py
**Acceptance Criteria:** Test multi-recipe meal plan → aggregated list.

## Phase 5: Search & Nutrition (T018-T020)

### T018: Full-Text Search
**Estimate:** 60 min **Dependencies:** T009 **Files:** search service, GET /api/search
**Acceptance Criteria:** Search by name/description/tags; ranking; tests.

### T019: Nutrition Data Integration
**Estimate:** 60 min **Dependencies:** T007 **Files:** nutrition service, cache, optional worker
**Acceptance Criteria:** Fetch/store per ingredient; aggregate per recipe; mock or USDA.

### T020: Nutrition API Endpoints
**Estimate:** 45 min **Dependencies:** T019 **Files:** routers/nutrition.py or extend recipes
**Acceptance Criteria:** GET nutrition for recipe/meal plan; 200/404.

## Phase 6: Quality & Documentation (T021-T023)

### T021: Test Enhancement to 95%
**Estimate:** 60 min **Dependencies:** T010, T014, T017 **Files:** test-enhancer agent run
**Acceptance Criteria:** Coverage ≥95%; edge cases documented.

### T022: Security and Error Review
**Estimate:** 45 min **Dependencies:** T021 **Files:** Security checklist, error messages
**Acceptance Criteria:** No sensitive data in responses; HTTPException consistency.

### T023: Documentation and README
**Estimate:** 45 min **Dependencies:** T022 **Files:** README, doc-updater agent, OpenAPI
**Acceptance Criteria:** README features; API docs current; ADR if applicable.


# dependency-graph.md
# RecipeBox Task Dependencies

## Critical Path: Foundation Phase

T001 → T002 → T003 → T004 → T005 → T006 (all models and migration).

## Parallel After Foundation

- **Track A (Recipe API):** T006 → T007 → T008 → T009 → T010
- **Track B (Meal Planning):** T006 → T011 → T012 → T013 → T014

T010 and T014 complete → Phase 4: T015 → T016 → T017. Then T018 → T019 → T020. Finally T021 → T022 → T023.

## Time Savings

Sequential estimate ~15–18 h. With parallel T007–T014: ~8–9 h calendar. Savings ~40–50%.


# task-executor.md
name: task-executor
description: Implements RecipeBox tasks following test-first workflow
tools: Read, Write, Edit, Bash, Grep
model: sonnet

You are implementing RecipeBox using FastAPI + SQLAlchemy + PostgreSQL.

Constitution: CLAUDE.md (auto-loaded)

Process:
1. Read task acceptance criteria from @specs/recipebox/tasks.md for the given T00X.
2. Review relevant sections of @specs/recipebox/specification.md and technical-plan.
3. Identify existing models/repos/services and integration points.
4. Write failing tests first (pytest) that match acceptance criteria.
5. Implement code to pass tests (repository → service → route as needed).
6. Run: pytest tests/ -v; pytest --cov=src/recipebox --cov-fail-under=90 tests/; fix any failures.
7. Report: "T00X complete. Validation: N tests passed, coverage X%. Files: [list]. Ready: commit message."

Standards from CLAUDE.md:
- Repository pattern (no direct DB in API)
- Type hints mandatory
- Google docstrings
- HTTPException errors
- 90%+ coverage


# recipe-validator.md

name: recipe-validator
description: Validates RecipeBox business logic correctness
tools: Read, Bash
model: sonnet

You validate RecipeBox business logic is correct.

Focus areas:
1. Serving scaling: Scale recipe from 4 to 8 servings; all ingredient amounts double; assert with pytest.
2. Meal plan date validation: Reject date in past; reject date >30 days ahead; accept today and next 30 days.
3. Shopping list aggregation: Meal plan with 2 recipes sharing an ingredient; generated list aggregates amounts and groups by category; assert counts and totals.
4. Error handling: Recipe not found returns 404; invalid payload returns 422; assert status and message shape.


# test-enhancer.md
name: test-enhancer
description: Enhances RecipeBox test coverage to 95%+
tools: Read, Write, Bash
model: sonnet

You enhance test coverage by adding edge cases.

Process:
1. Run coverage report: pytest --cov=src/recipebox --cov-report=term-missing tests/
2. Identify uncovered lines and critical paths (services, aggregation, validation).
3. Add tests for edge cases: zero servings, empty lists, invalid dates, duplicate ingredients, constraint violations.
4. Re-run coverage; iterate until ≥95% for target modules.
5. Document added cases in a short summary.

RecipeBox focus: serving scaling (0, negative, very large); meal plan date bounds (past, >30 days); shopping list aggregation (single recipe, many recipes, same ingredient multiple units); 404/422 error paths.


# doc-updater.md
name: doc-updater
description: Updates RecipeBox documentation
tools: Read, Write, Grep
model: sonnet

You keep documentation current with implementation.

Process:
1. Read OpenAPI/schema from app (or routes) and compare to README/API docs.
2. Update README endpoints list and example requests/responses to match current behavior.
3. Regenerate or adjust OpenAPI description if endpoints or status codes changed.
4. Ensure no stale references to removed endpoints or old field names.


# orchestrate-recipebox.md
# Orchestrate: RecipeBox Implementation

Implement RecipeBox using agent orchestration (Course 3 pattern).

Context:
@specs/recipebox/specification.md
@specs/recipebox/tasks.md
@CLAUDE.md

## Phase 1: Foundation (T001-T006) - Critical Path

For each of T001–T006: Delegate to task-executor with "Execute T00X from @specs/recipebox/tasks.md". Validate: pytest, coverage ≥90%. Commit atomically. Checkpoint after T006: all models and migration applied; tests green.

## Phase 2 & 3: Parallel Workflows

Track A: T007→T008→T009→T010 (RecipeRepository, RecipeService, schemas, API, integration tests). Track B: T011→T012→T013→T014 (MealPlan equivalent). Run sequentially in one session or in two sessions to simulate parallelism. Checkpoint after both: Recipe and MealPlan APIs both working.

## Phase 4: Shopping Lists (T015-T017)

T015 (ShoppingListService aggregation) → T016 (endpoints) → T017 (integration tests). Checkpoint: generate list from multi-recipe meal plan; assert aggregated items.

## Phase 5: Search & Nutrition (T018-T020)

T018 full-text search → T019 nutrition integration → T020 nutrition endpoints. Checkpoint: search returns ranked results; nutrition aggregated for recipe/plan.

## Phase 6: Quality & Documentation (T021-T023)

T021 test-enhancer to 95% → T022 security/error review → T023 doc-updater and README. Final checkpoint: coverage report, no HIGH/CRITICAL issues, docs current.

## Efficiency Metrics

Log agent time and validation time per task. Compare total to estimated manual (e.g. 15–18 h). Typical orchestrated total ~8–9 h; document efficiency gain (e.g. 40–50%).


```

```

---

### 📋 2. Core Implementation Strategy (`specs/recipebox/technical-plan.md`)

```markdown
# RecipeBox Technical Plan

## Stack
* Python 3.11+ (FastAPI 0.100+)
* PostgreSQL 15+ & SQLAlchemy 2.0 (Async Extension)
* Redis 7+ (Task Queue & Cache State Driver)
* pytest Framework with automated coverage floor controls

---

## Architecture


### Repository Pattern
All database communication is handled by specialized classes inside `src/repositories/`. API route endpoints and business services interact exclusively with repository interfaces. Direct database session manipulation, query filtering, or raw transaction execution are strictly prohibited outside the data access layer.

### Dependency Injection
Database contexts, repository wrappers, and user authentication scopes are injected into route parameters using FastAPI's native `Depends()` mechanism. Service layers use standard constructor dependency injection to load required repositories, making it easy to swap in test doubles during development.

### API-First Design
System route behaviors and validation bounds are established before implementation using Pydantic schemas. Incoming payloads undergo input checking via Pydantic model configurations, business rules are evaluated within service layers, and data structures are persisted via repositories.

### Background Jobs
Volumetric lookups and third-party macro calculations are handled out-of-band using a background worker pattern. The system pushes task identifiers to a Redis queue, allowing a standalone worker process (`NutritionFetchWorker`) to fetch data, update the database, and manage cache states without blocking user threads.

---

## Data Models (SQLAlchemy Declarative)

* **User:** `id` (UUID, PK), `email` (VARCHAR(255), UQ, Indexed), `password_hash` (VARCHAR(255)), `name` (VARCHAR(100)), `created_at` (TZ TIMESTAMP).
* **Recipe:** `id` (UUID, PK), `user_id` (UUID, FK users.id, Indexed), `name` (VARCHAR(200)), `description` (VARCHAR(1000)), `instructions` (TEXT), `prep_time_minutes` (INT), `cook_time_minutes` (INT), `servings` (INT, default 4), `tags` (VARCHAR(50)[]), `created_at` (TZ TIMESTAMP), `updated_at` (TZ TIMESTAMP).
* **Ingredient:** `id` (UUID, PK), `name` (VARCHAR(200), UQ, Indexed), `category` (VARCHAR(50)), `calories_per_100g` (NUMERIC(10,2)), `protein_per_100g` (NUMERIC(10,2)), `carbs_per_100g` (NUMERIC(10,2)), `fat_per_100g` (NUMERIC(10,2)).
* **RecipeIngredient:** `id` (UUID, PK), `recipe_id` (UUID, FK recipes.id, UQ Bound 1), `ingredient_id` (UUID, FK ingredients.id, UQ Bound 2), `amount` (NUMERIC(10,2)), `unit` (VARCHAR(20)).
* **MealPlan:** `id` (UUID, PK), `user_id` (UUID, FK users.id, UQ Bound 1), `week_start_date` (DATE, UQ Bound 2).
* **MealPlanItem:** `id` (UUID, PK), `meal_plan_id` (UUID, FK meal_plans.id), `recipe_id` (UUID, FK recipes.id), `date` (DATE), `meal_type` (VARCHAR(20)), `target_servings` (INT).
* **ShoppingList:** `id` (UUID, PK), `meal_plan_id` (UUID, FK meal_plans.id, UQ), `updated_at` (TZ TIMESTAMP).
* **ShoppingListItem:** `id` (UUID, PK), `shopping_list_id` (UUID, FK shopping_lists.id, UQ Bound 1), `ingredient_name` (VARCHAR(200), UQ Bound 2), `category` (VARCHAR(50)), `total_amount` (NUMERIC(10,2)), `unit` (VARCHAR(20), UQ Bound 3), `checked` (BOOLEAN, default False).

---

## Repositories
* **UserRepository:** `create_user()`, `get_user_by_id()`, `get_user_by_email()`.
* **RecipeRepository:** `create_recipe()`, `get_recipe_by_id()`, `get_all_user_recipes(user_id, search_query, tag_filter)`, `update_recipe()`, `delete_recipe()`.
* **IngredientRepository:** `get_or_create_ingredient()`, `get_ingredient_by_name()`, `list_all_ingredients()`.
* **MealPlanRepository:** `create_meal_plan()`, `get_meal_plan_by_id()`, `get_user_meal_plan_by_date(user_id, week_start_date)`, `delete_meal_plan()`.
* **ShoppingListRepository:** `save_shopping_list()`, `get_shopping_list_by_meal_plan(meal_plan_id)`, `update_item_checked_status()`.

---

## Services

### RecipeService
Handles recipe mutations and data lookups. Includes the `scale_ingredients(recipe_id, target_servings)` method, which scales base quantities using the serving factor formula:
$$\text{Scaled Amount} = \text{Base Amount} \times \left(\frac{\text{Target Servings}}{\text{Base Servings}}\right)$$

### MealPlanService
Manages weekly schedule tracking, enforces calendar validation rules (e.g., ensuring `week_start_date` lands on a Monday), and logs menu items within a 30-day planning window.

### ShoppingListService
Coordinates the **6-Step Shopping List Aggregation Algorithm**:
1. **Initialize Map Container:** Open an empty hash map memory cache structured as: `(ingredient_name, unit) -> {amount: Decimal, category: String}`.
2. **Collect Meal Plan Items:** Pull all scheduled items linked to the target `MealPlan` ID.
3. **Fetch Ingredient Details:** Iterate through all recipes in the plan and load their component ingredients via the repository layer.
4. **Scale Quantities:** Apply the serving scaling formula to calculate scaled ingredient amounts for each meal item.
5. **Convert & Accumulate Units:** Group identical ingredients together. If units are compatible but distinct (e.g., teaspoons and tablespoons), normalize them to the primary unit using our standard multipliers ($3\text{ tsp} = 1\text{ tbsp}$) before adding. Keep incompatible unit lines separate.
6. **Persist Consolidated Records:** Clear old generated lines and batch-save the aggregated results into the `ShoppingListItem` table, sorted by grocery category.

### NutritionService
Manages nutritional calculations and coordinates async background tasks. It updates recipe macro fields by summing the nutritional values of component ingredients scaled to a standard 100g weight baseline.

---

## API Routes
* **Recipes Group:** `POST /api/recipes` (21 Created), `GET /api/recipes` (200 List), `GET /api/recipes/{id}` (200 OK / 404), `PATCH /api/recipes/{id}` (200 OK), `DELETE /api/recipes/{id}` (204 No Content).
* **Meal Plans Group:** `POST /api/meal-plans` (201 Created / 400), `GET /api/meal-plans/{id}` (200 OK), `DELETE /api/meal-plans/{id}` (204).
* **Shopping Lists Group:** `GET /api/meal-plans/{id}/shopping-list` (200 OK / 403 / 404), `PATCH /api/shopping-list-items/{item_id}/toggle` (200 OK).

---

## Background Worker
The `NutritionFetchWorker` runs as a standalone process that polls a Redis task queue for background jobs. It connects asynchronously to the external Edamam API to parse ingredient strings, updates the global `Ingredient` macro metrics, saves the result to the Redis cache with a 30-day TTL, and updates affected recipe profiles.

---

## Testing Strategy
* **Unit Testing:** Uses `pytest` and an in-memory SQLite database setup to test repositories and services in isolation. It uses mock objects to simulate external API responses and network boundaries.
* **Integration Testing:** Uses FastAPI's `TestClient` to test the API endpoints against a sandbox database. It evaluates the entire request-response lifecycle, confirming header parsing, input validation checking, multi-tenant data isolation, and accurate HTTP error status responses.

```

## Documenting Architecture Decisions

# Introduction To Technical Documentation

Welcome to the final lesson of our course! So far, we have built a powerful API, organized specialized AI agents, and ensured our code is secure and fast. However, there is a common saying in software engineering: *"Code tells you how; documentation tells you why."*

In a production environment, you are rarely working alone. Even if you are, you might forget why you chose a specific database or a specific way of handling logins six months from now. If the "why" isn't recorded, future developers might change your code and accidentally break something important because they didn't understand your original plan.

In this lesson, we will learn how to capture these big decisions using **Architecture Decision Records (ADRs)**. We will also build specialized AI agents to help draft ADRs and audit documentation automatically. By the end of this lesson, you will have a complete documentation workflow for production-ready APIs.

---

## The Anatomy Of An ADR

An Architecture Decision Record (ADR) is a short document that captures a single significant technical choice. We don't write them for every line of code, but we do write them for big choices that affect the whole project.

Let's look at the structure of an effective ADR:

```markdown
# ADR-001: Repository Pattern for Data Access

**Status:** Accepted  
**Date:** 2024-11-15  
**Deciders:** Engineering team  

## Context
[What problem are we solving? What forces are at play?]

## Decision
[What specific action are we taking?]

## Alternatives Considered
[What other options did we evaluate and why didn't we choose them?]

## Consequences

### Positive
[Benefits of this decision]

### Negative
[Drawbacks we're accepting]

### Neutral
[Other impacts to be aware of]

```

The **Status** field is important because it tracks the lifecycle of the decision: `Proposed` ➔ `Accepted` ➔ `Deprecated` ➔ `Superseded`. ADRs are immutable—we never edit them. If we change our mind, we create a new ADR that supersedes the old one.

## AI Agents for Documentation

Just as we used specialized agents for implementation and testing, we can create agents for documentation tasks. There are two main types:

* **ADR Writer Agent:** Drafts architecture decision records by analyzing code and context. It can generate the basic structure and standard content, which humans then refine with project-specific details and nuanced trade-offs.
* **Documentation Auditor Agent:** Automatically checks that documentation matches implementation. It compares OpenAPI specs to actual endpoints, verifies README completeness, and identifies missing ADRs.

This division of labor is efficient: AI handles repetitive structure and obvious facts, while humans add judgment, context, and strategic thinking.

---

## Human-AI Collaboration in Documentation

The most effective documentation workflow uses both AI and human strengths:

### AI excels at:

* Generating consistent document structure
* Listing standard benefits/drawbacks
* Comparing files for completeness
* Identifying obvious gaps

### Humans excel at:

* Understanding project-specific context
* Evaluating subtle trade-offs
* Making strategic decisions
* Adding nuanced explanations

### A typical workflow looks like this:

1. AI agent drafts ADR with standard structure.
2. Human reviews and adds project-specific "why".
3. AI agent audits all documentation for gaps.
4. Human prioritizes and fixes critical issues.

### 📋 2. Foundational Architecture Decision Record (`ADR-001: Repository Pattern.md`)

This file represents the finalized, human-refined architectural record capturing the transition to decoupled data access layers.

```markdown
# ADR-001: Repository Pattern for Data Access

**Status:** Accepted  
**Date:** 2026-07-13  
**Deciders:** Engineering Team  

## Context
The TaskMaster API development cycle has hit a major bottleneck due to tight coupling between our HTTP route layers and the underlying database ORM models. Currently, our core endpoint routers directly execute inline SQLAlchemy queries, manage database transactions, and handle payload validation loops simultaneously. 

This tight coupling has caused several production issues over our last 3-week sprint cycle:
1. **Testing Bottlenecks:** Writing tests requires mocking complex SQLAlchemy connection pools and session transactions, which has increased unit test setup time by roughly 45% per module.
2. **High Regression Risk:** Minor schema modifications to any model force developers to manually trace and update queries scattered across multiple router files, increasing the risk of regression.
3. **Database Performance Gaps:** It is incredibly difficult to apply global query optimizations, database index checks, or eager loading strategies (`joinedload`) across scattered, un-abstracted query paths.

## Decision
We have decided to isolate all database persistence and query logic behind a dedicated data access layer by implementing the **Repository Pattern**. 

All endpoints must interact exclusively with service business layers, which in turn delegate database operations to isolated repository objects. No SQLAlchemy session operations (`db.add()`, `db.commit()`, `.query()`) are permitted to leak into the `src/api/` routing layer.

### Concrete Implementation Example

```python
# src/repositories/task_repository.py
from sqlalchemy.orm import Session
from src.models.task import Task

class TaskRepository:
    """Encapsulates all database CRUD logic for the Task entity."""
    
    def __init__(self, db: Session):
        self.db = db

    def get_task_by_id(self, task_id: int) -> Task | None:
        return self.db.query(Task).filter(Task.id == task_id).first()

    def create_task(self, user_id: int, title: str, description: str) -> Task:
        task = Task(user_id=user_id, title=title, description=description)
        self.db.add(task)
        self.db.commit()
        self.db.refresh(task)
        return task

```

---

## Alternatives Considered

### Alternative 1: Inline Active Record Pattern

* **Description:** Keep query logic directly inside database model helper classes (e.g., `task.save()`, `task.delete()`).
* **Rejection Reason:** While this keeps files consolidated, it fails to separate database transactions from HTTP route contexts. It does not solve our core testing issues and makes it difficult to switch out underlying database management tools in the future.

### Alternative 2: Direct Execution of Raw SQL Queries via AsyncPG

* **Description:** Bypass the SQLAlchemy ORM layer completely and execute raw SQL query strings through an asynchronous connection pool layer.
* **Rejection Reason:** Raw SQL text blocks are highly prone to syntax errors that automated tools cannot catch, significantly increasing our maintenance overhead. Additionally, this approach removes our automatic schema migration features via Alembic, forcing developers to manage database states manually.

---

## Consequences

### Positive

* **Total Layer Separation:** Changes to database schemas or query choices are isolated entirely within the `src/repositories/` folder, protecting our endpoint routes from regressions.
* **Trivialized Unit Testing:** Service layers can now be fully tested in isolation within 3 minutes by mocking simple repository class methods, completely removing the need to mock complex SQLAlchemy session configurations.
* **Centralized Performance Tuning:** Query profiling, cache integrations, and eager-loading optimizations can be applied directly within centralized methods, instantly improving performance across all calling endpoints.

### Negative

* **Increased File Overhead:** Implementing this pattern across our 8 core entity modules requires creating 8 new repository files and 8 test files. This introduces 16 new files to maintain across the codebase.
* **Additional Code Boilerplate:** Simple CRUD mutations that previously required a single line of inline code now must be explicitly routed through a multi-layer delegation pass (`Router ➔ Service ➔ Repository`).

### Neutral

* **Team Onboarding and Training:** Developers must be trained to follow this strict architectural pattern. Any code change that leaks raw database queries directly into API endpoints will be caught and rejected at the CI/CD gate.

```markdown
---
name: doc-auditor
description: Audits documentation for completeness and accuracy against implementation
tools: Read, Write, Grep, Bash
model: sonnet
---

You are a documentation quality auditor.

## Your Role
Your purpose is to act as an automated gatekeeper for production-grade documentation. You analyze code paths, routes, and persistence layers, cross-referencing them against external documentation files (OpenAPI YAML specs, project READMEs, and Architecture Decision Records) to catch discrepancies, missing specs, and undocumented design choices.

## Process

1. **Audit OpenAPI Specification**
   - Use `Grep` or `Bash` to extract actual route functions out of the application package directory:
     ```bash
     grep -r "@router." src/api/
     ```
   - Read `docs/openapi.yaml` or matching schema targets.
   - Map live application routes to documented endpoints and list missing path contracts.

2. **Audit README Completeness**
   - Check that the root `README.md` contains the 3 required boilerplate anchors: Features List, Local Setup Guide, and Testing Operations.
   - Scan package models and repositories to extract implemented capabilities and check if they are absent from the README.

3. **Audit ADR Coverage**
   - Search the code for major infrastructure dependencies, architectural decisions, and key technical choices (e.g., JWT Authentication, SQLAlchemy ORM, Pydantic, FastAPI):
     ```bash
     grep -Ei "jwt|sqlalchemy|pydantic|fastapi" src/
     ```
   - Cross-reference these with files in the `docs/adr/` directory to spot major architectural choices that lack a documented ADR.

4. **Audit Test Documentation**
   - Verify the presence of configuration baseline files (`pytest.ini` or equivalent).
   - Confirm that structural test coverage execution targets ($\ge 95\%$) are explicitly detailed.

5. **Generate Audit Report**
   - Synthesize all findings into a structured document under the filename `documentation-audit-report.md`.
   - Organize issues strictly by severity layers, including exact path locations and clear fix instructions.

## Standards
- **Severity Classification:**
  - `CRITICAL`: Missing authentication/authorization endpoint contracts or completely missing setup files.
  - `HIGH`: Implemented endpoints or models missing from OpenAPI or missing ADRs for core architectural decisions.
  - `MEDIUM`: Typos, minor formatting variations, or missing code usage examples.
- **Actionability:** Every finding must provide the exact file location and a concrete fix suggestion.
- **Readiness Metric:** The final status can only be marked `PRODUCTION READY` if there are 0 CRITICAL and 0 HIGH findings remaining.

## Output
Deliver a comprehensive markdown audit summary file detailing the complete production readiness assessment of the project's documentation stack.

```

```

---

### 📋 4. Architectural System Standard (`ADR-002-jwt-authentication.md`)

```markdown
# ADR-002: Stateless JWT Authentication Framework

**Status:** Accepted  
**Date:** 2026-07-14  
**Deciders:** Engineering Team

## Context
The TaskMaster ecosystem is scaling rapidly, and we need a secure, multi-tenant authentication system. Our user base is growing past 10,000 active members, and our roadmap requires supporting three distinct client runtimes simultaneously over the next 6 weeks: a web interface, a mobile application, and third-party API integrations.

Our current proof-of-concept setup relies on mock authentication, which is a major security risk. To support our scaling requirements, the authentication framework must satisfy the following constraints:
1. **Horizontally Scalable:** The system must not require synchronous database lookups to validate tokens on every incoming request, as this would create a significant database bottleneck.
2. **Cross-Platform Compatibility:** The authentication mechanism must rely on standard HTTP headers, avoiding reliance on web-only features like session cookies to ensure compatibility with our mobile app.
3. **Multi-Tenant Isolation:** Every token payload must securely carry authenticated tenant claims (`user_id`) that are tamper-proof and cryptographically signed.

## Decision
We have decided to implement a **Stateless JWT (JSON Web Token) Authentication Framework**. 

When a user logs in successfully, the backend issues a cryptographically signed token containing the user's identity claims. Clients store this token locally and include it in the `Authorization: Bearer <token>` header for all subsequent API requests. The server validates incoming tokens entirely in-memory using a shared symmetric `JWT_SECRET_KEY` and the `HS256` signing algorithm, avoiding unnecessary database lookups.

### Concrete Implementation Example

```python
# src/utils/jwt.py
from datetime import datetime, timedelta
from jose import jwt, JWTError
from src.config import settings

def create_access_token(user_id: int) -> str:
    """Generates a cryptographically signed stateless access token."""
    expire = datetime.utcnow() + timedelta(minutes=settings.ACCESS_TOKEN_EXPIRE_MINUTES)
    payload = {
        "sub": str(user_id),
        "exp": expire
    }
    return jwt.encode(payload, settings.JWT_SECRET_KEY, algorithm=settings.JWT_ALGORITHM)

def verify_access_token(token: str) -> int | None:
    """Validates the signature and expiration bounds of a token in-memory."""
    try:
        payload = jwt.decode(token, settings.JWT_SECRET_KEY, algorithms=[settings.JWT_ALGORITHM])
        user_id: str = payload.get("sub")
        if user_id is None:
            return None
        return int(user_id)
    except JWTError:
        return None

```

## Alternatives Considered

### Alternative 1: Server-Side Stateful Session Identifiers

* **Description:** Store session IDs in a centralized database or memory store (like Redis), mapping them to a web cookie on the client.
* **Rejection Reason:** This approach is heavily reliant on web browser cookie standards, making it difficult to implement securely on mobile applications. It also requires a database or memory-store lookup for every single incoming API request, which would increase our p95 latency under heavy concurrent load.

### Alternative 2: Asymmetric Cryptographic Signing (RS256 - Public/Private Keys)

* **Description:** Sign tokens using a private key and validate them across internal service clusters using a public key.
* **Rejection Reason:** While RS256 provides excellent security boundaries for microservices, our current single-instance deployment does not require public/private key pairs. Implementing it now would introduce unnecessary operational complexity around key management, without providing any immediate benefit for our monolithic architecture.

---

## Consequences

### Positive

* **High Efficiency Performance:** Validating signatures in-memory avoids database round-trips for token verification, saving ~50ms per API request based on our load benchmarks.
* **Excellent Mobile Compatibility:** Standard Bearer header token structures work seamlessly across web, iOS, Android, and third-party developer integration environments.
* **Stateless Scaling:** Because the server does not store session states, we can scale our web instances horizontally behind a load balancer without needing a shared session database.

### Negative

* **Difficult Token Revocation:** Once a token is signed and issued, it remains valid until it naturally expires. If a token is compromised within its 15-minute lifespan, it cannot be easily revoked without implementing a complex token blacklist in our database.
* **Secret Key Vulnerability:** The security of the entire system depends on the secrecy of our `JWT_SECRET_KEY`. If this key is leaked or compromised, an attacker can forge tokens for any user account in the system.

### Neutral

* **Client Token Management:** Frontend developers are responsible for safely storing tokens in-memory or within secure device storage, and must implement automatic token refresh flows before expiry.

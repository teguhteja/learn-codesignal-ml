```javascript
function generateStory() {
    const promptInput = document.getElementById('prompt').value;
    const tone = document.getElementById('tone').value;
    const loadingMessage = document.getElementById('loading-message');
    const generateButton = document.getElementById('generate-button');
    const storyContainer = document.getElementById('story-container');
    
    if (!promptInput) {
        alert('Please enter a story theme');
        return;
    }
    
    loadingMessage.style.display = 'block';
    generateButton.style.display = 'none';
    storyContainer.innerHTML = '';
    
    fetch('/api/generate_story', {
        method: 'POST',
        headers: { 'Content-Type': 'application/json' },
        body: JSON.stringify({ prompt: promptInput, tone: tone })
    })
    .then(response => response.json())
    .then(data => {
        loadingMessage.style.display = 'none';
        generateButton.style.display = 'block';
        
        if (data.error) {
            alert(data.error);
        } else {
            const pre = document.createElement('pre');
            pre.textContent = data.story.content;
            storyContainer.appendChild(pre);
        }
    })
    .catch(error => {
        loadingMessage.style.display = 'none';
        generateButton.style.display = 'block';
        alert('Error generating story. Please try again.');
        console.error('Error:', error);
    });
}

function fetchHistory() {
    const historyLoadingMessage = document.getElementById('history-loading-message');
    const historyButton = document.getElementById('history-button');
    const historyContainer = document.getElementById('history-container');
    
    historyLoadingMessage.style.display = 'block';
    historyButton.style.display = 'none';
    historyContainer.innerHTML = '';
    
    fetch('/api/get_stories')
    .then(response => response.json())
    .then(data => {
        historyLoadingMessage.style.display = 'none';
        historyButton.style.display = 'block';
        
        if (data.stories.length === 0) {
            historyContainer.innerHTML = '<p>No stories found.</p>';
            return;
        }
        
        data.stories.forEach(story => {
            const pre = document.createElement('pre');
            pre.textContent = story.content;
            historyContainer.appendChild(pre);
        });
    })
    // Add a catch() block to handle errors during the fetch operation
    .catch(error => {
        // Reset the UI states
        historyLoadingMessage.style.display = 'none';
        historyButton.style.display = 'block';
        
        // Inform user and log error
        alert('Error fetching story history. Please try again.');
        console.error('Error:', error);
    });
}

function openTab(tabName) {
    const tabContents = document.getElementsByClassName('tab-content');
    for (let i = 0; i < tabContents.length; i++) {
        tabContents[i].style.display = 'none';
    }
    document.getElementById(tabName).style.display = 'flex';
}

```

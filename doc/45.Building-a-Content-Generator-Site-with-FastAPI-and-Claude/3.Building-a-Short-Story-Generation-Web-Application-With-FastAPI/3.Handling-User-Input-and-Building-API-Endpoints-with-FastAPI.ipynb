## Implementing Tab Navigation

Our application has two main tabs: **"Generate Story"** and **"View History."** We need to implement a function that will show the selected tab and hide the others when a user clicks on a tab button.

Let's create a file named `script.js` in the `static/` folder of our FastAPI application. This is the same folder where we placed our `style.css` file in the previous lesson.

```text
app/
├── templates/
│   └── index.html
├── static/
│   ├── style.css
│   └── script.js
└── main.py

```

Let's start by implementing tab navigation in `script.js`. First, gather all elements with the class `tab-content` to hide them initially:

```javascript
function openTab(tabName) {
    document.querySelectorAll('.tab-content').forEach(tab => {
        tab.style.display = 'none';
    });

```

`document.querySelectorAll('.tab-content')` selects all elements with the class `tab-content`. We then use `.forEach()` to loop through each element and set its `display` style to `'none'`.

Next, display the selected tab by setting its `display` style to `'block'`:

```javascript
    document.getElementById(tabName).style.display = 'block';
}

```

This approach is backward compatible and scalable. If you decide to add more `.tab-content` sections in the future (for example, to introduce new features), you won't need to change this function — it will automatically handle hiding and displaying sections dynamically.

## Introducing API Calls with Fetch

We use the native **Fetch API** to send a `POST` request to `/api/generate_story`:

```javascript
fetch('/api/generate_story', {
    method: 'POST',
    headers: { 'Content-Type': 'application/json' },
    body: JSON.stringify({ prompt: promptInput, style: style })
})

```

`JSON.stringify` converts the client's form data into a JSON payload sent over the HTTP body.

---

## Handling Responses & Resolving Promises

We extend the `fetch` Promise chain to parse the payload and render the generated text:

```javascript
.then(response => response.json())
.then(data => {
    loadingMessage.style.display = 'none';
    generateButton.style.display = 'block';
    
    if (data.error) {
        alert(data.error);
    } else {
        const pre = document.createElement('pre');
        pre.textContent = data.story;
        storyContainer.appendChild(pre);
    }
})
.catch(error => {
    loadingMessage.style.display = 'none';
    generateButton.style.display = 'block';
    alert('Error generating story. Please try again.');
    console.error('Error:', error);
});

```

* **Data Rendering:** Re-enables input controls and appends the text content wrapped in a `<pre>` block to preserve formatting.
* **Error Handling:** The `.catch()` block catches network failures or server exceptions, preventing UI freezes by resetting loading states.

Here is the complete JavaScript code extending the `fetch` promise chain to parse the JSON response, render the story inside a `<pre>` element, and handle errors gracefully:

```javascript
function openTab(tabName) {
    document.querySelectorAll('.tab-content').forEach(tab => {
        tab.style.display = 'none';
    });
    document.getElementById(tabName).style.display = 'flex';
}

function generateStory() {
    const promptInput = document.getElementById('prompt').value;
    const tone = document.getElementById('tone').value;
    const loadingMessage = document.getElementById('loading-message');
    const generateButton = document.getElementById('generate-button');
    const storyContainer = document.getElementById('story-container');
    
    if (!promptInput) {
        alert('Please enter a story theme');
        return;
    }
    
    loadingMessage.style.display = 'block';
    generateButton.style.display = 'none';
    storyContainer.innerHTML = '';
    
    fetch('/api/generate_story', {
        method: 'POST',
        headers: { 'Content-Type': 'application/json' },
        body: JSON.stringify({ prompt: promptInput, tone: tone })
    })
    // Parse the JSON response
    .then(response => response.json())
    // Process the returned data
    .then(data => {
        loadingMessage.style.display = 'none';
        generateButton.style.display = 'block';

        if (data.error) {
            alert(data.error);
        } else {
            const pre = document.createElement('pre');
            // Safely extract content whether it's nested (data.story.content) or flat (data.story)
            pre.textContent = typeof data.story === 'object' && data.story !== null 
                ? (data.story.content || data.story.story) 
                : data.story;
            storyContainer.appendChild(pre);
        }
    })
    // Handle network failures or runtime exceptions
    .catch(error => {
        loadingMessage.style.display = 'none';
        generateButton.style.display = 'block';
        alert('Error generating story. Please try again.');
        console.error('Error:', error);
    });
}

```

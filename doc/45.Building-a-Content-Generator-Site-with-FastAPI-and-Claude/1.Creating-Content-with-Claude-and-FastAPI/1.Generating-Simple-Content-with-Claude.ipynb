## Initializing the Anthropic Client

To start generating content with Claude, we first need to initialize the Anthropic client. This client acts as a bridge between your code and the Claude model.

```python
import os
from anthropic import Anthropic

# Initialize the Anthropic client
client = Anthropic(api_key=os.environ.get("ANTHROPIC_API_KEY"))

```

In this snippet:

* We import the `os` module to access environment variables.
* We import the `Anthropic` class from the `anthropic` library.
* We create an instance of the `Anthropic` client using an API key stored in an environment variable. This key is essential for authenticating your requests to the Claude model.

## Crafting a Simple Prompt

The next step is to craft a simple prompt. A prompt is a piece of text that instructs the AI on what kind of content to generate. A well-crafted prompt is clear and concise, guiding the AI to produce the desired output.

```python
# Define a simple prompt to generate content
prompt = "Write a short story about a brave knight."

```

Here, the prompt is a straightforward instruction for the AI to generate a short story about a brave knight. The clarity of the prompt helps the AI understand the task and produce relevant content.

---

## Generating Content with Claude

Now that we have our prompt, let's use the Claude model to generate content. We'll break this process down into manageable steps.

```python
# Generate content using the Claude model
response = client.messages.create(
    model='claude-3-5-sonnet-20241022',
    max_tokens=500,  # Limit the content length
    messages=[
        {
            "role": "user",
            "content": prompt,
        }
    ]
)

```

In this code:

* We call the `create` method on the `client.messages` object to generate content.
* The `model` parameter specifies the version of the Claude model we are using.
* The `max_tokens` parameter limits the length of the generated content to 500 tokens.
* The `messages` parameter is a list of dictionaries, where each dictionary represents a message. Here, we have a single message with the role of `"user"` and the content being our prompt.

---

## Processing and Displaying Generated Content

Once the content is generated, we need to process and display it. This step involves extracting the content from the response and printing it.

```python
# Process and display the generated content
generated_content = response.content
print(generated_content)

```

In this snippet:

* We extract the generated content from the `response` object.
* We print the content to the console, allowing us to see the AI-generated story.

Here is the complete, placeholder-free implementation comparing two API calls with different `max_tokens` values:

```python
import os
from anthropic import Anthropic

# Initialize the Anthropic client
client = Anthropic(api_key=os.environ.get("ANTHROPIC_API_KEY"))

# Define a simple prompt to generate content
prompt = "Write a short story about a brave knight."

# Generate content using the Claude model with default parameters
response = client.messages.create(
    model='claude-sonnet-4-5-20250929',
    max_tokens=500,  # Default token limit
    messages=[
        {
            "role": "user",
            "content": prompt,
        }
    ]
)

# Create a second API call with a much lower max_tokens value to force a brief response
response_short = client.messages.create(
    model='claude-sonnet-4-5-20250929',
    max_tokens=80,  # Short token limit
    messages=[
        {
            "role": "user",
            "content": prompt,
        }
    ]
)

# Extract content from both responses
generated_content = response.content[0].text
generated_content_short = response_short.content[0].text

# Display both generated contents with clear labels
print("DEFAULT VERSION (max_tokens=500):")
print("-" * 50)
print(generated_content)
print("\n")

print("SHORT VERSION (max_tokens=80):")
print("-" * 50)
print(generated_content_short)
print("\n")

# Compare the outputs by character length
len_default = len(generated_content)
len_short = len(generated_content_short)

print("=" * 50)
print("OUTPUT COMPARISON METRICS:")
print(f"Default Version Length: {len_default} characters")
print(f"Short Version Length:   {len_short} characters")
print(f"Difference:             {len_default - len_short} characters longer in default mode")
print("=" * 50)

```

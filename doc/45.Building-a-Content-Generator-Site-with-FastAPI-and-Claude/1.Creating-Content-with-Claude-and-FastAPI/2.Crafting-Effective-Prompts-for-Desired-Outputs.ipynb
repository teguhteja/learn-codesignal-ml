## Understanding the Role of Prompts

Prompts are the instructions or questions you provide to an AI model to guide its content generation. The clarity and specificity of your prompts can significantly impact the quality of the output.

```python
import os
from anthropic import Anthropic

# Initialize the Anthropic client
client = Anthropic(api_key=os.environ.get("ANTHROPIC_API_KEY"))


def generate_content(prompt: str) -> str:
    """Helper function to execute Claude API calls with basic error handling.
    
    Args:
        prompt: The input prompt string for Claude.
        
    Returns:
        The extracted generated text response.
    """
    try:
        response = client.messages.create(
            model='claude-sonnet-4-5-20250929',
            max_tokens=1000,
            messages=[
                {
                    "role": "user",
                    "content": prompt,
                }
            ]
        )
        return response.content[0].text
    except Exception as e:
        return f"Error executing API request for prompt: {e}"


# Define prompts with varying detail
prompts = [
    "A poem about the sea",
    "A detailed article on climate change",
]

# Generate content for each prompt
print("Generating content for each prompt...\n")
print("=" * 60)

# Iterate through prompts using the reusable generate_content helper
for prompt in prompts:
    generated_content = generate_content(prompt)
    
    # Display results with clean structural formatting
    print(f"PROMPT: {prompt}\n")
    print("OUTPUT:")
    print(generated_content)
    print("=" * 60)

# Concluding message
print("\nREFACTORING BENEFIT:")
print("Using a dedicated helper function encapsulates the API request logic, simplifies error handling, and eliminates redundant code across prompt executions.")

```

Here is the complete, robust implementation equipped with structured `try/except` exception handling, validation checks for empty responses, and clear output formatting for both valid and malformed prompts:

```python
import os
import anthropic
from anthropic import Anthropic

# Initialize the Anthropic client
client = Anthropic(api_key=os.environ.get("ANTHROPIC_API_KEY"))


def generate_content(prompt):
    """
    Generate content using Claude based on the given prompt with robust error handling.
    
    Args:
        prompt (str): The prompt to send to Claude
        
    Returns:
        str: The generated content or a descriptive error message
    """
    # Guard clause to handle empty or invalid input string upfront
    if not prompt or not isinstance(prompt, str) or not prompt.strip():
        return "ERROR: The prompt provided is empty or invalid."

    try:
        response = client.messages.create(
            model='claude-sonnet-4-5-20250929',
            max_tokens=1000,
            messages=[
                {
                    "role": "user",
                    "content": prompt,
                }
            ]
        )
        
        # Validate that the response contains valid text blocks
        if not response or not response.content or len(response.content) == 0:
            return "ERROR: Received an empty or unexpected response payload from Claude API."
            
        return response.content[0].text

    except anthropic.APIError as e:
        # Handles SDK-specific API exceptions (e.g., rate limits, invalid requests, authentication failure)
        return f"API ERROR ({e.status_code}): {e.message}"
    except Exception as e:
        # Fallback for unexpected python runtime errors (e.g., network disconnects, timeouts)
        return f"UNEXPECTED ERROR: {str(e)}"


# Define prompts including a malformed/empty test case to trigger error handling
prompts = [
    "A humorous blog post about cats",
    "A technical guide on setting up a web server",
    "",  # Malformed test case: empty string
]

# Generate content for each prompt
print("Generating content for each prompt...\n")
print("=" * 60)

for prompt in prompts:
    # Execute API call via our safe wrapper function
    generated_content = generate_content(prompt)
    
    # Display prompt status
    if not prompt.strip():
        print("PROMPT: [EMPTY OR MALFORMED PROMPT DETECTED]\n")
    else:
        print(f"PROMPT: {prompt}\n")
        
    print("RESULT:")
    print(generated_content)
    print("=" * 60)

# Concluding summary on error handling benefits
print("\nBENEFITS OF ERROR HANDLING:")
print("1. Graceful Resilience: Prevents the entire execution pipeline from crashing when a single request fails.")
print("2. Improved UX & Debugging: Provides explicit error messages indicating whether an issue is caused by empty inputs, network timeouts, or API limits.")

```

## Creating the StoryManager Class

Let's start by defining the structure for our Story Manager. This component will be responsible for managing our stories:

```python
class StoryManager:
    def __init__(self):
        self.stories = []

```

Here, we define a class named `StoryManager`. Inside this class, we set up a collection called `stories` that will hold our generated stories. This collection is initialized as empty, ready to store stories as they are created.

> **Note:** This implementation keeps stories in memory only during the program’s execution. Once the program stops, the stored stories are lost. For long-term persistence, we’ll later need to store them in a database or file.

---

## Adding Stories to the Manager

Next, we need a way to add stories to our manager. Let's create a method that allows us to do this:

```python
def add_story(self, prompt, story):
    self.stories.append({"id": len(self.stories), "prompt": prompt, "story": story})
    return self.stories[-1]

```

The `add_story` method takes two pieces of information: the `prompt` and the `story` itself. It creates a new entry containing a unique identifier (based on the current number of stories), the prompt, and the story. This entry is added to the collection of stories. The method then returns the newly added story, allowing us to confirm that it was successfully stored.

---

## Retrieving Stories

To make our Story Manager useful, we need to be able to retrieve stories. Let's implement methods for this purpose:

```python
def get_stories(self):
    return self.stories

def get_story_by_id(self, id):
    return self.stories[id] if 0 <= id < len(self.stories) else None

```

* The `get_stories` method returns the entire collection of stories, allowing us to access all stored stories at once.
* The `get_story_by_id` method retrieves a specific story using its unique identifier. This is useful when we want to access a particular story without going through the entire collection.

# StoryManager with Keyword Search - Full Code

## story_manager.py

```python
# story_manager.py
class StoryManager:
    def __init__(self):
        self.stories = []  # List to store generated stories
    
    def add_story(self, prompt, story):
        """Save story and return the id"""
        self.stories.append({"id": len(self.stories), "prompt": prompt, "story": story})
        return self.stories[-1]
    
    def get_stories(self):
        """Retrieve all generated stories."""
        return self.stories
    
    def get_story_by_id(self, id):
        """Retrieve a story by id."""
        if id < 0 or id >= len(self.stories):
            return {"error": f"Story with ID {id} not found. Valid IDs range from 0 to {len(self.stories)-1 if self.stories else 0}."}
        return self.stories[id]
    
    def search_story_by_keyword(self, keyword):
        """
        Search for stories containing the specified keyword.
        Returns a list of stories where the keyword appears in either prompt or story text.
        Search is case-insensitive.
        """
        # Convert keyword to lowercase for case-insensitive search
        search_term = keyword.lower()
        matching_stories = []
        
        for story in self.stories:
            # Check if keyword appears in prompt or story text (case-insensitive)
            if search_term in story["prompt"].lower() or search_term in story["story"].lower():
                matching_stories.append(story)
        
        return matching_stories
```

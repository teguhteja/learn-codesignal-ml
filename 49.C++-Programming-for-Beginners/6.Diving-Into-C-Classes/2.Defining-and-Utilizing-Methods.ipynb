## Defining and Utilizing Methods

Welcome! Today, we're diving into an essential topic in C++ programming: **defining and utilizing methods**. Methods, also known as **member functions**, are functions inside a class that can access its members. By the end of this lesson, you'll know how to define methods and call these methods to perform operations.

---

### Importance of Methods

Why do we need methods? Imagine you have a robot. This robot can do tasks like walking, talking, and picking up items. In programming, these tasks are **methods** that belong to the robot class. Methods help organize our code, making it easier to manage and understand.

---

### Defining a Method

Let's define a simple class with a method. In C++, methods are defined in a class.

Here's an example:

```cpp
#include <iostream>

class Robot {
public:
    // Method to make the robot talk
    void talk() {
        std::cout << "Hello, I am a robot!" << std::endl;  // Output: Hello, I am a robot!
    }
};
```

Here, we have a `Robot` class with a `talk()` method that prints a message to the console.

---

### Implementing Methods Outside the Class

To keep the class definition clean, methods are usually implemented outside the class using the **scope resolution operator** `::`.

Here's an example:

```cpp
#include <iostream>

class Robot {
public:
    void talk();
};

void Robot::talk() {
    std::cout << "Hello, I am a robot!" << std::endl;  // Output: Hello, I am a robot!
}
```

In this example, the `talk()` method is **declared** inside the class but **implemented** outside, keeping the definition simpler.

| Style | Where the body lives | Syntax |
|-------|----------------------|--------|
| **Inside the class** | In the class braces | `void talk() { ... }` |
| **Outside the class** | After the class | `void Robot::talk() { ... }` |

---

### Example Code Walkthrough: Part 1

You can also add **parameters** to the methods. Let's create a `Calculator` class with a method to add numbers.

```cpp
#include <iostream>

class Calculator {
public:
    // Method to add two numbers
    int add(int a, int b) {
        return a + b;
    }
};
```

In this code, we define a `Calculator` class with an `add` method.

---

### Example Code Walkthrough: Part 2

Let's add more functionality to the `Calculator` class by including a `subtract` method.

```cpp
#include <iostream>

class Calculator {
public:
    // Method to add two numbers
    int add(int a, int b) {
        return a + b;
    }

    // Method to subtract two numbers
    int subtract(int a, int b) {
        return a - b;
    }
};
```

---

### Utilizing a Method

To use a method, **create an instance** of the class and **call the method** on it with the dot operator. Let's see an example:

```cpp
int main() {
    Calculator calc;
    int result1 = calc.add(3, 5);
    int result2 = calc.subtract(10, 4);

    std::cout << "Addition: " << result1 << std::endl;  // Output: Addition: 8
    std::cout << "Subtraction: " << result2 << std::endl;  // Output: Subtraction: 6

    return 0;
}
```

Here, we create an instance of our `Calculator` class. Then, we utilize its methods `add` and `subtract` to get results. As you can see, these methods work just like functions, but they are part of the class, which makes the code more structured.

---

### Lesson Summary

| Concept | Key Point |
|---------|-----------|
| **Method** | A function defined inside a class (also called a member function) |
| **Defining inside** | Write the full body within the class braces |
| **Defining outside** | Declare in the class, then implement with `ReturnType ClassName::method() { ... }` |
| **Parameters / return** | Methods take parameters and return values just like ordinary functions |
| **Calling** | Create an instance, then use dot notation: `object.method(args)` |

## Modify the Launch Method in Spaceship Class

Space Explorer, let's improve the `Spaceship` class by making a small change. Move the implementation of the `launch` method outside the class definition. This will help you become comfortable with defining and using methods outside the class!

---

### Starter Code

```cpp
#include <iostream>

class Spaceship {
public:
    void launch() {
        std::cout << "Spaceship is launching!" << std::endl;
    }
};

int main() {
    // Create an object
    Spaceship myShip;
    // Call the method
    myShip.launch();
    return 0;
}
```

---

### Building the Solution

- **Keep only the declaration inside the class:** replace the method body with a semicolon: `void launch();`.
- **Implement it outside the class** with the scope resolution operator: `void Spaceship::launch() { ... }`. The `Spaceship::` prefix tells the compiler this function belongs to the `Spaceship` class.
- **Move the body unchanged:** the `std::cout` line goes inside the new definition.
- **`main` stays the same:** calling `myShip.launch()` works exactly as before, because only where the body is written has changed.

---

### Completed Code

```cpp
#include <iostream>

class Spaceship {
public:
    void launch();
};

void Spaceship::launch() {
    std::cout << "Spaceship is launching!" << std::endl;
}

int main() {
    // Create an object
    Spaceship myShip;
    // Call the method
    myShip.launch();
    return 0;
}
```

---

### Output

```
Spaceship is launching!
```

---

### Output Trace

| Step | What happens |
|------|--------------|
| `Spaceship myShip;` | creates an object of the `Spaceship` class |
| `myShip.launch();` | calls the method declared in the class |
| `Spaceship::launch()` runs | the definition outside the class prints the message |
| Output | `Spaceship is launching!` |

> **Key takeaway:** Splitting a method into a declaration inside the class and a definition outside it changes nothing about how it behaves or how you call it. The only requirement is the `ClassName::` prefix on the definition, which is how the compiler links the outside body back to the class.

---

### Starter Code

```cpp
#include <iostream>

class Calculator {
public:
    // Method to add two numbers
    int add(int a, int b) {
        return a + b;
    }

    // Method to subtract two numbers
    int subtract(int a, int b) {
        return a - b;
    }

    // Add method prototypes for multiply and divide here
};

int main() {
    Calculator calc;
    int result1 = calc.add(3, 5);
    int result2 = calc.subtract(10, 4);

    // Declare variables for the results of the new methods
    int result3;
    int result4;

    // Calculate and store results for multiply and divide methods
    result3 = calc.multiply(3, 5);  // Define multiply method in the class
    result4 = calc.divide(10, 2);   // Define divide method in the class

    std::cout << "Addition: " << result1 << std::endl;        // Output: Addition: 8
    std::cout << "Subtraction: " << result2 << std::endl;     // Output: Subtraction: 6

    // Print results for the new methods
    std::cout << "Multiplication: " << result3 << std::endl;  // Should output: Multiplication: 15
    std::cout << "Division: " << result4 << std::endl;        // Should output: Division: 5

    return 0;
}
```

---

### Building the Solution

- **Add the prototypes inside the class** where the comment says so: `int multiply(int a, int b);` and `int divide(int a, int b);`. A prototype is the method signature ending in `;`, with no body.
- **Implement them outside the class** with the scope resolution operator, as in the lesson:
  - `int Calculator::multiply(int a, int b) { return a * b; }`
  - `int Calculator::divide(int a, int b) { return a / b; }`
- **`main` already calls both methods**, so once they exist, it compiles and prints the results. With `int` operands, `/` is integer division, which is fine here because `10 / 2 = 5` exactly.

---

### Completed Code

```cpp
#include <iostream>

class Calculator {
public:
    // Method to add two numbers
    int add(int a, int b) {
        return a + b;
    }

    // Method to subtract two numbers
    int subtract(int a, int b) {
        return a - b;
    }

    // Method prototypes for multiply and divide
    int multiply(int a, int b);
    int divide(int a, int b);
};

// Method to multiply two numbers
int Calculator::multiply(int a, int b) {
    return a * b;
}

// Method to divide two numbers (b is never zero)
int Calculator::divide(int a, int b) {
    return a / b;
}

int main() {
    Calculator calc;
    int result1 = calc.add(3, 5);
    int result2 = calc.subtract(10, 4);

    // Declare variables for the results of the new methods
    int result3;
    int result4;

    // Calculate and store results for multiply and divide methods
    result3 = calc.multiply(3, 5);
    result4 = calc.divide(10, 2);

    std::cout << "Addition: " << result1 << std::endl;
    std::cout << "Subtraction: " << result2 << std::endl;

    // Print results for the new methods
    std::cout << "Multiplication: " << result3 << std::endl;
    std::cout << "Division: " << result4 << std::endl;

    return 0;
}
```

---

### Output

```
Addition: 8
Subtraction: 6
Multiplication: 15
Division: 5
```

---

### Output Trace

| Call | Computation | Result |
|------|-------------|--------|
| `calc.add(3, 5)` | `3 + 5` | `8` |
| `calc.subtract(10, 4)` | `10 - 4` | `6` |
| `calc.multiply(3, 5)` | `3 * 5` | `15` |
| `calc.divide(10, 2)` | `10 / 2` | `5` |

> **Key takeaway:** A class can mix both styles: short methods defined inside the class body and new ones declared as prototypes inside and implemented outside with `Calculator::`. Also remember that dividing two `int` values truncates the fraction (`7 / 2` gives `3`), which is why the task promises `b` is never zero and uses numbers that divide evenly.

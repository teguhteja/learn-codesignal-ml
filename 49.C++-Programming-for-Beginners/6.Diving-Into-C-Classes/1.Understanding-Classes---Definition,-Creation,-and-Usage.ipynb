## Understanding Classes - Definition, Creation, and Usage

Hello! Today, we're diving into an exciting topic: **classes** in C++. This lesson aims to introduce you to the concept of classes, how to create them, and how to use them in real-life scenarios. By the end of this lesson, you'll understand the definition of classes, how to create a class in C++, and how to use this class in your code. Ready? Let's get started!

---

### Definition of Classes

So, what exactly is a class? In simple terms, a **class** is like a blueprint or template for creating objects. Just like blueprints describe the structure and behavior of a house, a class describes the **attributes** (data members) and **behaviors** (member functions) of objects.

To put this in a real-life context, think of a "Car" as a class. It can have attributes like color, brand, and speed. The class is not some specific car; it is rather a description of what a car is and what information we know about a car.

```cpp
#include <string>

class Car {
public:
    // Attributes
    std::string color;
    std::string brand;
    int speed;
};
```

---

### Components of a Class

Classes in C++ are made up of **attributes** and **methods**. Attributes store information about the object, while methods define what actions the object can perform. We will learn about methods in the next lesson.

Here's our `Car` class example:

| Attribute | Type | Meaning |
|-----------|------|---------|
| `color` | `std::string` | the car's color |
| `brand` | `std::string` | the car's brand |
| `speed` | `int` | the car's current speed |

Understanding these components is crucial, as they form the basis of creating and using classes.

---

### Creating a Class

Creating a class in C++ is straightforward. Start with the `class` keyword, followed by the class name, and then its attributes and methods inside curly braces. Here's our example once again:

```cpp
#include <iostream>
#include <string>

class Car {
public:
    std::string color;
    std::string brand;
    int speed;
};
```

Note how we define car attributes the same way we define regular variables, but inside the class. Now these variables are parts of the class.

Pay attention to two things:

- The class declaration **always ends with `;`**.
- We start the class with the `public:` keyword. It is the **access modifier**, and we will discuss its meaning and other options in one of the following lessons. For now, always use `public:`.

---

### Using a Class

Once you've created a class, you can create objects from that class. **Objects** are instances of the class and can have unique values for the attributes defined by the class. In our example, `myCar` is an object of the `Car` class. We can set its attributes like `color`, `brand`, and `speed`.

```cpp
int main() {
    Car myCar; // Creating an object of class Car
    myCar.color = "Red";
    myCar.brand = "Toyota";
    myCar.speed = 0;

    // Working with class attributes
    myCar.speed += 10; // Increase speed
    std::cout << "Speed: " << myCar.speed << std::endl; // Print out the speed
}
```

We can work with class attributes just like we do with regular variables. The only difference is that the class attributes are a part of an object, accessed with the **dot operator** (`.`). In the example above, we increased the speed of `myCar` by 10 and then printed it out using `cout`.

---

### Creating Multiple Objects

Creating multiple objects from the same class is straightforward. Each object can have its unique values for the attributes defined in the class.

Let's extend our `Car` example by creating two different `Car` objects:

```cpp
int main() {
    Car myCar; // First object of class Car
    myCar.color = "Red";
    myCar.brand = "Toyota";
    myCar.speed = 0;

    Car yourCar; // Second object of class Car
    yourCar.color = "Blue";
    yourCar.brand = "Honda";
    yourCar.speed = 20;

    // Working with class attributes
    myCar.speed += 10; // Increase speed of myCar
    yourCar.speed += 30; // Increase speed of yourCar

    std::cout << "MyCar - Brand: " << myCar.brand << ", Color: " << myCar.color << ", Speed: " << myCar.speed << std::endl;
    std::cout << "YourCar - Brand: " << yourCar.brand << ", Color: " << yourCar.color << ", Speed: " << yourCar.speed << std::endl;
}
```

In this example, `myCar` and `yourCar` are two different objects created from the `Car` class. Each object has its attributes set to different values. We can interact with these objects independently.

Running the code would result in:

```
MyCar - Brand: Toyota, Color: Red, Speed: 10
YourCar - Brand: Honda, Color: Blue, Speed: 50
```

This demonstrates how classes allow us to create multiple objects with unique attributes, enabling flexible and scalable code design.

### Lesson Summary

| Concept | Key Point |
|---------|-----------|
| **Class** | A blueprint/template that describes the attributes and behaviors of objects |
| **Attribute** | A variable declared inside a class that stores information about an object |
| **Method** | A function inside a class that defines an action the object can perform (covered next lesson) |
| **Object** | An instance of a class, with its own values for the attributes |
| **Syntax** | `class Name { public: ... };` — note the semicolon at the end |
| **Dot operator** | `object.attribute` accesses or modifies an object's attribute |

## Creating and Using a Car Class in C++

Have you ever wondered how the attributes of cars are managed in a garage system? In this task, we'll simulate creating a car object, setting its attributes, and displaying them.

The code creates the `myFirstCar` object from the `Car` class. It sets its color to `"Blue"`, brand to `"Ford"`, and speed to `50`. Finally, the code prints out the details of `myFirstCar` to check if the attributes are set correctly.

Click Run to see it in action!

---

### Code

```cpp
#include <iostream>
#include <string>

class Car {
public:
    std::string color;
    std::string brand;
    int speed;
};

int main() {
    Car myFirstCar; // Creating an object of class Car

    // Setting Parameters
    myFirstCar.color = "Blue";
    myFirstCar.brand = "Ford";
    myFirstCar.speed = 50;

    // Printing Information
    std::cout << "Brand: " << myFirstCar.brand << ", Color: " << myFirstCar.color << ", Speed: " << myFirstCar.speed << std::endl;
}
```

---

### How It Works

- **`class Car { public: ... };`** — defines the blueprint with three attributes: `color`, `brand`, and `speed`. The class ends with a semicolon.
- **`Car myFirstCar;`** — creates an **object** (an instance) of the `Car` class.
- **`myFirstCar.color = "Blue";`** etc. — uses the dot operator to set each attribute on this particular object.
- **`std::cout << ...`** — reads the attributes back through the same dot operator to confirm they were stored correctly.

---

### Output

```
Brand: Ford, Color: Blue, Speed: 50
```

---

### Output Trace

| Statement | `myFirstCar.color` | `myFirstCar.brand` | `myFirstCar.speed` |
|-----------|----------------------|----------------------|----------------------|
| `Car myFirstCar;` | *(empty string)* | *(empty string)* | *(uninitialized)* |
| `myFirstCar.color = "Blue";` | `"Blue"` | *(empty)* | *(uninitialized)* |
| `myFirstCar.brand = "Ford";` | `"Blue"` | `"Ford"` | *(uninitialized)* |
| `myFirstCar.speed = 50;` | `"Blue"` | `"Ford"` | `50` |

> **Key takeaway:** A class defines the attributes; an object holds the actual values. After `Car myFirstCar;` the `int` attribute holds no meaningful value until you assign one, so set every attribute (via the dot operator) before reading it.

### Building the Solution

- The starter code already assigns `"Blue"`, `"Ford"`, and `50` to `car1`. The task is to replace them with **different** values describing a new car.
- Assign new values with the dot operator, exactly as before: `car1.color = ...;`, `car1.brand = ...;`, `car1.speed = ...;`.

---

### Completed Code

```cpp
#include <iostream>
#include <string>

class Car {
public:
    std::string color;
    std::string brand;
    int speed;
};

int main() {
    Car car1;
    car1.color = "Red";
    car1.brand = "Toyota";
    car1.speed = 80;

    std::cout << "Brand: " << car1.brand << std::endl;
    std::cout << "Color: " << car1.color << std::endl;
    std::cout << "Speed: " << car1.speed << std::endl;
}
```

---

> **Key takeaway:** An object's attributes are ordinary variables reached through the dot operator, so you can overwrite them at any time. The `cout` statements simply read the current values, which is why they print the new car after the assignments change.

## Encapsulation in C++

Hello! Today, we're learning about **encapsulation** in C++. Encapsulation bundles data and the methods that operate on that data into a single unit called a **class**. It's used to protect our data from access or modifications from outside the class.

Our goal is to understand encapsulation, how to implement it using `private` and `public` members in a class, and why it helps keep our code safe and organized. We'll use an example: solving a quadratic equation and learn to protect critical values.

---

### Understanding Encapsulation

Encapsulation is a core principle of object-oriented programming. It combines data and methods into one unit, called a class. This helps us **hide the internal state** of the object and only expose what's necessary.

Think of encapsulation like a pill capsule. Inside the capsule, there are different ingredients (data), but when you swallow it, you don't interact with these ingredients directly. Similarly, encapsulation helps control how data is accessed and modified.

In C++, we use `private` and `public` **access specifiers** to implement encapsulation. **Private** members are variables and methods that cannot be accessed directly from outside the class. We use them to hide the internal state of the object.

---

### The Private Access Specifier

```cpp
#include <iostream>

class Example {
private:
    int hiddenData;
};

int main() {
    Example obj;
    obj.hiddenData = 10;  // Will cause an error, because hiddenData is private!
    return 0;
}
```

In this example, `hiddenData` is a private member of the `Example` class. It is not accessible directly outside the class, and trying to access `obj.hiddenData` will result in a **compilation error**.

---

### Step-by-Step Code Walkthrough: Part 1

Let's go through the full code example to see how encapsulation is implemented to solve a quadratic equation.

```cpp
#include <iostream>
#include <cmath>
#include <stdexcept>  // For invalid_argument

class QuadraticEquation {
private:
    int a;
    int b;
    int c;

public:
    // Constructor
    QuadraticEquation(int a, int b, int c) {
        if (a != 0) {
            this->a = a;
        } else {
            throw std::invalid_argument("Coefficient a cannot be zero.");
        }
        this->b = b;
        this->c = c;
    }
```

Notice that we have declared `a`, `b`, and `c` as **private** members. This is the essence of encapsulation; we're hiding these members from direct access outside the class. Importantly, to protect our class from setting `a` to zero (which would result in a division by zero during calculation), we make `a` private. By doing this, we prevent unintended modifications from external code. Though `b` and `c` can take any values and don't require special protection, it is common to make all the class attributes private to ensure we can control access to them in the future.

---

### Step-by-Step Code Walkthrough: Part 2

But how do we set and get the attribute if it's private? Let's take a look at this code snippet to see:

```cpp
    // Public method to set the value of a
    void setA(int a) {
        if (a != 0) {
            this->a = a;
        } else {
            throw std::invalid_argument("Coefficient a cannot be zero.");
        }
    }

    // Public method to get the value of a should be const as it does not modify anything
    int getA() const {
        return a;
    }
```

We then provide public **getter** and **setter** methods for `a`. This allows controlled access and modification:

| Method | Role | Notes |
|--------|------|-------|
| **Getter** (`getA`) | Allows external code to read the value of `a` | Marked `const` because it must not modify the class fields; `const` guarantees the method won't change the object's parameters |
| **Setter** (`setA`) | Allows external code to modify the value of `a` | Only accepts a non-zero value, which validates the input and prevents invalid scenarios like division by zero |

Similarly, we can define methods `getB`, `setB`, `getC`, `setC` to access other fields.

---

### Step-by-Step Code Walkthrough: Part 3

Finally, we define the `solve` method that provides an answer:

```cpp
    // Public method to solve the quadratic equation should be const as it does not modify anything
    void solve() const {
        double discriminant = b*b - 4*a*c;
        if (discriminant > 0) {
            std::cout << "Two Real Roots: " << (-b + std::sqrt(discriminant)) / (2*a) << " and " << (-b - std::sqrt(discriminant)) / (2*a) << std::endl;
        } else if (discriminant == 0) {
            std::cout << "One Real Root: " << -b / (2*a) << std::endl;
        } else {
            std::cout << "No Real Roots" << std::endl;
        }
    }
};
```

Here, we can be sure that `a` will never equal zero, and the division by zero won't occur.

---

### Step-by-Step Code Walkthrough: Part 4

```cpp
int main() {
    // Creating an object of QuadraticEquation
    QuadraticEquation equation(1, -3, 2);
    equation.solve();  // Output: Two Real Roots: 2 and 1

    // Trying to update a
    equation.setA(2);
    equation.solve();  // Output: No Real Roots

    return 0;
}
```

In the `main` function, we create an object of `QuadraticEquation` with initial values `a = 1`, `b = -3`, and `c = 2`. Then, we call the `solve` method to output the roots. We demonstrate modifying `a` through the setter method, ensuring the update follows our rule of `a` being non-zero.

---

### Benefits of Encapsulation

Encapsulation offers several benefits:

| Benefit | Meaning |
|---------|---------|
| **Data Security** | Protects data from unintended modifications |
| **Code Maintainability** | Makes code easier to manage and understand; we can change the internal implementation without affecting the rest |
| **Controlled Modification** | Ensures changes to data only happen through controlled methods |

---

### Lesson Summary

| Concept | Key Point |
|---------|-----------|
| **Encapsulation** | Bundles data and methods into one class and hides the internal state |
| **`private`** | Members accessible only inside the class; outside access is a compilation error |

| **Getter** | Public `const` method that reads a private field |

Great job! We've learned that encapsulation bundles data and methods into a single unit (a class) while hiding the internal state from direct access. By using private and public members, we control how data is accessed and modified.

In our quadratic equation example, encapsulation protected the coefficient `a`, ensuring it wasn't set to zero.

### Starter Code

```cpp
#include <iostream>

class Box {
private:
    double length;
    double width;
    double height;

public:
    // Constructor
    Box(double l, double w, double h) : length(l), width(w), height(h) {}

    // Getter methods
    double getLength() const { return length; }
    double getWidth() const { return width; }
    double getHeight() const { return height; }

    // Setter methods
    void setLength(double l) { length = l; }
    void setWidth(double w) { width = w; }
    void setHeight(double h) { height = h; }
};

int main() {
    Box myBox(5.0, 3.0, 4.0);
    std::cout << "Box dimensions are: "
              << myBox.getLength() << " x "
              << myBox.getWidth() << " x "
              << myBox.getHeight() << std::endl;

    // Print box volume here using a new method
    return 0;
}
```

---

### Building the Solution

- **Add the method inside the class**, in the `public:` section: `double calculateVolume() const { return length * width * height; }`.
- **Return type `double`**, because all three dimensions are `double`.
- **Mark it `const`**: it only reads the private fields and never changes them, so `const` after the parameter list promises the object stays unmodified (the same reason the getters are `const`).
- **Use the private fields directly.** The method belongs to the class, so it can read `length`, `width`, and `height` without the getters.
- **Call it in `main`** and print the result: `myBox.calculateVolume()`. With `5.0 * 3.0 * 4.0 = 60`.

### Completed Code

```cpp
#include <iostream>

class Box {
private:
    double length;
    double width;
    double height;

public:
    // Constructor
    Box(double l, double w, double h) : length(l), width(w), height(h) {}

    // Getter methods
    double getLength() const { return length; }
    double getWidth() const { return width; }
    double getHeight() const { return height; }

    // Setter methods
    void setLength(double l) { length = l; }
    void setWidth(double w) { width = w; }
    void setHeight(double h) { height = h; }

    // Method to calculate the volume of the box
    double calculateVolume() const {
        return length * width * height;
    }
};

int main() {
    Box myBox(5.0, 3.0, 4.0);
    std::cout << "Box dimensions are: "
              << myBox.getLength() << " x "
              << myBox.getWidth() << " x "
              << myBox.getHeight() << std::endl;

    // Print box volume here using a new method
    std::cout << "Box volume is: " << myBox.calculateVolume() << std::endl;
    return 0;
}
```

## Encapsulated Circle Area Calculation

Hello, Space Explorer!

Here is a code snippet that calculates the area of a circle using encapsulation in a class. Unfortunately, there is a bug. Identify and fix the error to ensure the code runs correctly and the area is calculated.

Good luck!

---

### Buggy Code

```cpp
#include <iostream>
#include <stdexcept>

class Circle {
private:
    double radius;

public:
    Circle(double r) {
        if (r > 0) {
            radius = r;
        } else {
            throw std::invalid_argument("Radius must be positive.");
        }
    }

    double getRadius() const {
        return radius;
    }

    void setRadius(double r) {
        if (r > 0) {
            radius = r;
        } else {
            throw std::invalid_argument("Radius must be positive.");
        }
    }

    double calculateArea() const {
        return 3.14159 * radius * radius;
    }
};

int main() {
    Circle circle(5.0);
    std::cout << "Area: " << circle.radius << std::endl;
    return 0;
}
```

**Buggy behavior:** the program does not compile. `main` reads `circle.radius` directly, but `radius` is `private`, so the compiler reports an access error. Even if it were accessible, the line is labeled `Area` but would print the radius instead of the area.

---

### Finding the Bug

- **Bug: wrong access in `main`.** `circle.radius` reaches into a private field, which encapsulation forbids. It also prints the wrong value: the radius (`5`), not the area.
- **Fix:** call the public method that computes the area: `circle.calculateArea()`.
- **Nothing else changes.** The class is already correct: `radius` is private, the constructor and setter validate that the radius is positive, and `calculateArea()` is a `const` method that returns `3.14159 * radius * radius`.

---

### Fixed Code

```cpp
#include <iostream>
#include <stdexcept>

class Circle {
private:
    double radius;

public:
    Circle(double r) {
        if (r > 0) {
            radius = r;
        } else {
            throw std::invalid_argument("Radius must be positive.");
        }
    }

    double getRadius() const {
        return radius;
    }

    void setRadius(double r) {
        if (r > 0) {
            radius = r;
        } else {
            throw std::invalid_argument("Radius must be positive.");
        }
    }

    double calculateArea() const {
        return 3.14159 * radius * radius;
    }
};

int main() {
    Circle circle(5.0);
    std::cout << "Area: " << circle.calculateArea() << std::endl;
    return 0;
}
```

---

```
Area: 78.5398
```

|------|-------------|-------|
| `Circle circle(5.0);` | `5.0 > 0`, so `radius = 5.0` | `5.0` |
| `circle.calculateArea()` | `3.14159 * 5.0 * 5.0` | `78.53975` |

> **Key takeaway:** With a `private` field, code outside the class must go through public methods, and the compiler enforces it. ### Building the Solution

- **Setters validate, then store.** `setLength` checks `len > 0`. If true, it stores `length = len;`, otherwise it throws `std::invalid_argument` (the header `<stdexcept>` is already included). `setWidth` does the same for `width`.
- **Getters just return the field:** `return length;` and `return width;`. They are `const` because they don't change the object.

### Completed Code

```cpp
#include <iostream>
#include <stdexcept>

class Rectangle {
private:
    double length;
    double width;

public:
    // Constructor
    Rectangle(double length, double width) {
        setLength(length);
        setWidth(width);
    }

    // Setter for length
    void setLength(double len) {
        if (len > 0) {
            length = len;
        } else {
            throw std::invalid_argument("Length must be positive.");
        }
    }

    // Getter for length
    double getLength() const {
        return length;
    }

    // Setter for width
    void setWidth(double wid) {
        if (wid > 0) {
            width = wid;
        } else {
            throw std::invalid_argument("Width must be positive.");
        }
    }

    // Getter for width
    double getWidth() const {
        return width;
    }

    // Method to calculate area
    double area() const {
        return length * width;
    }
};

int main() {
    // Creating an object of Rectangle
    Rectangle rect(5, 3);

    // Output the length, width, and area
    std::cout << "Length: " << rect.getLength() << std::endl;
    std::cout << "Width: " << rect.getWidth() << std::endl;
    std::cout << "Area: " << rect.area() << std::endl;

    // Modifying length and width
    rect.setLength(10);
    rect.setWidth(4);
    std::cout << "Updated Length: " << rect.getLength() << std::endl;
    std::cout << "Updated Width: " << rect.getWidth() << std::endl;
    std::cout << "Updated Area: " << rect.area() << std::endl;

    return 0;
}
```

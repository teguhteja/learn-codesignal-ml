## Constructors in C++ Classes

Welcome to the lesson on **Constructors** in C++ classes!

Have you ever wondered how objects in your code are created and initialized? This is where constructors come into play. Constructors are special functions that are **automatically called** when an object of a class is created. They help set up the object with initial values and are essential for managing resources efficiently. In this lesson, our goal is to understand different types of constructors in C++: **default**, **parameterized**, and **copy** constructors, and learn how to implement them in a class.

---

### Understanding Constructors

So, what exactly is a constructor? A **constructor** is a special member function of a class that initializes objects. It has the **same name as the class** and does **not have a return type**, not even `void`.

Think of a constructor like setting up a new house. When you move in, you need to set up the furniture before you can comfortably live in it. Similarly, when an object is created, a constructor sets up the initial state of the object.

---

### Example of a Default Constructor

A **default constructor** is a constructor that either has no parameters or has parameters with default values. Its primary purpose is to initialize objects with default settings.

Let's see how to write a default constructor using an example:

```cpp
#include <iostream>

class MyClass {
public:
    // Default constructor
    MyClass() {
        std::cout << "Default constructor called!" << std::endl;
    }
};

int main() {
    MyClass obj; // Default constructor is called here
    return 0;
}
```

In this example, when we create an object `obj` of `MyClass`, the default constructor is called, displaying the message `"Default constructor called!"`.

---

### Parameterized Constructor

**Parameterized constructors** help when you need to initialize objects with specific values. These constructors accept arguments to initialize an object with provided values.

Let's extend our previous example:

```cpp
#include <iostream>

class MyClass {
public:
    int value;

    // Default constructor
    MyClass() {
        value = 0;
        std::cout << "Default constructor called!" << std::endl;
    }

    // Parameterized constructor
    MyClass(int param) {
        value = param;
        std::cout << "Parameterized constructor called! Value: " << value << std::endl;
    }
};

int main() {
    MyClass obj1; // Default constructor is called
    MyClass obj2(100); // Parameterized constructor is called
    return 0;
}
```

Here, `obj1` uses the default constructor, and `obj2` uses the parameterized constructor to initialize `value` to `100`.

---

### The `this` Keyword

In C++, `this` is a keyword that refers to the **current instance** of the class. Inside a class member function, `this` can be used to reference the calling object's members. This is particularly useful in constructors to avoid **naming conflicts** between parameters and class members.

Let's consider an example:

```cpp
#include <iostream>
#include <string>

class BankAccount {
public:
    std::string accountHolderName;
    int accountNumber;
    double balance;

    BankAccount(std::string param1, int param2, double param3) {
        accountHolderName = param1;
        accountNumber = param2;
        balance = param3;
    }
};

int main() {
    BankAccount acc("John Doe", 123456, 1000.50);
    std::cout << "Account Holder: " << acc.accountHolderName << ", Account Number: " << acc.accountNumber << ", Balance: " << acc.balance << std::endl;
    return 0;
}
```

Here, naming parameters `param1`, `param2` and `param3` is confusing and doesn't reflect the meaning of the variables. However, we can't call them `accountHolderName`, `accountNumber` and `balance`, because it will cause a naming conflict between the attributes and the parameters.

We can solve this problem using the `this` keyword:

```cpp
#include <iostream>
#include <string>

class BankAccount {
public:
    std::string accountHolderName;
    int accountNumber;
    double balance;

    // Parameterized constructor using 'this' keyword
    BankAccount(std::string accountHolderName, int accountNumber, double balance) {
        this->accountHolderName = accountHolderName;
        this->accountNumber = accountNumber;
        this->balance = balance;
    }
};

int main() {
    BankAccount acc("John Doe", 123456, 1000.50);
    std::cout << "Account Holder: " << acc.accountHolderName << ", Account Number: " << acc.accountNumber << ", Balance: " << acc.balance << std::endl;
    return 0;
}
```

In this example, `this->accountHolderName` specifies that we are referring to the **class member** `accountHolderName` instead of the constructor **parameter** `accountHolderName`. This way, we can call parameters the same way as the class members, making the code clearer.

---

### Copy Constructor

A **copy constructor** is used to initialize an object by copying another object of the same class. It is essential when you need to create a copy of an existing object.

Here's how a copy constructor works:

```cpp
#include <iostream>

class MyClass {
public:
    int value;

    // Default constructor
    MyClass() {
        value = 0;
        std::cout << "Default constructor called!" << std::endl;
    }

    // Parameterized constructor
    MyClass(int value) {
        this->value = value;
        std::cout << "Parameterized constructor called! Value: " << value << std::endl;
    }

    // Copy constructor
    MyClass(const MyClass &obj) {
        value = obj.value;
        std::cout << "Copy constructor called! Value: " << value << std::endl;
    }
};

int main() {
    MyClass obj1(100); // Parameterized constructor is called
    MyClass obj2 = obj1; // Copy constructor is called
    return 0;
}
```

Output:

```
Parameterized constructor called! Value: 100
Copy constructor called! Value: 100
```

In this example, `obj2` is created as a copy of `obj1` using the copy constructor. The value of `obj1` is copied to `obj2`.

---

### Defining Constructors Outside the Class

Just like member functions, constructors can also be defined outside the class. Here is an example:

```cpp
#include <iostream>

class Demo {
public:
    int value;
    Demo();
};

Demo::Demo() {
    value = 42;
    std::cout << "Constructor defined outside the class. Value: " << value << std::endl;
}

int main() {
    Demo demoObj;
    return 0;
}
```

In this example, the constructor for the `Demo` class is defined outside the class using the scope resolution operator `::`. When `demoObj` is created, the constructor is called and initializes `value` to `42`.

---

### Member Initializer List

C++ also allows constructors to initialize members using a **member initializer list**. Here's an example:

```cpp
#include <iostream>

class Wheel {
public:
    int size;

    // Constructor using member initializer list
    Wheel(int s) : size(s) {
        std::cout << "Wheel size: " << size << std::endl;
    }
};

int main() {
    Wheel myWheel(15); // Member initializer list sets the size
    return 0;
}
```

Output:

```
Wheel size: 15
```

In this example, `Wheel(int s) : size(s) {}` is a constructor that uses the member initializer list to assign the value of `s` to the member `size`. This can be more efficient and is often used to initialize constant members or base classes in derived classes.

---

### Comparison of Constructor Types

Let's quickly compare the three types of constructors:

| Type | Purpose | Parameters |
|------|---------|------------|
| **Default Constructor** | Initializes an object with default settings | None (or all have default values) |
| **Parameterized Constructor** | Initializes an object with specific values | Accepts arguments |
| **Copy Constructor** | Initializes an object by copying an existing object | A reference to an object of the same class (`const MyClass &obj`) |

Each type has its unique use case and is crucial for effective object initialization and management.

---

### Lesson Summary

| Concept | Key Point |
|---------|-----------|
| **Constructor** | Special member function with the class name and no return type, called automatically on object creation |
| **Default constructor** | Takes no parameters; sets default values |
| **Parameterized constructor** | Takes arguments to set specific initial values |
| **`this` keyword** | Points to the current object; `this->member` resolves name conflicts with parameters |
| **Copy constructor** | `ClassName(const ClassName &obj)`; builds a new object from an existing one |
| **Outside the class** | Declare inside, define with `ClassName::ClassName() { ... }` |
| **Member initializer list** | `Wheel(int s) : size(s) { }` initializes members before the body runs |

Congratulations! You've now learned about three different types of constructors in C++: default constructors, parameterized constructors, and copy constructors. Constructors play a vital role in ensuring that objects are set up correctly with initial values. They simplify the process of managing resources and initializing objects, making your code more efficient and robust.

```cpp
#include <iostream>
#include <string>

// Class representing a bank account
class BankAccount {
public:
    std::string accountHolderName;
    int accountNumber;
    double balance;

    // Parameterized constructor using this keyword
    BankAccount(std::string accountHolderName, int accountNumber, double balance) {
        this->accountHolderName = accountHolderName;
        this->accountNumber = accountNumber;
        this->balance = balance;
    }
};

int main() {
    // Create a BankAccount object using the parameterized constructor
    BankAccount myAccount("Alice", 12345, 1000.0);

    // Output the class fields
    std::cout << "Account Holder Name: " << myAccount.accountHolderName << std::endl;
    std::cout << "Account Number: " << myAccount.accountNumber << std::endl;
    std::cout << "Balance: " << myAccount.balance << std::endl;

    return 0;
}
```

---

### Output Trace

| Assignment | Left side (field) | Right side (parameter) | Value stored |
|------------|-------------------|------------------------|--------------|
| `this->accountHolderName = accountHolderName;` | `this->accountHolderName` | `accountHolderName` | `"Alice"` |
| `this->accountNumber = accountNumber;` | `this->accountNumber` | `accountNumber` | `12345` |
| `this->balance = balance;` | `this->balance` | `balance` | `1000.0` |

> **Key takeaway:** When a constructor parameter has the same name as a field, the bare name means the parameter, and `this->name` means the field. Writing `this->field = field;` is the standard way to keep parameter names meaningful. Without `this->`, a line like `accountNumber = accountNumber;` would just assign the parameter to itself and leave the field uninitialized. Also note that `cout` prints the `double` `1000.0` as `1000`.

## Implement the Copy Constructor for BankAccount Class

In this practice task, you need to implement the copy constructor for the `BankAccount` class. The copy constructor should initialize a new `BankAccount` object by copying an existing one. Follow the comments in the code to complete it.

---

### Starter Code

```cpp
#include <iostream>
#include <string>

class BankAccount {
public:
    std::string accountHolderName;
    int accountNumber;
    double balance;

    // Default constructor
    BankAccount() {
        this->accountHolderName = "Unknown";
        this->accountNumber = 0;
        this->balance = 0.0;
    }

    // Parameterized constructor
    BankAccount(std::string name, int number, double bal) {
        this->accountHolderName = name;
        this->accountNumber = number;
        this->balance = bal;
    }

    // TODO: Implement the copy constructor
    BankAccount(const BankAccount &account) {
        // Copy the accountHolderName, accountNumber, and balance from the given account
    }
};

int main() {
    BankAccount firstAccount("Alice", 12345, 1000.0); // Parameterized constructor is called
    BankAccount secondAccount = firstAccount; // Copy constructor is called

    // Print details to validate
    std::cout << "First Account -> Name: " << firstAccount.accountHolderName << ", Number: " << firstAccount.accountNumber << ", Balance: $" << firstAccount.balance << std::endl;
    std::cout << "Second Account -> Name: " << secondAccount.accountHolderName << ", Number: " << secondAccount.accountNumber << ", Balance: $" << secondAccount.balance << std::endl;

    return 0;
}
```

---

### Building the Solution

- **The signature is already there:** `BankAccount(const BankAccount &account)` takes the object to copy by **reference** (`&`) so no extra copy is made, and by `const` so the original cannot be changed.
- **Copy each field** from `account` into the new object, using `this->` for the new object and the dot operator for the source:
  - `this->accountHolderName = account.accountHolderName;`
  - `this->accountNumber = account.accountNumber;`
  - `this->balance = account.balance;`
- **Why it matters here:** `BankAccount secondAccount = firstAccount;` calls this constructor. With an empty body, the new object's `accountNumber` and `balance` would be uninitialized garbage.

---

### Completed Code

```cpp
#include <iostream>
#include <string>

class BankAccount {
public:
    std::string accountHolderName;
    int accountNumber;
    double balance;

    // Default constructor
    BankAccount() {
        this->accountHolderName = "Unknown";
        this->accountNumber = 0;
        this->balance = 0.0;
    }

    // Parameterized constructor
    BankAccount(std::string name, int number, double bal) {
        this->accountHolderName = name;
        this->accountNumber = number;
        this->balance = bal;
    }

    // Copy constructor
    BankAccount(const BankAccount &account) {
        this->accountHolderName = account.accountHolderName;
        this->accountNumber = account.accountNumber;
        this->balance = account.balance;
    }
};

int main() {
    BankAccount firstAccount("Alice", 12345, 1000.0); // Parameterized constructor is called
    BankAccount secondAccount = firstAccount; // Copy constructor is called

    // Print details to validate
    std::cout << "First Account -> Name: " << firstAccount.accountHolderName << ", Number: " << firstAccount.accountNumber << ", Balance: $" << firstAccount.balance << std::endl;
    std::cout << "Second Account -> Name: " << secondAccount.accountHolderName << ", Number: " << secondAccount.accountNumber << ", Balance: $" << secondAccount.balance << std::endl;

    return 0;
}
```

---

### Output

```
First Account -> Name: Alice, Number: 12345, Balance: $1000
Second Account -> Name: Alice, Number: 12345, Balance: $1000
```

---

### Output Trace

| Step | Constructor called | `accountHolderName` | `accountNumber` | `balance` |
|------|--------------------|----------------------|------------------|-----------|
| `firstAccount("Alice", 12345, 1000.0)` | parameterized | `"Alice"` | `12345` | `1000.0` |
| `secondAccount = firstAccount` | **copy** | `"Alice"` (copied) | `12345` (copied) | `1000.0` (copied) |

> **Key takeaway:** A copy constructor builds a new object from an existing one by copying every field, and it takes its argument as `const ClassName &` to avoid an endless chain of copies and to protect the original. The two objects end up with equal values but remain separate, so changing `secondAccount.balance` later would not affect `firstAccount`. Also note that `cout` prints the `double` `1000.0` as `1000`.

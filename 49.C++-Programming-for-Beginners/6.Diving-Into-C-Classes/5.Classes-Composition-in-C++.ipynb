## Classes Composition in C++

Welcome to the lesson on class composition in C++. You have now learned the basics of classes, methods, constructors, and encapsulation. Today, we are going to take a step further into object-oriented programming by exploring **class composition**. The goal of this lesson is to understand how to construct complex classes by combining simpler, reusable components.

Class composition is an essential design principle that enables you to create complex objects. We will learn how to build classes containing other objects and see practical applications of this approach.

---

### What is Composition?

A real-life analogy is a **car**. A car comprises various parts, such as an engine, wheels, and a transmission. These parts work together to make the car function.

Similarly, in programming, composition allows you to build a `Car` class that includes an `Engine` class, a `Wheel` class, and so on. This is often called a **"has-a" relationship**: a Car *has an* Engine.

---

### Basic Class Composition

Let's start with a simple example:

```cpp
#include <iostream>

class Engine {
public:
    void start() {
        std::cout << "Engine started";
    }
};

class Car {
private:
    Engine engine;  // Car has an Engine
public:
    void start() {
        engine.start();               // Start the engine first
        std::cout << " Car started";
    }
};

int main() {
    Car myCar;
    myCar.start();  // Outputs: Engine started Car started
    return 0;
}
```

In this example, the `Car` class contains an `Engine` object:

- The `Engine` class has a method called `start()`, which prints `"Engine started"`.
- The `Car` class has a **private data member** `engine` of type `Engine`.
- The `Car` class also has a `start()` method that calls the Engine's `start()` method before printing `"Car started"`.

When you create a `Car` object and call its `start()` method, it first starts the engine and then starts the car, showing how the `Car` uses functionality provided by the `Engine`.

---

### Advanced Class Composition: Part 1

Now let's look at a more advanced example, where a `Car` is composed of both `Engine` and `Wheel` objects:

```cpp
#include <iostream>

class Engine {
private:
    int horsepower;   // Engine has a horsepower attribute
public:
    Engine(int hp) : horsepower(hp) {}  // Constructor initializes horsepower
    int getHorsepower() const {
        return horsepower;
    }
};

class Wheel {
private:
    int size;  // Wheel has a size attribute
public:
    Wheel(int s) : size(s) {}  // Constructor initializes size
    int getSize() const {
        return size;
    }
};
```

Key points:

- The `Engine` class has a private attribute `horsepower` and a method `getHorsepower()` to retrieve it.
- The `Wheel` class has a private attribute `size` and a method `getSize()` to retrieve it.
- The `Car` class (next part) will have both `Engine` and `Wheel` objects.

---

### Advanced Class Composition: Part 2

Continuing from the previous snippet, we now construct the `Car` class:

```cpp
class Car {
private:
    Engine engine;
    Wheel wheels[4];  // Composition with multiple instances of a component
public:
    // Constructor initializes Engine and Wheel objects
    Car(int hp, int ws) : engine(hp), wheels{Wheel(ws), Wheel(ws), Wheel(ws), Wheel(ws)} {}

    void displaySpecifications() const {
        std::cout << "Engine: " << engine.getHorsepower() << " HP\n";
        std::cout << "Wheel size: " << wheels[0].getSize() << " inches\n";
    }
};

int main() {
    Car myCar(300, 18);
    myCar.displaySpecifications();  // Outputs: Engine: 300 HP \n Wheel size: 18 inches
    return 0;
}
```

In this example:

- The `Car` class contains an `Engine` object and an **array of four `Wheel` objects**.
- The constructor of `Car` initializes the `Engine` and `Wheel` objects using the **member initializer list**.
- The `displaySpecifications()` method prints the engine horsepower and wheel size.

This approach shows how you can aggregate multiple simpler objects to build a complex object like a `Car` that consists of an `Engine` and multiple `Wheel`s.

---

### Benefits of Class Composition

| Benefit | Description |
|---|---|
| **Reusability** | Reuse simple, well-defined classes to build larger, more complex classes without duplicating code. |
| **Maintainability** | Each class has a single responsibility, so it is easier to understand, test, and maintain. |
| **Flexibility** | Change the behavior of a complex object by modifying or substituting its components. |

Using our car example, suppose you want to upgrade the car's engine or change the wheel size. With composition, you can make these changes without affecting the other parts of the car.

---

### Lesson Summary

| Concept | Key Point |
|---|---|
| **Composition** | A class contains objects of other classes as members ("has-a" relationship). |
| **Component classes** | Small, focused classes such as `Engine` and `Wheel`. |
| **Member initializer list** | Initializes component objects that have no default constructor, e.g. `engine(hp)`. |
| **Arrays of components** | `Wheel wheels[4];` holds multiple instances of a component. |
| **Benefits** | Reusability, maintainability, and flexibility. |

## Class Composition with Car, Engine, and Wheel

### Code

```cpp
#include <iostream>

// Class to represent an Engine component
class Engine {
private:
    int horsepower; // Private member to store horsepower
public:
    // Constructor to initialize horsepower
    Engine(int hp) : horsepower(hp) {}
    
    // Getter function to access horsepower
    int getHorsepower() const {
        return horsepower;
    }
};

// Class to represent a Wheel component
class Wheel {
private:
    int size; // Private member to store wheel size
public:
    // Constructor to initialize wheel size
    Wheel(int s) : size(s) {}

    // Getter function to access wheel size
    int getSize() const {
        return size;
    }
};

// Class to represent a Car that contains an Engine and multiple Wheels
class Car {
private:
    Engine engine; // Engine object
    Wheel wheels[4];  // Array of 4 Wheel objects
public:
    // Constructor to initialize engine and wheels
    Car(int hp, int ws) : engine(hp), wheels{Wheel(ws), Wheel(ws), Wheel(ws), Wheel(ws)} {}

    // Function to display specifications of the car
    void displaySpecifications() const {
        std::cout << "Engine: " << engine.getHorsepower() << " HP\n";
        std::cout << "Wheel size: " << wheels[0].getSize() << " inches\n";
    }
};

int main() {
    // Creating a Car object with 300 horsepower engine and 18-inch wheels
    Car myCar(300, 18);
    // Display the specifications of the created car
    myCar.displaySpecifications();  // Outputs: Engine: 300 HP \n Wheel size: 18 inches
    return 0;
}
```

---

### How It Works

- **`Engine`** stores a private `horsepower` and exposes it through `getHorsepower()`.
- **`Wheel`** stores a private `size` and exposes it through `getSize()`.
- **`Car`** *has an* `Engine` and *has* an array of four `Wheel`s as private members.
- The `Car` constructor takes `hp` and `ws`, then uses the **member initializer list** to build each component: `engine(hp)` and four `Wheel(ws)` objects.
- `displaySpecifications()` reads from the components through their getters. All four wheels share the same size, so it prints only `wheels[0]`.
- `Engine` and `Wheel` have no default constructor, so the initializer list is **required**. The components can't be built any other way.

---

```
Engine: 300 HP
Wheel size: 18 inches
```

---

### Output Trace

|---|---|---|
| 1 | `Car myCar(300, 18);` | `engine` gets `horsepower = 300`, each of the 4 wheels gets `size = 18` |
| 2 | `myCar.displaySpecifications();` | Called on the `Car` |
| 3 | `engine.getHorsepower()` | Returns `300`, prints `Engine: 300 HP` |

> **Key takeaway:** Composition builds a complex object from simpler ones. `Car` doesn't repeat engine or wheel logic. It owns the components, initializes them in its constructor's initializer list, and delegates to their public getters.

### Code

```cpp
#include <iostream>

class Wheel {
public:
    void rotate() {
        std::cout << "Wheel rotating.\n";
    }
};

class Bike {
private:
    Wheel frontWheel;  // Bike has a front Wheel
    Wheel backWheel;   // Bike has a back Wheel
public:
    void ride() {
        frontWheel.rotate();
        backWheel.rotate();
        std::cout << "Bike is moving!";
    }
};

int main() {
    Bike myBike;
    myBike.ride(); 
    return 0;
}
```

---

### How It Works

- **`Wheel`** has one public method, `rotate()`, which prints `Wheel rotating.` followed by a newline.
- **`Bike`** has two private `Wheel` members, `frontWheel` and `backWheel`. This is composition: a Bike *has* two Wheels.
- `Wheel` has no constructor, so the compiler's default constructor builds both members automatically. No initializer list is needed.
- `Bike::ride()` calls `rotate()` on each wheel, then prints `Bike is moving!`.
- `main` never touches the wheels. It only calls `myBike.ride()`, and the Bike delegates the work to its components.

### Building the Solution

1. **Define `Point`.** Give it private `double x` and `double y`, plus a constructor `Point(double x, double y)` that uses `this->` to separate the parameters from the members.
2. **Add `getDistance`.** Take `const Point& other`. A reference avoids copying, and `const` promises not to modify it. Mark the method `const` too, since it doesn't change `this`.
3. **Apply the formula.** `x1` is `this->x` and `x2` is `other.x`. Compute `dx = other.x - this->x` and `dy = other.y - this->y`, then return `std::sqrt(dx * dx + dy * dy)` (from `<cmath>`).

### Completed Code

```cpp
#include <iostream>
#include <cmath>

class Point {
private:
    double x;
    double y;
public:
    Point(double x, double y) : x(x), y(y) {}

    double getDistance(const Point& other) const {
        double dx = other.x - this->x;
        double dy = other.y - this->y;
        return std::sqrt(dx * dx + dy * dy);
    }
};

int main() {
    Point p1(0, 0);
    Point p2(3, 4);
    std::cout << "Distance: " << p1.getDistance(p2) << std::endl;
    return 0;
}
```

## Navigating the World of C++ Strings

Welcome back to our C++ course! Today, we'll unravel the data structure known as **strings**. Strings are sequences of characters functioning like vectors, thus accommodating elements, enabling access to them at certain positions, modifying elements, and much more. Whenever you handle text data in your programs, strings are your go-to tool. Let's dive in!

Suppose you wish to store `"Happy Birthday Alice!"` on an invitation card in a program. This text can be stored as a string. Like a vector holding characters, a string specializes in holding text, highlighting the significance of textual information in programming.

---

### Declaring a String

Declaring a string is analogous to declaring an integer: you use the `string` keyword, followed by a string variable name. Let's declare a `msg` string for the invitation message!

```cpp
#include <iostream>
#include <string>  // Includes the string library to work with strings

int main() {
    std::string msg;
    std::cout << "The invitation message: " << msg << std::endl;  // Prints "The invitation message:"
    return 0;
}
```

We declared the string `msg`, which is currently empty, resulting in the output `"The invitation message:"`.

Strings in C++ must be surrounded by **double quotes**. Let's also declare and initialize a non-empty string:

```cpp
#include <iostream>
#include <string>  // Includes the string library to work with strings

int main() {
    std::string greeting = "Hello, World!";
    std::cout << greeting << std::endl;  // Prints "Hello, World!"
    return 0;
}
```

---

### Assigning a Value

We'll extend `msg`, crafting our party invitation using the assignment operator (`=`).

```cpp
#include <iostream>
#include <string>

int main() {
    std::string msg;
    msg = "Happy Birthday Alice!";
    std::cout << "The invitation message: " << msg << std::endl;  // Prints "The invitation message: Happy Birthday Alice!"
    return 0;
}
```

We've now populated `msg` to welcome Alice!

---

### Adding and Accessing Elements in a String

Fetching specific string elements is akin to vector elements, facilitated by the index within square brackets `[]`.

```cpp
#include <iostream>
#include <string>

int main() {
    std::string msg = "Happy Birthday Alice!";
    std::cout << "The first character of the string: " << msg[0] << std::endl;  // Prints "The first character of the string: H"
    return 0;
}
```

This time, we've retrieved the first character `'H'` from the invitation.

---

### Modifying Characters in a String

Did you accidentally invite "Alics"? Not a problem! Strings can modify characters just like vectors. Remember, the new element must be of type `char`, so it must be surrounded by **single quotes** (`'`).

```cpp
#include <iostream>
#include <string>

int main() {
    std::string msg = "Happy Birthday Alics!";
    msg[19] = 'e';  // Corrects the typo
    std::cout << "The invitation message: " << msg << std::endl;  // Prints "The invitation message: Happy Birthday Alice!"
    return 0;
}
```

Here, we've rectified the typo by changing the 20th character from `'s'` to `'e'`.

---

### Accessing the String's Length

Strings in C++ boast handy functions similar to vectors. One of them is `size()`, which is very helpful to learn the string's length. It returns the number of characters in the string.

```cpp
#include <iostream>
#include <string>

int main() {
    std::string msg = "Happy Birthday Alice!";
    std::cout << "Number of characters in msg: " << msg.size() << std::endl;  // Number of characters in msg: 21
    return 0;
}
```

Here, we find out the number of letters in the `msg` string using the `.size()` method. It works pretty much the same as the vector's analogous method.

---

### Exploring String Concatenation

We can use the `+` operator to append one string to another, combining them into a single new string. This is called **concatenation**. Here's how you can concatenate strings:

```cpp
#include <iostream>
#include <string>

int main() {
    std::string part1 = "Happy Birthday";
    std::string part2 = " Alice!";
    std::string msg = part1 + part2;  // Concatenates "Happy Birthday" and " Alice!" into "Happy Birthday Alice!"
    std::cout << "The invitation message: " << msg << std::endl;  // Prints "The invitation message: Happy Birthday Alice!"
    return 0;
}
```

In this example, we start with two separate strings `part1` and `part2`. By using the `+` operator, we concatenate them into a single string `msg`, resulting in the complete invitation message.

---

### Lesson Summary

| Concept | Key Point |
|---------|-----------|
| **`std::string`** | A sequence-of-characters type from `<string>`, declared like any other variable type |
| **Literal syntax** | String literals use **double quotes** (`"text"`); individual `char` values use **single quotes** (`'c'`) |
| **Assignment (`=`)** | Replaces a string's contents, just like with other types |
| **Access/modify** | `str[index]` — same `[]` syntax as vectors, zero-based |
| **`size()`** | Returns the number of characters in the string |
| **Concatenation (`+`)** | Joins two strings into a new, combined string |

## Personalize the Party Invitation

You're creating a party invitation in C++. Let's make a small change to further personalize the invitation. Add a prefix to the host's name before printing it, so it stands out as `"- Hosted by Cosmo"`. Use simple string concatenation to achieve this. This minor adjustment will help you become more proficient with string manipulation in C++.

---

### Starter Code

```cpp
#include <iostream>
#include <string>

int main() {
    std::string greeting = "Welcome to the party!";
    std::string event = " Birthday Bash";
    std::string host = "Hosted by Cosmo";

    std::cout << greeting + event << std::endl; // Concatenates greeting with event
    std::cout << host << std::endl; // Display the host string without prefix
    return 0;
}
```

---

### Building the Solution

- `host` currently holds just `"Hosted by Cosmo"` — printing it as-is gives no visual separation from the line above.
- Concatenate a `"- "` prefix onto `host` using the `+` operator: `"- " + host`.
- Print the concatenated result directly in the `std::cout` statement, the same way `greeting + event` was concatenated on the line above.

---

### Completed Code

```cpp
#include <iostream>
#include <string>

int main() {
    std::string greeting = "Welcome to the party!";
    std::string event = " Birthday Bash";
    std::string host = "Hosted by Cosmo";

    std::cout << greeting + event << std::endl; // Concatenates greeting with event
    std::cout << "- " + host << std::endl; // Adds a "- " prefix before the host string
    return 0;
}
```

---

### Output

```
Welcome to the party! Birthday Bash
- Hosted by Cosmo
```

---

### Output Trace

| Statement | Concatenation | Printed |
|-----------|----------------|---------|
| `greeting + event` | `"Welcome to the party!" + " Birthday Bash"` | `Welcome to the party! Birthday Bash` |
| `"- " + host` | `"- " + "Hosted by Cosmo"` | `- Hosted by Cosmo` |

> **Key takeaway:** The `+` operator can concatenate a string literal directly with a `std::string` variable, not just two variables. Prefixing `"- " + host` inline avoids needing a separate variable just to hold the combined result.

## Capitalize the First Character in a String

Let's put your string manipulation skills to the test, Galactic Pioneer!

Your task is to create a C++ program that capitalizes the first character of the string `msg`. Complete the missing parts of the code to achieve this.

You got this, Galactic Pioneer!

---

### Starter Code

```cpp
#include <iostream>
#include <string>

int main() {
    std::string msg = "hello world!";
    // TODO: Capitalize the first character of the string
    std::cout << msg << std::endl;  // Should print "Hello world!"
    return 0;
}
```

---

### Building the Solution

- `msg[0]` accesses the first character of the string, just like the first element of a vector.
- `toupper()` (from `<cctype>`) converts a single `char` to its uppercase form.
- Reassign the result back into `msg[0]` to modify the string in place: `msg[0] = toupper(msg[0]);`.

---

### Completed Code

```cpp
#include <iostream>
#include <string>
#include <cctype>

int main() {
    std::string msg = "hello world!";
    msg[0] = toupper(msg[0]);  // Capitalizes the first character
    std::cout << msg << std::endl;  // Should print "Hello world!"
    return 0;
}
```

---

### Output

```
Hello world!
```

---

### Output Trace

| Index | Before | After `msg[0] = toupper(msg[0]);` |
|-------|--------|--------------------------------------|
| 0     | `'h'`  | `'H'`                                  |
| 1-11  | `"ello world!"` | `"ello world!"` (unchanged)   |

> **Key takeaway:** `toupper(char)` returns the uppercase version of a single character without touching the rest of the string. Assigning its result back into `msg[0]` modifies just that one position, since `std::string` characters are mutable through `[]` indexing, exactly like arrays and vectors.

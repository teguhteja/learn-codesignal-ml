## Navigating Multidimensional Arrays in C++

Hello learner! Today's itinerary includes **multidimensional arrays** — a powerful C++ tool that acts as organized cabinets within rooms.

Often, we encounter complex data structures that require multidimensional arrays, which are ideal for storing intricate datasets such as a game board, a calendar, or an image.

---

### Unraveling Multidimensional Arrays

An array, akin to a collection of identical lockers, is an assembly of components of the same kind, accessible via a numerical index. A **multidimensional array** extends this concept into a table (two-dimensional) or a more complex geometrical shape.

Imagine a two-dimensional array (**2D array**) as a chessboard, an array of arrays. Each figure being an array is located by its row (array of the first dimension) and column (array of the second dimension) positions. Remember, arrays can be extended into any dimension in C++.

---

### Constructing and Initializing Multidimensional Arrays

C++ facilitates the declaration and initialization of multidimensional arrays. Consider this rudimentary example:

```cpp
#include <iostream>

int main() {
    int my_array[2][3] = {{1, 2, 3}, {4, 5, 6}}; // Declare and initialize a 2x3 array
    return 0;
}
```

This way, we create an array that stores two other arrays, each storing 3 elements. It can be represented as a table:

```
1 2 3
4 5 6
```

Each sub-array is a row of this table.

---

### Accessing and Tweaking Multidimensional Arrays

Upon populating our multidimensional array, we can access and tweak any component using its indices. A 2-dimensional array has two indices: the first one defines the **row**, and the second one the **column**. All indices start from `0`, as always.

```cpp
#include <iostream>

int main() {
    int my_array[2][3] = {{1, 2, 3}, {4, 5, 6}};

    // Accessing third element of the first row
    std::cout << "Third element in first row: " << my_array[0][2] << std::endl; // 3

    // Modifying third element in first row
    my_array[0][2] = 50;
    std::cout << "Third element in first row after modification: " << my_array[0][2] << std::endl; // 50

    return 0;
}
```

Note that, once declared, the size or data type of arrays cannot be changed.

---

### Looping Through Multidimensional Arrays

Mastering the traversal of a two-dimensional array is possible with **nested `for` loops**. Imagine the outer loop sweeping through the rows and the inner loop scrutinizing each column. Observe:

```cpp
#include <iostream>

int main() {
    int my_array[2][3] = {{1, 2, 3}, {4, 5, 6}};

    // Looping through the array
    for (int i = 0; i < 2; i++) {
        for (int j = 0; j < 3; j++) {
            std::cout << my_array[i][j] << " "; // Print element at position [i][j]
        }
        std::cout << std::endl;
    }

    return 0;
}
```

```
Output:
1 2 3
4 5 6
```

Here, two loop structures function together, allowing us to access every element in this array.

---

### Pros and Cons of Multidimensional Arrays

While multidimensional arrays are effective and versatile, they have their limitations. Their static nature (size and data type cannot be changed after initiation) and lack of built-in operations (such as insertions and deletions) can pose challenges at times.

---

### Lesson Summary

| Concept | Key Point |
|---------|-----------|
| **Multidimensional array** | An array of arrays — e.g. `int arr[rows][cols]` for a 2D table |
| **Declaration/initialization** | `int my_array[2][3] = {{1, 2, 3}, {4, 5, 6}};` |
| **Access/modify** | `arr[row][col]` — two zero-based indices |
| **Traversal** | Nested `for` loops: outer loop over rows, inner loop over columns |
| **Limitations** | Fixed size and type after declaration; no built-in insert/delete operations |

Well done! We've uncovered the foundation of multidimensional arrays — how to declare, initialize, navigate, and modify them, along with their advantages and limitations.

## Visualizing a Tic-Tac-Toe Board in C++

Have you ever wondered how to visually represent a game of tic-tac-toe in code? This scenario achieves that by initializing and displaying a tic-tac-toe board using a multidimensional array. Click Run to reveal how the game board looks in this coded portrayal!

```
#include <iostream>

int main() {
    // Initialize the board for a tic-tac-toe game
    char gameBoard[3][3] = {{'X', '.', 'O'}, {'.', 'X', '.'}, {'O', '.', 'X'}};

    // Output the current state of the game board
    for (int row = 0; row < 3; row++) {
        for (int col = 0; col < 3; col++) {
            std::cout << gameBoard[row][col] << " ";
        }
        std::cout << std::endl; // Newline at the end of each row
    }

    return 0;
}

```


## Modify the Tic-Tac-Toe Board for a Draw Outcome

Your current mission is to tweak our tic-tac-toe game board to show a different game outcome. Change the game board so the element in the top-right corner of the board is `O`.

---

### Starter Code

```cpp
#include <iostream>

int main() {
    // Initialize a 2D array representing a tic-tac-toe board
    char gameBoard[3][3] = {{'X', 'O', 'X'}, {'O', 'X', 'O'}, {'O', 'X', 'X'}};

    // Print the tic-tac-toe board
    for (int i = 0; i < 3; i++) {
        for (int j = 0; j < 3; j++) {
            std::cout << gameBoard[i][j] << " ";
        }
        std::cout << std::endl;
    }

    return 0;
}
```

---

### Building the Solution

- The **top-right corner** of a 3×3 board is **row `0`, column `2`** — accessed as `gameBoard[0][2]`.
- It currently holds `'X'`; reassign it with `gameBoard[0][2] = 'O';` before the board is printed.
- The rest of the board and the printing loops stay unchanged.

---

### Completed Code

```cpp
#include <iostream>

int main() {
    // Initialize a 2D array representing a tic-tac-toe board
    char gameBoard[3][3] = {{'X', 'O', 'X'}, {'O', 'X', 'O'}, {'O', 'X', 'X'}};

    // Change the top-right corner to 'O'
    gameBoard[0][2] = 'O';

    // Print the tic-tac-toe board
    for (int i = 0; i < 3; i++) {
        for (int j = 0; j < 3; j++) {
            std::cout << gameBoard[i][j] << " ";
        }
        std::cout << std::endl;
    }

    return 0;
}
```

---

### Output

```
X O O
O X O
O X X
```

---

### Output Trace

| Position | Before | After `gameBoard[0][2] = 'O';` |
|-----------|--------|-----------------------------------|
| `[0][0]` | `X` | `X` |
| `[0][1]` | `O` | `O` |
| `[0][2]` | **`X`** | **`O`** |
| `[1][0]`–`[2][2]` | unchanged | unchanged |

> **Key takeaway:** A specific cell in a 2D array is reached with two indices, `arr[row][col]`. Reassigning just one cell — here the top-right corner — leaves every other element untouched, since arrays store each position independently.

## Debugging the Tic-Tac-Toe Board Display

The tic-tac-toe board is not displaying correctly, and it appears there's a small mistake in the code. Can you find and fix the issue so the board prints out as intended? Remember, the correct output should showcase the tic-tac-toe board's current state.

---

### Buggy Code

```cpp
#include <iostream>

int main() {
    char ticTacToeBoard[3][3] = {
        {'X', 'O', 'X'},
        {'O', 'X', 'O'},
        {'O', ' ', 'X'}
    }; // Initialize a 3x3 tic-tac-toe board

    // Print the current state of the tic-tac-toe board
    for (int i = 0; i < 3; i++) {
        for (int j = 0; j < 2; j++) {
            std::cout << ticTacToeBoard[i][j] << " ";
        }
        std::cout << std::endl;
    }
    return 0;
}
```

**Buggy behavior:** the inner loop's condition `j < 2` only visits columns `0` and `1`, so the **third column** of every row (`ticTacToeBoard[i][2]`) never gets printed — each row shows just 2 characters instead of 3.

---

### Finding the Bug

- The board is a `3x3` array, so each row has `3` columns, valid indices `0`, `1`, `2`.
- `j < 2` stops the loop after `j = 1`, skipping index `2`.
- **Fix:** change the inner loop's bound from `j < 2` to `j < 3`.

---

### Fixed Code

```cpp
#include <iostream>

int main() {
    char ticTacToeBoard[3][3] = {
        {'X', 'O', 'X'},
        {'O', 'X', 'O'},
        {'O', ' ', 'X'}
    }; // Initialize a 3x3 tic-tac-toe board

    // Print the current state of the tic-tac-toe board
    for (int i = 0; i < 3; i++) {
        for (int j = 0; j < 3; j++) {
            std::cout << ticTacToeBoard[i][j] << " ";
        }
        std::cout << std::endl;
    }
    return 0;
}
```

---

### Output

```
X O X
O X O
O   X
```

---

### Output Trace

| Row | `j < 2` (buggy) | `j < 3` (fixed) |
|-----|-------------------|--------------------|
| 0   | `X O`             | `X O X`              |
| 1   | `O X`             | `O X O`              |
| 2   | `O  `             | `O   X`              |

> **Key takeaway:** The inner loop's bound must match the number of **columns** in the array, not an arbitrary smaller number. Using `j < 2` on a 3-column row silently truncates every row's output by one element — a classic off-by-one bound mistake.

## Understanding Sets in C++

Hey there! Today, we're exploring **sets** in C++ — a useful way to handle collections of unique items. By the end of this lesson, you'll know how to define sets, add and remove items, check for specific items, and iterate through all the items in a set. Let's dive in!

A **set** in C++ is a container holding **unique** elements, just like a stamp collection without duplicates. Sets are perfect for tracking unique items without worrying about duplicates — like a class attendance list ensuring no name repeats.

To use sets in C++, include the set library with this line:

```cpp
#include <set>
```

---

### Defining a Set

To create an empty set that stores integers, define it like this:

```cpp
#include <set>

std::set<int> mySet;
```

This declares a set named `mySet` for integers.

---

### Adding Elements

Add elements using the `insert` function:

```cpp
#include <set>
#include <iostream>

int main() {
    std::set<int> mySet;

    mySet.insert(10);
    mySet.insert(20);
    mySet.insert(30);

    return 0;
}
```

If you try to insert `10` again, it won't be added as it's a duplicate:

```cpp
mySet.insert(10); // Duplicate, won't be added
```

However, it won't cause any errors.

---

### Removing Elements

Remove an item with the `erase` function:

```cpp
#include <set>
#include <iostream>

int main() {
    std::set<int> mySet;

    mySet.insert(10);
    mySet.insert(20);
    mySet.insert(30);

    // Remove element 20
    mySet.erase(20);

    return 0;
}
```

---

### Checking for Elements: `find` Function

Use `find` to check if an element is in the set. `find` is a more complex function that returns an **iterator** pointing to the set's element — we'll cover the details of iterators in the "Intermediate Programming in C++" path. For now, just know that if `find` doesn't locate the element, it returns something equal to `set.end()`, so here is how we can make the comparison:

```cpp
#include <set>
#include <iostream>

int main() {
    std::set<int> mySet;

    mySet.insert(10);
    mySet.insert(20);
    mySet.insert(30);

    // Check if 10 is in the set
    if (mySet.find(10) != mySet.end()) {
        std::cout << "10 is in the set" << std::endl;  // Output: 10 is in the set
    }

    return 0;
}
```

It works, but seems a bit tricky and hard to remember, right? Let's try out another approach!

---

### Checking for Elements: `count` Function

Alternatively, use the `count` function:

```cpp
#include <set>
#include <iostream>

int main() {
    std::set<int> mySet;

    mySet.insert(10);
    mySet.insert(20);
    mySet.insert(30);

    // Check if 20 is in the set
    if (mySet.count(20) > 0) {
        std::cout << "20 is in the set" << std::endl;  // Output: 20 is in the set
    }

    return 0;
}
```

The set can only have one instance of each element, so `.count()` will always return either `0` or `1`.

---

### Iterating Through a Set

Finally, we can iterate the set using a range-based `for` loop:

```cpp
#include <set>
#include <iostream>

int main() {
    std::set<int> mySet;

    mySet.insert(20);
    mySet.insert(10);
    mySet.insert(30);

    // Print all elements
    for (int item : mySet) {
        std::cout << item << " ";
    }

    return 0;
}
```

This code prints: `10 20 30`. Note that the set always stores elements in **ascending order** — the order of adding elements to the set doesn't matter.

---

### Lesson Summary

| Concept | Key Point |
|---------|-----------|
| **`std::set<T>`** | Stores unique elements from `<set>`, automatically kept in ascending order |
| **`insert(x)`** | Adds `x`; silently does nothing if `x` is already present |
| **`erase(x)`** | Removes `x` from the set |
| **`find(x)`** | Returns an iterator to `x`, or `set.end()` if `x` isn't present |
| **`count(x)`** | Returns `1` if `x` is present, `0` otherwise — simpler than `find` |
| **Traversal** | Range-based `for (int item : mySet)` — always visits elements in ascending order |

Sets are great for handling collections of unique items efficiently.

## Check Element Existence Using Count Function

Great job so far! Now, let's change the way we check for elements in the set. Modify the provided code to use the `count` function instead of the `find` function. This will solidify your understanding of the `count` function.

Let's code!

---

### Starter Code

```cpp
#include <set>
#include <iostream>

int main() {
    std::set<int> mySet;

    mySet.insert(10);
    mySet.insert(20);
    mySet.insert(30);

    // Check if 10 is in the set using find
    if (mySet.find(10) != mySet.end()) {
        std::cout << "10 is in the set" << std::endl;  // Output: 10 is in the set
    }

    return 0;
}
```

---

### Building the Solution

- `mySet.count(10)` returns `1` if `10` is present in the set, `0` otherwise — no iterator comparison needed.
- Replace the condition `mySet.find(10) != mySet.end()` with `mySet.count(10) > 0`.
- The rest of the program (inserting elements, printing the message) stays the same.

---

### Completed Code

```cpp
#include <set>
#include <iostream>

int main() {
    std::set<int> mySet;

    mySet.insert(10);
    mySet.insert(20);
    mySet.insert(30);

    // Check if 10 is in the set using count
    if (mySet.count(10) > 0) {
        std::cout << "10 is in the set" << std::endl;  // Output: 10 is in the set
    }

    return 0;
}
```

---

### Output

```
10 is in the set
```

---

### Output Trace

| Expression | Value |
|------------|-------|
| `mySet` | `{10, 20, 30}` |
| `mySet.count(10)` | `1` |
| `mySet.count(10) > 0` | `true` |

> **Key takeaway:** `count(x) > 0` and `find(x) != end()` answer the exact same question — "is `x` in the set?" — but `count` avoids dealing with iterators entirely, making it the more readable choice for a simple existence check.

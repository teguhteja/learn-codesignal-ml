## Understanding and Using Arrays in C++

Welcome to the thrilling world of C++ data structures, with a focus on **arrays**! Arrays, which are collections of similar items housed in contiguous memory spots, are crucial tools often used for efficient storage and manipulation of data in C++.

Our objective is simple: we aim to understand what arrays are, and learn how to **declare, initialize, access, modify, and traverse** them in C++.

---

### Understanding Arrays

An array in C++ is akin to a row of post-office boxes. Each slot, known as an **index**, holds an item of the array's type. This characteristic makes arrays remarkably useful when we need to operate on multiple entities of the same type.

Arrays can be declared in several ways. Let's consider an array to hold the test scores of five students:

```cpp
int scores[5]; // Array declaration without initialization
```

```cpp
int scores[5] = {89, 94, 79, 86, 92}; // Array declaration and initialization with scores
```

```cpp
int scores[] = {89, 94, 79, 86, 92}; // Compiler deduces size (5) based on the initializer list
```

`int` is the type of the array, denoting that it will store integer elements. The array, named `scores`, can hold five integer values, with one value for each student represented by a slot. The array requires a predefined size — it **can't** change its size dynamically.

---

### Common Mistake

Note that you can't define an array like this:

```cpp
int N;
std::cin >> N;
int array[N];
```

It won't work, because when the program starts, all variables are initialized, and only *then* the code executes. So `N` will not hold the user's value when the array is declared — the array in this code snippet can be of any size, most likely `0`.

The only proper way is:

```cpp
int max_N = 10000; // an example of the maximum possible N that the user will enter
int array[max_N];
```

Here, we initialize the array with the **maximum possible** `N`, and then use only a part of this array when the program executes.

---

### Accessing Array Elements

We can access elements within an array using their **index**, with `0` denoting the first element. For example, to print the score of the third student:

```cpp
std::cout << scores[2] << "\n"; // Access and print the third element via index `2`
```

> **Note:** Since valid array indices start from `0`, always ensure that the index is within the bounds of the array. Out-of-bounds indices can result in unpredictable behavior because C++ **doesn't** perform bounds checking!

---

### Understanding Indexing with Examples

Let's consider an array `numbers` initialized as:

```cpp
int numbers[] = {10, 20, 30, 40, 50};
```

- `numbers[0]` accesses the first element (`10`)
- `numbers[1]` accesses the second element (`20`)
- Accessing an out-of-bounds index, such as `numbers[10]`, produces **unexpected behavior**.

```cpp
std::cout << numbers[0] << "\n";  // prints 10
std::cout << numbers[1] << "\n";  // prints 20
std::cout << numbers[4] << "\n";  // prints 50
std::cout << numbers[10] << "\n"; // prints 126782 (could be anything — out of bounds!)
```

---

### Manipulating Arrays

Modifying array elements is as straightforward as accessing them. To update the score of the first student:

```cpp
scores[0] = 92; // Update first element at index '0'
```

To populate an uninitialized array, you might do something like this:

```cpp
int scores[5]; // Declare an uninitialized array

scores[0] = 89;
scores[1] = 94;
scores[2] = 79;
scores[3] = 86;
scores[4] = 92;
```

---

### Traversing Arrays

To display all of the scores, we can use a `for` loop:

```cpp
#include <iostream>

int main() {
    int scores[5] = {89, 94, 79, 86, 92}; // Array initialization with scores of students

    for (int i = 0; i < 5; i++) {
        std::cout << scores[i] << "\n"; // Print each score
    }

    return 0;
}
```

We can also use a **for-each** loop for cleaner code:

```cpp
#include <iostream>

int main() {
    int scores[5] = {89, 94, 79, 86, 92}; // Array initialization with scores of students

    for (int score : scores) {
        std::cout << score << "\n"; // Print each score
    }

    return 0;
}
```

`for (int score : scores) {}` automatically iterates through each element in the `scores` array, assigning each element's value to `score` in turn.

---

### Using Different Data Types in an Array

Arrays in C++ are **homogeneous** — they can only store one type of data. If you try to mix different data types, the compiler will throw an error:

```cpp
int numbers[5] = {1, 2, 3, 'c', 4.5}; // Compiler error — 'c' and 4.5 are not integers
```

---

### Lesson Summary

| Concept | Key Point |
|---------|-----------|
| **Array** | A fixed-size collection of same-typed elements in contiguous memory |
| **Declaration** | `int scores[5];` or `int scores[] = {...};` (size deduced) |
| **Fixed size** | Size must be a compile-time constant — cannot depend on user input at declaration |
| **Indexing** | Starts at `0`; no bounds checking — out-of-bounds access is undefined behavior |
| **Modification** | `scores[i] = value;` |
| **Traversal** | Classic `for (int i = 0; i < n; i++)` or `for (int val : arr)` (for-each) |
| **Homogeneity** | All elements must be the same type |

## Calculating Class Average Using Arrays in C++

In our journey through C++ data structures, let's explore how the concept of arrays is applied in a real-life scenario. Imagine you are a teacher needing to calculate the average grade for your class based on given scores. The provided code demonstrates how to sum up the grades in an array and calculate the average. Click Run to see how the class's average grade is computed!

```
#include <iostream>

int main() {
    int grades[5] = {89, 94, 79, 86, 92}; // Array initialization with scores of students
    float sum = 0;
    for(int i = 0; i < 5; i++) {
        sum += grades[i]; // Summing all the grades
    }
    std::cout << "Average grade: " << sum / 5 << "\n"; // Calculating and printing the average grade
    return 0;
}

```

## Identify and Correct the Array Index Error

There's a small mistake in this program that updates and displays a student's score. Can you find and fix it to ensure the program correctly updates the second student's score and prints it out? Run the code, and observe the output to identify what might be wrong.

---

### Buggy Code

```cpp
#include <iostream>

int main() {
    int studentScores[3] = {88, 76, 91}; // Initialize an array with 3 student test scores
    studentScores[85] = 1; // Update the second student's score to 85
    std::cout << "Updated score of the second student: " << studentScores[1] << std::endl; // Print the updated score of the second student
    return 0;
}
```

**Buggy behavior:** `studentScores[85] = 1;` writes to index `85`, which is **far out of bounds** for a 3-element array — this is undefined behavior (it may crash, corrupt other memory, or silently do nothing observable). Meanwhile `studentScores[1]` is never actually updated, so it still prints `76`.

---

### Finding the Bug

- The index and the value are **swapped**: `studentScores[85] = 1;` should be `studentScores[1] = 85;`.
- The intent is "set index `1` (the second student) to the value `85`" — but the code accidentally sets index `85` to the value `1`.
- **Fix:** swap the index and the value.

---

### Fixed Code

```cpp
#include <iostream>

int main() {
    int studentScores[3] = {88, 76, 91}; // Initialize an array with 3 student test scores
    studentScores[1] = 85; // Update the second student's score to 85
    std::cout << "Updated score of the second student: " << studentScores[1] << std::endl; // Print the updated score of the second student
    return 0;
}
```

---

### Output

```
Updated score of the second student: 85
```

---

### Output Trace

| Index | Initial value | After `studentScores[1] = 85;` |
|-------|----------------|----------------------------------|
| 0     | 88             | 88                                |
| 1     | 76             | **85**                            |
| 2     | 91             | 91                                |

> **Key takeaway:** `array[index] = value;` reads left-to-right as "the element *at* `index` gets set *to* `value`." Swapping the two — `array[value] = index;` — writes to a completely different (and possibly invalid) memory location. Always double-check which side of `[...]` is the index and which side of `=` is the value.

Now, it's time to delve deeper. We've explored array manipulation, but let's challenge ourselves with updating an array based on new data. Update a student's grade, and then display all the grades using a loop. Pay attention to syntax and the functioning of array indexes!

---

### Starter Code

```cpp
#include <iostream>

int main() {
    int classroomGrades[5] = {76, 85, 90, 67, 59}; // Array initialization with grades of students
    // TODO: Update the grade for the fourth student based on a new score
    
    // TODO: Write a loop to display each student's grade
    return 0; // Return statement to signify successful execution
}
```

---

### Building the Solution

- **Fourth student's index:** the fourth student sits at index `3` (indices `0, 1, 2, 3, 4` map to students `1st, 2nd, 3rd, 4th, 5th`).
- **Update:** `classroomGrades[3] = 88;` — assign a new score for the fourth student.
- **Display loop:** `for (int i = 0; i < 5; i++)` — iterate over all five indices, printing each grade.

---

### Completed Code

```cpp
#include <iostream>

int main() {
    int classroomGrades[5] = {76, 85, 90, 67, 59}; // Array initialization with grades of students
    classroomGrades[3] = 88; // Update the grade for the fourth student based on a new score

    for (int i = 0; i < 5; i++) {
        std::cout << "Student " << (i + 1) << ": " << classroomGrades[i] << "\n";
    }

    return 0; // Return statement to signify successful execution
}
```

---

### Output

```
Student 1: 76
Student 2: 85
Student 3: 90
Student 4: 88
Student 5: 59
```

---

| 4     | 5th     | 59             | 59             |

> **Key takeaway:** The "fourth student" corresponds to index `3`, not `4` — a direct application of zero-based indexing (`student N` is at index `N - 1`). Combining a targeted update (`array[i] = newValue;`) with a traversal loop is the standard pattern for modifying one entry and then confirming the whole array's state.

### Building the Solution

- **Declare and initialize:** `int classroomGrades[5] = {78, 85, 90, 67, 72};` — five grades, one per student.
- **Third student's index:** index `2` (indices `0, 1, 2, 3, 4` map to students `1st..5th`).
- **Add extra points:** `classroomGrades[2] += 5;` — adds `5` to the existing value in place.

### Completed Code

```cpp
#include <iostream>

int main() {
    int classroomGrades[5] = {78, 85, 90, 67, 72}; // Declare an array with 5 integer grades

    classroomGrades[2] += 5; // Add 5 extra points to the third student's grade

    std::cout << "Updated grade of third student: " << classroomGrades[2] << std::endl;

    return 0;
}
```

> **Key takeaway:** `array[i] += value;` is shorthand for `array[i] = array[i] + value;` — it reads the current element, adds to it, and writes the result back to the same slot. This is the idiomatic way to "boost" an existing array value rather than overwrite it entirely.

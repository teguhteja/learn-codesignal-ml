## Exploring the World of C++ Vectors

Welcome! Today's lesson focuses on **vectors**, which are dynamic equivalents of arrays in C++. We'll discuss how to declare, initialize, access, and manipulate them, and we'll delve into other unique elements.

---

### The Basics of Vectors

Vectors are **dynamic arrays** that form part of the C++ Standard Library. Unlike arrays, which require a pre-defined size, vectors allow elements to be populated as needed, resizing as they grow.

To use vectors, you must include the vector library:

```cpp
#include <vector>
```

Vectors are declared using the `std::vector` template followed by the type of elements they will store:

```cpp
std::vector<int> myVector; // Declares a vector of integers named myVector
```

You can use the following options to initialize a vector:

```cpp
// Initialization with a specified size and default values (0s)
std::vector<int> myVector2(5);

// Initialization with a specified size and a specified value
std::vector<int> myVector3(5, 10); // Vector of size 5 with all elements set to 10

// Initialization using an initializer list
std::vector<int> myVector4 {10, 20, 30, 40, 50};

// Initialization with another vector (copy initialization)
std::vector<int> myVector5(myVector4);
```

---

### Vectors and Arrays Comparison

| Aspect | Array | Vector |
|--------|-------|--------|
| **Size** | Static — fixed at compile time | Dynamic — can grow and shrink at runtime |
| **Memory** | Contiguous, manually sized | Contiguous, automatically managed |
| **Functionality** | No built-in helper functions | Built-in `push_back()`, `pop_back()`, `insert()`, `erase()`, etc. |

---

### Accessing and Modifying Vector Elements

Like arrays, vectors store elements in contiguous storage locations, allowing efficient access using indices. Elements can be accessed like array elements (`vector[index]`). Indices can also be used to modify vector values.

```cpp
#include <iostream>
#include <vector>

int main() {
    std::vector<int> myVector {10, 20, 30, 40, 50};

    // Accessing the second element
    std::cout << myVector[1] << std::endl; // 20

    // Modifying the second element
    myVector[1] = 99; // Indexing starts from 0. Thus, the second element corresponds to index 1.

    // Accessing the modified second element
    std::cout << myVector[1] << std::endl; // 99

    return 0;
}
```

---

### Special Vector Operations

Vectors support several built-in operations, including `size()`, `empty()`, `clear()`. These operations streamline the process of working with vectors.

```cpp
#include <iostream>
#include <vector>

int main() {
    std::vector<int> myVector {10, 20, 30, 40, 50};

    // Check if the vector is empty
    if (myVector.empty()) {
        std::cout << "Vector is empty!" << std::endl;
    } else {
        std::cout << "Vector is not empty!" << std::endl;
    }

    // Clear the vector
    myVector.clear(); // Removes all elements.

    // Check the size of the vector after clearing
    std::cout << "Size after clear operation: " << myVector.size() << std::endl; // 0

    return 0;
}
```

- **`size()`:** returns the number of elements in the vector.
- **`empty()`:** checks whether the vector is empty or not.
- **`clear()`:** removes all elements from the vector.

---

### Using `push_back()` to Append Elements and Populate Vectors

The `push_back()` function appends elements to the end of the vector, allowing us to populate an uninitialized vector.

```cpp
#include <iostream>
#include <vector>

int main() {
    std::vector<int> myVector;

    // Using push_back to populate the vector
    myVector.push_back(10);
    myVector.push_back(20);
    myVector.push_back(30);

    // Printing the vector
    for (int i = 0; i < myVector.size(); i++) {
        std::cout << myVector[i] << " "; // 10 20 30
    }
    std::cout << std::endl;

    return 0;
}
```

---

### Removing Elements with `pop_back()`

The `pop_back()` function removes the **last** element of the vector.

```cpp
#include <iostream>
#include <vector>

int main() {
    std::vector<int> myVector {10, 20, 30, 40, 50};

    // Remove the last element
    myVector.pop_back();

    // Printing the vector
    for (int i = 0; i < myVector.size(); i++) {
        std::cout << myVector[i] << " "; // 10 20 30 40
    }
    std::cout << std::endl;

    return 0;
}
```

---

### Working With a Vector Using a Range-Based `for` Loop

Similar to arrays, we can traverse a vector to access or modify its elements using a `for` loop. A more modern and concise approach is using the **range-based `for` loop**, which eliminates the need for explicitly iterating through indexes:

```cpp
#include <iostream>
#include <vector>

int main() {
    std::vector<int> newVector {10, 20, 30, 40, 50};

    for (int element : newVector)
        std::cout << element << " "; // 10 20 30 40 50
    std::cout << std::endl;

    return 0;
}
```

---

### Mixing Data Types in a Vector

Vectors in C++ are **type-safe**, meaning they can only store elements of a single data type. Attempting to use mixed data types will result in a **compile-time error**:

```cpp
#include <iostream>
#include <vector>

int main() {
    // The following line will cause a compile-time error:
    // std::vector<int> myVector {10, 20.5, 30};

    return 0;
}
```

## Vector Modification in a Library System

Watch how we can manage and visualize changes in a library's book collection with vectors. The given code starts with a list of book IDs, modifies an ID, and then displays both the initial and updated lists. Can you see how the book ID changes after running the code?

---

### Code

```cpp
#include <iostream>
#include <vector>

int main() {
    // Create a vector to store the book IDs in the library
    std::vector<int> bookIDs {101, 102, 103};

    // Display the initial list of book IDs
    std::cout << "Initial book IDs: ";
    for (int id : bookIDs) {
        std::cout << id << " ";
    }
    std::cout << "\n";

    // Modify the second book ID
    bookIDs[1] = 105;

    // Display the updated list of book IDs
    std::cout << "Updated book IDs: ";
    for (int id : bookIDs) {
        std::cout << id << " ";
    }
    std::cout << "\n";

    return 0;
}
```

---

### How It Works

- **`std::vector<int> bookIDs {101, 102, 103};`** — initializes a vector with three book IDs using an initializer list.
- **First range-based `for` loop:** prints every element in `bookIDs` as it currently stands — the "before" snapshot.
- **`bookIDs[1] = 105;`** — modifies the element at index `1` (the **second** book ID) just like modifying an array element.
- **Second range-based `for` loop:** prints the vector again, now showing the updated value in place of the old one.

---

### Output

```
Initial book IDs: 101 102 103 
Updated book IDs: 101 105 103 
```

---

### Output Trace

| Index | Initial value | After `bookIDs[1] = 105;` |
|-------|----------------|------------------------------|
| 0     | 101            | 101                           |
| 1     | 102            | **105**                       |
| 2     | 103            | 103                           |

> **Key takeaway:** Vectors support the same `[]` indexing syntax as arrays for both reading and writing elements — `bookIDs[1] = 105;` reassigns the value at index `1` in place, without changing the vector's size. The range-based `for` loop is a convenient way to display the full contents before and after a modification.

### Finding the Bug

- The comment says "Add a new book" — that's the job of **`push_back()`**, not `pop_back()`.
- `push_back(105)` appends `105` to the end of the vector, growing it from `4` to `5` elements.
- **Fix:** change `library.pop_back(105);` to `library.push_back(105);`.

---

### Fixed Code

```cpp
#include <iostream>
#include <vector>

int main() {
    // Declare a vector to simulate a small library database where each integer represents a book ID
    std::vector<int> library {101, 102, 103, 104};

    // Add a new book to the library
    library.push_back(105);

    // Print out all the book IDs in the library using a range-based for loop
    for (int bookID : library)
        std::cout << bookID << " ";

    return 0;
}
```

```
101 102 103 104 105 
```

---

## Convert Index-Based Loop to Range-Based Loop

Awesome progress, Space Explorer! Let's improve our vector handling by using modern techniques. Replace the loop that uses an index to iterate over vector elements with a range-based `for` loop. This change will simplify the code and improve readability.

Let's code!

---

### Starter Code

```cpp
#include <iostream>
#include <vector>

int main() {
    std::vector<int> myVector {1, 2, 3, 4, 5};

    // Traditional for loop with index
    for (int i = 0; i < myVector.size(); i++) {
        std::cout << myVector[i] << " ";
    }
    std::cout << std::endl;

    return 0;
}
```

---

### Converting the Loop

- **Index-based loop:** `for (int i = 0; i < myVector.size(); i++)` then `myVector[i]` — needs an explicit counter and bounds check.
- **Range-based loop:** `for (int element : myVector)` — iterates directly over each value, no index bookkeeping required.
- **Replace** `myVector[i]` with the loop variable `element` directly inside the loop body.

---

### Completed Code

```cpp
#include <iostream>
#include <vector>

int main() {
    std::vector<int> myVector {1, 2, 3, 4, 5};

    // Range-based for loop
    for (int element : myVector) {
        std::cout << element << " ";
    }
    std::cout << std::endl;

    return 0;
}
```

---

### Output

```
1 2 3 4 5 
```

---

### Output Trace

| Iteration | `element` | Printed |
|-----------|-----------|---------|
| 1         | 1         | `1 `    |
| 2         | 2         | `2 `    |
| 3         | 3         | `3 `    |
| 4         | 4         | `4 `    |
| 5         | 5         | `5 `    |

> **Key takeaway:** A range-based `for` loop removes the need to manage an index variable and call `.size()` in the loop condition — the compiler handles iteration internally. It produces the exact same output as the index-based version, but with less room for off-by-one mistakes.

## Fix Vector Out-Of-Bounds Access

Awesome job so far, Galactic Pioneer!

Your task is to fix the bug within the code. It is currently trying to access the last element of a vector, but something is wrong. Identify and correct the bug.

Let's code!

---

### Buggy Code

```cpp
#include <iostream>
#include <vector>

int main() {
    std::vector<int> myVector {10, 20, 30, 40, 50};

    // Trying to access the last vector element
    std::cout << "The last element: " << myVector[5] << std::endl;

    return 0;
}
```

**Buggy behavior:** `myVector` has `5` elements at valid indices `0`–`4`. `myVector[5]` reads **one past the end** — out-of-bounds access, which is undefined behavior (it may print garbage, crash, or appear to "work" by accident).

---

### Finding the Bug

- The vector has `5` elements, so the **last** valid index is `size() - 1 = 4`, not `5`.
- `myVector[5]` is out of bounds.
- **Fix:** change the index from `5` to `4`, or better, use `myVector.size() - 1` so it stays correct even if the vector's size changes.

---

### Fixed Code

```cpp
#include <iostream>
#include <vector>

int main() {
    std::vector<int> myVector {10, 20, 30, 40, 50};

    // Access the last vector element using size() - 1
    std::cout << "The last element: " << myVector[myVector.size() - 1] << std::endl;

    return 0;
}
```

---

### Output

```
The last element: 50
```

---

### Output Trace

| Index | Value |
|-------|-------|
| 0     | 10    |
| 1     | 20    |
| 2     | 30    |
| 3     | 40    |
| **4** | **50** |
| 5     | (out of bounds — doesn't exist) |

> **Key takeaway:** The last valid index of any vector (or array) is always `size() - 1`, never `size()`. Writing `myVector[myVector.size() - 1]` instead of a hard-coded number keeps the code correct even as the vector grows or shrinks.

```cpp
#include <iostream>
#include <vector>

int main() {
    std::vector<std::string> names {"Alice", "Bob", "Charlie"};

    // Check if names vector is empty
    if (names.empty()) {
        std::cout << "Names vector is empty!" << std::endl;
    } else {
        std::cout << "The vector has names in it!" << std::endl;
    }

    // Clear all elements from the names vector
    names.clear();

    // Check if names vector is empty once again
    if (names.empty()) {
        std::cout << "Names vector is empty!" << std::endl;
    } else {
        std::cout << "The vector has names in it!" << std::endl;
    }

    return 0;
}
```

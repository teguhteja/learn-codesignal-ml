### Getting Started with C++ Syntax

A C++ program comprises one or more function definitions. The program's operation commences from the main function.

Here's the basic structure of a program:

```cpp
#include <iostream>

int main() {
   std::cout << "Hello, C++!";
   
   return 0;
}
```

The first line includes the `iostream` library, which equips the program for input and output operations.

Don't worry about understanding every line of the code just yet. It will be covered in more detail in the following lessons. For now, you need to know that when a C++ file is run, the `main` function runs, and that `std::cout << "Hello, C++!";` prints "Hello, C++!" to the console.

### Delving into Code Comments

Comments in the code resemble footnotes in a book. They clarify the code, making it more comprehensible for the developer and others. These annotations do not affect the execution of the code.

Here are examples of single-line and multi-line comments, respectively:

```cpp
// Single-line comment

/* 
Multi-line
comment 
*/
```

Comments significantly aid in dissecting complex code, and it's good practice to annotate your code with descriptive comments.

### Practical Coding

Now, let's integrate these learnings into a straightforward C++ code. We will meticulously comment each section of the code for clarity.

```cpp
// We include the input-output stream library
#include <iostream>

// Begin defining the main function
int main() {
   // We print a message on the console
   std::cout << "Hello, CodeSignal!";
   
   // We signal successful program termination by returning 0
   return 0;
}
```

## Send a Space Camp Greeting in C++

Great job running your first C++ program, Space Voyager! Now, it's time to further test your skills. Can you fill in the missing pieces to send a space-themed greeting from our program? Remember how we include libraries and print messages.

```cpp
// Space greetings from a friendly astronaut!
#include <iostream>

int main() {
    // TODO: Use the output stream to send a greeting to space camp.
    return 0;
}
```

Here is the completed program with the missing `std::cout` line filled in:

```cpp
// Space greetings from a friendly astronaut!
#include <iostream>

int main() {
    // Use the output stream to send a greeting to space camp.
    std::cout << "Hello, Space Camp! Greetings from your friendly astronaut!" << std::endl;
    return 0;
}
```

**How it works:**

- `#include <iostream>` brings in the standard input/output stream library, which gives us access to `std::cout`.
- `std::cout` is the standard **character output stream** — it writes text to the console.
- The `<<` operator feeds the string literal `"Hello, Space Camp! Greetings from your friendly astronaut!"` into the output stream.
- `std::endl` flushes the output buffer and appends a newline, so the prompt lands on its own line after the program finishes.
- `return 0;` signals to the operating system that the program terminated successfully.

Running this program prints:

```
Hello, Space Camp! Greetings from your friendly astronaut!
```

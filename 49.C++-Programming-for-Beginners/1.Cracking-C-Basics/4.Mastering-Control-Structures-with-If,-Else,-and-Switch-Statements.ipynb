### The Anatomy of the `if` Statement

The `if` statement checks a given condition inside parentheses, which must result in a **boolean value** (true or false):

```cpp
if (condition) {
    // Code to be executed if the condition is true
}
```

- The **condition** is placed between the parentheses `()`.
- If the condition evaluates to `true`, the code block within the curly braces `{}` executes.
- This block will not run if the condition is `false`.

### `if` Statement in Action

Consider a scenario at a restaurant. When you are hungry, you order food. This situation can be modeled in C++ using the `if` statement, as illustrated below:

```cpp
#include <iostream> 

int main() {
   bool isHungry = true;

   if (isHungry) {
       // The following line will only be executed if isHungry is true
       std::cout << "Order food!";
   }

   return 0;
}
```

The `if` statement checks the condition: `isHungry`. If it's true, the statement within the block is executed. In this case, it prints `"Order food!"`.

---

### Digging Deeper into `if-else` Structures

The `if-else` statement in C++ tests a condition, then executes a certain task if the condition is met and performs a different task if not fulfilled. For instance, in a restaurant, you have a choice between the special dish or a juice, based on whether you're hungry or not:

```cpp
#include <iostream> 

int main() {
   bool isHungry = false;

   if (isHungry) {
       // Executed when isHungry is true
       std::cout << "Order the special dish!";
   } else {
       // Executed when isHungry is false
       std::cout << "Just order a juice.";
   }
 
   return 0;
}
```

---

### Exploring Comparison Operators within Control Structures

Before diving deeper into control structures, it's essential to understand the role of **comparison operators**. These operators allow us to compare two values, resulting in a boolean value (true or false). This outcome determines the flow of control structures like `if`, `else if`, and `switch`.

Here are the most commonly used comparison operators in C++:

- `==` — checks if two values are equal
- `!=` — checks if two values are not equal
- `>` — checks if the left value is greater than the right
- `<` — checks if the left value is less than the right
- `>=` — checks if the left value is greater than or equal to the right
- `<=` — checks if the left value is less than or equal to the right

### Comparison Operators in Action

Imagine you are building an application that assigns tasks based on priority levels. Tasks with a priority level higher than 5 need immediate attention, whereas tasks with a priority less than or equal to 5 can be scheduled for later. Here's how you can utilize the comparison operators:

```cpp
#include <iostream>

int main() {
    int taskPriority = 6;

    if (taskPriority > 5) {
        std::cout << "This task needs immediate attention!";
    } else {
        std::cout << "This task can be scheduled for later.";
    }

    return 0;
}
```

This code will print: `"This task needs immediate attention!"` because `taskPriority > 5` evaluates to `true`.

Understanding and utilizing comparison operators are fundamental in controlling the flow of your programs, ensuring you create efficient and effective C++ applications.

### Looking at Multiple Conditions Using `else if`

An `else if` statement is useful when handling more than two possibilities. Imagine visiting an amusement park where the ride choices depend on your age. For children under 10, it's the merry-go-round; for those aged 10 to 18, the roller coaster is the thrill, and for adults, the horror house awaits. Let's see this in action:

```cpp
#include <iostream> 

int main() {
   int age = 15;

   if (age < 10) {
       std::cout << "Head to the merry-go-round!";
   } else if (age <= 18) {
       std::cout << "Time for the roller coaster!";  // Executed when age is between 10 and 18
   } else {
       std::cout << "Dare the horror house!";
   }

   return 0;
}
```

### Understanding the `switch` Statement

The **`switch` statement** evaluates an expression once and compares its value against a list of case values. It executes a set of statements associated with that case. It's an efficient alternative to a long series of `if-else` statements when dealing with conditions based on a single variable.

```cpp
#include <iostream>

int main() {
   int dayOfWeek = 5;  // Let's assume it's a Friday

   switch(dayOfWeek) {
       case 1:  // If dayOfWeek is 1,
       case 2:  // or 2, 3, 4, 5
       case 3:
       case 4:
       case 5:  // then this code executes
           std::cout << "Work, work, work!";  // Prints message for weekdays
           break;  // Exits the switch statement
       case 6:  // If dayOfWeek is 6,
       case 7:  // or 7,
           std::cout << "Relax and chill!";  // Prints message for weekends
           break;  // Prevents fall-through
       default:  // If none of the above cases match,
           std::cout << "Invalid day!";  // this executes
   }

   return 0;
}
```

This will print: `"Work, work, work!"`

**Key concepts:**

- A **case value** is one of the possible values the switch expression (`dayOfWeek` in our case) can have.
- When the value of `dayOfWeek` matches a case, the code block following that case executes.
- A **`break` statement** ends execution within the switch construct. Without `break`, execution flows into the next case block (this is known as **fall-through**), which may lead to unexpected behaviors.
- The **`default` case** executes if none of the case values match the switch expression value. In this example, `default:` catches any `dayOfWeek` value outside 1 through 7, displaying `"Invalid day!"` as feedback.

## Identify and Correct the Code Error

Your task is to create a simple program that decides what to order at different times of the day. However, the code has a small bug that prevents it from running correctly. Can you identify and fix it, so it accurately tells you what to order based on the time of day?

**Starter code (with bug):**

```cpp
#include <iostream>

int main() {
    int time = 20; // Let's say time is in military format (0-23 hours)

    if (time < 10) {
        std::cout << "Breakfast time - order pancakes!";
    } else if (time < 16) {
        std::cout << "Lunch time - order a sandwich!"
    } else {
        std::cout << "Dinner time - order pasta!";
    }

    return 0;
}
```

### Identifying the Bug

If you try to compile this code, the compiler will report an **error on the line with `"Lunch time - order a sandwich!"`** — specifically, something like `expected ';' before '}'`.

The problem is that **a semicolon `;` is missing at the end of that statement**. In C++, every executable statement must end with a semicolon. Without it, the compiler doesn't know where one statement ends and the next begins, so it fails before the program can even run.

### Completed Code (fixed)

Add the missing `;` to the end of the `Lunch time` line:

```cpp
#include <iostream>

int main() {
    int time = 20; // Let's say time is in military format (0-23 hours)

    if (time < 10) {
        std::cout << "Breakfast time - order pancakes!";
    } else if (time < 16) {
        std::cout << "Lunch time - order a sandwich!";
    } else {
        std::cout << "Dinner time - order pasta!";
    }

    return 0;
}
```

**Output (with `time = 20`):**

```
Dinner time - order pasta!
```

Since `time = 20` is not less than 10, and not less than 16, the program skips both `if` and `else if`, falling through to the final `else` branch and printing the dinner message. Always remember to terminate your `std::cout` statements (and all C++ statements) with a semicolon!

### Hints

- The `switch` is built around an `int` variable named `restaurantChoice`.
- Other cuisines already use values `1` (Italian) and `2` (Japanese). Pick the next sequential value — `3` — for Mexican cuisine.
- Each case needs a **`break;`** at the end to avoid falling through to the next case.
- A fun, themed greeting works well — for example: `"You chose Mexican cuisine. ¡Buen provecho!"`

### Completed Code

Add the new `case` between the Japanese case and `default`:

```cpp
#include <iostream>

int main() {
    int restaurantChoice = 3; // Let's assume the user chose option 3

    switch (restaurantChoice) {
        case 1:
            std::cout << "You chose Italian cuisine. Buon appetito!";
            break;
        case 2:
            std::cout << "You chose Japanese cuisine. Itadakimasu!";
            break;
        case 3:
            std::cout << "You chose Mexican cuisine. ¡Buen provecho!";
            break;
        default:
            std::cout << "Invalid choice. Please choose a valid cuisine.";
    }

    return 0;
}
```

**Output (with `restaurantChoice = 3`):**

```
You chose Mexican cuisine. ¡Buen provecho!
```

Execution jumps to `case 3`, prints the message, hits `break;`, and exits the switch. Without the `break;`, execution would fall through into `default:`, printing the "Invalid choice" message right after — which would clearly be wrong!

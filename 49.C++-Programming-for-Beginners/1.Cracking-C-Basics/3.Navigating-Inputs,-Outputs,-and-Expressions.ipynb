## Navigating Inputs, Outputs, and Expressions


Welcome back, coding explorer! Today, we're taking a new turn they are simply mathematical formulas operating on variables and values. Take a glimpse of this mathematical wizardry:

```cpp
#include <iostream> 

int main() {
  int x = 10;
  int y = 5;

  std::cout << x * y;  // prints 50

  return 0;
}
```

Here, we've used a multiplication operation `x * y`, and `std::cout <<` delivers the result as 50!

### Applying Variables within Expressions

Combining all these insights, let's apply variables within expressions. Marvel at this:

```cpp
#include <iostream> 

int main() {
  int x = 15, y = 3; // Defining two variables on the same line by splitting them with a comma

  int res = 2 * (x + y * y) ;  // results in 48

  std::cout << "The result is: " << res; // displays: The result is: 48

  return 0;
}
```

In this snippet, we define two variables, `x` and `y`, assign constant values to them, use them in an expression to calculate the result, and then echo the result. In the same way, you can use other mathematical operations with variables such as:

* `-` for subtraction,
* `/` for division,
* `%` for the modulus operation (which returns the remainder of a division).

## Counting Fruits in the Basket

Wondering how quickly we can count fruits in our basket? The provided code effortlessly sums up the apples and bananas, demonstrating a basic operation with variables and output. Click Run to see what happens!

```
#include <iostream>

int main() {
    int apples = 3, bananas = 5;
    int fruitTotal = apples + bananas;
    std::cout << "Total number of fruits: " << fruitTotal << std::endl;
    return 0;
}
```

## Modify Fruit Market Inventory Code

In our fruit market scenario, you have successfully calculated the total number of fruits. Now, can you modify the code to display the difference between the number of apples and oranges? It's a small change but will help you practice modifying existing code.


```
#include <iostream>

int main() {
    int apples = 5, oranges = 3;
    // Display the total number of fruits in the fruit market.
    std::cout << "Total fruits in the market: " << apples + oranges << std::endl;
}

```

Here is the modified code that calculates and displays the difference between the number of apples and oranges:

```cpp
#include <iostream>

int main() {
    int apples = 5, oranges = 3;
    // Display the difference between apples and oranges.
    std::cout << "Difference between apples and oranges: " << apples - oranges << std::endl;
    return 0;
}

```

## Fruit Market Inventory Calculation

Great job so far! In this exercise, you have a small piece of code that calculates the total number of fruits by adding the number of oranges and bananas at a fruit market. However, it seems there is a mistake because the code fails to compile. Can you identify and fix the error to ensure it displays the correct total number of fruits?

```
#include <iostream>

int main() {
    int totalFruits = oranges + bananas;
    int oranges = 3, bananas = 5;
    std::cout << "Total number of fruits: " << totalFruits << std::endl;
}
```

The error occurs because the variables `oranges` and `bananas` are used to calculate `totalFruits` before they are declared and initialized. In C++, variables must be declared before they can be used.

Here is the corrected code:

```cpp
#include <iostream>

int main() {
    int oranges = 3, bananas = 5;
    int totalFruits = oranges + bananas;
    std::cout << "Total number of fruits: " << totalFruits << std::endl;
    return 0;
}

```

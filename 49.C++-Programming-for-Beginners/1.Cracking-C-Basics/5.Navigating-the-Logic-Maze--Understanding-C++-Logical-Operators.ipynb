## Navigating the Logic Maze: Understanding C++ Logical Operators

Hello, budding programmer! Let's dive into the topic of **Logical Operators** in C++. These special symbols — `&&`, `||`, and `!` — guide our program's decision-making process. In this lesson, we aim to demystify these operators and explain how they function in control structures such as `if` and `switch` statements.

---

### Logical Operators in C++

Jumping right in, we've identified three logical operators:

- **Logical AND (`&&`)** — yields `true` if **both** conditions are `true`
- **Logical OR (`||`)** — yields `true` if **any one** condition is `true`
- **Logical NOT (`!`)** — **reverses** the truth value of a condition

These operators work in conjunction with Boolean values (`true` and `false`) in a program. Here are some examples:

```cpp
#include <iostream> 

int main() {
   bool isDay = true;   // Variable indicating it is day
   bool isNight = false; // Variable indicating it is not night

   // Using Logical AND
   std::cout << "Is it day AND night? " << (isDay && isNight);  // Outputs 0 (false)

   // Using Logical OR
   std::cout << "Is it day OR night? " << (isDay || isNight);   // Outputs 1 (true)

   // Using Logical NOT
   std::cout << "Is it NOT day? " << (!isDay);                   // Outputs 0 (false)

   return 0;
}
```

---

### Understanding Truth Tables

To depict all possible logical operator outcomes, we turn to **Truth Tables**. Let's review each of them:

#### AND (`&&`)

| A       | B       | A && B  |
|---------|---------|---------|
| `true`  | `true`  | `true`  |
| `true`  | `false` | `false` |
| `false` | `true`  | `false` |
| `false` | `false` | `false` |

#### OR (`||`)

| A       | B       | A \|\| B |
|---------|---------|----------|
| `true`  | `true`  | `true`   |
| `true`  | `false` | `true`   |
| `false` | `true`  | `true`   |
| `false` | `false` | `false`  |

#### NOT (`!`)

| A       | !A      |
|---------|---------|
| `true`  | `false` |
| `false` | `true`  |

These tables neatly illustrate the outcomes of the logical operators based on all possible combinations of Boolean values.

### Application of Logical Operators in Control Structures

Now, let's put theory into practice by learning how to use logical operators within control structures.

Consider a scenario in which a student can go on a field trip **only if** it's a weekday **and** they have their parents' permission. In this case, the `if` condition checks both `isWeekday` and `hasPermission` using the Logical AND operator:

```cpp
#include <iostream> 

int main() {
   bool isWeekday = true;       // Variable indicating it's a weekday
   bool hasPermission = true;   // Variable indicating student has parents' permission

   if (isWeekday && hasPermission) {  // If both conditions are true
       std::cout << "The student can go to the field trip.";
   }
   else {
       std::cout << "The student can't go to the field trip.";
   }

   return 0;
}
```

This prints: `"The student can go to the field trip."`

#### Alternative Implementation Using Nested `if` Statements: A Cautionary Note

To illustrate an alternative but **not recommended** approach, let's rewrite the previous example using nested `if` statements instead of utilizing the logical AND operator directly. Though this method achieves the same result, it is typically **less efficient** and can make the code **harder to read**, especially as conditions become more complex:

```cpp
#include <iostream> 

int main() {
   bool isWeekday = true;       // Variable indicating it's a weekday
   bool hasPermission = true;   // Variable indicating student has parents' permission

   if (isWeekday) {  // First, check if it's a weekday
       if (hasPermission) {  // Inside, check if the student has permission
           std::cout << "The student can go to the field trip.";
       }
       else {
           std::cout << "The student can't go to the field trip.";
       }
   }
   else {
        std::cout << "The student can't go to the field trip.";
   }

   return 0;
}
```

This also prints: `"The student can go to the field trip."`

Although this nested `if` structure achieves the task, it introduces **additional complexity** without providing any clear benefit over using the logical AND (`&&`) operator. With each added condition requiring another level of nesting, code readability and maintainability can deteriorate quickly. Therefore, while it's important to know that such an approach exists and may occasionally be useful in very specific scenarios, generally, **it's advisable to use logical operators** for cleaner and more efficient code.

### Takeaways and Real-World Application

Logical operators play a significant role in programming, facilitating decision-making. Consider the scenario of a **digital library system** that allows access to a user if they are either a premium member or if a specific book has no access restrictions:

```cpp
#include <iostream> 

int main() {
    bool isPremiumMember = false;        // Variable indicating user is not a premium member
    bool bookHasNoRestrictions = true;   // Variable indicating the book has no access restrictions

    if (isPremiumMember || bookHasNoRestrictions) {  // If either condition is true
        std::cout << "User can access the book.";
    } 
    else {
        std::cout << "User cannot access the book.";
    }

    return 0;
}
```

This prints: `"User can access the book."`

This example showcases the practicality of the `OR` operator in designing conditions that enable broader access or opportunities, reflecting its importance in building **versatile and user-friendly systems**.

## Logical Operators in a Classroom Scenario

In the world of education, students often face various requirements to attend certain classes. The given code illustrates whether a student is allowed to attend a class based on two conditions: if they have done their homework or if they have a notebook. It uses logical OR to check these conditions. Can a student without his notebook but who has done his homework attend the class? Run the code to find out!

```cpp
#include <iostream>

int main() {
    bool didHomework = true; // Indicates student did their homework
    bool hasNotebook = false; // Indicates student forgot their notebook

    // Checking if student can attend the class
    if (didHomework || hasNotebook) {
        std::cout << "Student is allowed to attend the class.";
    } else {
        std::cout << "Student is not allowed to attend the class.";
    }

    return 0;
}
```

## Adjust Conditions with Logical AND Operator

Revise the conditions in our school scenario to reflect a different rule: A student can only play outside if they have finished their homework **and** have not yet read a book. Employ the Logical AND operator to modify the condition in the code. This adjustment to logical operators will provide another opportunity to apply your newfound skills.

**Starter code:**

```cpp
#include <iostream>

int main() {
    bool hasHomework = true;   // Constant value simulating condition
    bool hasReadBook = false;  // Constant value simulating another condition

    // Check if the student can play outside using Logical OR
    if (!hasHomework || hasReadBook) { 
        std::cout << "The student can play outside."; 
    }
    else {
        std::cout << "The student needs to finish homework or read the book.";
    }
}
```

### Hints

The starter code uses `||` (OR) with `!hasHomework` and `hasReadBook` — which is the *opposite* logic of what the prompt asks. The new rule is:

- The student has **finished** their homework → `!hasHomework` becomes `false`, so we need `hasHomework`
- The student has **not yet read** a book → `!hasReadBook` becomes `true`, so we use `!hasReadBook`
- We need **both** to be true → use `&&` (AND)

So the new condition becomes `hasHomework && !hasReadBook`.

### Completed Code

```cpp
#include <iostream>

int main() {
    bool hasHomework = true;   // Constant value simulating condition
    bool hasReadBook = false;  // Constant value simulating another condition

    // Check if the student can play outside using Logical AND
    if (hasHomework && !hasReadBook) { 
        std::cout << "The student can play outside."; 
    }
    else {
        std::cout << "The student needs to finish homework or read the book.";
    }
}
```

**Output (with `hasHomework = true`, `hasReadBook = false`):**

```
The student can play outside.
```

Walk-through: `hasHomework` is `true`, `!hasReadBook` is also `true` (because `hasReadBook` is `false`), so the `&&` produces `true` and the first branch runs.

You can verify other combinations too:

- `hasHomework = false`, `hasReadBook = false` → falls into `else` (homework not done)
- `hasHomework = true`,  `hasReadBook = true`  → falls into `else` (already read a book)

## Debug the School Club Eligibility Checker Program

Oh no, there seems to be a small bug in our school club participation checker program. Can you determine what needs to be fixed so that the program correctly displays whether a student can join the school club?

### Starter Code

```cpp
#include <iostream>

int main() {
    bool completedHomework = true;   // Constant value to simulate student's homework status
    bool hasPermissionSlip = false;  // Constant value to simulate permission slip status

    if (completedHomework == hasPermissionSlip) {
        std::cout << "The student can join the school club.";
    } else {
        std::cout << "The student cannot join the school club.";
    }

    return 0;
}
```

### Bug Identified

The condition `completedHomework == hasPermissionSlip` uses the **equality operator (`==`)**. This checks whether the two booleans have the **same** value — not whether the student meets **both** requirements.

With the current values:

- `completedHomework = true`
- `hasPermissionSlip = false`
- `true == false` evaluates to `false` ❌

So the program incorrectly prints "The student cannot join the school club.", even though the student has done their homework.

### Hints

- Joining a club typically requires **both** conditions to be true (homework done **and** permission slip signed)
- Replace `==` with `&&` (Logical AND) so the condition is `true` only when **both** booleans are `true`
- Logical AND `&&` is the right operator when a student needs to satisfy multiple requirements simultaneously

### Completed Code

```cpp
#include <iostream>

int main() {
    bool completedHomework = true;   // Student finished their homework
    bool hasPermissionSlip = false;  // Student has not yet returned the slip

    // Both requirements must be true for the student to join
    if (completedHomework && hasPermissionSlip) {
        std::cout << "The student can join the school club.";
    } else {
        std::cout << "The student cannot join the school club.";
    }

    return 0;
}
```

**Output (with `completedHomework = true`, `hasPermissionSlip = false`):**

```
The student cannot join the school club.
```

Walk-through: `completedHomework` is `true` but `hasPermissionSlip` is `false`, so `true && false` evaluates to `false`. The `else` branch runs — correctly indicating the student can't join until the permission slip is returned.

### Verification With Other Values

| `completedHomework` | `hasPermissionSlip` | `&&` Result | Output                                |
|---------------------|---------------------|-------------|---------------------------------------|
| `true`              | `true`              | `true`      | The student can join the school club. |
| `true`              | `false`             | `false`     | The student cannot join the school club. |
| `false`             | `true`              | `false`     | The student cannot join the school club. |
| `false`             | `false`             | `false`     | The student cannot join the school club. |

Only the first row lets the student in — exactly what we'd expect from a "must meet both requirements" rule.

### Takeaway

`==` compares two values for equality; `&&` checks whether multiple conditions are *all* simultaneously true. When your program needs **every** requirement to be satisfied, reach for `&&`, not `==`.

## School Trip Eligibility Checker in C++

Ready for a creative challenge using C++? Imagine you're coding a school program to check if a student can go on a trip. There are conditions to meet, such as having completed their homework and wearing the correct uniform. Use logical operators and control structures to determine the outcome. Remember, practicing writing code from scratch polishes your skills!

### Starter Code

```cpp
#include <iostream>

int main() {
  // TODO: Declare a Boolean variable to represent if the student did their homework
  // TODO: Declare another Boolean variable to represent if the student's uniform is correct
  // TODO: Write an if-else condition using logical operators to check if the student can go on the school trip
  // If the student did their homework but the uniform is not correct, print: "Student can go on the trip, but needs to correct the uniform."
  // Otherwise, print: "Student cannot go on the trip."
  return 0;
}
```

### Hints

- You'll need **two Boolean variables**: one for homework status (`didHomework`) and one for uniform correctness (`uniformCorrect`).
- There are **three** meaningful outcomes to consider, so use `if / else if / else`:
  1. `didHomework && uniformCorrect` → both done — student goes (and is fully ready)
  2. `didHomework && !uniformCorrect` → homework done, uniform wrong — the *specific* message in the prompt
  3. otherwise → student can't go
- Use the **Logical AND (`&&`)** to require multiple conditions, and **Logical NOT (`!`)** to test "uniform is NOT correct".

### Completed Code

```cpp
#include <iostream>

int main() {
    // Boolean flags describing the student's readiness
    bool didHomework      = true;   // Student finished their homework
    bool uniformCorrect   = false;  // Student's uniform is NOT yet correct

    if (didHomework && uniformCorrect) {
        // Both requirements met — student goes on the trip with no issues
        std::cout << "Student can go on the trip.";
    }
    else if (didHomework && !uniformCorrect) {
        // Homework is done but uniform still needs fixing
        std::cout << "Student can go on the trip, but needs to correct the uniform.";
    }
    else {
        // Homework wasn't completed — trip is off, regardless of the uniform
        std::cout << "Student cannot go on the trip.";
    }

    return 0;
}
```

**Output (with `didHomework = true`, `uniformCorrect = false`):**

```
Student can go on the trip, but needs to correct the uniform.
```

Walk-through: `didHomework` is `true`, `uniformCorrect` is `false`, so `true && uniformCorrect` is `false`. The first `if` is skipped. Then `didHomework && !uniformCorrect` evaluates to `true && true`, which is `true`, so the `else if` branch runs and prints the special "needs to correct uniform" message.

### Verification With Other Combinations

| `didHomework` | `uniformCorrect` | Branch Taken                                  | Output |
|---------------|------------------|-----------------------------------------------|--------|
| `true`        | `true`           | `if` (both met)                               | `Student can go on the trip.` |
| `true`        | `false`          | `else if` (homework done, uniform wrong)      | `Student can go on the trip, but needs to correct the uniform.` |
| `false`       | `true`           | `else` (homework missing — trips off)         | `Student cannot go on the trip.` |
| `false`       | `false`          | `else` (homework missing — trips off)         | `Student cannot go on the trip.` |

Notice how the **homework** requirement is the *gatekeeper*: without it, the trip is off — no uniform tweak can rescue the situation.

### Takeaway

`if / else if / else` lets you branch into **multiple** distinct outcomes, each with its own message. Combined with `&&` and `!`, you can encode nuanced rules like "X is required, but Y can be patched later" — perfect for real-world eligibility checks where missing one critical thing blocks progress, while missing a secondary thing only triggers a warning.

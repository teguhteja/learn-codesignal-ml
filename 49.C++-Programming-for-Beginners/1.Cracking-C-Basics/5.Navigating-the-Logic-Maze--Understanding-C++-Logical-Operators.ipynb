## Navigating the Logic Maze: Understanding C++ Logical Operators

Hello, budding programmer! Let's dive into the topic of **Logical Operators** in C++. These special symbols — `&&`, `||`, and `!` — guide our program's decision-making process. In this lesson, we aim to demystify these operators and explain how they function in control structures such as `if` and `switch` statements.

---

### Logical Operators in C++

Jumping right in, we've identified three logical operators:

- **Logical AND (`&&`)** — yields `true` if **both** conditions are `true`
- **Logical OR (`||`)** — yields `true` if **any one** condition is `true`
- **Logical NOT (`!`)** — **reverses** the truth value of a condition

These operators work in conjunction with Boolean values (`true` and `false`) in a program. Here are some examples:

```cpp
#include <iostream> 

int main() {
   bool isDay = true;   // Variable indicating it is day
   bool isNight = false; // Variable indicating it is not night

   // Using Logical AND
   std::cout << "Is it day AND night? " << (isDay && isNight);  // Outputs 0 (false)

   // Using Logical OR
   std::cout << "Is it day OR night? " << (isDay || isNight);   // Outputs 1 (true)

   // Using Logical NOT
   std::cout << "Is it NOT day? " << (!isDay);                   // Outputs 0 (false)

   return 0;
}
```

### Understanding Truth Tables

To depict all possible logical operator outcomes, we turn to **Truth Tables**. Let's review each of them:

#### AND (`&&`)

| A       | B       | A && B  |
|---------|---------|---------|
| `true`  | `true`  | `true`  |
| `true`  | `false` | `false` |
| `false` | `true`  | `false` |
| `false` | `false` | `false` |

#### OR (`||`)

| A       | B       | A \|\| B |
|---------|---------|----------|
| `true`  | `true`  | `true`   |
| `true`  | `false` | `true`   |
| `false` | `true`  | `true`   |
| `false` | `false` | `false`  |

#### NOT (`!`)

| A       | !A      |
|---------|---------|
| `true`  | `false` |
| `false` | `true`  |

These tables neatly illustrate the outcomes of the logical operators based on all possible combinations of Boolean values.

### Application of Logical Operators in Control Structures

Now, let's put theory into practice by learning how to use logical operators within control structures.

Consider a scenario in which a student can go on a field trip **only if** it's a weekday **and** they have their parents' permission. In this case, the `if` condition checks both `isWeekday` and `hasPermission` using the Logical AND operator:

```cpp
#include <iostream> 

int main() {
   bool isWeekday = true;       // Variable indicating it's a weekday
   bool hasPermission = true;   // Variable indicating student has parents' permission

   if (isWeekday && hasPermission) {  // If both conditions are true
       std::cout << "The student can go to the field trip.";
   }
   else {
       std::cout << "The student can't go to the field trip.";
   }

   return 0;
}
```

This prints: `"The student can go to the field trip."`

#### Alternative Implementation Using Nested `if` Statements: A Cautionary Note

To illustrate an alternative but **not recommended** approach, let's rewrite the previous example using nested `if` statements instead of utilizing the logical AND operator directly. Though this method achieves the same result, it is typically **less efficient** and can make the code **harder to read**, especially as conditions become more complex:

```cpp
#include <iostream> 

int main() {
   bool isWeekday = true;       // Variable indicating it's a weekday
   bool hasPermission = true;   // Variable indicating student has parents' permission

   if (isWeekday) {  // First, check if it's a weekday
       if (hasPermission) {  // Inside, check if the student has permission
           std::cout << "The student can go to the field trip.";
       }
       else {
           std::cout << "The student can't go to the field trip.";
       }
   }
   else {
        std::cout << "The student can't go to the field trip.";
   }

   return 0;
}
```

This also prints: `"The student can go to the field trip."`

Although this nested `if` structure achieves the task, it introduces **additional complexity** without providing any clear benefit over using the logical AND (`&&`) operator. With each added condition requiring another level of nesting, code readability and maintainability can deteriorate quickly. Therefore, while it's important to know that such an approach exists and may occasionally be useful in very specific scenarios, generally, **it's advisable to use logical operators** for cleaner and more efficient code.

### Takeaways and Real-World Application

Logical operators play a significant role in programming, facilitating decision-making. Consider the scenario of a **digital library system** that allows access to a user if they are either a premium member or if a specific book has no access restrictions:

```cpp
#include <iostream> 

int main() {
    bool isPremiumMember = false;        // Variable indicating user is not a premium member
    bool bookHasNoRestrictions = true;   // Variable indicating the book has no access restrictions

    if (isPremiumMember || bookHasNoRestrictions) {  // If either condition is true
        std::cout << "User can access the book.";
    } 
    else {
        std::cout << "User cannot access the book.";
    }

    return 0;
}
```

This prints: `"User can access the book."`

This example showcases the practicality of the `OR` operator in designing conditions that enable broader access or opportunities, reflecting its importance in building **versatile and user-friendly systems**.

## Logical Operators in a Classroom Scenario

In the world of education, students often face various requirements to attend certain classes. The given code illustrates whether a student is allowed to attend a class based on two conditions: if they have done their homework or if they have a notebook. It uses logical OR to check these conditions. Can a student without his notebook but who has done his homework attend the class? Run the code to find out!

```cpp
#include <iostream>

int main() {
    bool didHomework = true; // Indicates student did their homework
    bool hasNotebook = false; // Indicates student forgot their notebook

    // Checking if student can attend the class
    if (didHomework || hasNotebook) {
        std::cout << "Student is allowed to attend the class.";
    } else {
        std::cout << "Student is not allowed to attend the class.";
    }

    return 0;
}
```

## Adjust Conditions with Logical AND Operator

Revise the conditions in our school scenario to reflect a different rule: A student can only play outside if they have finished their homework **and** have not yet read a book. Employ the Logical AND operator to modify the condition in the code. This adjustment to logical operators will provide another opportunity to apply your newfound skills.

**Starter code:**

```cpp
#include <iostream>

int main() {
    bool hasHomework = true;   // Constant value simulating condition
    bool hasReadBook = false;  // Constant value simulating another condition

    // Check if the student can play outside using Logical OR
    if (!hasHomework || hasReadBook) { 
        std::cout << "The student can play outside."; 
    }
    else {
        std::cout << "The student needs to finish homework or read the book.";
    }
}
```

### Hints

The starter code uses `||` (OR) with `!hasHomework` and `hasReadBook` — which is the *opposite* logic of what the prompt asks. The new rule is:

- The student has **finished** their homework → `!hasHomework` becomes `false`, so we need `hasHomework`
- The student has **not yet read** a book → `!hasReadBook` becomes `true`, so we use `!hasReadBook`
- We need **both** to be true → use `&&` (AND)

So the new condition becomes `hasHomework && !hasReadBook`.

### Completed Code

```cpp
#include <iostream>

int main() {
    bool hasHomework = true;   // Constant value simulating condition
    bool hasReadBook = false;  // Constant value simulating another condition

    // Check if the student can play outside using Logical AND
    if (hasHomework && !hasReadBook) { 
        std::cout << "The student can play outside."; 
    }
    else {
        std::cout << "The student needs to finish homework or read the book.";
    }
}
```

**Output (with `hasHomework = true`, `hasReadBook = false`):**

```
The student can play outside.
```

Walk-through: `hasHomework` is `true`, `!hasReadBook` is also `true` (because `hasReadBook` is `false`), so the `&&` produces `true` and the first branch runs.

You can verify other combinations too:

- `hasHomework = false`, `hasReadBook = false` → falls into `else` (homework not done)
- `hasHomework = true`,  `hasReadBook = true`  → falls into `else` (already read a book)

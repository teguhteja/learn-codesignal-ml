## Understanding Scope and Object Lifecycle in C++

Welcome to our captivating lesson on understanding objects and variables **scope** in C++. In programming, the term "scope" defines the territory where a variable or an object can be accessed or is valid. This concept is crucial for writing efficient and bug-free code.

Suppose you're organizing your room. You would likely place your frequently used items, such as your glasses or books, near your bed or study table, while more occasionally used items like your suitcases are neatly tucked away in your closet. Similarly, understanding the scope of objects and variables helps you organize your code by influencing where and how your variables and objects can be accessed.

```cpp
#include <iostream>

int main() {
    int x = 10; // x has a scope within the main function
    std::cout << "x: " << x << "\n";  // x: 10
    return 0;
}
```

In the above code, the variable `x` has a specific scope, residing within the `main` function.

---

### Defining Scope in C++

To understand the concept of "scope", we need to delve deeper into the territory where a variable can be accessed. The importance of defining appropriate scopes in your code is similar to setting boundaries in a playground. You are free to move anywhere within the playground, but you can't venture beyond it.

```cpp
#include <iostream>

int globalVar = 42;

int main() {
    int localVar = 10; // localVar has a scope within the main function
    std::cout << "localVar: " << localVar << "\n";  // localVar: 10
    std::cout << "globalVar: " << globalVar << "\n";  // globalVar: 42
    return 0;
}
```

There are two main kinds of scope in C++: **local** and **global**. As suggested by their names, `localVar` in our example is accessible only within the `main` function, whereas `globalVar`, being a global variable, can be accessed from any part of the code.

---

### Understanding Local and Global Variables

Imagine your town's local laws as local variables and your country's laws as global variables. Being aware of this distinction helps you avoid clashes between local and global variables and maintains clear and readable code.

```cpp
#include <iostream>

int myVar = 42;

int main() {
    int myVar = 10; // myVar here is local to the main function
    std::cout << "Local myVar: " << myVar << "\n";  // Local myVar: 10
    std::cout << "Global myVar: " << ::myVar << "\n";  // Global myVar: 42
    return 0;
}
```

In our example, `myVar` is declared twice: once in the global scope and again in the `main` function. Notice that we use the **scope resolution operator** `::` with `myVar` to access the global variable.

---

### Lifecycle of Objects within Scope

Objects in C++, akin to human beings, have a life cycle — creation (birth), modification (life), and deletion (death). Scope is a significant factor governing this lifecycle.

```cpp
#include <iostream>

void myFunction() {
    int localVar = 10;
    std::cout << "Inside myFunction, localVar: " << localVar << "\n";  // Inside myFunction, localVar: 10
}

int main() {
    myFunction();
    std::cout << localVar << std::endl;  // error: 'localVar' was not declared in this scope
    return 0;
}
```

In our code, `localVar` is created within `myFunction`, exists during the function's execution, and gets destroyed upon concluding `myFunction`. This makes `localVar` inaccessible in `main`.

---

### Scope Rules in C++

Now, let's delve into the rules of scope. They dictate how we can interact with different scopes.

```cpp
#include <iostream>

int myVar = 42;

int main() {
    int myVar = 10;
    if (true) {
        int myVar = 100;
        std::cout << "Inside if block, myVar: " << myVar << "\n";  // Inside if block, myVar: 100
    }
    std::cout << "Inside main, myVar: " << myVar << "\n";  // Inside main, myVar: 10
    std::cout << "Global myVar: " << ::myVar << "\n";  // Global myVar: 42
    return 0;
}
```

In this code, we declare `myVar` three times: in the global scope, in the `main` function, and inside the `if` block. While each `myVar` has its own scope, they are distinct entities.

---

### Lesson Summary

| Concept | Key Point |
|---------|-----------|
| **Scope** | The region of code where a variable or object can be accessed |
| **Local variable** | Declared inside a function or block; exists only until that scope ends |
| **Global variable** | Declared outside all functions; accessible from anywhere in the file |
| **Shadowing** | A local variable with the same name hides the outer/global one within its scope |
| **`::` operator** | The scope resolution operator, used to reach the global variable (`::myVar`) when it's shadowed |
| **Object lifecycle** | A local object is created when its scope begins and destroyed when it ends |

## Global and Local Variables in Book Inventory Management

In book inventory management, global and local variables can manage different information. The given code shows how a local variable `totalBooks` inside a function shadows the global variable `totalBooks`. This can be useful for specific calculations or data handling within a function.

Click Run to see what happens!

---

### Code

```cpp
#include <iostream>

int totalBooks = 100; // Global scope variable

void libraryProcedure() {
    int totalBooks = 150; // Local scope variable, shadows global variable
    std::cout << "Books in procedure: " << totalBooks << "\n";
}

int main() {
    libraryProcedure();
    // Accessing global variable
    std::cout << "Total books in library: " << totalBooks << "\n";

    return 0;
}
```

---

### How It Works

- **`int totalBooks = 100;`** — declared outside every function, so it is a **global** variable visible throughout the file.
- **`int totalBooks = 150;`** inside `libraryProcedure` — declares a new **local** variable with the same name. Inside this function, the local one **shadows** the global, so `totalBooks` refers to `150`.
- When `libraryProcedure` ends, its local `totalBooks` is destroyed; the global one was never changed.
- **`main`** has no local `totalBooks`, so it reads the global value `100`.

---

### Output

```
Books in procedure: 150
Total books in library: 100
```

---

### Output Trace

| Where | Which `totalBooks` | Value |
|-------|----------------------|-------|
| Inside `libraryProcedure` | Local (shadows global) | `150` |
| After `libraryProcedure` returns, in `main` | Global | `100` |

> **Key takeaway:** A local variable with the same name as a global one **shadows** it only within its own scope, and modifying or creating it never touches the global. Once the function ends, the local is destroyed and the global value is what `main` sees.

### Buggy Code

```cpp
#include <iostream>

void addBooksToSection(int booksInSection) {
    totalBooksInLibrary += booksInSection; // Error: Trying to use totalBooksInLibrary variable
    std::cout << "Books in this section: " << booksInSection << "\n";
}

int main() {
    int totalBooksInLibrary = 0; // Should be global
    addBooksToSection(50);
    addBooksToSection(20);
    addBooksToSection(30);
    std::cout << "Total books in library: " << totalBooksInLibrary << "\n";
    return 0;
}
```

**Buggy behavior:** `totalBooksInLibrary` is declared as a **local** variable inside `main`, but `addBooksToSection` also uses it. A function cannot see another function's local variables, so this is a compile-time error: `'totalBooksInLibrary' was not declared in this scope`.

---

### Finding the Bug

- `totalBooksInLibrary` needs to be shared between `main` and `addBooksToSection`, so it must live in a scope both can see — the **global** scope.
- Local variables exist only inside the function that declares them, so `addBooksToSection` can never reach the one in `main`.
- **Fix:** move `int totalBooksInLibrary = 0;` out of `main` and declare it **above** the functions, at global scope.

---

### Fixed Code

```cpp
#include <iostream>

int totalBooksInLibrary = 0; // Global: visible to every function

void addBooksToSection(int booksInSection) {
    totalBooksInLibrary += booksInSection;
    std::cout << "Books in this section: " << booksInSection << "\n";
}

int main() {
    addBooksToSection(50);
    addBooksToSection(20);
    addBooksToSection(30);
    std::cout << "Total books in library: " << totalBooksInLibrary << "\n";
    return 0;
}
```

---

### Output

```
Books in this section: 50
Books in this section: 20
Books in this section: 30
Total books in library: 100
```

---

### Output Trace

| Call | `booksInSection` (local) | `totalBooksInLibrary` (global) |
|------|----------------------------|-----------------------------------|
| Initial | — | `0` |
| `addBooksToSection(50)` | `50` | `50` |
| `addBooksToSection(20)` | `20` | `70` |
| `addBooksToSection(30)` | `30` | `100` |

> **Key takeaway:** A variable that several functions need to read or update must be declared at **global scope**, before the functions that use it. Per-call data like `booksInSection` stays local, since it only matters inside one function call, while the running total persists across calls as a global.

## Mastering Recursion: Simplifying Complexity in C++

### Introduction to Recursion

Greetings, coder!

Today, we're unraveling the world of **recursion**. Recursion is a scenario in which a function calls itself. Think of Russian Matryoshka dolls — each doll contains a smaller doll inside, which mirrors the concept of recursion.

Recursion is valuable when you can break down a problem into smaller, yet similar, simpler problems. It plays a critical role in various tasks such as sorting and searching algorithms.

The lesson for today involves understanding recursion, implementing it in C++, comparing it with iteration, and practicing debugging it.

---

### Understanding Recursion

Recursion in programming occurs when a function solves a problem by resolving smaller instances of the same problem. This is akin to tracing a family tree — each person reports their parent, who, in turn, does the same. This process of lineage tracing serves as a good metaphor for recursion.

However, it is paramount to define a proper **base case** in recursion to bring it to an end and avoid infinite loops.

---

### Implementing Recursion in C++

Now, we will examine recursion in C++ using factorials as an example. The factorial of `n` is `n` times the factorial of `n-1`.

Here is the C++ implementation:

```cpp
#include <iostream>

int factorial(int n) {
    // Base case:
    if (n == 1) {
        return 1;
    }
    // Recursive case:
    return n * factorial(n - 1);
}
```

In the implementation above, the function `factorial()` calls itself to compute the factorial of `n`. Here are the key things to pay attention to:

- **Base case:** within the `factorial` function, the base case is checked first. The base case is the condition under which the function stops calling itself. If `n` is `1`, the function returns `1`. This prevents infinite recursion and serves as the termination condition.
- **Recursive case:** if `n` is not `1`, the function proceeds to the recursive case. It returns `n` multiplied by the result of `factorial(n - 1)`. This means the function calls itself with `n` reduced by `1`, breaking down the problem into smaller subproblems until it reaches the base case.

---

### Example of the Function Call

Here is how we can call our recursive function:

```cpp
int main() {
    // The factorial of 5 is 120:
    std::cout << "The factorial of 5 is: " << factorial(5);
    return 0;
}
```

The result of `factorial(5)` is calculated as `5 * 4 * 3 * 2 * 1`, which equals `120`. In our program, it works in the following way:

1. We call `factorial(5)`, starting the recursion.
2. The result of `factorial(5)` is `5 * factorial(4)` according to the function's recursive case.
3. The result of `factorial(4)` is `4 * factorial(3)`. Overall, the result of the function at this point is `5 * 4 * factorial(3)`.
4. Similarly, we get to `5 * 4 * 3 * 2 * factorial(1)`. The `factorial(1)` call triggers the base case and returns simply `1`. Thus, the overall return value is `5 * 4 * 3 * 2 * 1`, which is our correct answer.

| Call | Returns |
|------|---------|
| `factorial(1)` | `1` (base case) |
| `factorial(2)` | `2 * 1 = 2` |
| `factorial(3)` | `3 * 2 = 6` |
| `factorial(4)` | `4 * 6 = 24` |
| `factorial(5)` | `5 * 24 = 120` |

---

### Comparison: Recursion vs Iteration

While recursion simplifies code, it can consume more memory than iteration. Iterative solutions might seem more complex at a glance, but they are more efficient. For instance, we can use a `for` loop to calculate the factorial:

```cpp
#include <iostream>

int factorial(int n) {
    int result = 1;
    // Loop through numbers from 1 to n:
    for (int i = 1; i <= n; i++) {
        // Each time, multiply the result by the current number:
        result *= i;
    }
    return result;
}

int main() {
    // The factorial of 5 is 120 (the same as with the recursive method):
    std::cout << "The factorial of 5 is: " << factorial(5);
    return 0;
}
```

---

### Stopping Recursion: The Base Case

Defining a base case for ending recursion is crucial. Without a properly defined base case, you might find yourself in an endless loop.

Let's see what happens when a base case isn't defined:

```cpp
#include <iostream>

int factorial(int n) {
    return n * factorial(n - 1);
}

int main() {
    // Recursion doesn't stop, causing a stack overflow error:
    std::cout << "The factorial of 5 is: " << factorial(5);
    return 0;
}
```

This code results in a runtime error due to a **stack overflow**, as `factorial()` calls itself indefinitely, thus causing infinite recursion.

---

### Understanding Recursive Calls and the Call Stack

When dealing with recursion, understanding the **call stack** is essential. In C++, when a function calls itself, each call is added to the stack. Once the base case is reached, the calls start returning until no calls are left in the stack. This process is known as "backtracking."

This is precisely why recursive functions may consume more memory than iterative ones, possibly leading to a stack overflow.

Learning recursion involves the possibility of making mistakes such as forgetting the base case or making a recursive call that doesn't progress towards it.

When debugging recursive functions, start with small inputs and use `std::cout` statements at critical junctures to keep track of variable values and state changes.

---

### Lesson Summary

| Concept | Key Point |
|---------|-----------|
| **Recursion** | A function that calls itself to solve smaller instances of the same problem |
| **Base case** | The condition that stops the recursion; without it, recursion never ends |
| **Recursive case** | The part where the function calls itself with a smaller/simpler input |
| **Call stack** | Each recursive call is added to the stack; calls return in reverse order once the base case is hit |
| **Recursion vs iteration** | Recursion is often simpler to write; iteration is usually more memory-efficient |
| **Stack overflow** | Caused by infinite or very deep recursion, e.g. a missing base case |

## Counting Digits with Recursion in C++

In the given code, we have implemented a function to count the number of digits in an integer using recursion.

Study the implementation, understand how base cases stop infinite recursion, and run the code to observe the output. Click Run to see how it works!

Have you ever needed to determine the length of a number? This method of counting digits can be particularly useful in certain applications like data validation or analysis.

---

### Code

```cpp
#include <iostream>

// Recursive function to count digits
int countDigits(int n) {
    // Base case: if n is less than 10, it has only one digit
    if (n < 10) {
        return 1;
    }
    // Recursive case: remove the last digit and count the rest
    return 1 + countDigits(n / 10);
}

int main() {
    int number = 6789;
    std::cout << "The number of digits in " << number << " is: " << countDigits(number);
    return 0;
}
```

---

### How It Works

- **Base case:** `if (n < 10) return 1;` — a number below `10` has exactly one digit, so the recursion stops here.
- **Recursive case:** `1 + countDigits(n / 10)` — integer division by `10` drops the last digit, and the `1` counts the digit just removed.
- Each call works on a number one digit shorter, so the input always moves toward the base case.

---

### Output

```
The number of digits in 6789 is: 4
```

---

### Output Trace

| Call | `n` | Base case? | Returns |
|------|-----|-------------|---------|
| `countDigits(6789)` | `6789` | No | `1 + countDigits(678)` |
| `countDigits(678)` | `678` | No | `1 + countDigits(67)` |
| `countDigits(67)` | `67` | No | `1 + countDigits(6)` |
| `countDigits(6)` | `6` | **Yes** | `1` |

Unwinding the stack: `countDigits(67) = 1 + 1 = 2`, `countDigits(678) = 1 + 2 = 3`, `countDigits(6789) = 1 + 3 = 4`.

> **Key takeaway:** Integer division by `10` strips the last digit, which makes it a natural "smaller problem" for recursion. The base case (`n < 10`) guarantees the recursion ends, and the `1 +` on the way back up adds one for each digit removed.

### Buggy Code

```cpp
#include <iostream>

// Function to calculate factorial using recursion
int factorial(int n) {
    // Base case
    if (n == 0) {
        return 0;
    }
    // Recursive case
    return factorial(n - 1);
}

int main() {
    // Calculate factorial of 5
    std::cout << "The factorial of 5 is: " << factorial(5);
    return 0;
}
```

**Buggy behavior:** there are two mistakes, and either alone gives a wrong answer. The recursive case never multiplies by `n`, and the base case returns `0` instead of `1`. As written, `factorial(5)` just calls itself down to `factorial(0)` and returns `0`, so the program prints `The factorial of 5 is: 0`.

---

### Finding the Bug

- **Base case:** the factorial of `0` is defined as `1`, not `0`. Returning `0` would also wipe out any product built on top of it, since multiplying by `0` always gives `0`.
- **Recursive case:** `factorial(n - 1)` ignores the current `n`. The definition is `n! = n * (n - 1)!`, so the call must be multiplied by `n`.
- **Fix:** change the base case to `return 1;` and the recursive case to `return n * factorial(n - 1);`.

---

### Fixed Code

```cpp
#include <iostream>

// Function to calculate factorial using recursion
int factorial(int n) {
    // Base case
    if (n == 0) {
        return 1;
    }
    // Recursive case
    return n * factorial(n - 1);
}

int main() {
    // Calculate factorial of 5
    std::cout << "The factorial of 5 is: " << factorial(5);
    return 0;
}
```

---

### Building the Solution

- **Base case:** `if (num == 0) return 0;` — once every digit has been stripped away, there is nothing left to add. (Unlike a "sum of all digits" function, this one can't stop at a single digit, because a lone digit might be odd and must contribute `0`.)
- **Even last digit:** `(num % 10) % 2 == 0` — add that digit and recurse on the rest: `return (num % 10) + sumEvenDigits(num / 10);`.
- **Odd last digit:** skip it and just recurse on the rest: `return sumEvenDigits(num / 10);`.

---

### Completed Code

```cpp
#include <iostream>

int sumEvenDigits(int num) {
    // Base case: no digits left
    if (num == 0) {
        return 0;
    }
    // Recursive case with check for even digits
    if ((num % 10) % 2 == 0) {
        // Add the last digit to the sum of even digits of the remaining number
        return (num % 10) + sumEvenDigits(num / 10);
    }
    // Last digit is odd: skip it and continue with the remaining digits
    return sumEvenDigits(num / 10);
}

int main() {
    // Testing the sumEvenDigits function
    std::cout << "Sum of even digits in 123456: " << sumEvenDigits(123456) << std::endl;
    return 0;
}
```

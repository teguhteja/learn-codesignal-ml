## Mastering the Basics of Functions in C++

Welcome to our lesson on C++ **functions**! Functions form the building blocks of any program by encapsulating chunks of code into reusable modules, which are key to constructing complex code. Let's explore further!

---

### Understanding Functions

Have you ever used a recipe? If so, you've employed a real-world function, albeit an ad hoc one. A **function** is a set of instructions designed to accomplish a specific task. Functions help package blocks of code into reusable components, a concept known as **code reusability**.

---

### Function Syntax and Definition

Every C++ function consists of a **name**, a **return type**, and a **parameter list**, forming the following syntax:

```cpp
return_type function_name( parameter list ) {
   // function body
}
```

Let's break down each part.

---

### Function: Example 1

Let's consider a function **without arguments** that prints a greeting in the console. It doesn't take any parameters, and a `void` return type specifies it doesn't return a value.

Example function:

```cpp
void greet() {
    std::cout << "Hello, C++ learner!" << std::endl;
}
```

Example function call:

```cpp
#include <iostream>

int main() {
    greet();  // Hello, C++ learner!
    return 0;
}
```

While this function might seem not very useful, it already brings out the most significant advantage of using functions: **reusability**! We can call this function as many times as we want without rewriting its code:

```cpp
#include <iostream>

int main() {
    greet();  // Hello, C++ learner!
    greet();  // Hello, C++ learner!
    greet();  // Hello, C++ learner!
    return 0;
}
```

---

### Function: Example 2

Let's consider a function **with an argument**. It will still have a `void` return type, meaning it doesn't return a value.

Example function:

```cpp
void greet(string name) {
    std::cout << "Hello, " << name << "!" << std::endl;
}
```

This function prints out a personalized greeting, using the provided name. Here is an example code:

```cpp
#include <iostream>
#include <string>

int main() {
    greet("Bob");  // Hello, Bob!
    greet("Alice");  // Hello, Alice!
    greet("C++ learner");  // Hello, C++ learner!
    return 0;
}
```

Now, our program is more friendly!

---

### Function: Example 3

Let's consider a function with **multiple parameters**:

```cpp
void printSum(int x, int y) {
    int sum = x + y;
    std::cout << "The sum is: " << sum << std::endl;
}
```

This function takes two numbers as arguments. It calculates and prints their sum.

Example function call:

```cpp
#include <iostream>

int main() {
    printSum(5, 6);  // The sum is: 11
    printSum(3, 4);  // The sum is: 7
    return 0;
}
```

---

### Function: Example 4

While printing the sum this easy is awesome, it is not the optimal design for your code. A function is supposed to be as general as possible: we must be able to reuse it in different scenarios. With that said, it is more common to define a function that both **accepts arguments and returns a value**.

Example function:

```cpp
int addNumbers(int x, int y) {
   int sum = x + y;
   return sum;
}
```

Here is how we call it:

```cpp
#include <iostream>

int main() {
   int result = addNumbers(5, 6);
   std::cout << "The sum is: " << result;  // The sum is: 11
   return 0;
}
```

Note that we still print the sum, but not directly in the function. We first put the result into a variable and then print it. The advantage of it is that we can now use the function in cases where we don't need to print the sum. For example:

```cpp
#include <iostream>

int main() {
   int result = addNumbers(5, 6) * 3;
   std::cout << "The result is: " << result;  // The result is: 33
   return 0;
}
```

Here, we use the result of our function as an intermediate step for some calculation.

---

### Lesson Summary

| Concept | Key Point |
|---------|-----------|
| **Function** | A reusable block of code that performs a specific task |
| **Syntax** | `return_type function_name(parameter list) { body }` |
| **`void` return type** | The function does its work but returns no value |
| **Parameters** | Values passed in so one function works with different inputs |
| **`return`** | Sends a result back to the caller so it can be stored or used in further calculations |
| **Reusability** | Define once, call as many times as needed |

Today, we've learned the basics, components, and applications of functions. We've understood how to declare, define, and call a function, and how to understand function arguments and return values. ## Using Functions to Multiply Numbers in C++

This task will serve as an example of defining a function for you. Click Run to see the result!

---

### Code

```cpp
#include <iostream>

// Function to multiply two numbers and return the result
int multiply(int a, int b) {
    return a * b;
}

// Main function to call the multiply function and display the result
int main() {
    int result = multiply(7, 5);
    std::cout << "The product is: " << result << std::endl;  // The product is: 35
    return 0;
}
```

---

### How It Works

- **`int multiply(int a, int b)`** — defines a function named `multiply` that takes two `int` parameters and returns an `int`.
- **`return a * b;`** — computes the product and sends it back to the caller.
- **`int result = multiply(7, 5);`** — calls the function with the arguments `7` and `5`, and stores the returned value in `result`.
- **`std::cout << ... << result`** — prints the stored result, rather than printing inside the function itself.

---

### Output

```
The product is: 35
```

---

### Output Trace

| Step | `a` | `b` | Returned | `result` |
|------|-----|-----|----------|----------|
| `multiply(7, 5)` called | `7` | `5` | — | — |
| `return a * b;` | `7` | `5` | `35` | — |
| Assigned in `main` | — | — | — | **`35`** |

> **Key takeaway:** A function that **returns** its result instead of printing it is more reusable — the caller decides what to do with the value, whether that's printing it, storing it, or using it in further calculations.

## Function Modification

Let's use what you've learned about functions! Change the given `displayGreeting` function to accept an additional parameter `string date` to include today's date, concatenate this information with a custom message, and print the result. Rename the function to `displayMessage`.

Here is an example of an updated output: `"Hello, Alice! Today's date is 2023-10-01."`

---

### Starter Code

```cpp
#include <iostream>
#include <string>

void displayGreeting(std::string name) {
   std::cout << "Hello, " << name << "!" << std::endl;
}

int main() {
   displayGreeting("Alice");  // function call with a single name parameter
   return 0;
}
```

---

### Building the Solution

- **Rename:** change `displayGreeting` to `displayMessage`, both in the definition and in the call inside `main`.
- **Add the parameter:** the signature becomes `void displayMessage(std::string name, std::string date)`.
- **Extend the message:** after the greeting, append `" Today's date is "`, the `date`, and a final period, all chained with `<<`.
- **Update the call:** pass both arguments — `displayMessage("Alice", "2023-10-01");`.

---

### Completed Code

```cpp
#include <iostream>
#include <string>

void displayMessage(std::string name, std::string date) {
   std::cout << "Hello, " << name << "! Today's date is " << date << "." << std::endl;
}

int main() {
   displayMessage("Alice", "2023-10-01");  // function call with name and date parameters
   return 0;
}
```

---

### Output

```
Hello, Alice! Today's date is 2023-10-01.
```

---

### Output Trace

| Part | Value |
|------|-------|
| `name` | `"Alice"` |
| `date` | `"2023-10-01"` |
| Printed | `Hello, Alice! Today's date is 2023-10-01.` |

> **Key takeaway:** A function's behavior is extended by adding parameters to its signature, and every call site must then supply a matching argument for each one. Renaming a function means updating the definition **and** every place that calls it.

## Fix Arithmetic Operation in Function

Great! Let's do some debugging.

Identify and fix the bug in the code below. The code is supposed to multiply two numbers and print the result.

---

### Buggy Code

```cpp
#include <iostream>

// Function that multiplies two numbers and returns the result
int multiply(int a, int b) {
    return a * b;
}

int main() {
    int num1 = 4;
    int num2 = 5;
    int product = multiply(num1);
    std::cout << "The product is: " << product << std::endl;
    return 0;
}
```

**Buggy behavior:** `multiply` is defined with **two** parameters (`int a, int b`), but it's called with only **one** argument (`multiply(num1)`). This is a compile-time error: too few arguments to the function call. `num2` is declared but never used.

---

### Finding the Bug

- The function signature `int multiply(int a, int b)` requires exactly two arguments.
- The call `multiply(num1)` supplies just one, so the compiler can't match it to the function.
- **Fix:** pass the second argument too — change `multiply(num1)` to `multiply(num1, num2)`.

---

### Fixed Code

```cpp
#include <iostream>

// Function that multiplies two numbers and returns the result
int multiply(int a, int b) {
    return a * b;
}

int main() {
    int num1 = 4;
    int num2 = 5;
    int product = multiply(num1, num2);
    std::cout << "The product is: " << product << std::endl;
    return 0;
}
```

---

### Output

```
The product is: 20
```

---

### Output Trace

| Step | `a` | `b` | Returned | `product` |
|------|-----|-----|----------|-----------|
| `multiply(num1, num2)` called | `4` | `5` | — | — |
| `return a * b;` | `4` | `5` | `20` | — |
| Assigned in `main` | — | — | — | **`20`** |

> **Key takeaway:** A function call must supply **exactly one argument for each parameter** in the function's signature, in the right order. A mismatch in the number of arguments is caught by the compiler as an error.

## Implement the sumOfSquares Function

Great job so far, space voyager! Now let's dig a little deeper into function implementation. You'll need to complete the `calcDistance` function. Follow the TODO comments to fill in the missing code and calculate the result.

As a reminder, the distance $d$ between two points $(x_1, y_1)$ and $(x_2, y_2)$ in a plane is calculated by the formula: $d = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$

---

### Starter Code

```cpp
#include <iostream>
#include <cmath> // for std::sqrt

double calcDistance(double x1, double y1, double x2, double y2) {
    // TODO: Return the distance between the two points
}

int main() {
    // Calling the calcDistance function with example coordinates and printing the result
    std::cout << "The distance is: " << calcDistance(1.0, 2.0, 4.0, 6.0) << std::endl;
    return 0;
}
```

---

### Building the Solution

- **Differences:** compute `dx = x2 - x1` and `dy = y2 - y1`.
- **Squares and sum:** square each difference and add them: `dx * dx + dy * dy`.
- **Square root:** wrap the sum in `std::sqrt(...)` (from `<cmath>`) and `return` it. A function declared `double` must return a value, otherwise the result is undefined.

---

### Completed Code

```cpp
#include <iostream>
#include <cmath> // for std::sqrt

double calcDistance(double x1, double y1, double x2, double y2) {
    double dx = x2 - x1;
    double dy = y2 - y1;
    return std::sqrt(dx * dx + dy * dy);
}

int main() {
    // Calling the calcDistance function with example coordinates and printing the result
    std::cout << "The distance is: " << calcDistance(1.0, 2.0, 4.0, 6.0) << std::endl;
    return 0;
}
```

---

### Output

```
The distance is: 5
```

---

### Output Trace

| Step | Value |
|------|-------|
| `dx = x2 - x1` | `4.0 - 1.0 = 3.0` |
| `dy = y2 - y1` | `6.0 - 2.0 = 4.0` |
| `dx * dx + dy * dy` | `9.0 + 16.0 = 25.0` |
| `std::sqrt(25.0)` | `5.0` |
| Printed | `5` (`cout` drops the trailing `.0`) |

> **Key takeaway:** A non-`void` function must `return` a value of its declared type. Breaking a formula into named intermediate values (`dx`, `dy`) keeps it readable, and `std::sqrt` from `<cmath>` handles the final square root — here the points form a 3-4-5 right triangle, so the distance is exactly `5`.

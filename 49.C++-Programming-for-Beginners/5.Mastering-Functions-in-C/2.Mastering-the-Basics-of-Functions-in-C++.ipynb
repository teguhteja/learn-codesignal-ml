## Mastering the Basics of Functions in C++

Welcome to our lesson on C++ **functions**! Functions form the building blocks of any program by encapsulating chunks of code into reusable modules, which are key to constructing complex code. Let's explore further!

---

### Understanding Functions

Have you ever used a recipe? If so, you've employed a real-world function, albeit an ad hoc one. A **function** is a set of instructions designed to accomplish a specific task. Functions help package blocks of code into reusable components, a concept known as **code reusability**.

---

### Function Syntax and Definition

Every C++ function consists of a **name**, a **return type**, and a **parameter list**, forming the following syntax:

```cpp
return_type function_name( parameter list ) {
   // function body
}
```

Let's break down each part.

---

### Function: Example 1

Let's consider a function **without arguments** that prints a greeting in the console. It doesn't take any parameters, and a `void` return type specifies it doesn't return a value.

Example function:

```cpp
void greet() {
    std::cout << "Hello, C++ learner!" << std::endl;
}
```

Example function call:

```cpp
#include <iostream>

int main() {
    greet();  // Hello, C++ learner!
    return 0;
}
```

While this function might seem not very useful, it already brings out the most significant advantage of using functions: **reusability**! We can call this function as many times as we want without rewriting its code:

```cpp
#include <iostream>

int main() {
    greet();  // Hello, C++ learner!
    greet();  // Hello, C++ learner!
    greet();  // Hello, C++ learner!
    return 0;
}
```

---

### Function: Example 2

Let's consider a function **with an argument**. It will still have a `void` return type, meaning it doesn't return a value.

Example function:

```cpp
void greet(string name) {
    std::cout << "Hello, " << name << "!" << std::endl;
}
```

This function prints out a personalized greeting, using the provided name. Here is an example code:

```cpp
#include <iostream>
#include <string>

int main() {
    greet("Bob");  // Hello, Bob!
    greet("Alice");  // Hello, Alice!
    greet("C++ learner");  // Hello, C++ learner!
    return 0;
}
```

Now, our program is more friendly!

---

### Function: Example 3

Let's consider a function with **multiple parameters**:

```cpp
void printSum(int x, int y) {
    int sum = x + y;
    std::cout << "The sum is: " << sum << std::endl;
}
```

This function takes two numbers as arguments. It calculates and prints their sum.

Example function call:

```cpp
#include <iostream>

int main() {
    printSum(5, 6);  // The sum is: 11
    printSum(3, 4);  // The sum is: 7
    return 0;
}
```

---

### Function: Example 4

While printing the sum this easy is awesome, it is not the optimal design for your code. A function is supposed to be as general as possible: we must be able to reuse it in different scenarios. With that said, it is more common to define a function that both **accepts arguments and returns a value**.

Example function:

```cpp
int addNumbers(int x, int y) {
   int sum = x + y;
   return sum;
}
```

Here is how we call it:

```cpp
#include <iostream>

int main() {
   int result = addNumbers(5, 6);
   std::cout << "The sum is: " << result;  // The sum is: 11
   return 0;
}
```

Note that we still print the sum, but not directly in the function. We first put the result into a variable and then print it. The advantage of it is that we can now use the function in cases where we don't need to print the sum. For example:

```cpp
#include <iostream>

int main() {
   int result = addNumbers(5, 6) * 3;
   std::cout << "The result is: " << result;  // The result is: 33
   return 0;
}
```

Here, we use the result of our function as an intermediate step for some calculation.

---

### Lesson Summary

| Concept | Key Point |
|---------|-----------|
| **Function** | A reusable block of code that performs a specific task |
| **Syntax** | `return_type function_name(parameter list) { body }` |
| **`void` return type** | The function does its work but returns no value |
| **Parameters** | Values passed in so one function works with different inputs |
| **`return`** | Sends a result back to the caller so it can be stored or used in further calculations |
| **Reusability** | Define once, call as many times as needed |

## Using Functions to Multiply Numbers in C++

This task will serve as an example of defining a function for you. Click Run to see the result!

---

### Code

```cpp
#include <iostream>

// Function to multiply two numbers and return the result
int multiply(int a, int b) {
    return a * b;
}

// Main function to call the multiply function and display the result
int main() {
    int result = multiply(7, 5);
    std::cout << "The product is: " << result << std::endl;  // The product is: 35
    return 0;
}
```

---

### How It Works

- **`int multiply(int a, int b)`** — defines a function named `multiply` that takes two `int` parameters and returns an `int`.
- **`return a * b;`** — computes the product and sends it back to the caller.
- **`int result = multiply(7, 5);`** — calls the function with the arguments `7` and `5`, and stores the returned value in `result`.
- **`std::cout << ... << result`** — prints the stored result, rather than printing inside the function itself.

---

### Output

```
The product is: 35
```

---

### Output Trace

| Step | `a` | `b` | Returned | `result` |
|------|-----|-----|----------|----------|
| `multiply(7, 5)` called | `7` | `5` | — | — |
| `return a * b;` | `7` | `5` | `35` | — |
| Assigned in `main` | — | — | — | **`35`** |

> **Key takeaway:** A function that **returns** its result instead of printing it is more reusable — the caller decides what to do with the value, whether that's printing it, storing it, or using it in further calculations.

## Delving into C++ Do-While Loops

Welcome! In today's lesson, we are focusing on the **do-while loop** in C++. Unique among loops, the do-while loop runs its code block **once before checking the condition**. This exclusive feature can be incredibly useful in certain code constructs. Through this lesson, you will understand the syntax, operations, and practical applications of the do-while loop.

---

### Understanding the Do-While Loop Syntax

The do-while loop has a straightforward syntax as follows:

```cpp
do {
   // code block to be executed
} while (condition);
```

Here, `do` initiates the loop. Following that, there lies a code block enclosed within `{}`. Subsequently, we have `while` with the condition in `()`. Remember, the condition is checked **only after** the loop's code block has been executed at least once. Also, **don't forget the semicolon `;` at the end**.

Here's a simple example:

```cpp
#include <iostream>

int main() {
   int counter = 0;
   do {
      counter++;
      std::cout << "This is loop iteration number: " << counter << std::endl; // printed three times
   } while (counter < 3);

   return 0;
}
```

**Output:**

```
This is loop iteration number: 1
This is loop iteration number: 2
This is loop iteration number: 3
```

Walk-through: `counter` starts at `0`, but it is incremented **before** the print. So the first line already says `1`. The condition `counter < 3` keeps the loop running until `counter` reaches `3`.

---

### Do-While Loop Executing at Least Once

An important characteristic of the do-while loop is that it **always executes the code block at least once**, regardless of the condition. Even if the condition is initially `false`, the loop will still run the body one time before evaluating the condition.

Here's an example demonstrating this behavior where the condition is `false` from the beginning:

```cpp
#include <iostream>

int main() {
   int start = 10;
   do {
      std::cout << "This will print at least once, even if the loop condition is false." << std::endl;
   } while (start < 5); // Condition is false from the beginning

   return 0;
}
```

**Output:**

```
This will print at least once, even if the loop condition is false.
```

In this example, the message inside the loop will be printed **exactly once** because the condition `start < 5` is `false` from the start. Despite the false condition, the code block within the `do` statement executes **before** the condition is evaluated.

---

### Real-Life Examples of Do-While Loop

To demonstrate the do-while loop in a real-life example, let's use a **number guessing game**:

```cpp
#include <iostream>

int main() {
   int number = 5;
   int guess;
   do {
      std::cout << "Guess the number (between 1 and 10): "; // Ask the user for input
      std::cin >> guess; // Get user input
   } while (guess != number);

   std::cout << "Congrats! You guessed it right."; // Congratulate the user on guessing the number correctly

   return 0;
}
```

In this scenario, the user is asked to input their guess **until** they guess the correct number. The do-while structure guarantees that the user is prompted at least once — perfect for input-validation flows where you *must* ask before you can compare.

---

### Loop Control Variables in Do-While Loop

In loop control, **updating variables is crucial** to prevent an infinite loop. If not updated, an infinite loop may occur. Here's an instructive code snippet illustrating how this problem could arise:

```cpp
#include <iostream>

int main() {
   int number = 5;
   int guess = 0;
   do {
      std::cout << "Guess the number (between 1 and 10): "; // missing cin command → infinite loop
   } while (guess != number);

   std::cout << "Congrats! You guessed it right."; // This line will never be printed due to an infinite loop

   return 0;
}
```

Without `std::cin >> guess;` inside the loop, `guess` never changes. Because `guess = 0` and `number = 5`, `guess != number` stays `true` forever — the loop prints the prompt endlessly.

---

### Catching Common Mistakes

When working with do-while loops, it is common to forget to place a **semicolon at the end**. Unlike other loops, a do-while loop always ends with a semicolon:

```cpp
do {
   // code block
} while (condition); // note the trailing semicolon here
```

Forgetting the `;` is a syntax error — the compiler treats `while (condition)` as a *statement* and expects a terminator. Compare this with the `while` loop, where the trailing `;` would create a different (and likely unintended) loop body.

#### Quick Comparison

| Loop Type        | Trailing semicolon required? | Runs body before condition? |
|------------------|------------------------------|------------------------------|
| `while`          | **No**                       | No (condition first)         |
| `do { } while ();` | **Yes**                      | **Yes** (always at least once) |

## Sequential Number Guessing Game Simulation

Let's simulate a tiny **sequential guessing game**: the program starts with a guess, then increments the guess until it lands on the secret number. This is a clean way to demonstrate the **do-while** loop because the body must run **at least once** — we always take at least one guess before checking.

### Starter Code

```cpp
#include <iostream>

int main() {
    int secretNumber = 7;
    int currentGuess = 0;
    int attempts = 0;

    // TODO: Use a do-while loop to increment currentGuess until it matches secretNumber
    // Inside the loop:
    //   1. Print the current guess (e.g. "Guessing: X")
    //   2. Increment currentGuess by 1
    //   3. Increment attempts so we can report how many tries it took

    std::cout << "Found it!" << std::endl;
    return 0;
}
```

### Hints

- The loop guard should keep running **while** `currentGuess != secretNumber`.
- Use the **do-while** form: `do { ... } while (currentGuess != secretNumber);` — note the trailing `;`.
- Increment `currentGuess` **inside** the loop so the guess actually progresses (otherwise it loops forever).
- Track how many iterations ran by incrementing `attempts` each pass — useful for reporting the result.

### Completed Code

```cpp
#include <iostream>

int main() {
    int secretNumber  = 7;
    int currentGuess  = 0;
    int attempts      = 0;

    // Keep guessing until we hit the secret number
    do {
        currentGuess += 1;        // take the next guess
        attempts     += 1;        // count this attempt
        std::cout << "Guessing: " << currentGuess << std::endl;
    } while (currentGuess != secretNumber);

    std::cout << "Found it in " << attempts << " attempts!" << std::endl;

    return 0;
}
```

**Output:**

```
Guessing: 1
Guessing: 2
Guessing: 3
Guessing: 4
Guessing: 5
Guessing: 6
Guessing: 7
Found it in 7 attempts!
```

Walk-through: each iteration increments `currentGuess` first, then prints it. The loop keeps running **while** `currentGuess != 7`. After printing `7`, the condition fails and the loop exits — exactly **7 attempts** were needed because we started at `0` and incremented one-by-one until we hit the secret.

### Anatomy of the Loop

| Piece                              | What it does                                                    |
|------------------------------------|-----------------------------------------------------------------|
| `do { ... } while (cond);`         | Runs the body **once before** checking the condition.            |
| `currentGuess += 1;`               | Drives the loop forward — without it, the condition never changes. |
| `attempts += 1;`                   | Counts how many iterations ran (useful for reporting).            |
| `while (currentGuess != secretNumber)` | Loop guard — repeats **while** the guess is wrong.            |
| Trailing `;` after `while (cond)`  | Required by the `do-while` syntax.                              |

### Stretch Ideas

- **Randomize** the secret with `std::rand() % 10 + 1` (between 1 and 10) so the count varies each run.
- Add a **maximum attempts** cap using a nested `if` inside the loop that breaks out via a `bool found` flag.

### Takeaway

A `do-while` loop is ideal when the work must happen **at least once** before you can decide whether to repeat. Sequential searches, input prompts, and menu redraws all fit this pattern: do the action, then check whether to repeat.

## Adjust the Guess Increment in the Number Guessing Game

Great progress, Space Explorer! Now, let's fine-tune our guessing game. Initially, our guess improved by 1 with each try. Your task is to modify the **step size** of the increment within the loop. Change the code so that the guess increases by **2** instead. This adjustment requires your latest understanding of controlling the do-while loop.

### Starter Code

```cpp
#include <iostream>

int main() {
  int secretNumber = 7, guess = 5;
  do {
    std::cout << "The guess is not the secret number." << std::endl;
    guess++; // Change the guess in each iteration
  } while (guess < secretNumber);

  std::cout << "Congratulations! You've guessed the secret number: " << guess << std::endl;
  return 0;
}
```

**Output of the starter (with `guess++`):**

```
The guess is not the secret number.
The guess is not the secret number.
Congratulations! You've guessed the secret number: 7
```

The current code works — but it takes **two** iterations to get from `5` to `7`. The exercise asks us to speed things up by jumping in bigger steps.

### Bug Identified

The line `guess++;` only increases `guess` by `1` per iteration. With `guess = 5` and `secretNumber = 7`, the loop runs the body **twice** (5 → 6 → 7) before the condition `guess < 7` fails. The exercise wants a **bigger step** so the loop converges in fewer iterations.

### Hints

- Replace `guess++` (which adds `1`) with a step that adds **`2`** instead.
- The compound assignment `guess += 2` is the cleanest way to express "add 2 to `guess`".
- Keep the rest of the loop unchanged — only the increment step needs adjusting.
- Verify the trace mentally: starting at `5`, one `+= 2` jump lands on `7`, so the loop body should run **exactly once** before exiting.

### Completed Code

```cpp
#include <iostream>

int main() {
    int secretNumber = 7, guess = 5;
    do {
        std::cout << "The guess is not the secret number." << std::endl;
        guess += 2; // bigger step: jump by 2 instead of 1
    } while (guess < secretNumber);

    std::cout << "Congratulations! You've guessed the secret number: " << guess << std::endl;
    return 0;
}
```

**Output (with `guess += 2`):**

```
The guess is not the secret number.
Congratulations! You've guessed the secret number: 7
```

Walk-through:

- Iteration 1: enter the loop, print `"The guess is not the secret number."`, then `guess += 2` makes `guess = 7`. The condition `7 < 7` evaluates to `false`, so the loop exits.
- After the loop, `guess` is `7` — exactly the secret number — so the closing line prints `7`.

Notice how **one big step** collapsed two iterations into one. The body still ran **at least once** (a do-while guarantee), and the final value of `guess` is still correct.

### Before vs After

| Version       | Increment | Iterations | Final `guess` |
|---------------|-----------|------------|----------------|
| Starter       | `guess++`  (`+1`)     | 2          | `7`            |
| Fixed         | `guess += 2` (`+2`)   | 1          | `7`            |

### Edge Cases Worth Considering

- If `secretNumber = 8` and we still jump by `2` from `5`, we'd land on `7`, then `9` — overshooting the target. A `do-while` based on strict `<` will keep looping until `guess` is **at least** the secret. Watch out for this when changing step sizes — sometimes the loop needs a `>=` guard or a `break`.
- If the starting `guess` is already greater than `secretNumber`, the body runs **once** (do-while guarantee) but no increment is strictly needed.

### Stretch Ideas

- Make the **step size configurable** by adding `const int step = 2;` at the top and using `guess += step;`.
- After the loop, check whether `guess == secretNumber` and print `"You overshot!"` if not.
- Replace `guess += 2` with `std::rand() % 5 + 1` for a randomized step — but guard against infinite loops!

### Takeaway

The **update step** inside a loop controls how fast the loop variable moves toward (or past) the loop guard. Tweak that step and you change how many iterations run, even if everything else stays the same — useful for tuning performance or simulating different "speeds" of progress.

## Mastering Nested Loops in C++

Welcome to the lesson on **nested loops** — loops placed inside other loops. They arise naturally whenever you need to repeat something *for each iteration* of an outer process, such as printing a grid, building a multiplication table, or traversing a 2D array.

---

### What Is a Nested Loop?

A **nested loop** is any loop whose body contains another loop. The **inner loop** completes all its iterations for every single iteration of the **outer loop**.

A real-world analogy: a digital clock. The **minute hand** loops from `0` to `59`. Each time it completes a full cycle, the **hour hand** advances by one — exactly the outer-loop / inner-loop relationship.

---

### Syntax of Nested `for` Loops

```cpp
for (int i = 0; i < n1; i++) {       // Outer loop — runs n1 times
    for (int j = 0; j < n2; j++) {   // Inner loop — runs n2 times per outer iteration
        // code executed n1 × n2 times total
    }
    // code executed once per outer iteration
}
```

- `n1` controls how many times the outer loop runs.
- `n2` controls how many times the inner loop runs **per outer iteration**.
- Total inner-loop executions: `n1 × n2`.

---

### First Example — Asterisk Grid

```cpp
#include <iostream>

int main() {
    int rows = 5, cols = 5;

    for (int i = 0; i < rows; i++) {       // outer: one pass per row
        for (int j = 0; j < cols; j++) {   // inner: one star per column
            std::cout << "* ";
        }
        std::cout << "\n";                 // newline after each row
    }

    return 0;
}
```

**Output:**

```
* * * * *
* * * * *
* * * * *
* * * * *
* * * * *
```

> **Note:** `"\n"` and `std::endl` both move the cursor to the next line. They are interchangeable for simple output; `"\n"` is slightly faster since it doesn't flush the buffer.

---

### Common Pitfall — Wrong Control Variable in Update

A classic nested-loop bug: incrementing the **outer** variable inside the **inner** loop instead of the inner variable.

```cpp
for (int i = 1; i <= 10; i++) {
    for (int j = 0; j <= 10; i++) {  // BUG: i++ should be j++
        std::cout << i << " x " << j << " = " << i * j << "\n";
    }
}
```

`i` grows inside the inner loop, so `j` never advances — the inner loop runs forever (infinite loop). The fix:

```cpp
for (int j = 0; j <= 10; j++) {  // correct: increment j
```

**Rule:** Every loop's update statement must increment **its own** control variable.

---

### Application — Traversing a 2D Array

Nested loops are the standard way to iterate over a **2D array** (rows × columns):

```cpp
#include <iostream>

int main() {
    int arr[2][3] = {{1, 2, 3}, {4, 5, 6}};

    for (int i = 0; i < 2; i++) {
        for (int j = 0; j < 3; j++) {
            std::cout << "arr[" << i << "][" << j << "] = " << arr[i][j] << "\n";
        }
    }

    return 0;
}
```

**Output:**

```
arr[0][0] = 1
arr[0][1] = 2
arr[0][2] = 3
arr[1][0] = 4
arr[1][1] = 5
arr[1][2] = 6
```

The outer loop steps through **rows** (`i`), the inner loop through **columns** (`j`) — visiting every element exactly once.

---

### Lesson Summary

| Concept | Key Point |
|---------|-----------|
| **Nested loop** | A loop inside another loop |
| **Execution count** | Inner body runs `n1 × n2` times |
| **Control variables** | Each loop must update its **own** variable |
| **`\n` vs `std::endl`** | Both print a newline; `\n` doesn't flush the buffer |
| **Common bug** | Incrementing the wrong variable → infinite inner loop |
| **Use cases** | Grids, patterns, multiplication tables, 2D arrays |

Practice exercises follow — focus on tracking both `i` and `j` separately as you trace through each loop. Happy coding!

## Simulate a Digital Clock with Nested Loops

Use nested `for` loops to simulate a **12-hour clock** — hours `1` to `12`, and for each hour every minute in **5-minute intervals** (`0, 5, 10, …, 55`). Print each time in `H:MM` format (no leading zero on hours).

---

### Starter Code

```cpp
#include <iostream>

int main() {
    // TODO: outer loop over hours (1–12)
    //   TODO: inner loop over minutes (0–55, step 5)
    //     TODO: print time in H:MM format
    return 0;
}
```

---

### Building the Solution

- **Outer loop:** `for (int hour = 1; hour <= 12; hour++)` — starts at `1`, no leading zero needed
- **Inner loop:** `for (int minute = 0; minute <= 55; minute += 5)` — step of `5`
- **Minute formatting:** print `"0"` before minutes less than `10`; hours print as-is

---

### Completed Code

```cpp
#include <iostream>

int main() {
    for (int hour = 1; hour <= 12; hour++) {
        for (int minute = 0; minute <= 55; minute += 5) {
            std::cout << hour
                      << ":"
                      << (minute < 10 ? "0" : "") << minute
                      << "\n";
        }
    }
    return 0;
}
```

---

### Output (first two hours shown)

```
1:00
1:05
1:10
1:15
1:20
1:25
1:30
1:35
1:40
1:45
1:50
1:55
2:00
2:05
...
12:55
```

---

### Output Trace

| `hour` | `minute` | Printed  |
|--------|----------|----------|
| 1      | 0        | `1:00`   |
| 1      | 5        | `1:05`   |
| 1      | 10       | `1:10`   |
| 1      | …        | …        |
| 1      | 55       | `1:55`   |
| 2      | 0        | `2:00`   |
| …      | …        | …        |
| 12     | 55       | `12:55`  |

Total lines: `12 hours × 12 steps = 144`.

> **Key takeaway:** The inner loop's **step size** (`minute += 5`) controls how many iterations run per outer loop pass. Changing the step from `1` to `5` reduces inner iterations from `60` to `12` — the outer loop structure stays identical.

## Adjust the Digital Clock from 24-Hour to 12-Hour Format

Take the 24-hour clock starter below and modify it to a **12-hour format** with 5-minute intervals:

- Hours run from `1` to `12` (no leading zero)
- Minutes step by `5` (`0, 5, 10, …, 55`)

---

### Starter Code (24-hour format)

```cpp
#include <iostream>

int main() {
    for (int hour = 0; hour <= 23; hour++) {
        for (int minute = 0; minute <= 59; minute++) {
            std::cout << (hour < 10 ? "0" : "") << hour
                      << ":"
                      << (minute < 10 ? "0" : "") << minute
                      << "\n";
        }
    }
    return 0;
}
```

**Current output (first 3 lines):**

```
00:00
00:01
00:02
```

---

### What to Change

| What | From | To |
|------|------|----|
| Hour start | `hour = 0` | `hour = 1` |
| Hour end | `hour <= 23` | `hour <= 12` |
| Minute step | `minute++` | `minute += 5` |
| Minute end | `minute <= 59` | `minute <= 55` |
| Hour leading zero | `(hour < 10 ? "0" : "") << hour` | just `hour` |

---

### Completed Code

```cpp
#include <iostream>

int main() {
    for (int hour = 1; hour <= 12; hour++) {
        for (int minute = 0; minute <= 55; minute += 5) {
            std::cout << hour
                      << ":"
                      << (minute < 10 ? "0" : "") << minute
                      << "\n";
        }
    }
    return 0;
}
```

---

### Output (first and last hour shown)

```
1:00
1:05
1:10
1:15
1:20
1:25
1:30
1:35
1:40
1:45
1:50
1:55
...
12:00
12:05
...
12:55
```

Total lines: `12 hours × 12 steps = 144`.

> **Key takeaway:** Converting 24→12 hour format requires changing **three parts of the outer loop** (start, end, formatting) and **two parts of the inner loop** (step size, end condition). The nested structure itself stays unchanged — only the loop parameters shift.

## Debug the Digital Clock Display

Oh no, our 5-minute intervals digital clock code is supposed to display hours and minutes, but it's incorrectly printing the minutes for some times. Can you find and fix the bug so that it properly shows the correct time format, especially for minutes less than 10?

---

### Buggy Code

```cpp
#include <iostream>

int main() {
    // Digital clock: outer loop for hours, inner loop for minutes
    for (int hour = 0; hour < 12; hour++) { // 12-hour clock
        for (int minute = 0; minute < 60; minute += 5) { // 60 minutes in an hour
            // Print time in format H:MM
            std::cout << hour << ":";
            if (minute < 10) {
               std::cout << "";
            }
            std::cout << minute << " ";
        }
        std::cout << "\n"; // Next line after every hour
    }
    return 0;
}
```

**Buggy output (first hour):**

```
0:0 0:5 0:10 0:15 0:20 0:25 0:30 0:35 0:40 0:45 0:50 0:55
```

Minutes under `10` are missing their leading zero — `0:5` should read `0:05`.

---

### Finding the Bug

- The `if (minute < 10)` check correctly detects single-digit minutes.
- But the body prints `""` (an **empty string**) instead of `"0"` — so nothing is added before the minute digit.
- **Fix:** change `std::cout << "";` to `std::cout << "0";`.

---

### Fixed Code

```cpp
#include <iostream>

int main() {
    for (int hour = 0; hour < 12; hour++) {
        for (int minute = 0; minute < 60; minute += 5) {
            std::cout << hour << ":";
            if (minute < 10) {
               std::cout << "0";
            }
            std::cout << minute << " ";
        }
        std::cout << "\n";
    }
    return 0;
}
```

---

### Output Trace (first hour)

| `hour` | `minute` | `minute < 10`? | Printed |
|--------|----------|----------------|---------|
| 0      | 0        | yes → `"0"`    | `0:00`  |
| 0      | 5        | yes → `"0"`    | `0:05`  |
| 0      | 10       | no             | `0:10`  |
| 0      | 15       | no             | `0:15`  |
| …      | …        | …              | …       |
| 0      | 55       | no             | `0:55`  |

**Corrected output (first hour):**

```
0:00 0:05 0:10 0:15 0:20 0:25 0:30 0:35 0:40 0:45 0:50 0:55
```

> **Key takeaway:** Printing an empty string `""` silently does nothing — it's easy to mistake it for "print nothing extra" when you actually meant "print a placeholder character." Always double-check that a conditional format branch prints the **intended literal**, not just an empty one.

## Digital Clock Hour and Minute Match Simulator

Using nested `for` loops, simulate a **12-hour clock** (hours `1`–`12`, checking **every minute** `0`–`59`) but only print a time when the **hour matches the minute** (e.g. `1:01`, `2:02`, ..., `12:12`).

---

### Starter Code

```cpp
#include <iostream>

int main() {
    // TODO: outer loop over hours (1–12)
    //   TODO: inner loop over every minute (0–59)
    //     TODO: if hour == minute, print the time in H:MM format
    return 0;
}
```

---

### Building the Solution

- **Outer loop:** `for (int hour = 1; hour <= 12; hour++)`
- **Inner loop:** `for (int minute = 0; minute <= 59; minute++)` — must check **every** minute, not just multiples of `5`, or matches like `1:01` and `2:02` get skipped
- **Match condition:** `if (hour == minute)` — a match exists for every hour `1`–`12`, since each hour value also appears as a minute value in `0`–`59`
- **Minute formatting:** print `"0"` before minutes less than `10`

---

### Completed Code

```cpp
#include <iostream>

int main() {
    for (int hour = 1; hour <= 12; hour++) {
        for (int minute = 0; minute <= 59; minute++) {
            if (hour == minute) {
                std::cout << hour
                          << ":"
                          << (minute < 10 ? "0" : "") << minute
                          << "\n";
            }
        }
    }
    return 0;
}
```

---

### Output

```
1:01
2:02
3:03
4:04
5:05
6:06
7:07
8:08
9:09
10:10
11:11
12:12
```

---

### Output Trace

| `hour` | `minute` | `hour == minute`? | Printed |
|--------|----------|--------------------|---------|
| 1      | 0        | no                 | —       |
| 1      | 1        | **yes**            | `1:01`  |
| 1      | 2–59     | no                 | —       |
| 2      | 2        | **yes**            | `2:02`  |
| …      | …        | …                  | …       |
| 12     | 12       | **yes**            | `12:12` |

> **Key takeaway:** A step of `5` in the inner loop only visits minutes `0, 5, 10, …` — it silently skips values like `1`, `2`, `3`, `4`, so matches such as `1:01` or `2:02` never get checked. Whenever every value in a range matters (not just some), the inner loop's increment must be `1` (`minute++`), even though that means more total iterations (`12 × 60 = 720` instead of `12 × 12 = 144`).

## Simulate a Digital Clock in C++

Capstone exercise: use nested `for` loops to simulate a full **24-hour digital clock**, printing **every minute** of every hour (hours `00`–`23`, minutes `00`–`59`) in `HH:MM` format, with leading zeros on both hours and minutes.

---

### Starter Code

```cpp
#include <iostream>

int main() {
    // TODO: outer loop over hours (0–23)
    //   TODO: inner loop over every minute (0–59)
    //     TODO: print the time in HH:MM format, with leading zeros
    return 0;
}
```

---

### Building the Solution

- **Outer loop:** `for (int hour = 0; hour <= 23; hour++)` — full 24-hour range, starting at `0`
- **Inner loop:** `for (int minute = 0; minute <= 59; minute++)` — every minute, step of `1`
- **Hour formatting:** print `"0"` before hours less than `10` (`00`–`09`)
- **Minute formatting:** print `"0"` before minutes less than `10` (`00`–`09`)
- **Total iterations:** `24 hours × 60 minutes = 1440` lines of output

---

### Completed Code

```cpp
#include <iostream>

int main() {
    for (int hour = 0; hour <= 23; hour++) {
        for (int minute = 0; minute <= 59; minute++) {
            std::cout << (hour < 10 ? "0" : "") << hour
                      << ":"
                      << (minute < 10 ? "0" : "") << minute
                      << "\n";
        }
    }
    return 0;
}
```

---

### Output (first and last hour shown)

```
00:00
00:01
00:02
...
00:59
01:00
...
23:00
23:01
...
23:59
```

---

### Output Trace

| `hour` | `minute` | Printed |
|--------|----------|---------|
| 0      | 0        | `00:00` |
| 0      | 1        | `00:01` |
| 0      | …        | …       |
| 0      | 59       | `00:59` |
| 1      | 0        | `01:00` |
| …      | …        | …       |
| 23     | 59       | `23:59` |

Total lines: `24 hours × 60 minutes = 1440`.

> **Key takeaway:** A 24-hour clock needs leading-zero formatting on **both** the hour and the minute, since single-digit hours (`0`–`9`) are just as common as single-digit minutes. The ternary `(value < 10 ? "0" : "")` pattern applies identically to either field — only the loop's range (`0`–`23` vs `1`–`12`) changes between 24-hour and 12-hour formats.

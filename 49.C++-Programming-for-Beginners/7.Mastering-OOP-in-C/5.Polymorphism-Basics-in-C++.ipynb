## Polymorphism Basics in C++

**Polymorphism** is a key concept in C++ that allows objects from different classes to be treated as objects of a shared base class. It makes your code more flexible and reusable. The term "polymorphism" stems from Greek words meaning "many forms."

Our goal in this lesson is to understand the basics of polymorphism in C++ by working with a chess example.

---

### What is Polymorphism and Why Do We Need It?

Polymorphism allows you to design and implement programs that are more flexible and easier to extend over time. It enables one interface to be used for a general class of actions, with specific actions determined by the exact nature of the situation.

In C++, polymorphism is primarily achieved through the use of **base class pointers or references** to invoke methods in derived classes. This makes it possible to write code that can work with objects of different types and classes in a unified way. Sounds difficult? It really isn't! Let's see an example, it will help you understand the idea.

---

### Define an Abstract Class: `ChessPiece`

An **abstract class** in C++ is a class that cannot be instantiated on its own and is designed to be inherited by other classes. It serves as a blueprint for derived classes. An abstract class typically includes one or more **pure virtual functions**.

A pure virtual function is a function declaration that is specified in the base class but has no implementation in that class. Instead, it must be implemented by any non-abstract derived class. This is done by assigning `0` to the function declaration in the base class.

Additionally, it's a good practice to have a **virtual destructor** in an abstract class. This ensures that the destructor for derived classes will be called correctly, preventing memory leaks.

We'll define an abstract class `ChessPiece` with a pure virtual function `draw` and a virtual destructor. This function will be overridden by derived classes.

```cpp
#include <iostream>

class ChessPiece {
public:
    virtual ~ChessPiece() {}  // Virtual destructor
    virtual void draw() const = 0; // Pure virtual function
};
```

Here, `ChessPiece` serves as a blueprint for all chess pieces, and `draw` is a method that each piece must implement.

Note that you can't create an instance of a class having at least one pure virtual function, because this class is abstract. For example, this is not allowed:

```cpp
int main() {
    ChessPiece abstract_piece = new ChessPiece();
    abstract_piece->draw();
}
```

This code will cause an error.

---

### Derive Classes `Pawn` and `Knight` from `ChessPiece`

Next, we'll create two derived classes, `Pawn` and `Knight`, which will implement the `draw` method.

```cpp
class Pawn : public ChessPiece {
public:
    void draw() const override {
        std::cout << "_Pawn_" << std::endl;
    }
};

class Knight : public ChessPiece {
public:
    void draw() const override {
        std::cout << "_Knight_" << std::endl;
    }
};
```

Here, both classes `Pawn` and `Knight` provide their own implementations of the `draw` method.

---

### Store `Pawn` Objects and `Knight` Objects in One Array

What if we want to create an array to store all the chess pieces?

Using pointers to the base class allows you to create an array capable of holding objects of any derived class. This is crucial because C++ is a statically-typed language, and the compiler needs to know the type and size of each array element at compile-time. By storing pointers to the base class, you enable the compiler to treat all array elements as having the same type (pointers to the base class), thus achieving polymorphic behavior.

Let's see an example:

```cpp
int main() {
    ChessPiece* pieces[10]; // Array of base class pointers

    for (int i = 0; i < 8; ++i) {
        pieces[i] = new Pawn();
    }

    for (int i = 8; i < 10; ++i) {
        pieces[i] = new Knight();
    }
```

Here, we create an array `pieces` of `ChessPiece` pointers. Both `Pawn` and `Knight` are derived from `ChessPiece`, so both can be initialized with a `ChessPiece*` pointer. The code shows how to store 8 `Pawn` objects and 2 `Knight` objects in this array.

---

### Call the `draw` Method for Each Object in the Array

Finally, we'll loop through the array and call the `draw` method on each object. This demonstrates polymorphic behavior. (This snippet continues the same `main` function from above.)

```cpp
    for (int i = 0; i < 10; ++i) {
        pieces[i]->draw(); // Calls appropriate overridden function
    }

    for (int i = 0; i < 10; ++i) {
        delete pieces[i]; // Clean up the heap-allocated memory
    }

    return 0;
}
```

In this loop, each `pieces[i]->draw()` call invokes the `draw` method for the appropriate `Pawn` or `Knight` object. This way, we have a uniform code to work with each object. You can add more pieces of different types, and you won't need to modify this draw loop: it will automatically call the appropriate `draw` method for each piece.

---

### Non-abstract Base Class

Polymorphism works even when the base class is not abstract. Let's take a look at this example:

```cpp
class ChessPiece {
public:
    virtual void draw() {
        std::cout << "Can't draw an abstract piece!" << std::endl;
    }

    virtual ~ChessPiece() {}  // Virtual destructor
};

class Pawn : public ChessPiece {
public:
    void draw() {
        std::cout << "_Pawn_" << std::endl;
    }
};

class Knight : public ChessPiece {
public:
    void draw() {
        std::cout << "_Knight_" << std::endl;
    }
};
```

It will work in exactly the same way. But in this case, the `ChessPiece` class is **not abstract**, as its only method, `draw`, is defined. We can create instances of the `ChessPiece` class and operate them.

**Buggy behavior:** The code compiles, but all 10 calls print `Can't draw an abstract piece!` instead of `_Pawn_` / `_Knight_`.

---

### Finding the Bug

| Question | Answer |
|----------|--------|
| **What happens now?** | `draw()` is not `virtual`, so `pieces[i]->draw()` is resolved at compile time from the pointer's type (`ChessPiece*`). It always calls `ChessPiece::draw`, whatever object the pointer actually points to. |
| **What fixes it?** | Mark `ChessPiece::draw` as `virtual`. The call is then resolved at runtime from the real object, so `Pawn::draw` or `Knight::draw` runs. |
| **Why `override`?** | It makes the compiler check that the derived method really overrides a base `virtual` method. A typo or a signature mismatch becomes a compile error instead of a silent bug. |

---

### Fixed Code

```cpp
#include <iostream>

class ChessPiece {
public:
    virtual void draw() {
        std::cout << "Can't draw an abstract piece!" << std::endl;
    }
};

class Pawn : public ChessPiece {
public:
    void draw() override {
        std::cout << "_Pawn_" << std::endl;
    }
};

class Knight : public ChessPiece {
public:
    void draw() override {
        std::cout << "_Knight_" << std::endl;
    }
};

int main() {
    ChessPiece* pieces[10];

    for (int i = 0; i < 8; ++i) {
        pieces[i] = new Pawn();
    }

    for (int i = 8; i < 10; ++i) {
        pieces[i] = new Knight();
    }

    for (int i = 0; i < 10; ++i) {
        pieces[i]->draw();
    }

    for (int i = 0; i < 10; ++i) {
        delete pieces[i];
    }

    return 0;
}
```

```cpp
#include <iostream>
#include <cmath> // For M_PI

class Shape {
public:
    virtual ~Shape() {}  // Virtual destructor
    virtual double calculateArea() const = 0;  // Pure virtual function
};

class Circle : public Shape {
private:
    double radius;
public:
    Circle(double r) : radius(r) {}

    double calculateArea() const override {
        return M_PI * radius * radius;
    }
};

class Rectangle : public Shape {
private:
    double width;
    double height;
public:
    Rectangle(double w, double h) : width(w), height(h) {}

    double calculateArea() const override {
        return width * height;
    }
};

int main() {
    Shape* shapes[4]; // Array of base class pointers

    shapes[0] = new Circle(1);
    shapes[1] = new Circle(2);
    shapes[2] = new Rectangle(3, 4);
    shapes[3] = new Rectangle(5, 6);

    for (int i = 0; i < 4; ++i) {
        std::cout << "Area: " << shapes[i]->calculateArea() << std::endl;
    }

    for (int i = 0; i < 4; ++i) {
        delete shapes[i]; // Clean up the heap-allocated memory
    }

    return 0;
}
```

### Completed Code

```cpp
#include <iostream>
#include <string>
#include <cmath> // For M_PI

class Shape {
public:
    virtual double calculateArea() const = 0; // Pure virtual function
    virtual std::string to_string() const = 0; // Pure virtual function for description
};

class Circle : public Shape {
private:
    double radius;
public:
    Circle(double r) : radius(r) {}

    double calculateArea() const override {
        return M_PI * radius * radius; // Area of circle
    }

    std::string to_string() const override {
        return "Circle with radius " + std::to_string(radius);
    }
};

class Rectangle : public Shape {
private:
    double width;
    double height;
public:
    Rectangle(double w, double h) : width(w), height(h) {}

    double calculateArea() const override {
        return width * height; // Area of rectangle
    }

    std::string to_string() const override {
        return "Rectangle with width " + std::to_string(width) +
               " and height " + std::to_string(height);
    }
};

int main() {
    Shape* shapes[4]; // Array of base class pointers

    shapes[0] = new Circle(5.0);
    shapes[1] = new Rectangle(4.0, 6.0);
    shapes[2] = new Circle(3.0);
    shapes[3] = new Rectangle(2.0, 8.0);

    for (int i = 0; i < 4; ++i) {
        std::cout << shapes[i]->to_string() << " - Area: " << shapes[i]->calculateArea() << std::endl;
    }

    for (int i = 0; i < 4; ++i) {
        delete shapes[i]; // Clean up the heap-allocated memory
    }

    return 0;
}
```

| **2. Fill the array** | Use `new Warrior()` for two slots and `new Mage()` for the other two. A `Warrior*` and a `Mage*` both convert to `RPGCharacter*` because of public inheritance. |
| **3. Call `attack`** | A `for` loop from `0` to `3` calling `characters[i]->attack();`. Because `attack` is virtual, each object runs its own version. |
| **4. Clean up** | A second loop calling `delete characters[i];` frees each heap-allocated object. |

> The task didn't say in which order to place the characters, so I alternated them: `Warrior`, `Mage`, `Warrior`, `Mage`. Any order with two of each works.
>
> The task text had the typo `atttack`; I fixed it to `attack` in the description.

### Completed Code

```cpp
#include <iostream>

// Abstract class
class RPGCharacter {
public:
    virtual void attack() const = 0; // Pure virtual function
};

// Derived class Warrior
class Warrior : public RPGCharacter {
public:
    void attack() const override {
        std::cout << "Warrior attacks with a sword!" << std::endl;
    }
};

// Derived class Mage
class Mage : public RPGCharacter {
public:
    void attack() const override {
        std::cout << "Mage casts a fireball!" << std::endl;
    }
};

int main() {
    RPGCharacter* characters[4]; // Array of base class pointers

    characters[0] = new Warrior();
    characters[1] = new Mage();
    characters[2] = new Warrior();
    characters[3] = new Mage();

    for (int i = 0; i < 4; ++i) {
        characters[i]->attack(); // Calls the right attack for each character
    }

    for (int i = 0; i < 4; ++i) {
        delete characters[i]; // Clean up the heap-allocated memory
    }

    return 0;
}
```

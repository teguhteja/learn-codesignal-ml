## Access Specifiers in Inheritance

Welcome to our lesson on "Access Specifiers in Inheritance" in C++! Access specifiers (`public`, `protected`, and `private`) are key in object-oriented programming as they help control access to class members. This lesson will focus on understanding how these access specifiers work in inheritance.

By the end of this lesson, you'll know how access specifiers affect the accessibility of inherited class members.

---

### Access Specifiers Overview

In C++, access specifiers set the accessibility of class members (attributes and methods). The primary access specifiers are:

| Specifier | Who can access the member |
|-----------|---------------------------|
| **`public`** | Accessible from outside the class |
| **`protected`** | Accessible within the class and by derived class instances |
| **`private`** | Accessible only within the class declaring them |

These controls help ensure data encapsulation and security by preventing unauthorized access. We have already seen `public` and `private` specifiers in the previous course. Let's discuss `protected` in more detail.

---

### The Protected Specifier in Detail

The `protected` specifier in C++ allows members to be accessible within the same class, as well as any derived classes. However, these members are not accessible from code that is outside the class hierarchy.

Let's consider the following example to understand the `protected` specifier more clearly:

```cpp
#include <iostream>

class Parent {
    protected:
        int protectedVar;
    public:
        Parent() : protectedVar(0) {}
        void setProtectedVar(int val) {
            protectedVar = val;
        }
};

class Child : public Parent {
    public:
        void display() {
            std::cout << "protectedVar: " << protectedVar << std::endl;  // OK: Accessible within derived class
        }
};

int main() {
    Child obj;
    // obj.protectedVar = 10;  // Error: protectedVar is not accessible outside the class hierarchy
    obj.setProtectedVar(10);  // OK: Access through public member function of the base class
    obj.display();            // Output: protectedVar: 10
    return 0;
}
```

In this example:

- `protectedVar` is declared as `protected` in the `Parent` class.
- The `Child` class, which is derived from `Parent`, can access `protectedVar` directly within its member functions.
- From `main()`, direct access to `protectedVar` is not allowed, ensuring that it remains encapsulated within the class hierarchy.

**Advantages of Using Protected:**

- **Encapsulation with Controlled Exposure:** While `private` members totally encapsulate data, `protected` members allow derived classes to reuse and extend functionality.
- **Inheritance Flexibility:** Derived classes can build upon the base class without exposing sensitive data to the outside world.

---

### Public Inheritance Explained

As you might recall, the syntax of inheritance is the following:

```cpp
class Derived : public Base
```

Here, we specify the **inheritance access specifier**, which defines what happens to the class fields' specifiers in the derived class. So far, we only used the `public` inheritance access specifier.

When a class inherits from another class publicly, the access specifiers of the base class members are **preserved** in the derived class.

```cpp
#include <iostream>

class Base {
    public:
        int publicVar;
    protected:
        int protectedVar;
    private:
        int privateVar;
};

class Derived : public Base {
    public:
        void display() {
            publicVar = 1;       // OK: public in Base remains public in Derived
            protectedVar = 2;    // OK: protected in Base remains protected in Derived
            // privateVar = 3;   // Error: private in Base is not accessible in Derived
            std::cout << "publicVar: " << publicVar << " protectedVar: " << protectedVar << std::endl;
        }
};

int main() {
    Derived obj;
    obj.display();  // Output: publicVar: 1 protectedVar: 2
    return 0;
}
```

Here:

- `publicVar`: Accessible in `Derived` because it stays public.
- `protectedVar`: Accessible through member functions because it remains protected.
- `privateVar`: Inaccessible in `Derived` because it is private in `Base`.

---

### Protected Inheritance Explained

In **protected inheritance**, public and protected members of the base class become **protected** in the derived class.

```cpp
#include <iostream>

class Base {
    public:
        int publicVar;
    protected:
        int protectedVar;
    private:
        int privateVar;
};

class Derived : protected Base {
    public:
        void display() {
            publicVar = 1;       // OK: public in Base becomes protected in Derived
            protectedVar = 2;    // OK: protected in Base remains protected in Derived
            // privateVar = 3;   // Error: private in Base is not accessible in Derived
            std::cout << "publicVar: " << publicVar << " protectedVar: " << protectedVar << std::endl;
        }
};

int main() {
    Derived obj;
    // obj.publicVar = 1;     // Error: publicVar is now protected in Derived
    obj.display();  // Output: publicVar: 1 protectedVar: 2
    return 0;
}
```

Here:

- `publicVar`: Becomes protected, not accessible directly from `main`.
- `protectedVar`: Remains protected, accessible within `Derived`.
- `privateVar`: Inaccessible in `Derived`.

---

### Private Inheritance Explained

In **private inheritance**, public and protected members of the base class become **private** in the derived class.

```cpp
#include <iostream>

class Base {
    public:
        int publicVar;
    protected:
        int protectedVar;
    private:
        int privateVar;
};

class Derived : private Base {
    public:
        void display() {
            publicVar = 1;       // OK: public in Base becomes private in Derived
            protectedVar = 2;    // OK: protected in Base becomes private in Derived
            // privateVar = 3;   // Error: private in Base is not accessible in Derived
            std::cout << "publicVar: " << publicVar << " protectedVar: " << protectedVar << std::endl;
        }
};

int main() {
    Derived obj;
    // obj.publicVar = 1;     // Error: publicVar is now private in Derived
    obj.display();  // Output: publicVar: 1 protectedVar: 2
    return 0;
}
```

Here:

- `publicVar`: Becomes private in `Derived`, not accessible directly from `main`.
- `protectedVar`: Becomes private in `Derived`, not accessible from `main`.
- `privateVar`: Inaccessible in `Derived`.

---

### Lesson Summary

Awesome job! You've learned how access specifiers control member accessibility in inheritance. To recap:

| Inheritance type | Base `public` members become | Base `protected` members become | Base `private` members |
|------------------|------------------------------|---------------------------------|------------------------|
| **Public** (`: public Base`) | `public` | `protected` | Inaccessible |
| **Protected** (`: protected Base`) | `protected` | `protected` | Inaccessible |
| **Private** (`: private Base`) | `private` | `private` | Inaccessible |

These concepts help secure and organize your C++ programs.

## Accessing Protected Members in Derived Class

Hello, Stellar Navigator!

In this code, something is wrong with the base class' access specifier. It seems to make the name field inaccessible from the derived class. One important thing is that you should preserve encapsulation, meaning the name field shouldn't be accessible from the main function.

Use your knowledge to fix this problem and make this code work!

---

### Buggy Code

```cpp
#include <iostream>
#include <string>

class Person {
    private:
        std::string name;
    public:
        Person(std::string n) : name(n) {}
};

class Employee : public Person {
    public:
        Employee(std::string n) : Person(n) {}
        void display() {
            std::cout << "Employee Name: " << name << std::endl;
        }
};

int main() {
    Employee emp("John Doe");
    emp.display();
    return 0;
}
```

**Buggy behavior:** The code does not compile. `name` is `private` in `Person`, so `Employee::display()` cannot access it (error along the lines of `'std::string Person::name' is private within this context`).

---

### Finding the Bug

| Question | Answer |
|----------|--------|
| **What fails?** | `display()` in `Employee` reads `name`, but `private` members are accessible only inside the class that declares them, not in derived classes. |
| **Why not make it `public`?** | That would fix `display()`, but `main` could then write `emp.name`, which breaks the required encapsulation. |
| **Which specifier fits?** | `protected`: accessible in `Person` and its derived classes, but not from outside the hierarchy (such as `main`). |

The fix is a single word: change `private:` to `protected:` in `Person`.

---

### Fixed Code

```cpp
#include <iostream>
#include <string>

class Person {
    protected:
        std::string name;
    public:
        Person(std::string n) : name(n) {}
};

class Employee : public Person {
    public:
        Employee(std::string n) : Person(n) {}
        void display() {
            std::cout << "Employee Name: " << name << std::endl;
        }
};

int main() {
    Employee emp("John Doe");
    emp.display();
    return 0;
}
```

---

### Output

```
Employee Name: John Doe
```

### Output Trace

| Step | Code | What happens | Output |
|------|------|--------------|--------|
| 1 | `Employee emp("John Doe")` | `Employee` passes `"John Doe"` to `Person(n)`, which sets `name` | *(none)* |
| 2 | `emp.display()` | `name` is `protected` in `Person`, so `Employee` can read it | `Employee Name: John Doe` |
| 3 | *(not in the code)* `emp.name` in `main` | Would not compile: `protected` members are hidden from code outside the hierarchy | *(n/a)* |

> **Key takeaway:** `private` hides a member even from derived classes, while `protected` shares it with derived classes but still hides it from the outside world. Use `protected` when a derived class needs direct access and you still want to preserve encapsulation. Making `name` `public` would also compile, but would let `main` modify it.

### Starter Code

```cpp
#include <iostream>

class Vehicle {
    public:
        int speed;
    protected:
        int fuel;
    private:
        int engineStatus;
};

class Car : public Vehicle { // Change this inheritance specifier to protected
    public:
        void display() {
            speed = 60;
            fuel = 30;
            std::cout << "Speed: " << speed << " Fuel: " << fuel << std::endl;
        }
};

int main() {
    Car myCar;
    myCar.speed = 60;     // Currently works, will it work after the change?
    myCar.display();  // Output: Speed: 60 Fuel: 30
    return 0;
}
```

---

### Finding the Bug

After changing `public Vehicle` to `protected Vehicle`, the program no longer compiles.

| Question | Answer |
|----------|--------|
| **What fails?** | `myCar.speed = 60;` in `main`. With protected inheritance, the public member `speed` becomes `protected` in `Car`, so code outside the class hierarchy can't access it. |
| **Does `display()` still work?** | Yes. `Car`'s own methods can use `speed` (now protected) and `fuel` (still protected). |
| **How do we restore the old behavior?** | Add a **public setter** to `Car` that sets `speed` from inside the class, then call it from `main` instead of assigning directly. |

---

### Building the Solution

| Step | What to do |
|------|------------|
| **1. Change the inheritance** | `class Car : protected Vehicle` |
| **2. Add a setter** | In `Car`'s `public` section: `void setSpeed(int s) { speed = s; }`. It's allowed to touch `speed` because it is a member of `Car`. |
| **3. Update `main`** | Replace `myCar.speed = 60;` with `myCar.setSpeed(60);` |

> The task didn't name the setter, so `setSpeed` is my choice.

---

### Completed Code

```cpp
#include <iostream>

class Vehicle {
    public:
        int speed;
    protected:
        int fuel;
    private:
        int engineStatus;
};

class Car : protected Vehicle { // Changed to protected inheritance
    public:
        void setSpeed(int s) {
            speed = s;
        }

        void display() {
            speed = 60;
            fuel = 30;
            std::cout << "Speed: " << speed << " Fuel: " << fuel << std::endl;
        }
};

int main() {
    Car myCar;
    myCar.setSpeed(60);   // Direct access to speed is no longer allowed, so use the setter
    myCar.display();  // Output: Speed: 60 Fuel: 30
    return 0;
}
```

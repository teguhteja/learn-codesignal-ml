## Destructors in C++

Hello! Today, we are going to learn about a special type of function in C++ called a **destructor**. Destructors play a crucial role in managing resources, like memory, in programming. If you've ever played with toys and had to clean up, think of a destructor as the tool that helps clean up after the objects your program uses.

By the end of this lesson, you'll understand what destructors are, how to use them, and why they're important.

---

### Understanding Destructors

In C++, a destructor is a special member function of a class that runs when an object of that class **goes out of scope** or is **explicitly deleted**. Imagine you have a paper airplane, and when you're done playing with it, you need to recycle the paper. A destructor helps with that recycling process in programming.

Destructors are vital because they ensure that resources, like memory or files, are properly released. This prevents memory leaks.

---

### Destructor Syntax

A destructor has the same name as the class but is preceded by a tilde (`~`). It doesn't take any arguments and doesn't return any value. Additionally, a class can only have **one** destructor, so you can't overload it.

Here's a simple example:

```cpp
class MyClass {
public:
    // Constructor
    MyClass() {
        // Constructor code
    }

    // Destructor
    ~MyClass() {
       // Destructor code
    }
};
```

This class `MyClass` has a constructor (to set up the object) and a destructor (to clean up the object). If you don't define a destructor, the compiler provides a default one. However, it's good practice to always declare your own destructor if your class manages resources like dynamic memory, file handles, or network connections. This ensures that resources are correctly and explicitly released.

You can also define the destructor outside the class, just like with other methods:

```cpp
// Destructor implementation outside the class
MyClass::~MyClass() {
    // Destructor code
}
```

---

### Destructors in Action

Let's explore how destructors work with a practical example. We will create a class called `MyClass`, and our program will print messages when the constructor and destructor are called.

```cpp
#include <iostream>

class MyClass {
public:
    MyClass() : data(nullptr) {
        std::cout << "Constructor called!" << std::endl;
    }
    ~MyClass() {
        std::cout << "Destructor called!" << std::endl;
    }

private:
    int* data;
};

int main() {
    MyClass obj; // Constructor called here
    return 0; // Destructor called here
}
```

Here's what's happening:

- When `obj` is created, the constructor runs and prints `"Constructor called!"`.
- When `obj` goes out of scope (at the end of `main`), the destructor runs and prints `"Destructor called!"`.

---

### Key Points to Remember

| Point | Description |
|---|---|
| **Automatic calling** | Destructors are called automatically when an object goes out of scope or is deleted. |
| **Only one destructor** | A class can have only one destructor. |
| **Default destructor** | If you don't define a destructor, the compiler provides a default one. |
| **Write outside the class** | You can write the body of the destructor outside of the class definition. |
| **No throwing exceptions** | A destructor should not throw exceptions. It might not be safely caught if an exception is thrown during destruction. Destructors can be invoked during the stack unwinding of another exception, which leads to a call to `std::terminate`. |

---

### Practical Use Cases

Now, let's look at an example where a destructor is very useful: managing dynamic memory. Suppose we have a `MyClass` that allocates an array dynamically. The destructor will handle deleting this array.

```cpp
#include <iostream>

class MyClass {
private:
    int* data;
public:
    MyClass(int size) : data(new int[size]) {
        std::cout << "Memory allocated!" << std::endl;
    }
    ~MyClass() {
        delete[] data; // Deallocate memory
        std::cout << "Memory deallocated!" << std::endl;
    }
};

int main() {
    MyClass obj(10); // Memory allocated here
    // Memory deallocated here
}
```

This example shows how destructors help manage dynamic memory effectively. In real life, just as you might need to dispose of garbage properly to keep your room clean, using destructors ensures your program doesn't waste memory or other resources.

---

### Lesson Summary

Destructors are special functions that clean up when objects are no longer needed.

| Concept | Key Point |
|---|---|
| **When they run** | Automatically, when an object goes out of scope or is deleted. |
| **How many** | A class can have only one destructor. |
| **Purpose** | Critical for releasing resources and preventing memory leaks. |
| **Exceptions** | Should not throw exceptions, to ensure safe execution. |

Destructors make sure your programs are tidy and efficient, just like cleaning up after playing keeps your room neat.

---

### Building the Solution

1. **`Resource` destructor.** Add `~Resource()` that prints `"Resource released!"`.
2. **`Container` destructor.** Add `~Container()` that prints `"Container destroyed!"`.
3. **Leave `main` alone.** `c` goes out of scope at the end of `main`, so both destructors run automatically.

**Why the order matters.** `Container` has a `Resource` member (composition), so the member must exist before the container's constructor body runs:

| Phase | Order |
|---|---|
| **Construction** | Members first (`Resource`), then the container's own body (`Container`) |
| **Destruction** | The exact reverse: the container's own body first (`Container`), then its members (`Resource`) |

---

### Completed Code

```cpp
#include <iostream>
using namespace std;

class Resource {
public:
    // Constructor
    Resource() {
        cout << "Resource acquired!" << endl;
    }
    // Destructor
    ~Resource() {
        cout << "Resource released!" << endl;
    }
};

class Container {
    Resource resource;
public:
    // Constructor
    Container() {
        cout << "Container created!" << endl;
    }
    // Destructor
    ~Container() {
        cout << "Container destroyed!" << endl;
    }
};

int main() {
    Container c;
    // The destructors are called here
    return 0;
}
```

| 2 | `Resource resource;` is constructed | Prints `Resource acquired!` |
| 3 | `Container()` body runs | Prints `Container created!` |
| 4 | `return 0;` | `c` goes out of scope, destruction begins |
| 5 | `~Container()` body runs | Prints `Container destroyed!` |
| 6 | Members are destroyed after the body, so `~Resource()` runs | Prints `Resource released!` |

> **Key takeaway:** Destructors run in the reverse order of constructors. The container's constructor finishes after its members are built, and its destructor finishes before its members are destroyed, so the outer object is created last and destroyed first.

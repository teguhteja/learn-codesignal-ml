## Understanding Inheritance in C++

Hello there! Welcome to our lesson on understanding inheritance in C++. Today, we're going to explore an important concept in programming called **inheritance**.

---

### Why Inheritance?

Kids inherit certain traits from their parents, like eye color or hair color. In programming, inheritance works in a similar way. It allows one class (the **child class**) to inherit properties and behaviors from another class (the **parent class**). This helps us write more efficient and organized code by reusing existing code.

By the end of this lesson, you will understand what inheritance is in C++, how it works, and how to use it to create organized and reusable code.

---

### Inheritance Syntax and Basic Example: Part 1

In C++, we use the `:` symbol to indicate inheritance. For now, we'll use the `public` keyword to specify **public inheritance**. We'll discuss other inheritance options in the next lesson. Let's look at a basic example:

```cpp
#include <iostream>

class Cat {
public:
    void voice() {
        std::cout << "Meow!" << std::endl;
    }
};

class Lion : public Cat {
};

class Tiger : public Cat {
};
```

In this example:

- `Cat` is the **base class**.
- `Lion` and `Tiger` are **derived classes** inheriting from `Cat`.
- Both `Lion` and `Tiger` can use the `voice()` method from their parent class `Cat`.

Let's take a look at the `main` function example:

```cpp
int main() {
    Cat cat;
    Lion lion;
    Tiger tiger;

    cat.voice();   // Output: Meow!
    lion.voice();  // Output: Meow!
    tiger.voice(); // Output: Meow!

    return 0;
}
```

In `Lion` and `Tiger`, we don't have to define the `voice()` method; they have it by default because they are inherited from the `Cat` class.

---

### Inheritance Syntax and Basic Example: Part 2

Let's add some attributes to our base class:

```cpp
#include <iostream>

class Cat {
public:
    std::string color;
    int age;

    void voice() {
        std::cout << "Meow!" << std::endl;
    }
};

class Lion : public Cat {
};

class Tiger : public Cat {
};
```

Now, `Cat` is defined by its `color` and `age`. These attributes are also derived by the `Lion` and the `Tiger` classes.

Let's see how we can use them:

```cpp
int main() {
    Lion lion;
    Tiger tiger;

    lion.color = "Golden";
    lion.age = 5;

    tiger.color = "Orange";
    tiger.age = 3;

    std::cout << "Lion: " << lion.color << ", " << lion.age << " years old" << std::endl;   // Output: Lion: Golden, 5 years old
    std::cout << "Tiger: " << tiger.color << ", " << tiger.age << " years old" << std::endl; // Output: Tiger: Orange, 3 years old

    return 0;
}
```

Both `Lion` and `Tiger` can access the `color` and `age` attributes inherited from the `Cat` class.

---

### Overriding and Extending Base Class Functions: Part 1

Sometimes, we need to change the behavior of a base class method in a derived class. This is known as **function overriding**. To do this, we must mark the method in the base class as `virtual` and use the `override` keyword in the derived class. Using `override` helps to avoid errors by ensuring that the derived class method is actually overriding a base class method.

Example:

```cpp
#include <iostream>

class Cat {
public:
    virtual void voice() {
        std::cout << "Meow!" << std::endl;
    }
};

class Lion : public Cat {
public:
    void voice() override {
        std::cout << "Roar!" << std::endl;
    }
};

class Tiger : public Cat {
public:
    void voice() override {
        std::cout << "Grr!" << std::endl;
    }
};
```

`Tiger` and `Lion` classes redefine the `voice` method to have their own unique behavior.

---

### Overriding and Extending Base Class Functions: Part 2

Now, each type of cat has a unique voice:

```cpp
int main() {
    Cat cat;
    Lion lion;
    Tiger tiger;

    cat.voice();   // Output: Meow!
    lion.voice();  // Output: Roar!
    tiger.voice(); // Output: Grr!

    return 0;
}
```

---

### Extending Base Class with Unique Methods: Part 1

In some cases, we may need to add methods to a derived class that are not present in the base class. This way, we can **extend** the base class' functionality. This can be done simply by defining the new method in the derived class. Let's add a unique method to the `Lion` class:

```cpp
#include <iostream>

class Cat {
public:
    virtual void voice() {
        std::cout << "Meow!" << std::endl;
    }
};

class Lion : public Cat {
public:
    void voice() override {
        std::cout << "Roar!" << std::endl;
    }

    void hunt() {
        std::cout << "Lion is hunting!" << std::endl;
    }
};
```

In this example:

- The `Lion` class has a unique method called `hunt()`.
- This method is not available in the `Cat` class.

---

### Extending Base Class with Unique Methods: Part 2

Let's see how it works:

```cpp
int main() {
    Cat cat;
    Lion lion;

    cat.voice();  // Output: Meow!
    lion.voice(); // Output: Roar!
    lion.hunt();  // Output: Lion is hunting!
    // cat.hunt(); // Uncommenting this line will cause a compilation error as 'Cat' doesn't have a 'hunt' method

    return 0;
}
```

**Note:**

- When calling `lion.hunt()`, we get the output specific to the `Lion` class, showing how to extend the base class functionality with new methods.
- We can't call `cat.hunt()`, because `hunt()` is a new method of the extended `Lion` class.

---

### Constructors with Inheritance: Part 1

The **base class constructor is called first** when a derived class is instantiated. Let's see how constructors work in inheritance:

```cpp
#include <iostream>

class Cat {
public:
    std::string color;
    int age;

    Cat(std::string c, int a) : color(c), age(a) {
        std::cout << "Cat constructor called" << std::endl;
    }

    virtual void voice() {
        std::cout << "Meow!" << std::endl;
    }
};

class Lion : public Cat {
public:
    Lion(std::string c, int a) : Cat(c, a) {
        std::cout << "Lion constructor called" << std::endl;
    }

    void voice() override {
        std::cout << "Roar!" << std::endl;
    }
};
```

In this example:

- The `Cat` class has a constructor that initializes the `color` and `age` attributes.
- The `Lion` class inherits from `Cat` and also has its own constructor.
- The `Lion` constructor calls the `Cat` constructor using the **initializer list** syntax.

---

### Constructors with Inheritance: Part 2

Let's examine our constructors:

```cpp
int main() {
    Lion lion("Golden", 5);

    // Output:
    // Cat constructor called
    // Lion constructor called

    std::cout << "Lion: " << lion.color << ", " << lion.age << " years old" << std::endl; // Output: Lion: Golden, 5 years old
    lion.voice(); // Output: Roar!

    return 0;
}
```

When an object of the `Lion` class is created, the `Cat` constructor is called first, followed by the `Lion` constructor.

By calling the base class constructor from the derived class constructor, we ensure the base class attributes are properly initialized.

---

### Destructors in Inheritance: Part 1

Just as constructors are called in a specific order in an inheritance hierarchy, so are destructors. When a derived class object is destroyed, the **derived class destructor is called first**, followed by the base class destructor. This ensures that any additional cleanup required by derived classes is performed before the base class destructor is executed.

Example:

```cpp
#include <iostream>

class Cat {
public:
    Cat() {
        std::cout << "Cat constructor called" << std::endl;
    }
    virtual ~Cat() {
        std::cout << "Cat destructor called" << std::endl;
    }
    virtual void voice() {
        std::cout << "Meow!" << std::endl;
    }
};

class Lion : public Cat {
public:
    Lion() {
        std::cout << "Lion constructor called" << std::endl;
    }
    ~Lion() {
        std::cout << "Lion destructor called" << std::endl;
    }
    void voice() override {
        std::cout << "Roar!" << std::endl;
    }
};
```

The `Cat` destructor is marked as `virtual` to ensure that the derived class destructor is called first (even when deleting through a base class pointer).

---

### Destructors in Inheritance: Part 2

Let's see it in action:

```cpp
int main() {
    Lion lion;

    // Output:
    // Cat constructor called
    // Lion constructor called
    // Lion destructor called
    // Cat destructor called

    return 0;
}
```

In this example:

- When the `lion` object is destroyed, the `Lion` destructor is called first, followed by the `Cat` destructor.
- This ensures that any resources allocated by derived classes are properly released before the base class destructor runs.

## Inheritance with Books and Novels

Let's explore another example!

This code creates a base `Book` class and a specialized version — a `Textbook` class. The `Textbook` class will add some attributes and methods specific to textbooks, such as the subject of the textbook and a method to display citations needed for a chapter. Observe how inheritance helps us build more specialized classes from a base class.

---

### Code

```cpp
#include <iostream>
#include <vector>

// Base class
class Book {
public:
    std::string title;
    std::string author;
    int pages;

    // Constructor
    Book(std::string t, std::string a, int p) : title(t), author(a), pages(p) {
        std::cout << "Book constructor called" << std::endl;
    }

    // Method to display book information
    virtual void displayInfo() {
        std::cout << "Title: " << title << ", Author: " << author << ", Pages: " << pages << std::endl;
    }
};

// Derived class
class Textbook : public Book {
public:
    std::string subject;
    std::vector<std::string> citations;

    // Constructor
    Textbook(std::string t, std::string a, int p, std::string s, std::vector<std::string> c) 
        : Book(t, a, p), subject(s), citations(c) {
        std::cout << "Textbook constructor called" << std::endl;
    }

    // Overriding method to display textbook information
    void displayInfo() override {
        std::cout << "Title: " << title << ", Author: " << author << ", Pages: " << pages << ", Subject: " << subject << std::endl;
    }

    // New method specific to Textbook class
    void displayCitations(int chapter) {
        if (chapter > 0 && chapter <= citations.size()) {
            std::cout << "Citations for chapter " << chapter << ": " << citations[chapter - 1] << std::endl;
        } else {
            std::cout << "No citations available for chapter " << chapter << std::endl;
        }
    }
};

int main() {
    Book book("Introduction to Programming", "John Doe", 150);
    book.displayInfo();

    std::vector<std::string> citations = {"Citation 1", "Citation 2", "Citation 3"};
    Textbook textbook("Advanced Mathematics", "Jane Smith", 500, "Mathematics", citations);
    textbook.displayInfo();
    textbook.displayCitations(2);

    return 0;
}
```

---

### How It Works

This example uses all three inheritance ideas from the lesson:

| Idea | Where it appears |
|------|------------------|
| **Inheritance** | `class Textbook : public Book` — `Textbook` gets `title`, `author`, and `pages` for free |
| **Constructor chaining** | `Textbook(...) : Book(t, a, p), ...` — the base part is built first via the initializer list |
| **Overriding** | `Book::displayInfo()` is `virtual`; `Textbook::displayInfo() override` also prints the subject |
| **Extending** | `displayCitations(int chapter)` exists only in `Textbook` |

**Inside `displayCitations`:**

- Chapters are numbered from 1, but vector indexes start at 0, so it reads `citations[chapter - 1]`.
- The check `chapter > 0 && chapter <= citations.size()` guards against out-of-range access. Anything else prints the "No citations" message.

---

### Output

```
Book constructor called
Title: Introduction to Programming, Author: John Doe, Pages: 150
Book constructor called
Textbook constructor called
Title: Advanced Mathematics, Author: Jane Smith, Pages: 500, Subject: Mathematics
Citations for chapter 2: Citation 2
```

### Output Trace

| Step | Code | Output |
|------|------|--------|
| 1 | `Book book(...)` | `Book constructor called` |
| 2 | `book.displayInfo()` | `Title: Introduction to Programming, Author: John Doe, Pages: 150` |
| 3 | `Textbook textbook(...)` → base part | `Book constructor called` |

> **Key takeaway:** A derived class reuses everything from its base class, then specializes it: the base constructor runs first (steps 3–4), `override` swaps in new behavior (step 5), and new methods like `displayCitations` extend the class (step 6). Note that `chapter <= citations.size()` compares an `int` with an unsigned `size_t`; it is safe here because `chapter > 0` is checked first, but some compilers will warn about it.

> **Encapsulation + inheritance:** `balance` is `private`, so even a derived class can't read or change it directly. `SavingsAccount` must go through the base class's **public** methods (`deposit`, `withdraw`, `getBalance`). That's what protects the data from invalid changes, such as withdrawing more than the balance.

### Completed Code

```cpp
#include <iostream>

class BaseBankAccount {
private:
    double balance;

public:
    BaseBankAccount(double initialBalance) : balance(initialBalance) {}

    void deposit(double amount) {
        balance += amount;
    }

    void withdraw(double amount) {
        if (amount <= balance) {
            balance -= amount;
        } else {
            std::cout << "Insufficient funds!" << std::endl;
        }
    }

    double getBalance() const {
        return balance;
    }
};

class SavingsAccount : public BaseBankAccount {
public:
    SavingsAccount(double initialBalance) : BaseBankAccount(initialBalance) {}

    void displayBalance() {
        std::cout << "Balance: " << getBalance() << std::endl;
    }
};

int main() {
    // Create a SavingsAccount object
    SavingsAccount account(1000);

    // Deposit money
    account.deposit(500);
    account.displayBalance(); // Should display current balance

    // Withdraw money
    account.withdraw(200);
    account.displayBalance(); // Should display current balance

    // Attempt to withdraw more money than balance
    account.withdraw(1400); // Should display message for insufficient funds
    account.displayBalance(); // Should display current balance

    return 0;
}
```

> **Key takeaway:** A `private` member is hidden even from derived classes, so `SavingsAccount` reads the balance only through the inherited public `getBalance()`. ### Building the Solution

| TODO | What to do |
|------|------------|
| **1. `Cat` destructor** | Declare `virtual ~Cat()` and print `Cat destructor called`. `virtual` makes sure the derived destructor also runs when an object is deleted through a `Cat*`. |
| **2. `Lion` destructor** | Declare `~Lion()` and print `Lion destructor called`. Because `~Cat()` is `virtual`, `~Lion()` is automatically virtual too, so `override` is optional (`~Lion() override` also works). |

### Completed Code

```cpp
#include <iostream>

class Cat {
public:
    Cat() {
        std::cout << "Cat constructor called" << std::endl;
    }
    virtual ~Cat() {
        std::cout << "Cat destructor called" << std::endl;
    }
    virtual void voice() {
        std::cout << "Meow!" << std::endl;
    }
};

class Lion : public Cat {
public:
    Lion() {
        std::cout << "Lion constructor called" << std::endl;
    }
    ~Lion() {
        std::cout << "Lion destructor called" << std::endl;
    }
    void voice() override {
        std::cout << "Roar!" << std::endl;
    }
};

int main() {
    Lion lion;

    // Expected Output:
    // Cat constructor called
    // Lion constructor called
    // Lion destructor called
    // Cat destructor called

    return 0;
}
```

### Output Trace

| Step | Code | What happens | Output |
|------|------|--------------|--------|
| 1 | `Lion lion;` → base part | `Cat()` runs first | `Cat constructor called` |
| 2 | `Lion lion;` → derived part | `Lion()` body runs | `Lion constructor called` |
| 3 | `return 0;` → `lion` goes out of scope | `~Lion()` runs first | `Lion destructor called` |
| 4 | after `~Lion()` finishes | `~Cat()` runs last | `Cat destructor called` |

> **Key takeaway:** Destructors run in the reverse order of constructors: derived first, then base.

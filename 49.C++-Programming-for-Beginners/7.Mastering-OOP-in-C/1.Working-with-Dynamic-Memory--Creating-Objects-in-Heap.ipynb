## Working with Dynamic Memory: Creating Objects in Heap

Welcome! Today, we're diving into a crucial topic in C++: **working with dynamic memory and creating objects in the heap**. Understanding dynamic memory allocation allows for more flexible memory management, especially when you need to allocate space at runtime.

In some programs, we don't know how much memory we need until runtime. For instance, if you're creating a game that loads levels, you might not know how many objects a level contains until it loads. Dynamic memory helps by letting you allocate memory exactly when you need it.

Ready? Let's get started!

---

### Understanding Memory Management in C++

Think of your computer's memory as a **bookshelf (stack)** and a **storage room (heap)**. The stack is small but quick to access, like a bookshelf where you can easily pick up your favorite book. The heap is larger but slower to access, like a storage room full of boxes.

| Memory | Managed | Best for | Trade-off |
|---|---|---|---|
| **Stack** | Automatically | Local variables | Faster, but limited in size |
| **Heap** | Manually | Dynamic memory allocation | Larger and more flexible, but needs careful handling to avoid memory leaks |

---

### Dynamic Memory Allocation with `new` and `delete`

To allocate memory in the heap, use the `new` keyword. This creates an object or an array of objects and returns a **pointer** to the memory location.

One crucial thing to know is that you must **deallocate** the dynamic memory you used. If you don't, that memory stays blocked. This problem is known as a **memory leak**.

Let's start with a simple example:

```cpp
#include <iostream>

int main() {
    // Allocating an integer in the heap
    int* intPtr = new int(10);

    // Accessing the integer
    std::cout << "Value: " << *intPtr << std::endl; // Output: Value: 10

    // Free the allocated memory to prevent memory leaks
    delete intPtr;
    return 0;
}
```

In this example:

- `new int(10)`: allocates memory for an integer in the heap and initializes it with `10`.
- `int* intPtr`: stores the address of the allocated memory.
- `*intPtr`: dereferences the pointer to access the value.
- `delete intPtr`: deallocates the memory to prevent memory leaks.

---

### Dynamic Arrays

Dynamic arrays are another common use of heap memory. Allocating an array dynamically allows its size to be determined at runtime.

```cpp
#include <iostream>

int main() {
    // Allocating an array of integers in the heap
    int* arrayPtr = new int[5];

    // Initializing the array
    for (int i = 0; i < 5; ++i) {
        arrayPtr[i] = i * 10;
    }

    // Accessing and displaying the array elements
    for (int i = 0; i < 5; ++i) {
        std::cout << "Value at index " << i << ": " << arrayPtr[i] << std::endl;
    }

    // Free the allocated memory to prevent memory leaks
    delete[] arrayPtr;
    return 0;
}
```

In this example:

- `new int[5]`: allocates memory for an array of 5 integers in the heap.
- `arrayPtr[i]`: accesses and manipulates elements using the pointer.
- `delete[] arrayPtr`: deallocates the array memory to prevent memory leaks.

A memory leak is mostly dangerous when initializing dynamic arrays, as they could take a lot of RAM. Every call to `new` must be paired with a corresponding `delete`, and every `new[]` must be paired with a `delete[]`.

| Allocate | Free with |
|---|---|
| `new` | `delete` |
| `new[]` | `delete[]` |

---

### Creating Objects in the Heap

Next, let's create an object dynamically and understand how to access and use it. Consider this class:

```cpp
#include <iostream>

class Person {
private:
    std::string name;
    int age;

public:
    Person(std::string n, int a) : name(n), age(a) {}

    void display() {
        std::cout << "Name: " << name << ", Age: " << age << std::endl;
    }
};
```

We can create an object of this class in the heap using the `new` keyword:

```cpp
int main() {
    // Creating an object in the heap
    Person* personPtr = new Person("John", 30);

    // Free the allocated memory to prevent memory leaks
    delete personPtr;
    return 0;
}
```

When you allocate memory dynamically, you use **pointers** to access and use objects. A pointer stores the object's memory address.

---

### Accessing the Dynamic Object

Usually we access the object's methods and fields using the `.` operator. You might think that when you have a pointer to an object, you can access its members like this:

```cpp
int main() {
    *personPtr.display();
}
```

Here, we try to dereference the pointer and then use `.` to access the `display` method. However, it won't work, as the `.` operator has a **higher precedence** than the `*` operator. So, the correct way is this:

```cpp
int main() {
    // Creating an object in the heap
    Person* personPtr = new Person("John", 30);

    (*personPtr).display();  // Name: John, Age: 30

    // Free the allocated memory to prevent memory leaks
    delete personPtr;
    return 0;
}
```

---

### Arrow Operator

Using `*` and `.` can be a bit confusing. There is a shorter and easier way to do the same thing: the **arrow operator** `->`. It combines dereferencing the pointer and accessing the class member in one operator.

Instead of:

```cpp
(*personPtr).display();
```

You can do this:

```cpp
personPtr->display();
```

And this will work exactly the same!

Here is the complete code:

```cpp
int main() {
    // Creating an object in the heap
    Person* personPtr = new Person("John", 30);
    
    personPtr->display();  // Name: John, Age: 30

    // Free the allocated memory to prevent memory leaks
    delete personPtr;
    return 0;
}
```

---

### Avoiding Dangling Pointers

Calling a method on already deleted memory can cause **undefined behavior**. For example:

```cpp
#include <iostream>

int main() {
    // Creating an object in the heap
    Person* personPtr = new Person("John", 30);

    // Free the allocated memory
    delete personPtr;

    // Attempting to call a method on a deleted object
    // This can cause a crash or undefined behavior
    personPtr->display();  // Undefined behavior

    return 0;
}
```

After `delete personPtr;`, the memory that `personPtr` pointed to is deallocated, but the pointer still holds the old address. Attempting to access it afterward can lead to crashes or unexpected behavior. Hence, it's important to set the pointer to `nullptr` after deletion to avoid **dangling pointers**:

```cpp
delete personPtr;
personPtr = nullptr;  // Prevents accidental access
```

| Concept | Key Point |
|---|---|
| **Dynamic memory allocation** | Use `new` and `delete` for heap memory. |
| **Dynamic arrays** | Create and manage them with `new[]` and `delete[]`. |
| **Accessing heap objects** | Use pointers: `(*ptr).member` or the shorter `ptr->member`. |
| **Memory management** | Free allocated memory to prevent memory leaks. |
| **Dangling pointers** | Set pointers to `nullptr` after deallocation. |

## Fixing Dynamic Memory Management

Hello, Space Voyager!

The following code contains a **double-free error** (double delete). Memory that is already freed should not be freed again, as this can lead to undefined behavior and potentially crash your program. Please correct the code so that the memory allocated is properly managed and only freed once.

---

### Buggy Code

```cpp
#include <iostream>

int main() {
    // Allocate memory in the heap for an integer
    int* intPointer = new int(42);

    // Print the allocated value
    std::cout << "Value: " << *intPointer << std::endl;

    // Free the allocated memory
    delete intPointer;

    // Incorrect: Freeing the allocated memory again (double delete error)
    delete intPointer;

    return 0;
}
```

**Buggy behavior:** The first `delete` frees the memory. The second `delete` tries to free the same address again, which is **undefined behavior**. It typically crashes (for example, with a "double free detected" error) or corrupts the heap, even though `Value: 42` is printed first.

---

### Finding the Bug

1. `new int(42)` allocates one integer in the heap, and `intPointer` holds its address.
2. The first `delete intPointer;` returns that memory to the system. This is correct.
3. `intPointer` still holds the old address, so it is now a **dangling pointer**.
4. The second `delete intPointer;` frees an address that is already freed. This is the bug.

**The rule:** every `new` needs exactly one matching `delete`. Remove the second `delete`. As a safety habit, also set the pointer to `nullptr` after the first `delete`, because `delete nullptr;` is safe and does nothing.

---

### Fixed Code

```cpp
#include <iostream>

int main() {
    // Allocate memory in the heap for an integer
    int* intPointer = new int(42);

    // Print the allocated value
    std::cout << "Value: " << *intPointer << std::endl;

    // Free the allocated memory (only once)
    delete intPointer;
    intPointer = nullptr;  // Avoid a dangling pointer

    return 0;
}
```

---

### Output

```
Value: 42
```

---

### Output Trace

| Step | Code | Result |
|---|---|---|
| 1 | `int* intPointer = new int(42);` | Heap integer created with value `42` |
| 2 | `std::cout << "Value: " << *intPointer` | Prints `Value: 42` |
| 3 | `delete intPointer;` | Memory freed (once) |
| 4 | `intPointer = nullptr;` | Pointer no longer refers to freed memory |
| 5 | `return 0;` | Program ends cleanly, with no crash |

> **Key takeaway:** Each `new` is paired with exactly one `delete`. Deleting the same pointer twice is undefined behavior. Setting the pointer to `nullptr` right after `delete` prevents both dangling-pointer use and an accidental second delete.

## Allocating a Dynamic Array for Scores

### Building the Solution

3. **Free the array.** Before `return 0;`, add `delete[] scores;`. Memory from `new[]` must be released with `delete[]`, not plain `delete`. Then set `scores = nullptr;`.

### Completed Code

```cpp
#include <iostream>

int main() {
    // Size of the array
    int size = 5;

    // Allocate the scores array in the heap
    int* scores = new int[size];

    // Initializing the array with scores
    for (int i = 0; i < size; ++i) {
        scores[i] = (i + 1) * 10;
    }

    // Displaying the scores
    for (int i = 0; i < size; ++i) {
        std::cout << "Score " << i + 1 << ": " << scores[i] << std::endl;
    }

    // Free the dynamic array
    delete[] scores;
    scores = nullptr;

    return 0;
}
```

## Use the Arrow Operator to Access Class Methods

Now, let's make a small adjustment. Change the starter code to use the arrow `->` operator to access the `display` method instead of using `*` and `.`. This will help you become more comfortable with pointer syntax and the arrow operator.

---

### Starter Code

```cpp
#include <iostream>

class GameCharacter {
private:
    std::string name;
    int healthPoints;

public:
    GameCharacter(std::string n, int hp) : name(n), healthPoints(hp) {}

    void display() {
        std::cout << "Character: " << name << ", HP: " << healthPoints << std::endl;
    }
};

int main() {
    // Creating a GameCharacter object in the heap
    GameCharacter* charPtr = new GameCharacter("Knight", 100);
    
    // TODO: modify this line to use -> operator
    (*charPtr).display(); // Character: Knight, HP: 100

    // Free the allocated memory to prevent memory leaks
    delete charPtr;

    return 0;
}
```

---

### Building the Solution

1. **Find the target line.** `(*charPtr).display();` dereferences the pointer with `*`, then calls the method with `.`. The parentheses are needed because `.` binds tighter than `*`.
2. **Replace it with `->`.** `charPtr->display();` does both steps in one operator: dereference the pointer and access the member.
3. **Change nothing else.** The allocation and `delete charPtr;` stay the same.

| Style | Syntax |
|---|---|
| Dereference, then dot | `(*charPtr).display();` |
| Arrow operator | `charPtr->display();` |

---

### Completed Code

```cpp
#include <iostream>

class GameCharacter {
private:
    std::string name;
    int healthPoints;

public:
    GameCharacter(std::string n, int hp) : name(n), healthPoints(hp) {}

    void display() {
        std::cout << "Character: " << name << ", HP: " << healthPoints << std::endl;
    }
};

int main() {
    // Creating a GameCharacter object in the heap
    GameCharacter* charPtr = new GameCharacter("Knight", 100);
    
    // Use the -> operator to call display
    charPtr->display(); // Character: Knight, HP: 100

    // Free the allocated memory to prevent memory leaks
    delete charPtr;

    return 0;
}
```

---

### Output

```
Character: Knight, HP: 100
```

---

### Output Trace

| Step | Code | Result |
|---|---|---|
| 1 | `new GameCharacter("Knight", 100)` | Heap object with `name = "Knight"`, `healthPoints = 100` |
| 2 | `charPtr->display();` | Follows the pointer and calls `display()` on the object |
| 3 | `std::cout << ...` | Prints `Character: Knight, HP: 100` |
| 4 | `delete charPtr;` | Heap object freed |

> **Key takeaway:** `ptr->member` is shorthand for `(*ptr).member`. The behavior and output are identical, but `->` is shorter and easier to read.

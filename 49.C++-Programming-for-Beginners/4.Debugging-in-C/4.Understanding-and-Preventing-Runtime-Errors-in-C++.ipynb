## Understanding and Preventing Runtime Errors in C++

Welcome to this thrilling session on runtime errors in C++ programming. We're going to delve into runtime errors, understand their types and occurrences, and learn how to handle them. Mastering this will make your C++ code reliable and fail-safe.

Today, we'll dissect runtime errors, categorize them, and learn how to identify and prevent them in C++ programs.

---

### Understanding Runtime Errors

Runtime errors appear during your program's execution and prevent the program from running as expected. Such errors occur when commands, even though syntactically correct, are logically impossible to execute.

Runtime errors occur in various forms, including:

* **Null pointer dereferences**: Attempting to access a null pointer's property.
* **Division by zero**: Attempting to divide a number by zero.
* **Accessing uninitialized variables**: Using a variable that has not been assigned a value.

Below are examples for each type.

---

### Runtime Error Example: Null Pointer Dereference

```cpp
#include <iostream>

int main() {
    int* ptr = nullptr;
    std::cout << *ptr << std::endl;  // Null pointer dereference error.
    return 0;
}
```

In this example, `ptr` is assigned the value `nullptr`. Attempting to dereference the null pointer with `*ptr` results in a runtime error since the pointer does not point to a valid memory address.

---

### Runtime Error Example: Division by Zero

```cpp
#include <iostream>

int main() {
    int number = 1;
    int zero = 0;
    std::cout << number / zero << std::endl;  // Division by zero error.
    return 0;
}
```

Here, `number` is divided by zero. Division by zero is mathematically undefined and causes a runtime error in the program.

---

### Runtime Error Example: Accessing Uninitialized Variables

```cpp
#include <iostream>

int main() {
    int uninitVar;  // Variable declared but not initialized.
    std::cout << uninitVar << std::endl;  // Undefined behavior: accessing uninitialized variable.
    return 0;
}
```

This example declares `uninitVar` but does not initialize it. Accessing the uninitialized variable through `cout` leads to undefined behavior and potential runtime errors since its value is indeterminate.

---

### Fixing Runtime Errors

Programmatically preventing error conditions can help fix runtime errors:

```cpp
#include <iostream>

int main() {
    int number = 10;
    int zero = 0;
    if (zero != 0) {
        std::cout << number / zero << std::endl;
    } else {
        std::cout << "Cannot divide by zero." << std::endl;
    }
    return 0;
}
```

This code snippet introduces basic error handling by implementing an `if-else` condition. It first checks if the divisor is not zero before executing a division operation that could potentially result in a runtime error. This simple yet effective method prevents the common runtime error of dividing by zero. The division proceeds if the condition is met (the divisor is not zero); otherwise, a message is printed to inform that division by zero is impossible, thereby avoiding the error and making the program more robust.

## Avoiding Division by Zero in C++

Have you ever wondered how to avoid crashing your program when attempting a division by zero in C++? The given code showcases a simple yet effective way to handle potential runtime errors by checking the divisor before executing the division. This approach ensures your program runs smoothly without encountering a crash. Click Run to see how it elegantly handles both a safe division and an unsafe division by zero scenario!

---

### Code

```cpp
#include <iostream>

int main() {
    int dividend1 = 10;
    int divisor1 = 2;
    if (divisor1 == 0) {
        std::cout << "Oops! Division by zero is not allowed." << std::endl;
    } else {
        std::cout << "Result of division is " << dividend1 / divisor1 << std::endl;
    }

    int dividend2 = 5;
    int divisor2 = 0;
    if (divisor2 == 0) {
        std::cout << "Oops! Division by zero is not allowed." << std::endl;
    } else {
        std::cout << "Result of division is " << dividend2 / divisor2 << std::endl;
    }

    return 0;
}
```

---

### How It Works

- This is a demo, not a bug-fix: the code is already correct and needs no changes — just click Run.
- Each division is guarded by `if (divisor == 0)`, so the division in the `else` branch can only execute when the divisor is non-zero.
- First case: `10 / 2` — the divisor is `2`, so the `else` branch runs and prints the result `5`.
- Second case: `5 / 0` — the divisor is `0`, so the `if` branch runs and prints the warning instead of dividing, avoiding the runtime error.

---

### Output

```
Result of division is 5
Oops! Division by zero is not allowed.
```

> **Key takeaway:** check the divisor before dividing — integer division by zero is undefined behavior and typically crashes the program, while a simple `if` guard turns it into a handled, readable message.

## Fix the Division by Zero Check

Here's a code snippet that attempts to check whether we can perform division on two integers. However, there seems to be a hiccup affecting its goal. Dive into the code without any hints and fix the issue preventing it from working as expected. Pay attention to how division and conditional checks are handled in C++.

---

### Starter Code

```cpp
#include <iostream>

int main() {
    int dividend = 10;
    int divisor = 0;

    if (divisor == 0) {
        std::cout << "Result: " << dividend / divisor << std::endl;
    } else {
        std::cout << "Oops! Cannot divide by zero." << std::endl;
    }
    return 0;
}
```

---

### Building the Solution

- The condition is inverted: `if (divisor == 0)` runs the division exactly when the divisor *is* zero, so the program crashes with a division-by-zero runtime error, and the "Cannot divide by zero" message sits in the `else` branch where it can never help.
- Flip the condition to `divisor != 0` so the division only runs when the divisor is non-zero, and the warning is printed otherwise.

---

### Completed Code

```cpp
#include <iostream>

int main() {
    int dividend = 10;
    int divisor = 0;

    if (divisor != 0) {
        std::cout << "Result: " << dividend / divisor << std::endl;
    } else {
        std::cout << "Oops! Cannot divide by zero." << std::endl;
    }
    return 0;
}
```

---

### Output

```
Oops! Cannot divide by zero.
```

> **Key takeaway:** a safety check only protects you if it guards the risky operation correctly — an inverted condition (`== 0` instead of `!= 0`) sends the dangerous case straight into the division.

## Handling Division by Zero in C++

Can you write a piece of code in C++ that prevents your program from crashing when attempting to divide a number by zero? Instead of the program ending abruptly, it should output a friendly message.

Remember that if the denominator is not a zero, your program should output the division result! Let's code!

---

### Starter Code

```cpp
#include <iostream>

int main() {
    int numerator = 10;
    int denominator = 0;
    // TODO: If denominator is zero, print "Error: Cannot divide by zero." Otherwise, print the division result.
    return 0;
}
```

---

### Building the Solution

- Check the denominator before dividing: `if (denominator == 0)` prints the error message `Error: Cannot divide by zero.`.

## Check and Handle Null Pointer in C++

You’re now prepared to handle runtime errors in C++. Your task is to write a piece of C++ code that checks if a pointer is nullptr and handle it appropriately. Use your knowledge about null pointer dereferencing errors to complete this exercise.

---

### Starter Code

```cpp
#include <iostream>
#include <cstring>

int main() {
    char* text = nullptr;
    // TODO: Check if the 'text' pointer is not nullptr. If it isn't, print the length of the string it points to.
    // Otherwise, print "Text is nullptr and cannot be processed."
    return 0;
}
```

---

### Building the Solution

- Test the pointer before using it: `if (text != nullptr)`.
- Inside the `if`, print the string's length with `std::strlen(text)` (from `<cstring>`) — this is only safe because the pointer is known to be valid.
- In the `else` branch, print `Text is nullptr and cannot be processed.` instead of calling `strlen` on a null pointer, which would crash.

---

### Completed Code

```cpp
#include <iostream>
#include <cstring>

int main() {
    char* text = nullptr;

    if (text != nullptr) {
        std::cout << std::strlen(text) << std::endl;
    } else {
        std::cout << "Text is nullptr and cannot be processed." << std::endl;
    }

    return 0;
}
```

---

### Output

```
Text is nullptr and cannot be processed.
```

> **Key takeaway:** always check a pointer against `nullptr` before dereferencing it or passing it to functions like `strlen` — calling `strlen(nullptr)` is a null pointer dereference and crashes the program.

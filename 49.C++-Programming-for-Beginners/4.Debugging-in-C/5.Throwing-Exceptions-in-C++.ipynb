## Throwing Exceptions in C++

Hello! Today, we will learn about throwing exceptions in C++, which is vital for writing robust programs. Think of exceptions like traffic signals. When driving, if you see a red light, you stop to avoid collisions—a clear sign something needs attention. Similarly, exceptions signal problems in our code, allowing us to handle errors smoothly. By the end, you'll know how to throw and catch exceptions, making your programs more resilient.

---

### Throwing Exceptions

To "throw" an exception means to signal that something unusual has occurred. In C++, we use the `throw` keyword to do this. It’s like raising your hand to say, "Wait, there's a problem!"

Here’s a simple example. Imagine you ask a vending machine for a snack that's out of stock. The machine needs to inform you that it can't complete your request:

```cpp
#include <iostream>
#include <stdexcept> // For std::runtime_error


int main() {
    int stock = 0;
    if (stock == 0) {
        throw std::runtime_error("Out of stock!");
    }
    return 0;
}
```

Output:

```text
terminate called after throwing an instance of 'std::runtime_error'
  what():  Out of stock!
```

---

### Catching Exceptions: Part 1

Imagine we have a function that can throw an error:

```cpp
#include <iostream>
#include <stdexcept>

void give_order(int order, int stock) {
    if (order <= stock) {
        std::cout << "Here is your order!" << std::endl;
    } else {
        throw std::runtime_error("Out of stock!");
    }
}

int main() {
    give_order(1, 10);  // Here is your order!
}
```

In this case, everything is fine, and the function is successfully executed. But if we call it like this, it will throw an error:

```cpp
int main() {
    give_order(5, 0);
    // terminate called after throwing an instance of 'std::runtime_error'
    // what():  Out of stock!
}
```

This way, our function is designed to warn the program with an error that something went wrong. However, we don't want the program to terminate every time this error appears. Let's learn how to handle it.

---

### Catching Exceptions: Part 2

In C++, we can handle exceptions with a special `try-catch` block. The program tries to execute the code inside the `try`, and in case it fails with an exception, executes the code inside the `catch`.

Let’s modify our example:

```cpp
int main() {
    int order = 5;
    int stock = 3;

    try {
        give_order(order, stock);
    } catch (const std::runtime_error& e) {
        std::cout << "Error: " << e.what() << std::endl;
    }

    return 0;
}
```

When the `give_order` function is called with `order = 5` and `stock = 3`, it will throw a `std::runtime_error` because the order exceeds the stock.

The `catch` block catches exceptions of type `std::runtime_error`. If the exception is caught, it prints the error message using `e.what()` where `e` is the exception object caught. This prevents the program from crashing and allows you to handle the error however you want.

This way, your program continues to run and provides a meaningful handling of an error.

---

### Catching Multiple Exceptions: Part 1

Sometimes, different types of exceptions can occur, and it’s essential to handle each type uniquely. In C++, you can catch multiple exceptions by adding multiple `catch` blocks. Each `catch` block handles a specific type of exception.

Here’s a meaningful real-life example. Imagine a bank application where you need to process a transaction. Different errors can occur during the process, such as insufficient funds or an invalid account:

```cpp
#include <iostream>
#include <stdexcept>

void process_transaction(double amount, double balance, bool account_valid) {
    if (!account_valid) {
        throw std::invalid_argument("Invalid account.");
    }
    if (amount > balance) {
        throw std::runtime_error("Insufficient funds.");
    }
    std::cout << "Transaction processed successfully!" << std::endl;
}
```

The `process_transaction` function can throw either a `std::invalid_argument` if the account is invalid or a `std::runtime_error` if there are insufficient funds.

---

### Catching Multiple Exceptions: Part 2

```cpp
int main() {
    double amount = 50.0;
    double balance = 50.0;
    bool account_valid = true;

    try {
        process_transaction(amount, balance, account_valid);
    } catch (const std::invalid_argument& e) {
        std::cout << "Error: " << e.what() << std::endl;
    } catch (const std::runtime_error& e) {
        std::cout << "Error: " << e.what() << std::endl;
    }

    return 0;
}
```

In this code:

* In the `main` function, we try to call `process_transaction` with `account_valid = true`.
* If a `std::invalid_argument` is thrown, it is caught by the first `catch` block.
* If a `std::runtime_error` is thrown, it is caught by the second `catch` block.

This method allows you to handle different types of exceptions separately, providing more specific error management in your programs. In this case, the program will execute successfully and print out "Transaction processed successfully!". Let's look at other options.

---

### Catching Multiple Exceptions: Part 3

Firstly, it is possible that the account is not valid:

```cpp
int main() {
    double amount = 50.0;
    double balance = 50.0;
    bool account_valid = false; // Not valid!

    try {
        process_transaction(amount, balance, account_valid);
    } catch (const std::invalid_argument& e) {  // This exception is caught!
        std::cout << "Error: " << e.what() << std::endl;  // Error: Invalid account.
    } catch (const std::runtime_error& e) {
        std::cout << "Error: " << e.what() << std::endl;
    }

    return 0;
}
```

In this case, the first `catch` block works.

---

### Catching Multiple Exceptions: Part 4

Secondly, it is possible that there are not enough funds on the balance:

```cpp
int main() {
    double amount = 50.0;
    double balance = 25.0;  // Not enough!
    bool account_valid = true;

    try {
        process_transaction(amount, balance, account_valid);
    } catch (const std::invalid_argument& e) {
        std::cout << "Error: " << e.what() << std::endl;
    } catch (const std::runtime_error& e) {  // This exception is caught!
        std::cout << "Error: " << e.what() << std::endl;  // Error: Insufficient funds.
    }

    return 0;
}
```

In this case, the second `catch` block works.

---

### Catching Multiple Exceptions: Part 5

Lastly, there can be both problems at once:

```cpp
int main() {
    double amount = 50.0;
    double balance = 25.0;  // Not enough!
    bool account_valid = false;  // Not valid!

    try {
        process_transaction(amount, balance, account_valid);
    } catch (const std::invalid_argument& e) {  // This exception is caught!
        std::cout << "Error: " << e.what() << std::endl;  // Error: Invalid account.
    } catch (const std::runtime_error& e) {
        std::cout << "Error: " << e.what() << std::endl;
    }

    return 0;
}
```

In this case, the first `catch` block will work as it is the first one. The second `catch` block won't be executed.

## Handling Specific Exceptions in C++

Let's start coding! Modify the provided function to throw the std::invalid_argument exception when a negative number is passed to the function square_root instead of the std::runtime_error. This will help you practice handling specific exceptions more precisely.

Don't forget to update the error-handling logic accordingly!

---

### Starter Code

```cpp
#include <iostream>
#include <stdexcept>
#include <cmath>

double square_root(double num) {
    if (num < 0) {
        throw std::runtime_error("Negative input not allowed.");
    }
    return sqrt(num);
}

int main() {
    double num = -4.0;

    try {
        std::cout << "Square root: " << square_root(num) << std::endl;
    } catch (const std::runtime_error& e) {
        std::cout << "Error: " << e.what() << std::endl;
    }

    return 0;
}
```

---

### Building the Solution

- Change the thrown exception in `square_root` from `std::runtime_error` to `std::invalid_argument`, which better describes a bad input value. Keep the same message.
- Update the `catch` block in `main` to match: `catch (const std::invalid_argument& e)`. If it still caught `std::runtime_error`, the new exception would not be caught and the program would terminate.

---

### Completed Code

```cpp
#include <iostream>
#include <stdexcept>
#include <cmath>

double square_root(double num) {
    if (num < 0) {
        throw std::invalid_argument("Negative input not allowed.");
    }
    return sqrt(num);
}

int main() {
    double num = -4.0;

    try {
        std::cout << "Square root: " << square_root(num) << std::endl;
    } catch (const std::invalid_argument& e) {
        std::cout << "Error: " << e.what() << std::endl;
    }

    return 0;
}
```

---

### Output

```
Error: Negative input not allowed.
```

> **Key takeaway:** a `catch` block only handles the exception type it names (or a base class of it) — `std::invalid_argument` and `std::runtime_error` are sibling types, so when you change what is thrown you must change what is caught.

## Handling Division by Zero Exception in C++

Complete the function `do_division` to throw a `std::runtime_error` if the divisor is zero. Then, update the `main` function to catch this exception and print an appropriate error message.

---

### Starter Code

```cpp
#include <iostream>
#include <stdexcept>

double do_division(double numerator, double denominator) {
    // TODO: Check if the denominator is zero and throw a runtime error
    return numerator / denominator;
}

int main() {
    double numerator = 10.0;
    double denominator = 0.0;

    try {
        double result = do_division(numerator, denominator);
        std::cout << "Result: " << result << std::endl;
    } catch (const std::runtime_error& e) {
        // TODO: Catch the runtime error and print the error message
    }

    return 0;
}
```

---

### Building the Solution

- **Guard the division:** at the top of `do_division`, check `if (denominator == 0)` and `throw std::runtime_error("Division by zero.");` before the division ever runs.
- **Catch it:** inside the `catch` block, print the error using `e.what()`, the same pattern used throughout this lesson.

---

### Completed Code

```cpp
#include <iostream>
#include <stdexcept>

double do_division(double numerator, double denominator) {
    if (denominator == 0) {
        throw std::runtime_error("Division by zero.");
    }
    return numerator / denominator;
}

int main() {
    double numerator = 10.0;
    double denominator = 0.0;

    try {
        double result = do_division(numerator, denominator);
        std::cout << "Result: " << result << std::endl;
    } catch (const std::runtime_error& e) {
        std::cout << "Error: " << e.what() << std::endl;
    }

    return 0;
}
```

---

### Output

```
Error: Division by zero.
```

---

### Output Trace

| Step | Value |
|------|-------|
| `denominator == 0` | `true` |
| Exception thrown | `std::runtime_error("Division by zero.")` |
| Caught by | `catch (const std::runtime_error& e)` |
| Printed | `Error: Division by zero.` |

> **Key takeaway:** checking for an invalid input (like a zero divisor) and throwing *before* performing the risky operation prevents undefined behavior (dividing by zero on floating-point types produces `inf`/`nan` rather than crashing, but the check still communicates the error explicitly). Wrapping the call in `try`/`catch` lets the program report the problem and keep running instead of propagating an unhandled exception.

## Handle Multiple Exception Types for Negative and Out-of-Stock Orders

Excellent work so far! Your current task is to modify the `give_order` function to throw a `std::invalid_argument` exception when the order is negative. Then, update the `main` function to handle this new exception type. This will help you practice throwing and catching multiple exceptions.

---

### Starter Code

```cpp
#include <iostream>
#include <stdexcept>

void give_order(int order, int stock) {
    // TODO: if the order is negative, throw the std::invalid_argument exception
    if (order <= stock) {
        std::cout << "Here is your order!" << std::endl;
    } else {
        throw std::runtime_error("Out of stock!");
    }
}

int main() {
    int order = -5;
    int stock = 10;

    // TODO: also handle the std::invalid_argument exception using one more catch block
    try {
        give_order(order, stock);
    } catch (const std::runtime_error& e) {
        std::cout << "Error: " << e.what() << std::endl;
    }

    return 0;
}
```

---

### Building the Solution

- **Guard against a negative order:** at the top of `give_order`, check `if (order < 0)` and `throw std::invalid_argument("Order cannot be negative.");` before the existing stock check runs.
- **Add a second `catch` block:** `catch (const std::invalid_argument& e)` alongside the existing `catch (const std::runtime_error& e)` — each block only handles the exception type it names.
- With `order = -5`, the new check fires first, so the `std::invalid_argument` block catches it; the `std::runtime_error` block (for "Out of stock!") is never reached for this call.

---

### Completed Code

```cpp
#include <iostream>
#include <stdexcept>

void give_order(int order, int stock) {
    if (order < 0) {
        throw std::invalid_argument("Order cannot be negative.");
    }
    if (order <= stock) {
        std::cout << "Here is your order!" << std::endl;
    } else {
        throw std::runtime_error("Out of stock!");
    }
}

int main() {
    int order = -5;
    int stock = 10;

    try {
        give_order(order, stock);
    } catch (const std::invalid_argument& e) {
        std::cout << "Error: " << e.what() << std::endl;
    } catch (const std::runtime_error& e) {
        std::cout << "Error: " << e.what() << std::endl;
    }

    return 0;
}
```

---

### Output

```
Error: Order cannot be negative.
```

---

### Output Trace

| Check | Condition | Result |
|-------|-----------|--------|
| `order < 0` | `-5 < 0` → `true` | Throws `std::invalid_argument("Order cannot be negative.")` |
| Caught by | `catch (const std::invalid_argument& e)` | Prints `Error: Order cannot be negative.` |
| `order <= stock` | never reached | — |

> **Key takeaway:** when a function can fail for more than one reason, throw a distinct exception type for each cause and give each one its own `catch` block. The order of the checks inside the function matters too — validating `order < 0` before the stock comparison ensures a negative order is reported as an invalid argument rather than falling through to the stock logic.

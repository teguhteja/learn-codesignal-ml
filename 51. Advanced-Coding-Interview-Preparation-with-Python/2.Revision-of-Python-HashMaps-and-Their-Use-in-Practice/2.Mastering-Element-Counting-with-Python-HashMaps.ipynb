## Mastering Element Counting with Python HashMaps

---

### Understanding the Problem

We begin in a library, where we want to count book copies. With a small collection, we might be able to tally each one manually. However, as the collection grows, this approach becomes cumbersome and inefficient. A more efficient method uses a HashMap, or a dictionary in Python.

For a quick illustration, consider this list of colors:

```python
colors = ['red', 'blue', 'red', 'green', 'blue', 'blue']
```

If we count manually, `red` appears twice, `blue` appears thrice, and `green` appears once. We can employ HashMaps for a more efficient counting process.

---

### Introducing HashMaps

Simple yet powerful, HashMaps allow us to store and retrieve data using keys. The unique colors in our list act as keys, and the count of each color becomes its corresponding value. Let's demonstrate how we can count elements in our `colors` list using a Python dictionary:

```python
colors = ['red', 'blue', 'red', 'green', 'blue', 'blue']
color_dict = {}

# Start the loop to iterate over each color
for color in colors:

    # If the color is present in our dictionary, increment its value by 1
    if color in color_dict:
        color_dict[color] += 1

    # If the color isn't present, it means we're encountering this color in our list for the first time. In this case, we add it to our dictionary and set its value to 1
    else:
        color_dict[color] = 1

# At the end of the loop, print our dictionary with counts
print(color_dict)
# prints {'red': 2, 'blue': 3, 'green': 1}
```

When the above code executes, it displays the counts for each color: `{'red': 2, 'blue': 3, 'green': 1}`.

---

### Understanding the Above Solution

Here's how we created a dictionary to count our elements:

We began with an empty dictionary. Then, we went through our list, and for every occurring element, we checked if it was in our dictionary. If it was, we increased its value. If it was not, we added it to the dictionary with a value of `1`.

Consequently, this code efficiently counts the colors in our list, showcasing how performant counting can be, even as the list size increases!

Python dictionaries have a `get(key, default)` method, which is an alternative to checking whether a key already exists in a dictionary. It allows us to fetch the value of a key if it exists, or return a default value specified by you if it doesn't.

The code above can be also written like this:

```python
colors = ['red', 'blue', 'red', 'green', 'blue', 'blue']
color_dict = {}

# Start the loop to iterate over each color
for color in colors:

    # Get the value of the color key if it exists, otherwise use a default value of 0. Then increment the value by 1
    color_dict[color] = color_dict.get(color, 0) + 1

# At the end of the loop, print our dictionary with counts
print(color_dict)
# prints {'red': 2, 'blue': 3, 'green': 1}
```

---

### Time Complexity Analysis

The time complexity of our approach is $O(n)$, where `n` is the number of elements in our list. This is because we iterate over our list exactly once, performing constant-time operations for each element. Here is why:

- Dictionary accesses (both setting a value and getting a value) in Python are typically $O(1)$, constant time operations.
- The `for` loop iterates over each element in the list exactly once, so it's an $O(n)$ operation.

The total time complexity, therefore, remains $O(n)$ because the time taken is directly proportional to the number of items in the list. As the size of the list increases, the time taken scales linearly, making this approach efficient for larger collections.

It is also worth noting that the space complexity of this approach is $O(k)$, where `k` is the number of unique elements in the list. In the worst case scenario, where all elements are unique, the space complexity would be $O(n)$.

In conclusion, using hashmaps or dictionaries for counting is a time-efficient approach, especially when working with large datasets.

This approach can be applied to larger lists, strings, and nested collections to count elements. Counting is a ubiquitous task in areas like data analysis and natural language processing. You can employ this concept to count the frequency of words in sentences, characters in strings, or items in shopping lists.

### Starter Code

```python
# Python program based on the scenario of counting books in a library
books = ['mystery', 'sci-fi', 'mystery', 'fantasy', 'sci-fi', 'romance']
books_count = {}

for book in books:
    if book in books_count:
        books_count[book] += 1
    else:
        books_count[book] = 1

```


---

### Solution

The `if`/`else` block can be replaced by a single line using `dict.get(key, default)`:

```python
for book in books:
    books_count[book] = books_count.get(book, 0) + 1
```

- `books_count.get(book, 0)` returns the current count if `book` is already a key, or `0` if it is not.
- Adding `1` and assigning the result back either increments an existing count or creates the key with a value of `1`.
- No separate membership check (`in`) or branch is needed, so the loop body shrinks from four lines to one, with the same $O(n)$ time.

Running the script prints `{'mystery': 2, 'sci-fi': 2, 'fantasy': 1, 'romance': 1}`, the same result as the starter code.


````python
# Python program based on the scenario of counting books in a library
books = ['mystery', 'sci-fi', 'mystery', 'fantasy', 'sci-fi', 'romance']
books_count = {}

for book in books:
    books_count[book] = books_count.get(book, 0) + 1  # get() replaces the if/else check

print(books_count)
````

## Correct the Book Counter Code

You're tasked with creating a system to keep track of book copies in the library. However, after implementing your code, you notice the output is not what you expected. Can you determine what went wrong and fix it? Ensure the program accurately counts each book's copies.

---

### Starter Code

```python
# Create a dictionary to count the copies of books in a library
book_dict = {}

# List of books to count
books = ['The Hobbit', 'The Hobbit', 'The Lion King', 'The Hobbit', 'Cinderella']

# Counting the number of each book in the list
for book in books:
    book_dict[book] = book_dict.get(book, 1) + 1

# Print the dictionary to see how many copies of each book we have
print(book_dict)
```


---

### Solution

The bug is the default value passed to `get()`:

```python
book_dict[book] = book_dict.get(book, 0) + 1  # default must be 0, not 1
```

- With `get(book, 1)`, a book seen for the first time starts at `1` and then gets `+ 1`, so it is stored as `2`. Every book is over-counted by one.
- With `get(book, 0)`, a new book starts at `0`, and `+ 1` stores the first copy as `1`. Later copies keep incrementing from the stored count.
- The buggy version prints `{'The Hobbit': 4, 'The Lion King': 2, 'Cinderella': 2}`. The corrected version prints `{'The Hobbit': 3, 'The Lion King': 1, 'Cinderella': 1}`, which matches the list.


````python
# Create a dictionary to count the copies of books in a library
book_dict = {}

# List of books to count
books = ['The Hobbit', 'The Hobbit', 'The Lion King', 'The Hobbit', 'Cinderella']

# Counting the number of each book in the list
for book in books:
    book_dict[book] = book_dict.get(book, 0) + 1  # default 0, so the first copy counts as 1

# Print the dictionary to see how many copies of each book we have
print(book_dict)
````

### Solution

There are two `TODO`s: the loop body inside the function, and the call that prints the result.

```python
for letter in sentence:
    if letter.isalpha():
        letter_count[letter] = letter_count.get(letter, 0) + 1

```

- `letter.isalpha()` skips spaces, so only alphabetical characters are counted.
- `letter_count.get(letter, 0) + 1` increments an existing count or starts a new letter at `1`.
- The function returns `letter_count`, so `print(count_letter_frequency(sentence))` shows the result.

- Counting is case-sensitive, so `'O'` (from `Once`) and `'o'` (from `upon`) are separate keys.

The script prints `{'O': 1, 'n': 3, 'c': 1, 'e': 2, 'u': 1, 'p': 1, 'o': 1, 'a': 6, 't': 1, 'i': 3, 'm': 1, 'f': 1, 'r': 3, 'w': 1, 'y': 2, 'l': 1, 'b': 1}`.


````python
# Function to count the frequency of each letter in a given sentence
def count_letter_frequency(sentence):
    letter_count = {}
    for letter in sentence:
        # If the character is a letter, update its count in the dictionary
        # or add it with a count of 1 if it's not already there
        if letter.isalpha():
            letter_count[letter] = letter_count.get(letter, 0) + 1
    return letter_count

# Example usage with a predefined sentence
sentence = "Once upon a time in a faraway library"
# Call the function with the sentence variable and print the result
print(count_letter_frequency(sentence))
````

### Solution

There are four `TODO`s, and the task lets you pick the book titles. Together they build the list, create the dictionary, count, and print:

```python
books = ['The Hobbit', 'Dune', 'The Hobbit', 'Emma', 'Dune', 'The Hobbit']
book_count = {}

for book in books:
    if book in book_count:
        book_count[book] += 1
    else:
        book_count[book] = 1

print(book_count)
```

- `books` must contain duplicate titles, otherwise there is nothing to count. The titles here are one possible choice.
- `book_count = {}` starts empty, and each distinct title becomes a key whose value is its number of copies.
- If `book` is already a key, `book_count[book] += 1` increments it. Otherwise the `else` branch adds it with a count of `1`.
- Without the `if`/`else`, `book_count[book] += 1` on a new title raises a `KeyError`, because the key does not exist yet.
- The loop touches each item once with $O(1)$ dictionary operations, so the time is $O(n)$.

With the titles above, the script prints `{'The Hobbit': 3, 'Dune': 2, 'Emma': 1}`. Any other list with duplicates works the same way, and the `.get(book, 0) + 1` form from the earlier exercises is an equivalent shorter version of the loop.


````python
# Create and initialize a list 'books' with some duplicated book titles of your choice
books = ['The Hobbit', 'Dune', 'The Hobbit', 'Emma', 'Dune', 'The Hobbit']

# Create an empty dictionary 'book_count' to store the count of each book
book_count = {}

# Loop through each book in the 'books' list
for book in books:
    # Check if the book is already in dictionary, if so increment its count, otherwise add it with count 1
    if book in book_count:
        book_count[book] += 1
    else:
        book_count[book] = 1

# Finally, print the 'book_count' dictionary
print(book_count)
````

### Solving Real-World Task 1: Cataloging Books in a Library

Suppose you're asked to manage the cataloging of books in a library. In Python, we represent HashMaps using dictionaries. Here, the book ID serves as the key, while the details of the book, such as the title, author, and year of publication, are stored as values.

This approach allows us to add, search for, and remove books from our library catalog using just a few lines of Python code.

```python
library_catalog = {}  # Initializing HashMap

# Details of a book
book_id = "123"
book_details = {"title": "To Kill a Mockingbird", "author": "Harper Lee", "year_published": 1960}

library_catalog[book_id] = book_details  # Adding a book to library catalog

if book_id in library_catalog:  # Searching for a book
    print(library_catalog[book_id])

del library_catalog[book_id]  # Removing a book from library
```

As you can see, HashMaps make the task of cataloging books in the library simpler and more efficient!

### Solving Real-World Task 2: Counting Votes in an Election

Imagine a scenario in which we need to count votes in an election. We employ a HashMap, where each name is a unique key, and the frequency of that name serves as the associated value. Let's write some Python code to better understand this.

```python
votes_list = ["Alice", "Bob", "Alice", "Charlie", "Bob", "Alice"]  # Cast votes
vote_counts = {}  # Initializing a HashMap

# Counting the votes
for name in votes_list:
    if name in vote_counts:  # If name is already a key
        vote_counts[name] += 1  # increment its value (count)
    else:  # If name doesn't exist as a key
        vote_counts[name] = 1  # add it as a key and set value as 1

print(vote_counts)
# Prints {'Alice': 3, 'Bob': 2, 'Charlie': 1}
```

HashMaps facilitate the efficient counting of votes.

### Solving Real-World Task 3: Tracking Store Inventories

Finally, consider a task that involves managing a store's inventory. Here, we can use a HashMap in which product names are keys and quantities are values. This approach allows us to easily add new items, adjust the quantity of items, check whether an item is in stock, and much more.

```python
store_inventory = {"Apples": 100, "Bananas": 80, "Oranges": 90}  # Initializing an inventory

store_inventory["Pears"] = 50  # Adding a new product to inventory and setting its quantity

    print(prod, "are in stock.")

    print(prod, "are out of stock.")
```

Thus, when managing inventory data, HashMaps offer an efficient solution!

## Update Library System to Handle Borrowing Logic

In our library system, ensure that a book's copies are decremented only when copies are available. Modify the code to check if more than 0 copies are available before decreasing. This introduces a practical use of conditional expressions in real-world tasks.

---

### Starter Code

```python
# HashMap to hold the library system
library_system = {}

# Example: Adding a book
library_system['978-3-16-148410-0'] = {'title': 'The Little Prince', 'author': 'Antoine de Saint-Exupéry', 'copies': 3}

# Example: Borrowing a book (decrementing the number of copies)
library_system['978-3-16-148410-0']['copies'] -= 1  # Assuming the book is available
```

---

### Solution

The starter code has no `TODO`. The change is to wrap the borrowing line in a check, so the number of copies is decremented only when at least one copy is available:

```python
if library_system['978-3-16-148410-0']['copies'] > 0:
    library_system['978-3-16-148410-0']['copies'] -= 1
```

- `library_system['978-3-16-148410-0']['copies']` reads the number of copies from the nested dictionary, which is `3` here.
- The condition `> 0` is true only when a copy can be borrowed, so the decrement runs only then.
- Without the check, the line runs unconditionally. Once `copies` reaches `0`, another borrow would make it `-1`, which is not a valid number of copies.

After the change the book has `2` copies left. Running the borrow line again when `copies` is `0` leaves it at `0`.


````python
# HashMap to hold the library system
library_system = {}

# Example: Adding a book
library_system['978-3-16-148410-0'] = {'title': 'The Little Prince', 'author': 'Antoine de Saint-Exupéry', 'copies': 3}

# Example: Borrowing a book (decrementing the number of copies)
if library_system['978-3-16-148410-0']['copies'] > 0:  # only borrow when a copy is available
    library_system['978-3-16-148410-0']['copies'] -= 1
````

## Spot the Bug in Library Catalog Management

Great work on managing the library system! Now, let's test your skills in spotting a subtle bug. It appears there is an issue when attempting to display book information by its ID. Can you identify and resolve it to ensure the book details are correctly printed? Analyze the code and perform the necessary correction. Remember, the devil is in the details.

---

### Starter Code

```python
# Initializing a library catalog HashMap with book details
library_catalog = {
    '001': {'title': 'Wonder', 'author': 'R.J. Palacio', 'year_published': 2012},
    '002': {'title': 'Charlotte\'s Web', 'author': 'E.B. White', 'year_published': 1952},
}

# Function to display book information by ID
def show_book_info(book_id):
    if book_id in library_catalog():
        print(library_catalog[book_id])
    else:
        print("Book not found in the library catalog.")

# Display the info for book ID '001'
show_book_info('001')
```

---

### Solution

The bug is in the membership check. `library_catalog` is a dictionary, and the extra parentheses try to call it like a function:

```python
if book_id in library_catalog:  # no parentheses: check the keys of the dictionary
```

- `library_catalog()` raises `TypeError: 'dict' object is not callable`, so the function crashes before it prints anything.
- `book_id in library_catalog` checks whether `book_id` is a key of the dictionary, which is what the function needs.
- `'001'` is a key, so the function prints the details of the book. An unknown ID goes to the `else` branch and prints `Book not found in the library catalog.`

After the fix the script prints `{'title': 'Wonder', 'author': 'R.J. Palacio', 'year_published': 2012}`. The buggy version stopped with the `TypeError` instead.


````python
# Initializing a library catalog HashMap with book details
library_catalog = {
    '001': {'title': 'Wonder', 'author': 'R.J. Palacio', 'year_published': 2012},
    '002': {'title': 'Charlotte\'s Web', 'author': 'E.B. White', 'year_published': 1952},
}

# Function to display book information by ID
def show_book_info(book_id):
    if book_id in library_catalog:  # no parentheses: check the keys of the dictionary
        print(library_catalog[book_id])
    else:
        print("Book not found in the library catalog.")

# Display the info for book ID '001'
show_book_info('001')
````

## Update Library Inventory and Check Stock Status

Ready to put your skills to the test, Space Voyager? Let's enhance your learning adventure with HashMaps. Your mission is to manage a library's inventory by adding code to update the stock of an existing book and to check if a new book is already available. Utilize your understanding of HashMaps to complete this mission!

---

### Starter Code

```python
# Maintain a simple inventory of books in a library using a HashMap

library_inventory = {'BookA': 3, 'BookB': 5, 'BookC': 2}  # Existing books

library_inventory['BookD'] = 4  # Adding a new book with its initial count

# TODO: Update the stock for 'BookA' by adding 2 more to its current count

# TODO: Check if 'BookB' is in the inventory and print its stock count.
# If it's not available, print "[Book Name] is not in stock."
```

---

### Solution

There are two `TODO`s: add `2` more copies of `'BookA'`, and check whether `'BookB'` is in the inventory.

```python
library_inventory['BookA'] += 2

if 'BookB' in library_inventory:
    print("BookB stock:", library_inventory['BookB'])
else:
    print("BookB is not in stock.")
```

- `library_inventory['BookA'] += 2` reads the current count (`3`) and stores the new one, `5`.
- `'BookB' in library_inventory` checks the keys of the dictionary, so it is `True` when the book exists.
- Reading `library_inventory['BookB']` directly for a book that is missing would raise a `KeyError`. The `in` check avoids that and lets the `else` branch print the "is not in stock" message.

The script prints `BookB stock: 5`.


````python
# Maintain a simple inventory of books in a library using a HashMap

library_inventory = {'BookA': 3, 'BookB': 5, 'BookC': 2}  # Existing books

library_inventory['BookD'] = 4  # Adding a new book with its initial count

# Update the stock for 'BookA' by adding 2 more to its current count
library_inventory['BookA'] += 2  # 3 + 2 = 5

# Check if 'BookB' is in the inventory and print its stock count.
# If it's not available, print "[Book Name] is not in stock."
if 'BookB' in library_inventory:  # check the keys, so a missing book does not raise KeyError
    print("BookB stock:", library_inventory['BookB'])
else:
    print("BookB is not in stock.")
````

## Add Word Count Functionality to the HashMap

Great job on the last task! Now, let's see if you can add a missing piece. Use your knowledge from the lesson to determine how to count and display the number of appearances for each specified word in a provided string. Focus on analyzing the string and updating the HashMap accordingly.

---

### Starter Code

```python
# Initialize a HashMap to store the word counts
word_counts = {}

# List of words to count
words_to_count = ["apple", "banana", "cherry"]

# String in which to count words
string_to_search = "apple banana apple cherry apple banana"

# Split the string into individual words
words_in_string = string_to_search.split()

# TODO: Count the appearances of each word in the string
# and update word_counts accordingly

# Output the counts of each word
print(word_counts)
```

---

### Solution

There is one `TODO`: loop over the words in the string and count each one that appears in `words_to_count`, using the same pattern as the vote counter in the lesson.

```python
for word in words_in_string:
    if word in words_to_count:  # only count the specified words
        if word in word_counts:  # word already counted
            word_counts[word] += 1
        else:  # first time we see this word
            word_counts[word] = 1
```

- `words_in_string` is `['apple', 'banana', 'apple', 'cherry', 'apple', 'banana']`, so the loop visits each word once.
- `word in words_to_count` skips any word that is not on the list of words to count.
- `word in word_counts` decides between adding `1` to an existing count and creating the key with a count of `1`. Writing `word_counts[word] += 1` for a word that is not a key yet would raise a `KeyError`.

The script prints `{'apple': 3, 'banana': 2, 'cherry': 1}`.


````python
# Initialize a HashMap to store the word counts
word_counts = {}

# List of words to count
words_to_count = ["apple", "banana", "cherry"]

# String in which to count words
string_to_search = "apple banana apple cherry apple banana"

# Split the string into individual words
words_in_string = string_to_search.split()

# Count the appearances of each word in the string
# and update word_counts accordingly
for word in words_in_string:
    if word in words_to_count:  # only count the specified words
        if word in word_counts:  # word already counted: increment its count
            word_counts[word] += 1
        else:  # first time we see this word: add it with a count of 1
            word_counts[word] = 1

# Output the counts of each word
print(word_counts)
````

## Check Book Availability in Library Catalog

You've practiced with examples; now, it's time to test your skills in managing a library catalog using HashMaps. Let's create a Library Management System where you check if a specific book is available. Remember how HashMaps can greatly simplify such tasks? Use your knowledge to complete this challenge!

---

### Starter Code

```python
# Suppose you're tasked with checking the availability of a book by its ID in a library

# TODO: Define a 'library_catalog' dictionary with a few books. Each book should have id as key and details like title, author, and year_published as values

# TODO: Assign a variable 'book_id' with an id you want to check

# TODO: Write an if-else statement to check if the book_id exists in the library_catalog and print the book details or a message saying "Book not found in the library catalog."
```

---

### Solution

There are three `TODO`s: build the catalog, pick a `book_id`, and check whether that ID is in the catalog.

```python
library_catalog = {
    '001': {'title': 'Wonder', 'author': 'R.J. Palacio', 'year_published': 2012},
    '002': {'title': 'Charlotte\'s Web', 'author': 'E.B. White', 'year_published': 1952},
    '003': {'title': 'Matilda', 'author': 'Roald Dahl', 'year_published': 1988},
}

book_id = '002'

if book_id in library_catalog:
    print(library_catalog[book_id])
else:
    print("Book not found in the library catalog.")
```

- Each key is a book ID, and each value is a nested dictionary with `title`, `author` and `year_published`, as in the lesson.
- `book_id in library_catalog` checks the keys of the dictionary, so it is `True` only for an ID that exists.
- Reading `library_catalog[book_id]` without the check would raise a `KeyError` for an unknown ID. The `else` branch prints the "not found" message instead.

The script prints `{'title': "Charlotte's Web", 'author': 'E.B. White', 'year_published': 1952}`. With an ID such as `'999'` it prints `Book not found in the library catalog.`


````python
# Suppose you're tasked with checking the availability of a book by its ID in a library

# Define a 'library_catalog' dictionary with a few books: id as key, details as values
library_catalog = {
    '001': {'title': 'Wonder', 'author': 'R.J. Palacio', 'year_published': 2012},
    '002': {'title': 'Charlotte\'s Web', 'author': 'E.B. White', 'year_published': 1952},
    '003': {'title': 'Matilda', 'author': 'Roald Dahl', 'year_published': 1988},
}

# Assign a variable 'book_id' with an id you want to check
book_id = '002'

# Check if the book_id exists in the library_catalog and print the book details or a "not found" message
if book_id in library_catalog:  # check the keys, so an unknown ID does not raise KeyError
    print(library_catalog[book_id])
else:
    print("Book not found in the library catalog.")
````

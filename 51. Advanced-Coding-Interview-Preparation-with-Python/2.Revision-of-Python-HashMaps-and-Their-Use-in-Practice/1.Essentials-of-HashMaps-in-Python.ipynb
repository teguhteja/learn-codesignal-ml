## Essentials of HashMaps in Python

### Introduction to HashMaps

Hi, and welcome! Today, we'll explore HashMaps, a data structure that organizes data as key-value pairs, much like a treasure box with unique labels for each compartment.

Imagine dozens of toys in a box. If each toy had a unique label (the key), you could directly select a toy (the value) using the label. No rummaging required — that's the power of HashMaps! Today, we'll understand HashMaps and learn how to implement them in Python.

---

### Understanding HashMaps

HashMaps are special types of lists that utilize unique keys instead of indexes. When you know the key (toy's label), you can directly pick up the value (toy). That's how a HashMap works!

Consider an oversized library of books. With HashMaps (which act like the library catalog), you'll quickly locate any book using a unique number (key)!

---

### HashMaps in Python: Dictionaries

Python implements HashMaps through dictionaries. They hold data in key-value pairs, enclosed within `{}` (curly braces).

Let's create a dictionary, functioning as a catalog for a library:

```python
# Creating a catalog for the library using dictionaries
library_catalog = {'book1': 'A Tale of Two Cities',
                   'book2': 'To Kill a Mockingbird',
                   'book3': '1984'}
```

In this dictionary, `book1`, `book2`, and `book3` are keys, while the book titles serve as their respective values.

It's important to remember that while dictionary values can be mutable or immutable types, dictionary keys must be of an immutable type (such as strings, numbers, or tuples).

Additionally, dictionaries preserve item order as of Python 3.7 — which means that values are stored and retrieved in the same order they were added!

---

### HashMap Operations: Accessing, Updating, and Removing Elements

Dictionaries allow you to access, update, or remove elements:

- **Accessing Elements:** You can retrieve a book's title using its key in a straightforward way: `library_catalog['book1']` would return `'A Tale of Two Cities'`. But what happens if you try to access a key that isn't present in the dictionary? This would result in a `KeyError`.

  To prevent such errors, Python dictionaries provide the `get()` method. It fetches the value for a given key if it exists. If it doesn't, it simply returns `None`.

  ```python
  # Using get() to access a book's title
  book1 = library_catalog.get('book1')
  print(book1)  # Output: "A Tale of Two Cities"

  # Using get() to access a nonexistent key
  nonexistent_book = library_catalog.get('book100')
  print(nonexistent_book)  # Output: None
  ```

  In the example above, `get('book1')` retrieves the value for the key `book1`, but `get('book100')` does not raise an error even though `'book100'` does not exist in the dictionary. Instead, it returns `None`, allowing your program to continue running smoothly.

- **Adding or Updating Elements:** Whether you're adding a new book to the catalog or updating an existing book's title, you'll use the assignment operator (`=`). This syntax in Python's dictionaries allows for both updating existing key-value pairs and establishing new ones.

  If the specified key exists in the dictionary, the assigned value replaces the existing one. For updating a title: `library_catalog['book1'] = 'The Tell-Tale Heart'`.

  If the key doesn't exist in the dictionary yet, the operation creates a new key-value pair. For adding a new book: `library_catalog['book4'] = 'Pride and Prejudice'`.

- **Removing Elements:** If `'book1'` no longer exists, you can remove it using `del library_catalog['book1']`.

---

### Dictionary Methods: `items()`, `keys()`, `values()`, and others

Python's dictionaries offer several useful methods to interact with and manage your data:

- **Checking for a Key:** Ensure if a given book is present in your catalog using `'book1' in library_catalog`.

- **Accessing all Key-Value Pairs:** Use the `items()` method to retrieve all key-value pairs in the dictionary as tuples in a list-like object. This will come in handy when you need to examine all the data you have stored.

  ```python
  all_books = library_catalog.items()
  # all_books now holds: dict_items([('book1', 'A Tale of Two Cities'), ('book2', 'To Kill a Mockingbird'), ('book3', '1984')])
  ```

- **Accessing all Keys and Values:** The `keys()` and `values()` methods return list-like objects consisting of all keys and all values in the dictionary, respectively.

  ```python
  # Getting all keys
  all_keys = library_catalog.keys()
  # all_keys now holds: dict_keys(['book1', 'book2', 'book3'])

  # Getting all values
  all_values = library_catalog.values()
  # all_values now holds: dict_values(['A Tale of Two Cities', 'To Kill a Mockingbird', '1984'])
  ```

- **Iterating over the Dictionary:** Loop through your dictionary to access each key-value pair in turn:

  ```python
  # Looping over the dictionary
  for key, value in library_catalog.items():
      print(key, ":", value)
  ```

  When run, this code prints:

  ```text
  book1 : A Tale of Two Cities
  book2 : To Kill a Mockingbird
  book3 : 1984
  ```

  This looping method is useful if you need to process or analyze all entries in your dictionary in some way.

Keep in mind that dictionary methods return "list-like" objects, but these aren't actual lists. They provide a dynamic view on the dictionary's entries, which means that any changes to the dictionary will be reflected in these objects. If you need a real list, you can simply transform these list-like objects into an actual list by using `list()` like `list(library_catalog.keys())`.

---

### Diving Deeper: Understanding Time Complexity

HashMaps are popular because they save time! Operations like addition, update, and locating elements take constant time, $O(1)$, which means they require nearly the same amount of time regardless of the library size.

## Update the Library Catalog Entry

You've mastered updating titles in our library catalog! For your next mini-challenge, switch things up by changing the library catalog: update the title for `'book3'` to `'Little Women'`, instead of altering `'book1'`. Keep the librarian spirit alive and let's refine our catalog!

---

### Starter Code

```python
# Scenario: Library and book cataloging.

# Create a dictionary representing a small library catalog
library_catalog = {'book1': 'The Hobbit', 'book2': 'The Catcher in the Rye', 'book3': 'Pride and Prejudice'}

# Update the title for 'book1'
library_catalog['book1'] = 'The Lord of the Rings'

# Print the updated library catalog
for book, title in library_catalog.items():
    print(f'{book}: {title}')
```

---

### Solution

The starter code updates `'book1'`. The task asks for `'book3'` instead, so only the key and the new title change:

```python
library_catalog['book3'] = 'Little Women'
```

- `library_catalog['book3'] = ...` uses the assignment operator. Because `'book3'` already exists, the assigned value replaces `'Pride and Prejudice'` instead of creating a new pair.
- `'book1'` is no longer reassigned, so it keeps its original title, `'The Hobbit'`.
- Assignment on a dictionary is an $O(1)$ operation, and the item order stays the same.

When the script runs, the loop prints `book1: The Hobbit`, `book2: The Catcher in the Rye`, and `book3: Little Women`.

````python
# Scenario: Library and book cataloging.

# Create a dictionary representing a small library catalog
library_catalog = {'book1': 'The Hobbit', 'book2': 'The Catcher in the Rye', 'book3': 'Pride and Prejudice'}

# Update the title for 'book3'
library_catalog['book3'] = 'Little Women'

# Print the updated library catalog
for book, title in library_catalog.items():
    print(f'{book}: {title}')
````

## Library Catalog Error Correction

Great job on understanding dictionaries, Space Explorer! Now, let's further test your skills. Here's a piece of code that intends to print the title of a specific book from a library catalog, but it seems to miss the mark. Can you spot the issue and make the necessary fix to ensure it outputs the correct book title? Dive in and show what you've learned!

---

### Starter Code

```python
# Simple HashMap example to catalog books and print a specific book title
library_catalog = {
    'book1': 'The Little Prince',
    'book2': 'The Hobbit'
    'book3': 'The Catcher in the Rye',
}

# Access and print the title of book2 from the catalog
print(library_catalog['book2'])
```

---

### Solution

The dictionary literal is missing a comma after the `'book2'` entry, so Python cannot parse it. Adding the comma fixes it:

```python
'book2': 'The Hobbit',
```

- Entries in a dictionary literal must be separated by commas. Without the comma, `'The Hobbit'` is followed directly by `'book3'`, and Python raises a `SyntaxError` (`invalid syntax. Perhaps you forgot a comma?`) before any line runs.
- With the comma in place, `library_catalog` is built with three key-value pairs, and `library_catalog['book2']` looks up the value for `'book2'`.
- The trailing comma after the last entry is allowed in Python, so it can stay.

When the script runs, it prints `The Hobbit`.

````python
# Simple HashMap example to catalog books and print a specific book title
library_catalog = {
    'book1': 'The Little Prince',
    'book2': 'The Hobbit',  # added the missing comma
    'book3': 'The Catcher in the Rye',
}

# Access and print the title of book2 from the catalog
print(library_catalog['book2'])
````

## Check Book Availability in Library Catalog

Bravo for learning about dictionaries in Python! Now, let's test your knowledge. Your task is to check if a certain book exists in our library catalog. If it does, simply print `"Book is available"`. If not, notify us that it's not in the catalog. Remember to use dictionary operations to accomplish this.

---

### Starter Code

```python
# Library Catalog
library_catalog = {
    'Moby Dick': 'Herman Melville',
    'The Great Gatsby': 'F. Scott Fitzgerald'
}

# TODO: Check if '1984' is in the library_catalog. If yes, print "Book is available". If no, print it's not in the catalog.
```

---

### Solution

The `# TODO` asks for a membership check on the dictionary, which is what the `in` operator does:

```python
if '1984' in library_catalog:
    print("Book is available")
else:
    print("Book is not in the catalog")
```

- `'1984' in library_catalog` checks whether `'1984'` is one of the dictionary's **keys**. Here the keys are the book titles (`'Moby Dick'`, `'The Great Gatsby'`), and the values are the authors.
- The check returns a boolean, so it fits directly in an `if` / `else`. It is an $O(1)$ lookup and does not raise a `KeyError`, unlike `library_catalog['1984']` on a missing key.
- Without the `in` check, accessing `library_catalog['1984']` directly would raise a `KeyError` because that key does not exist.

`'1984'` is not a key in this catalog, so the script prints `Book is not in the catalog`.

````python
# Library Catalog
library_catalog = {
    'Moby Dick': 'Herman Melville',
    'The Great Gatsby': 'F. Scott Fitzgerald'
}

# Check if '1984' is in the library_catalog using the `in` operator
if '1984' in library_catalog:
    print("Book is available")
else:
    print("Book is not in the catalog")
````

## Update and Access Elements in a Library Catalog Dictionary

Great job learning about HashMaps! Now, let's practice updating and accessing elements within a dictionary. Your task is to update the catalog with a new title for the `'mystery'` genre, and then print it out. Remember, updating and accessing elements are crucial operations.

---

### Starter Code

```python
# Create a dictionary representing a small library catalog
library_catalog = {
    'mystery': 'Sherlock Holmes',
    'sci-fi': 'Dune',
    'fantasy': 'Harry Potter'
}

# TODO: Update the mystery book's title to 'Murder on the Orient Express' in the library_catalog

# TODO: Print the new title of the mystery book
```

---

### Solution

There are two `# TODO` lines: one update and one access. Both use the key `'mystery'`:

```python
library_catalog['mystery'] = 'Murder on the Orient Express'
print(library_catalog['mystery'])
```

- `library_catalog['mystery'] = ...` uses the assignment operator. Because the key `'mystery'` already exists, its value `'Sherlock Holmes'` is replaced instead of a new pair being created.
- `library_catalog['mystery']` reads the value back by key, so the `print` shows the new title. The update has to come first, otherwise the old title would be printed.
- Both operations are $O(1)$, and the other entries (`'sci-fi'`, `'fantasy'`) are left unchanged.

When the script runs, it prints `Murder on the Orient Express`.

````python
# Create a dictionary representing a small library catalog
library_catalog = {
    'mystery': 'Sherlock Holmes',
    'sci-fi': 'Dune',
    'fantasy': 'Harry Potter'
}

# Update the mystery book's title to 'Murder on the Orient Express' in the library_catalog
library_catalog['mystery'] = 'Murder on the Orient Express'

# Print the new title of the mystery book
print(library_catalog['mystery'])
````

## Updating a Library Catalog Using HashMaps in Python

You're on the final step of understanding HashMaps using Python dictionaries. Dive into creating a simple library catalog: update book titles and print the updated catalog. Start from scratch and craft each part carefully.

---

### Starter Code

```python
# TODO: Create a dictionary named library_catalog with 'book1' and 'book2' as keys and their titles as values

# TODO: Update the title for 'book1' in the dictionary

# TODO: Print the updated title for 'book1'
```

---

### Solution

There are three `# TODO` lines: create the dictionary, update one entry, and print it. The task doesn't give specific titles, so any titles work. This solution reuses the ones from the lesson:

```python
library_catalog = {'book1': 'A Tale of Two Cities', 'book2': 'To Kill a Mockingbird'}
library_catalog['book1'] = 'The Tell-Tale Heart'
print(library_catalog['book1'])
```

- A dictionary literal `{key: value, ...}` builds `library_catalog` with the two required keys, `'book1'` and `'book2'`, each mapped to a title.
- `library_catalog['book1'] = ...` uses the assignment operator. Because `'book1'` already exists, its value is replaced instead of a new pair being created.
- `library_catalog['book1']` reads the value back by key, so the `print` shows the updated title. The dictionary has to be created first, otherwise the update and the print would raise a `NameError`.
- Each operation is $O(1)$.

When the script runs, it prints `The Tell-Tale Heart`.

````python
# Create a dictionary named library_catalog with 'book1' and 'book2' as keys and their titles as values
library_catalog = {'book1': 'A Tale of Two Cities', 'book2': 'To Kill a Mockingbird'}

# Update the title for 'book1' in the dictionary
library_catalog['book1'] = 'The Tell-Tale Heart'

# Print the updated title for 'book1'
print(library_catalog['book1'])
````

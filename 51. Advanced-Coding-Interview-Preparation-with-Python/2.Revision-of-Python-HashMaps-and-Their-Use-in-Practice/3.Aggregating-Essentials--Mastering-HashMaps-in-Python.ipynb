### Data Aggregation Using HashMaps

Let's unveil how HashMaps assist us in data aggregation. Picture a Python dictionary wherein the keys signify different fruit types, and the values reflect their respective quantities. A HashMap could efficiently total all the quantities, providing insights into the Sum, Count, Max, Min, and Average operations.

---

### Practice: Summing Values in a HashMap

Let's delve into a hands-on example using a fruit basket represented as a dictionary:

```python
fruit_basket = {"apples": 5, "bananas": 4, "oranges": 8}
# A dictionary representing our fruit basket
```

We can tally the total quantity of fruits by summing the values in our dictionary with Python's `sum` function:

```python
total_fruits = sum(fruit_basket.values())
# Sums up the fruit quantities
print("The total number of fruits in the basket is:", total_fruits)
# It outputs: "The total number of fruits in the basket is: 17"
```

---

### Practice: Counting Elements in a HashMap

Just as easily, we can count the number of fruit types in our basket, which corresponds to the number of keys in our dictionary.

```python
count_fruits = len(fruit_basket)  # The count operation
print("The number of fruit types in the basket is:", count_fruits)
# It outputs: "The number of fruit types in the basket is: 3"
```

---

### Practice: Maximum and Minimum Values in a HashMap

Python's built-in functions, `max` and `min`, are very handy to find the highest and lowest values in a HashMap. Let's find out which fruit has the most and least quantity in our basket.

```python
# Using the `max` function
max_fruit = max(fruit_basket, key=fruit_basket.get)
# The expression for finding the maximum
print("The fruit with the most quantity is:", max_fruit)
# It outputs: "The fruit with the most quantity is: oranges"

# Using the `min` function
min_fruit = min(fruit_basket, key=fruit_basket.get)
# The expression for finding the minimum
print("The fruit with the least quantity is:", min_fruit)
# It outputs: "The fruit with the least quantity is: bananas"
```

---

### Practice: Averaging Values in a HashMap

Similar to finding the total quantity of fruits, we can calculate the average number of each type using the `len()` and `sum()` functions. Here, we divide the total quantity of fruits by the number of fruit types to determine the average.

```python
average_fruits = sum(fruit_basket.values()) / len(fruit_basket)
# The expression for finding the average
print("The average number of each type of fruit in the basket is:", average_fruits)
# It outputs: "The average number of each type of fruit in the basket is: 5.67"
```

## Calculate the Average Quantity in Inventory

In your grocery shop inventory, you've successfully calculated the maximum quantity of items. Let's now apply a new skill: change the code to calculate the average quantity of items in the inventory. Utilize the `sum` and `len` functions to find your answer. Ready to give it a try?

---

### Starter Code

```python
# A dictionary simulating a grocery shop inventory where key is the item and value is its quantity
inventory = {"potatoes": 30, "carrots": 25, "onions": 15}

# Finding the maximum quantity among the inventory
max_quantity = max(inventory.values())
print("The maximum quantity of any item in the shop is:", max_quantity)
```

---

### Solution

The starter code finds the maximum, so the change is to compute the average instead. There is no `TODO`; the two lines that use `max` are replaced:

```python
average_quantity = sum(inventory.values()) / len(inventory)
print("The average quantity of items in the shop is:", average_quantity)
```

- `inventory.values()` gives the quantities `30`, `25` and `15`, and `sum(...)` adds them to `70`.
- `len(inventory)` is the number of items (keys), which is `3`.
- Dividing the total by the number of items gives the average, `70 / 3`.
- Using `max(inventory.values())` would still return the largest quantity (`30`), not the average.

The script prints `The average quantity of items in the shop is: 23.333333333333332`.


````python
# A dictionary simulating a grocery shop inventory where key is the item and value is its quantity
inventory = {"potatoes": 30, "carrots": 25, "onions": 15}

# Finding the average quantity among the inventory
average_quantity = sum(inventory.values()) / len(inventory)  # total quantity divided by number of items
print("The average quantity of items in the shop is:", average_quantity)
````

## Inventory Maximum Quantity Calculation Correction

In this exercise, you have a small program intended to calculate and display the minimum quantity of a single type of fruit in a grocery shop's inventory. However, there's a bug preventing it from working correctly. Your task is to find and fix the bug without adding or removing any lines of code.

---

### Starter Code

```python
# A dictionary simulating the inventory of a grocery shop
inventory = {"apples": 30, "bananas": 45, "cherries": 25}

# Calculate the minimum quantity of a single fruit type in the inventory
min_quantity = min(inventory)

print("The minimum quantity of a single type of fruit in the inventory is:", min_quantity)
```

---

### Solution

The bug is what `min()` is given. Passing the dictionary itself makes `min` compare its keys, not its quantities:

```python
min_quantity = min(inventory.values())  # compare the quantities, not the keys
```

- `min(inventory)` iterates over the keys, so it compares the strings `"apples"`, `"bananas"` and `"cherries"` alphabetically and returns `"apples"`.
- `inventory.values()` gives the quantities `30`, `45` and `25`, so `min(...)` returns the smallest quantity, `25`.
- The fix changes one expression on the same line, so no lines are added or removed, as the task requires.

After the fix the script prints `The minimum quantity of a single type of fruit in the inventory is: 25`. The buggy version printed `apples` instead.


````python
# A dictionary simulating the inventory of a grocery shop
inventory = {"apples": 30, "bananas": 45, "cherries": 25}

# Calculate the minimum quantity of a single fruit type in the inventory
min_quantity = min(inventory.values())  # compare the quantities, not the keys

print("The minimum quantity of a single type of fruit in the inventory is:", min_quantity)
````

## Inventory Stock Level Comparison to Average

Great job on your progress so far! Now, let's take it a step further. You've mastered managing a grocery shop inventory. Your next challenge is to determine how each item's stock level compares to the average stock level in the store. Write some code to calculate the difference between each item's count and the average, and print out whether each item has more, less, or exactly the average amount of stock.

---

### Starter Code

```python
# Creating a simple inventory containing a count of items in a grocery shop
inventory = {"milk": 24, "bread": 36, "eggs": 48}

# TODO: Calculate the average count of the inventory items

# Finding how much each item's count differs from the average
for item, count in inventory.items():
    difference_from_average = count - average_count
    # TODO: Check if the count is greater, less, or equal to the average and set the appropriate message
    print(f"{item.capitalize()} has {status_message}")
```

---

### Solution

There are two `TODO`s: the average count, and the message that says how each item compares to it.

```python
average_count = sum(inventory.values()) / len(inventory)

if difference_from_average > 0:
    status_message = f"more than the average stock by {difference_from_average}"
elif difference_from_average < 0:
    status_message = f"less than the average stock by {abs(difference_from_average)}"
else:
    status_message = "exactly the average stock"
```

- `sum(inventory.values())` adds the counts `24`, `36` and `48` to `108`, and `len(inventory)` is `3`, so `average_count` is `36.0`.
- `average_count` must be defined before the loop. Without it, `count - average_count` raises a `NameError`.
- `difference_from_average` is positive when the item is above the average, negative when below and `0` when equal, so an `if` / `elif` / `else` chain covers all three cases.
- `abs(...)` keeps the "less than" message free of a minus sign.
- `status_message` must be set in every branch. Without it, the `print` raises a `NameError`.

The script prints `Milk has less than the average stock by 12.0`, `Bread has exactly the average stock` and `Eggs has more than the average stock by 12.0`.


````python
# Creating a simple inventory containing a count of items in a grocery shop
inventory = {"milk": 24, "bread": 36, "eggs": 48}

# Calculate the average count of the inventory items
average_count = sum(inventory.values()) / len(inventory)  # total count divided by number of items

# Finding how much each item's count differs from the average
for item, count in inventory.items():
    difference_from_average = count - average_count
    # Check if the count is greater, less, or equal to the average and set the appropriate message
    if difference_from_average > 0:
        status_message = f"more than the average stock by {difference_from_average}"
    elif difference_from_average < 0:
        status_message = f"less than the average stock by {abs(difference_from_average)}"
    else:
        status_message = "exactly the average stock"
    print(f"{item.capitalize()} has {status_message}")
````

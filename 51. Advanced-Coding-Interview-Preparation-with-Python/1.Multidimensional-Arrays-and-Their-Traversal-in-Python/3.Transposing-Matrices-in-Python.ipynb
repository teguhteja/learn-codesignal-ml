### Task Statement

To begin, let's elaborate on the task at hand. You are required to write a Python function named `transformMatrix()`. This function will accept a 2D list (which represents a matrix) that contains integers as inputs. Your responsibility is to return another 2D list, which is the transposed version of the given matrix.

Remember, when we mention 'transposing a matrix', we are referring to the process of switching its rows and columns. In other words, all the rows of the original matrix should convert into columns in the transposed matrix, and vice versa.

For instance, if the original matrix (input 2D list) is:

```python
[
    [1, 2, 3],
    [4, 5, 6]
]
```

Then the transposed matrix (output 2D list) will be:

```python
[
    [1, 4],
    [2, 5],
    [3, 6]
]
```

Let me remind you that it is vital for your result to maintain the integrity of the data type that is present in the original matrix. In layman's terms, the values seen in the input matrix are integers, and they should be integers in the output matrix as well.

---

### Solution Building: Step 1

The initial step of building our solution involves determining the dimensions of the matrix. We need to know the number of rows and columns present in it. The Python `len()` function can offer this information. The number of rows is simply the length of the matrix (the outer list), and the number of columns matches the length of any of the inner lists.

```python
def transformMatrix(matrix):
    rows = len(matrix)
    cols = len(matrix[0]) if rows > 0 else 0
```

---

### Solution Building: Step 2

The subsequent step is to create a "placeholder" for the transposed matrix that is in alignment with its required dimensions. This will be a new 2D list, but with the number of rows and columns swapped. Initially, we can populate this matrix with all zeros.

```python
def transformMatrix(matrix):
    rows = len(matrix)
    cols = len(matrix[0]) if rows > 0 else 0
    result = [[0 for _ in range(rows)] for _ in range(cols)]
```

---

### Solution Building: Step 3

It's time to get to the crux of the matter - transposing the matrix. For every element in the original matrix, we want to move it from the `i`th row and `j`th column to the `j`th row and `i`th column of the transposed matrix. A straightforward nested `for` loop can effortlessly execute this swap for all elements of the matrix.

```python
def transformMatrix(matrix):
    rows = len(matrix)
    cols = len(matrix[0]) if rows > 0 else 0
    result = [[0 for _ in range(rows)] for _ in range(cols)]

    for i in range(rows):
        for j in range(cols):
            result[j][i] = matrix[i][j]

    return result

# Initial matrix (2D list)
matrix = [
    [1, 2, 3],
    [4, 5, 6]
]

# Call our function on the matrix and output the result
transposed = transformMatrix(matrix)
print(transposed)  # Outputs: [[1, 4], [2, 5], [3, 6]]
```

This step concludes our solution!

---

### Alternative Solution

Another solution could be written by using only the short array creation syntax:

```python
def transformMatrix(matrix):
    rows = len(matrix)
    cols = len(matrix[0]) if rows > 0 else 0
    result = [[matrix[j][i] for j in range(rows)] for i in range(cols)]

    return result
```

Here we fill the matrix directly at the creation, and the `j`-th column of the `i`-th row of the new matrix is assigned the value `matrix[j][i]`.

## Restaurant Seating Arrangement Transposition

Wonderful progress, Space Voyager! Here's a practical task involving the restaurant seating arrangements you've been learning about. Your task is to fix a minor error in the code that's supposed to transpose the seating arrangement of a restaurant. However, the output isn't as expected. Review the code carefully and adjust it to ensure it accurately transposes the matrix. Good luck!

---

### Starter Code

```python
# A function to transpose a matrix representing restaurant seating arrangement
def transpose_seating(seating):
    rows, cols = len(seating), len(seating[0])
    return [[seating[i][j] for j in range(rows)] for i in range(cols)]

# Sample restaurant seating arrangement represented as a 2D matrix
restaurant_seating = [
    [10, 11, 12],
    [20, 21, 22]
]

# Trying to transpose the seating arrangement
transposed_seating = transpose_seating(restaurant_seating)
print(transposed_seating)  # Output isn't as expected. Can you identify the fix?
```

---

### Solution

The bug is the index order inside the list comprehension. The outer loop `i` runs over `cols` (the new rows) and the inner loop `j` runs over `rows` (the new columns), so the element to read is the one in row `j` and column `i`. The fix is to swap the indices:

```python
return [[seating[j][i] for j in range(rows)] for i in range(cols)]
```

- The transposed matrix has `cols` rows and `rows` columns, so `i` ranges over `cols` and `j` ranges over `rows`.
- Row `i` of the result collects column `i` of the original: `seating[0][i]`, `seating[1][i]`, and so on.
- With `seating[i][j]`, `i` goes up to `cols - 1` (2 here) but `seating` only has `rows` (2) rows, so `seating[2]` raises an `IndexError` for the 2x3 sample.
- This matches the alternative solution from the lesson, which uses `matrix[j][i]`.

For the 2x3 `restaurant_seating`, the script should print `[[10, 20], [11, 21], [12, 22]]`. I did not run it; this follows from tracing the code by hand.

````python
# A function to transpose a matrix representing restaurant seating arrangement
def transpose_seating(seating):
    rows, cols = len(seating), len(seating[0])
    # fixed: read row j, column i (was seating[i][j])
    return [[seating[j][i] for j in range(rows)] for i in range(cols)]

# Sample restaurant seating arrangement represented as a 2D matrix
restaurant_seating = [
    [10, 11, 12],
    [20, 21, 22]
]

# Trying to transpose the seating arrangement
transposed_seating = transpose_seating(restaurant_seating)
print(transposed_seating)  # [[10, 20], [11, 21], [12, 22]]
````

## Transpose the Seating Chart in Reverse Order

You've just learned how to transpose a matrix. Now, let's add a twist! Adjust the transpose function to transpose the seating chart in reverse order, starting from the last column and moving to the first. This slight modification will provide a fresh perspective on transposing matrices. Go ahead and give it a try!

---

### Starter Code

```python
# Given a constant matrix that represents tables in a restaurant
seating_chart = [
    [101, 102, 103, 104],
    [201, 202, 203, 204],
    [301, 302, 303, 304]
]

# Function to transpose the seating chart (Matrix) for new arrangement
def transpose(chart):
    return [[chart[j][i] for j in range(len(chart))] for i in range(len(chart[0]))]

# Example usage:
transposed_seating = transpose(seating_chart)
print(transposed_seating)  # Output will be the transposed matrix
# The expected output - [[104, 204, 304], [103, 203, 303], [102, 202, 302], [101, 201, 301]]
```

---

### Solution

This starter code has no `TODO`; the task is to change the transpose so the new rows come out from the last column to the first. Only the outer loop of the list comprehension needs to change: instead of `i` counting up from `0`, it counts down from the last column index to `0`:

```python
return [[chart[j][i] for j in range(len(chart))] for i in range(len(chart[0]) - 1, -1, -1)]
```

- `i` indexes the original columns, and each `i` produces one row of the result, so the order of `i` is the order of the output rows.
- `range(len(chart[0]) - 1, -1, -1)` starts at the last column (`3` for the 4-column chart) and stops after `0`, because the stop value `-1` is excluded.
- The inner part `[chart[j][i] for j in range(len(chart))]` is unchanged, so each output row still reads its column from top to bottom.
- Without the change, the function returns the normal transpose `[[101, 201, 301], [102, 202, 302], ...]`, which does not match the expected output.

For the 3x4 `seating_chart`, the script should print `[[104, 204, 304], [103, 203, 303], [102, 202, 302], [101, 201, 301]]`, which matches the expected output in the starter code. I did not run it; this follows from tracing the code by hand.

````python
# Given a constant matrix that represents tables in a restaurant
seating_chart = [
    [101, 102, 103, 104],
    [201, 202, 203, 204],
    [301, 302, 303, 304]
]

# Function to transpose the seating chart (Matrix) for new arrangement
def transpose(chart):
    # reversed: the outer index i goes from the last column down to the first
    return [[chart[j][i] for j in range(len(chart))] for i in range(len(chart[0]) - 1, -1, -1)]

# Example usage:
transposed_seating = transpose(seating_chart)
print(transposed_seating)  # Output will be the transposed matrix
# The expected output - [[104, 204, 304], [103, 203, 303], [102, 202, 302], [101, 201, 301]]
````

### Solution

There are two `TODO`s: the one-liner body of `transposeRestaurantSeating` (marked by `pass`), and the call that transposes `original_seating` and prints each row. The one-liner uses `zip` with argument unpacking:

```python
def transposeRestaurantSeating(seatingArrangement):
    return [list(row) for row in zip(*seatingArrangement)]
```

```python
transposed_seating = transposeRestaurantSeating(original_seating)
for row in transposed_seating:
    print(row)
```

- `*seatingArrangement` unpacks the rows, so `zip` receives `[1, 2, 3]`, `[4, 5, 6]` and `[7, 8, 9]` as separate arguments.
- `zip` pairs up the i-th item of every row, giving `(1, 4, 7)`, `(2, 5, 8)`, `(3, 6, 9)`. Each tuple is a column of the original, which is a row of the transpose.
- `list(row)` turns every tuple back into a list, so the result is a 2D list like the input.
- `pass` returns `None`, so without the one-liner the loop would fail with `TypeError: 'NoneType' object is not iterable`.
- An empty `seatingArrangement` gives `zip()` with no arguments, which yields nothing, so the result is `[]`.

For the 3x3 `original_seating`, the script prints `[1, 4, 7]`, `[2, 5, 8]` and `[3, 6, 9]`, one row per line. I ran it and got exactly that output.

````python
def transposeRestaurantSeating(seatingArrangement):
    # one-liner transposition: zip(*rows) pairs up the i-th item of every row
    return [list(row) for row in zip(*seatingArrangement)]

# Example Seating Arrangement
original_seating = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# Call the function to transpose the seating and print each row of the transposed seating.
transposed_seating = transposeRestaurantSeating(original_seating)
for row in transposed_seating:
    print(row)
````

### Solution

There are two `TODO`s: the loop body that fills `new_matrix`, and the call that stores the result in `transformed_matrix`. The loop body needs an inner loop over `j`, and the element at `(i, j)` moves to `(size - 1 - j, size - 1 - i)`:

```python
for i in range(size):
    for j in range(size):
        new_matrix[size - 1 - j][size - 1 - i] = matrix[i][j]
```

```python

```

- The mirror of position `(i, j)` across the secondary diagonal is `(size - 1 - j, size - 1 - i)`. Compare this with the normal transpose, which maps `(i, j)` to `(j, i)`.
- Cells on the secondary diagonal satisfy `i + j == size - 1`, so `size - 1 - j == i` and `size - 1 - i == j`. They map to themselves and stay in place.
- This matches the swaps in the task. In a 4x4 matrix, `matrix[0][0]` goes to `new_matrix[3][3]`, and `matrix[0][1]` goes to `new_matrix[2][3]`.
- Writing into `new_matrix` instead of `matrix` avoids overwriting elements before they are read.
- Without the inner loop, only one cell per row is handled. Without the call, `print(transformed_matrix)` raises a `NameError`.

For the 3x3 `square_matrix`, the script prints `[[9, 6, 3], [8, 5, 2], [7, 4, 1]]`, which matches the expected output. I ran it and got exactly that. A 4x4 check also gave `[[16, 12, 8, 4], [15, 11, 7, 3], [14, 10, 6, 2], [13, 9, 5, 1]]`.

````python
def reflectOverSecondaryDiagonal(matrix):
    size = len(matrix)
    new_matrix = [[0 for _ in range(size)] for _ in range(size)]
    for i in range(size):
        # element (i, j) goes to (size - 1 - j, size - 1 - i)
        for j in range(size):
            new_matrix[size - 1 - j][size - 1 - i] = matrix[i][j]
    return new_matrix

# Example square matrix to reflect over the secondary diagonal
square_matrix = [
 [1, 2, 3],
 [4, 5, 6],
 [7, 8, 9]
]

# Call the function and print the transformed matrix
transformed_matrix = reflectOverSecondaryDiagonal(square_matrix)
print(transformed_matrix)
# Output should be:
# [
#  [9, 6, 3],
#  [8, 5, 2],
#  [7, 4, 1]
# ]
````

## Transpose the Restaurant Seating Arrangement

Time to apply what you've learned about transposing matrices! Imagine you are assisting in redesigning the seating arrangement of a restaurant to better accommodate guests. Your task is to write Python code from scratch, which takes an initial seating arrangement (a 2D list) and transposes it. Remember, transposing switches the rows and columns, creating a new perspective on seating.

---

### Starter Code

```python
# TODO: Define a function 'transpose_seating' that takes a 2D list 'arrangement' as its parameter.
    # TODO: Create a new 2D list 'transposed' with dimensions swapped (columns become rows and vice versa) filled initially with zeros.

    # TODO: Use nested loops to transpose each element from the original seating arrangement to the new arrangement 'transposed'.

# Restaurant seating before transposition (rows)
seating_before = [
 [1, 2],
 [3, 4],
 [5, 6]
]

# TODO: Print the 'transposed' arrangement to see the new seating arrangement.
```

---

### Solution

There is no starter code to fix here; all four `TODO`s are written from scratch. The function reads the dimensions, builds a zero-filled matrix with rows and columns swapped, and copies each element from `(i, j)` to `(j, i)` with nested loops. The result is then printed:

```python
def transpose_seating(arrangement):
    rows = len(arrangement)
    cols = len(arrangement[0]) if rows > 0 else 0
    transposed = [[0 for _ in range(rows)] for _ in range(cols)]

    for i in range(rows):
        for j in range(cols):
            transposed[j][i] = arrangement[i][j]

    return transposed
```

```python
transposed = transpose_seating(seating_before)
print(transposed)
```

- `rows` is the number of seating rows and `cols` is the number of seats per row. The `if rows > 0 else 0` guard stops `arrangement[0]` from failing on an empty list.
- `transposed` has `cols` rows and `rows` columns, so a 3x2 input gets a 2x3 zero-filled list.
- `transposed[j][i] = arrangement[i][j]` moves the element in row `i`, column `j` to row `j`, column `i`.
- The function returns `transposed`, because a variable created inside the function is not visible to the final `print`. The caller stores the returned list in its own `transposed`.
- Without the nested loops, `transposed` would stay all zeros.

For the 3x2 `seating_before`, the script prints `[[1, 3, 5], [2, 4, 6]]`. I ran it and got exactly that.

````python
def transpose_seating(arrangement):
    rows = len(arrangement)
    cols = len(arrangement[0]) if rows > 0 else 0
    # New 2D list with dimensions swapped (columns become rows), filled with zeros
    transposed = [[0 for _ in range(rows)] for _ in range(cols)]

    # Move each element from (i, j) to (j, i)
    for i in range(rows):
        for j in range(cols):
            transposed[j][i] = arrangement[i][j]

    return transposed

# Restaurant seating before transposition (rows)
seating_before = [
 [1, 2],
 [3, 4],
 [5, 6]
]

# Print the 'transposed' arrangement to see the new seating arrangement.
transposed = transpose_seating(seating_before)
print(transposed)
````

### Solution

This starter code has no `TODO`; the task is to change the four-direction check into a horizontal-only check. Remove the two vertical checks (`i - 1` and `i + 1`) and keep the two horizontal ones (`j - 1` and `j + 1`):

```python
if board[i][j] == 'E' and (
    is_empty_neighbor(board, i, j - 1) or
    is_empty_neighbor(board, i, j + 1)

    result.append((i, j))
```

- `is_empty_neighbor(board, i - 1, j)` and `is_empty_neighbor(board, i + 1, j)` look at the cells above and below. Removing them stops vertical moves from counting.
- `is_empty_neighbor(board, i, j - 1)` and `is_empty_neighbor(board, i, j + 1)` look at the cells to the left and right, so a cell is added to `result` only when one of them is empty.
- `is_empty_neighbor` already returns `False` outside the board, so the cells in the first and last columns need no extra boundary checks.
- Without the change, cells such as `(1, 2)`, `(2, 1)` and `(3, 1)` are still reported, because they have an empty neighbor above or below.

For the 4x4 `board`, only `(0, 1)` and `(0, 2)` have an empty cell directly to their left or right, so the script prints `[(0, 1), (0, 2)]`. I ran it and got exactly that.

````python
# Function to check if a neighboring position is empty ('E')
def is_empty_neighbor(board, x, y):
    rows, cols = len(board), len(board[0])

    # Check that (x, y) is within board boundaries
    if 0 <= x < rows and 0 <= y < cols:
        return board[x][y] == 'E'
    return False

def main():
    board = [
        ['P', 'E', 'E', 'P'],
        ['E', 'P', 'E', 'P'],
        ['P', 'E', 'P', 'P'],
        ['P', 'E', 'P', 'E']
    ]
    result = []
    rows, cols = len(board), len(board[0])

    for i in range(rows):
        for j in range(cols):
            # Check for empty spot with an empty neighbor on the left or right only
            if board[i][j] == 'E' and (
                is_empty_neighbor(board, i, j - 1) or
                is_empty_neighbor(board, i, j + 1)
            ):
                result.append((i, j))
    print(result)

if __name__ == "__main__":
    main()
````

### Solution

The starter code already checks the cell above. Add the three other directions (down, left, right) in the same style, each guarded by a boundary check, and then call the function for `(0, 1)`:

```python
def is_position_strategic(board, row, col):
```

There are two `TODO`s. In `evaluate_move`, return `True` only when the cell itself is `'E'` **and** at least one of its up, down, left or right neighbors is `'E'`. Then use a list comprehension to collect every `(i, j)` for which `evaluate_move` is `True`:

- `and` combines the two requirements (empty cell, empty neighbor), and `or` accepts the cell as soon as one neighbor is empty.

````python
def evaluate_move(board, row, col):
    # the cell must be empty and have at least one empty neighbor (up, down, left, right)
    return board[row][col] == 'E' and (
        (row > 0 and board[row - 1][col] == 'E') or
        (row < len(board) - 1 and board[row + 1][col] == 'E') or
        (col > 0 and board[row][col - 1] == 'E') or
        (col < len(board[0]) - 1 and board[row][col + 1] == 'E')
    )

# Assuming a constant 2D array representing a board
board = [
    ['P', 'E', 'E', 'P'],
    ['E', 'P', 'E', 'P'],
    ['P', 'E', 'P', 'P'],
    ['P', 'E', 'P', 'E']
]

# Find all valid move positions with a list comprehension
valid_moves = [(i, j) for i in range(len(board)) for j in range(len(board[0])) if evaluate_move(board, i, j)]
print(valid_moves)
````

```python
def countSubmatricesWithE(board):
    count = 0

    for i in range(len(board) - 2):
        for j in range(len(board[0]) - 2):
```

- A 3x3 submatrix whose top-left corner is `(i, j)` covers rows `i` to `i + 2` and columns `j` to `j + 2`. So `i` can go up to `len(board) - 3` and `j` up to `len(board[0]) - 3`, which is why the loops use `range(len(board) - 2)` and `range(len(board[0]) - 2)`.
- The four corners are `(i, j)`, `(i, j + 2)`, `(i + 2, j)` and `(i + 2, j + 2)`. The middle row and column of the submatrix are never read, so only the corners matter.
- `and` requires all four corners to be `'E'`; one `'P'` is enough to skip that submatrix.
- Submatrices may overlap, so a single `'E'` can be a corner of several of them, and each one is counted separately.
- Boards smaller than 3x3 give `range` of a zero or negative number, so the loops do not run and the function returns `0`.
- Without the `- 2` limits, `board[i + 2][j + 2]` would raise an `IndexError` near the bottom and right edges.

````python
def countSubmatricesWithE(board):
    # count of 3x3 submatrices that have 'E' in all four corners
    count = 0

    # (i, j) is the top-left corner of a 3x3 submatrix, so it must leave 2 rows and 2 columns
    for i in range(len(board) - 2):
        for j in range(len(board[0]) - 2):
            # the corners are (i, j), (i, j + 2), (i + 2, j) and (i + 2, j + 2)
            if (board[i][j] == 'E' and board[i][j + 2] == 'E' and
                    board[i + 2][j] == 'E' and board[i + 2][j + 2] == 'E'):
                count += 1

    return count

# A matrix 'board' with 'E' for empty and 'P' for piece positions
board = [
    ['E', 'P', 'E', 'P', 'E'],
    ['P', 'P', 'P', 'P', 'P'],
    ['E', 'P', 'E', 'P', 'E'],
    ['P', 'P', 'P', 'P', 'P'],
    ['E', 'P', 'E', 'P', 'E']
]

# Count the submatrices and print the result
print(countSubmatricesWithE(board))
````

## Creating a Basic FastAPI Application

Let's start by creating a simple FastAPI application. This will serve as the foundation for our API.

First, import the `FastAPI` class from the `fastapi` module and create an instance of it:

```python
from fastapi import FastAPI

app = FastAPI()

```

Here, `app` is an instance of the `FastAPI` class, which represents our web application.

Next, let's define a basic route that will render the homepage:

```python
from fastapi.responses import HTMLResponse

@app.get("/", response_class=HTMLResponse)
async def index():
    return "<h1>Welcome to the Short Story Generator!</h1>"

```

In this code snippet, `@app.get("/")` is a decorator that tells FastAPI to execute the `index` function when the root URL (`/`) is accessed. The function returns a simple welcome message as HTML.

## Implementing API Endpoints

Now, let's implement the API endpoints that will handle story generation and retrieval.

### POST Endpoint: Generating Stories

We'll start by creating a POST endpoint to generate stories. This endpoint will accept user input and return a generated story:

```python
from fastapi import Request

@app.post("/api/generate_story")
async def generate_story(request: Request):
    data = await request.json()
    user_input = data.get("user_input")
    return story_generator_controller.generate_story(user_input)

```

* `@app.post("/api/generate_story")`: Defines a route for the `/api/generate_story` URL, specifying that it accepts POST requests.
* `await request.json()`: Extracts the JSON payload from the request body asynchronously.
* `data.get("user_input")`: Retrieves the `user_input` field from the JSON payload.
* `story_generator_controller.generate_story(user_input)`: Generates a story based on the user input.

### GET Endpoint: Retrieving Stories

Next, we'll create a GET endpoint to retrieve all generated stories:

```python
@app.get("/api/get_stories")
async def get_stories():
    return story_generator_controller.get_stories()

```

* `@app.get("/api/get_stories")`: Defines a route for the `/api/get_stories` URL, specifying that it accepts GET requests.
* `story_generator_controller.get_stories()`: Retrieves all generated stories from memory.

That assertion error usually happens because of how FastAPI's `CORSMiddleware` handles origin matching when `allow_credentials` is set.

If `allow_origins=["*"]` is combined with `allow_credentials=True`, Starlette/FastAPI will **not** return `Access-Control-Allow-Origin: *` for security reasons (or might drop the wildcard header depending on whether an `Origin` header was present in the incoming test request).

To fix this so `response.headers.get("access-control-allow-origin") == "*"` passes the check, set **`allow_credentials=False`** (or omit `allow_credentials` completely).

---

### Fixed `main.py` Code

Here is the exact code snippet to replace in your main application file:

```python
from fastapi import FastAPI, Request
from fastapi.responses import HTMLResponse
from fastapi.templating import Jinja2Templates
from fastapi.middleware.cors import CORSMiddleware
from controllers.story_generator_controller import StoryGeneratorController
import uvicorn

app = FastAPI()

# Add CORS middleware with allow_credentials set to False (or omitted)
# so that the wildcard "*" origin header is explicitly returned.
app.add_middleware(
    CORSMiddleware,
    allow_origins=["*"],
    allow_credentials=False,
    allow_methods=["*"],
    allow_headers=["*"],
)

templates = Jinja2Templates(directory="app/templates")

story_generator_controller = StoryGeneratorController()

@app.get("/", response_class=HTMLResponse)
async def index(request: Request):
    return templates.TemplateResponse("story_generator.html", {"request": request})

@app.post("/api/generate_story")
async def generate_story(request: Request):
    json_data = await request.json()
    user_input = json_data.get('user_input')
    return story_generator_controller.generate_story(user_input)

@app.get("/api/get_stories")
async def get_stories():
    return story_generator_controller.get_stories()

@app.get("/api/health")
async def health_check():
    return {"status": "ok"}

if __name__ == '__main__':
    uvicorn.run("main:app", host="0.0.0.0", port=3000, reload=True)

```

### Why this fixes it:

When `allow_credentials` is set to `True`, the CORS specification strictly forbids using the wildcard `*` in the `Access-Control-Allow-Origin` header. By switching `allow_credentials` to `False`, `CORSMiddleware` safely responds with `Access-Control-Allow-Origin: *`, allowing your automated test assertion to pass.

Here is the complete implementation with the custom 404 error handler registered using `@app.exception_handler` and `StarletteHTTPException`:

```python
from fastapi import FastAPI, Request, status
from fastapi.exceptions import HTTPException
from starlette.exceptions import HTTPException as StarletteHTTPException
from fastapi.responses import HTMLResponse, JSONResponse
from fastapi.templating import Jinja2Templates
from controllers.story_generator_controller import StoryGeneratorController
import uvicorn
import json

app = FastAPI()
templates = Jinja2Templates(directory="app/templates")

story_generator_controller = StoryGeneratorController()

@app.get("/", response_class=HTMLResponse)
async def index(request: Request):
    return templates.TemplateResponse("story_generator.html", {"request": request})

@app.post("/api/generate_story")
async def generate_story(request: Request):
    try:
        json_data = await request.json()
        if 'user_input' not in json_data:
            return JSONResponse(
                status_code=400,
                content={"error": "Missing required field: user_input"}
            )
        
        user_input = json_data.get('user_input')
        return story_generator_controller.generate_story(user_input)
    except json.JSONDecodeError:
        return JSONResponse(
            status_code=400,
            content={"error": "Invalid JSON"}
        )

@app.get("/api/get_stories")
async def get_stories():
    return story_generator_controller.get_stories()

@app.get("/api/health")
async def health_check():
    return {"status": "ok"}

# Global exception handler for 404 (Not Found) errors
@app.exception_handler(StarletteHTTPException)
async def custom_http_exception_handler(request: Request, exc: StarletteHTTPException):
    if exc.status_code == status.HTTP_404_NOT_FOUND:
        return JSONResponse(
            status_code=status.HTTP_404_NOT_FOUND,
            content={"error": "Resource or endpoint not found"}
        )
    return JSONResponse(
        status_code=exc.status_code,
        content={"error": exc.detail}
    )

if __name__ == '__main__':
    uvicorn.run("main:app", host="0.0.0.0", port=3000, reload=True)

```

### Key Highlights:

* **`StarletteHTTPException` Catching:** Standard route mismatches (404s) in FastAPI are raised internally as `StarletteHTTPException`. Decorating with `StarletteHTTPException` captures these default 404 routes properly.
* **Consistent Response Structure:** Any non-existent route will now respond with `{ "error": "Resource or endpoint not found" }` and an HTTP `404` status code, keeping your entire API's response format unified.

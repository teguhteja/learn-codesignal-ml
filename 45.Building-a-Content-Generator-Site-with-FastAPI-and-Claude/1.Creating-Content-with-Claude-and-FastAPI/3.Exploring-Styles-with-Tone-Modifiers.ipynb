Understanding Style and Tone Modifiers

In content generation, style and tone modifiers are used to adjust the way content is presented. Style refers to the format or genre of the content, such as a news article or a children's story. Tone, on the other hand, refers to the emotional quality or attitude of the content, such as formal or playful.

Different styles and tones can significantly impact how the content is perceived by the audience. For example, a motivational speech with an inspiring tone can uplift and encourage, while a product review with a critical tone can provide a balanced and analytical perspective.

Here is the complete code with execution timing included for each API call:

```python
import os
from anthropic import Anthropic
import time

# Initialize the Anthropic client
client = Anthropic(api_key=os.environ.get("ANTHROPIC_API_KEY"))

# Define prompts combining styles with tone modifiers
style_tone_prompts = [
    "A motivational speech, inspiring tone",
    "A comprehensive product review of the latest high-end smartphone, including an in-depth analysis of its design, build quality, display characteristics (such as resolution, brightness, and color accuracy), processor performance in real-world and benchmark scenarios, battery life under various usage patterns, camera capabilities with sample scenarios (low light, portrait, landscape), software features and user interface experience, connectivity options, comparison with previous models and main competitors, pricing and value for money, and a critical tone highlighting both strengths and weaknesses with specific examples and recommendations for potential buyers",
]

# Time the content generation for each prompt and print the time taken
for prompt in style_tone_prompts:
    start_time = time.time()
    
    response = client.messages.create(
        model='claude-sonnet-4-5-20250929',
        max_tokens=800,
        messages=[
            {
                "role": "user",
                "content": prompt,
            }
        ]
    )
    
    end_time = time.time()
    elapsed_time = end_time - start_time

    # Process and display the generated content and execution timing
    generated_content = response.content[0].text
    print(f"Prompt: {prompt[:80]}...")  # Truncated display for legibility
    print(generated_content)
    print(f"\n⏱️ Time taken: {elapsed_time:.2f} seconds")
    print("-" * 50)

```

---

### 💡 Which prompt takes longer?

The **second (longer) prompt** will almost certainly take longer to process and complete. Here is why:

1. **Output Token Volume (Primary Factor):** LLMs generate content sequentially token-by-token. A detailed prompt asking for an in-depth, multi-category smartphone review will push Claude to output closer to the `800` token limit, whereas a simple motivational speech prompt might finish in far fewer tokens.
2. **Prompt Processing (TTFT):** Larger input prompts take slightly more time for the server to encode during the prefill phase before generating the first token (Time To First Token).

Here is the complete, refactored implementation that separates data generation from display presentation by storing results in a structured list first:

```python
import os
from anthropic import Anthropic

# Initialize the Anthropic client
client = Anthropic(api_key=os.environ.get("ANTHROPIC_API_KEY"))

# Define prompts combining styles with tone modifiers
style_tone_prompts = [
    "A motivational speech, inspiring tone",
    "A news article, formal tone",
]

# List to store generated content before presentation
results = []

# Generate content for each combined prompt
for prompt in style_tone_prompts:
    response = client.messages.create(
        model='claude-sonnet-4-5-20250929',
        max_tokens=800,
        messages=[
            {
                "role": "user",
                "content": prompt,
            }
        ]
    )
    
    generated_content = response.content[0].text
    
    # Store both the prompt and generated content as a dictionary
    results.append({
        "prompt": prompt,
        "content": generated_content
    })

# Display loop to present all stored content at once
print("=" * 60)
print("STORED GENERATION RESULTS")
print("=" * 60 + "\n")

for idx, item in enumerate(results, start=1):
    print(f"[{idx}] PROMPT: {item['prompt']}\n")
    print("GENERATED CONTENT:")
    print(item['content'])
    print("-" * 60 + "\n")

```

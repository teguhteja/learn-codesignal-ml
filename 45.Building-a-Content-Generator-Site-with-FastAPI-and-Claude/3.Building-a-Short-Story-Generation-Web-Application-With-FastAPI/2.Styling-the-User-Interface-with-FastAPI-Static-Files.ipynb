## Styling the User Interface with FastAPI Static Files

# Lesson 2: Styling the Web Application Interface

Welcome to the second lesson of our course on building a story generation web application with FastAPI!

In the previous lesson, you set up the foundational HTML structure for your application, organized the project layout, and created the main interface with tabs, forms, and containers. While the structure is in place, the application currently lacks visual appeal and usability.

In this lesson, you'll learn how to add styling to your application using FastAPI's approach to serving static files. Styling allows you to control the appearance of your HTML elements — such as colors, fonts, spacing, and layout — making your interface more attractive and user-friendly.

---

## Defining Basic Page Style

Let's start by setting up the basic page style for your application. This will ensure a consistent look and feel across your interface.

### Set the Font-Family and Background Color

Begin by defining the font family and background color for the entire page. In your project, you will write these styles in the `style.css` file, which is served as a static file.

```css
body {
    font-family: 'Segoe UI', sans-serif;
    background-color: #f4f4f4;
}

```

* The `font-family` property specifies the typeface.
* `background-color` sets the page's background color.

### Remove Default Margins and Add Padding

Next, remove the default margins and add padding to the body to ensure content is not flush against the edges.

```css
body {
    margin: 0;
    padding: 20px;
}

```

* `margin: 0;` removes any default spacing around the body.
* `padding: 20px;` adds space inside the body boundaries.

### Use Flex Layout for Centering Content

To center content both vertically and horizontally, use a flex layout. This will make the page more visually appealing and easier to navigate.

```css
body {
    display: flex;
    flex-direction: column;
    align-items: center;
    min-height: 80vh;
}

```

* `display: flex;` enables Flexbox positioning.
* `flex-direction: column;` arranges content vertically.
* `align-items: center;` centers content horizontally.
* `min-height: 80vh;` ensures the body takes up most of the viewport height.

---

## Creating a Responsive Container

Now, let's create a central white block to hold your content using the `.container` class.

### Define the Container's Appearance

Set the background color, padding, and border radius to create a visually distinct container card.

```css
.container {
    display: flex;
    flex-direction: column;
    align-items: center;
    width: 100%;
    max-width: 400px;
    background: white;
    padding: 30px;
    border-radius: 12px;
    box-shadow: 0 0 10px rgba(0, 0, 0, 0.1);
}

```

* `background: white;` sets the container's background color.
* `padding: 30px;` adds space inside the container.
* `border-radius: 12px;` softens the corners.
* `box-shadow` adds a subtle elevation effect.
* `width: 100%;` and `max-width: 400px;` ensure the container adapts to different screen sizes while maintaining readability.

---

## Implementing Tab Navigation System

Let's style the tab navigation system using the `.tabs` and `.tab-button` selectors.

### Set Up the Tabs

Use flexbox to align the tabs and add spacing below them.

```css
.tabs {
    display: flex;
    justify-content: center;
    gap: 10px;
    margin-bottom: 20px;
}

```

`display: flex;` and `justify-content: center;` center the tabs, while `gap: 10px;` and `margin-bottom: 20px;` add space between and below the tabs.

### Style the Tab Buttons

Define the appearance of the tab buttons, including padding, margin, background color, and text color.

```css
.tab-button {
    padding: 10px 20px;
    margin: 0 10px;
    background-color: #0056b3;
    color: white;
    border: none;
    cursor: pointer;
    border-radius: 5px;
    transition: background-color 0.3s;
}

```

`padding` and `margin` control spacing, `background-color` and `color` define the button's theme, and `cursor: pointer;` changes the cursor on hover. `border-radius` and `transition` add smooth visual feedback.

### Add Hover Effect

Apply a transition hover state to enhance interactivity.

```css
.tab-button:hover {
    background-color: #007bff;
}

```

The `:hover` pseudo-class changes the background color on hover, providing clear visual feedback.

## Styling Form Elements

Next, style the form elements, including inputs, selects, and action buttons.

### Style Inputs and Selects

Ensure inputs and selects use full width with appropriate padding and border boundaries.

```css
input, select {
    width: 100%;
    max-width: 100%;
    box-sizing: border-box;
    padding: 10px;
    margin-top: 10px;
    margin-bottom: 20px;
    border: 1px solid #ccc;
    border-radius: 5px;
}

```

`width: 100%;` ensures full responsiveness, `padding` adds breathing room inside inputs, and `border` + `border-radius` define clean input fields.

### Style the Primary Button

Define the main submission button's default and interactive states.

```css
button {
    width: 100%;
    max-width: 100%;
    box-sizing: border-box;
    padding: 10px 20px;
    margin-top: 10px;
    background-color: #007bff;
    color: white;
    border: none;
    cursor: pointer;
    border-radius: 5px;
    transition: background-color 0.3s;
}

button:hover {
    background-color: #0056b3;
}

```

## Enhancing Media Display Areas

Finally, add styles for media display containers, including `#story-container` and `#history-container`.

### Add Margins to Containers

Add a top margin to separate output sections from inputs above them.

```css
#story-container, #history-container {
    margin-top: 20px;
}

```

### Style Preformatted Story Text

If you display generated stories or history entries in preformatted text blocks (`<pre>`), style them for readability:

```css
pre {
    text-align: left;
    background-color: #f9f9f9;
    padding: 10px;
    border-radius: 5px;
    box-shadow: 0 0 5px rgba(0, 0, 0, 0.1);
    white-space: pre-wrap;
    word-break: break-word;
    overflow-x: auto;
    max-width: 100%;
    box-sizing: border-box;
}

```

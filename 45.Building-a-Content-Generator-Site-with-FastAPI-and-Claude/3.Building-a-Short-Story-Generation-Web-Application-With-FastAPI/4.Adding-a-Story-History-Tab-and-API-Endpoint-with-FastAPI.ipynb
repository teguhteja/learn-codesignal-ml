## Updating the HTML Structure for the History Tab

To begin, let's update `index.html` to include the necessary UI elements for the history tab so users can load and view previously generated stories.

First, add a button to trigger fetching the story history:

```html
<button id="history-button" onclick="fetchHistory()">Load Previous Stories</button>

```

Next, include a paragraph element for the loading message:

```html
<p id="history-loading-message" style="display: none;">Loading history... Please wait.</p>

```

> **Why Loading Indicators Matter:** When making API calls, providing clear feedback prevents users from wondering if their action was registered or if the application froze. Proper error handling similarly prevents confusion when something goes wrong.

Finally, add an empty container where the fetched stories will appear:

```html
<div id="history-container"></div>

```

By setting explicit element IDs (`history-button`, `history-loading-message`, `history-container`), you ensure these elements can easily be accessed and manipulated with JavaScript.

---

## Initializing the `fetchHistory()` Function

Now, let's create and initialize the `fetchHistory()` function in `app/static/script.js`.

Start by retrieving references to the DOM elements:

```javascript
const historyLoadingMessage = document.getElementById('history-loading-message');
const historyButton = document.getElementById('history-button');
const historyContainer = document.getElementById('history-container');

```

Next, update the UI state and clear any old data before fetching new entries:

```javascript
historyLoadingMessage.style.display = 'block';
historyButton.style.display = 'none';
historyContainer.innerHTML = '';

```

---

## Implementing the API Call and UI Feedback

With the UI prepared, send a `GET` request to the `/api/get_stories` endpoint:

```javascript
fetch('/api/get_stories')
    .then(response => response.json())
    .then(data => {
        historyLoadingMessage.style.display = 'none';
        historyButton.style.display = 'block';
        // Data processing logic will go here
    })

```

Toggling the UI controls back to their default state after fetching ensures a smooth user experience.

---

## Handling API Response Data

Now, handle the returned data array by checking if stories exist:

```javascript
if (data.stories.length === 0) {
    historyContainer.innerHTML = '<p>No stories found.</p>';
    return;
}

data.stories.forEach(story => {
    const pre = document.createElement('pre');
    pre.textContent = story.story;
    historyContainer.appendChild(pre);
});

```

This dynamically creates `<pre>` blocks for every historical story returned from the server.

---

## Enhancing Error Handling

To ensure robustness, attach a `.catch()` block to catch any issues during the fetch request:

```javascript
    .catch(error => {
        historyLoadingMessage.style.display = 'none';
        historyButton.style.display = 'block';
        alert('Error fetching story history. Please try again.');
        console.error('Error:', error);
    });

```
